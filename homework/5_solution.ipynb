### **作業一：有效的數獨**

```  
輸出: false
(說明: 左上角3×3子框內有兩個8)
```  

#### **要求：**  
1. 每行必須包含1-9且不重複  
2. 每列必須包含1-9且不重複  
3. 每個3×3子框必須包含1-9且不重複  
4. 只需驗證已填數字的格子（忽略標記為'.'的格子）  

#### **訓練技能：**  
- 二維陣列遍歷  
- 哈希集合檢測重複  
- 網格子區域劃分（3×3子框計算）  


### **Solution Code**  
```python
def isValidSudoku(board):
    rows = [set() for _ in range(9)]
    cols = [set() for _ in range(9)]
    boxes = [set() for _ in range(9)]

    for i in range(9):
        for j in range(9):
            num = board[i][j]
            if num != '.':
                box_index = (i // 3) * 3 + (j // 3)
                if num in rows[i] or num in cols[j] or num in boxes[box_index]:
                    return False
                rows[i].add(num)
                cols[j].add(num)
                boxes[box_index].add(num)
    return True

# Valid board
board1 = [
    ["5","3",".",".","7",".",".",".","."],
    ["6",".",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    [".",".",".",".","8",".",".","7","9"]
]

# Invalid board (duplicate 8 in top-left sub-box)
board2 = [
    ["8","3",".",".","7",".",".",".","."],
    ["6",".",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    [".",".",".",".","8",".",".","7","9"]
]

print(isValidSudoku(board1))  # Expected: True
print(isValidSudoku(board2))  # Expected: False
```

### **Homework 2: Container With Most Water**  
**Task:** Find two vertical lines that, together with the x-axis, form a container that contains the most water.  
![](5_2.png)

#### **Input/Output Example:**  
```  
Input: height = [1,8,6,2,5,4,8,3,7]  
Output: 49  
Explanation: The maximum area of water that can be contained is 49.

Input: height = [1,1]  
Output: 1  
```

#### **Requirements:**  
1. You must find two lines that form the container with the maximum area.
2. The area is calculated as the width between the two lines multiplied by the height of the shorter line.
3. You cannot slant the container; it must be vertical.

#### **Skills Tested:**  
- Two-pointer technique  
- Area calculation based on height and width  
- Efficient searching with O(n) complexity  

---

### **作業二：容器能裝最多的水**  
**任務：** 找出兩條垂直線，與x軸一起形成一個容器，該容器能裝最多的水。  

#### **輸入/輸出範例：**  
```  
輸入: height = [1,8,6,2,5,4,8,3,7]  
輸出: 49  
說明: 能裝最多水的面積為49。

輸入: height = [1,1]  
輸出: 1  
```

#### **要求：**  
1. 需找到形成最大面積的兩條線。
2. 面積計算為兩條線之間的寬度乘以較短線的高度。
3. 容器必須是垂直的，不能傾斜。

#### **訓練技能：**  
- 雙指針技巧  
- 基於高度和寬度的面積計算  
- O(n)複雜度的高效搜索  



### Code Solution

```python
def maxArea(height):
    left, right = 0, len(height) - 1
    max_water = 0
    
    while left < right:
        # Calculate the area with the current left and right boundaries
        width = right - left
        current_height = min(height[left], height[right])
        current_area = width * current_height
        
        # Update the maximum area found
        max_water = max(max_water, current_area)
        
        # Move the pointer pointing to the shorter line
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
            
    return max_water
```

The time complexity of the solution for the "Container With Most Water" problem is **O(n)**, where **n** is the length of the input array `height`. 

### Explanation:

1. **Single Pass**: The algorithm uses a two-pointer approach, initializing one pointer at the start (`left`) and the other at the end (`right`) of the array. The pointers move towards each other, which means each element is considered only once.

2. **Pointer Movement**: In each iteration of the while loop:
   - The area is calculated based on the heights at the `left` and `right` pointers.
   - The pointer pointing to the shorter line is moved inward, which is a crucial step since moving the taller line does not potentially increase the area (the height is limited by the shorter line).

3. **Termination**: The loop continues until the `left` pointer meets or surpasses the `right` pointer. This ensures that every height is evaluated once, leading to linear time complexity.

### Summary:
- **Best Case**: O(n)
- **Worst Case**: O(n)
- **Average Case**: O(n)

Thus, the overall time complexity is **O(n)**, making the solution efficient for large input sizes.

### 時間覆雜度分析

對於“容器能裝最多的水”問題的解決方案，時間覆雜度為 **O(n)**，其中 **n** 是輸入數組 `height` 的長度。

### 解釋：

1. **單次遍歷**：該算法使用雙指針方法，初始化一個指針在數組的開始位置（`left`），另一個指針在結束位置（`right`）。這兩個指針向中間移動，這意味著每個元素只被考慮一次。

2. **指針移動**：在每次叠代的 while 循環中：
   - 根據 `left` 和 `right` 指針處的高度計算面積。
   - 移動指向較短線的指針，這是關鍵步驟，因為移動較高的線不會潛在地增加面積（高度受限於較短的線）。

3. **終止條件**：循環繼續，直到 `left` 指針與 `right` 指針相遇或交叉。這確保每個高度都被評估一次，從而導致線性時間覆雜度。

### 總結：
- **最佳情況**：O(n)
- **最壞情況**：O(n)
- **平均情況**：O(n)

因此，整體時間覆雜度為 **O(n)**，使得該解決方案對於大輸入規模非常高效。

Here’s a solution for the "Container With Most Water" problem with a time complexity of **O(n²)**. This solution uses a nested loop to evaluate all possible pairs of lines.

### Code Solution

```python
def maxArea(height):
    max_water = 0
    n = len(height)
    
    for i in range(n):
        for j in range(i + 1, n):
            width = j - i
            current_height = min(height[i], height[j])
            current_area = width * current_height
            max_water = max(max_water, current_area)
    
    return max_water
```

### Explanation

1. **Nested Loops**: The outer loop iterates through each line (indexed by `i`), and the inner loop iterates through all subsequent lines (indexed by `j`).
2. **Area Calculation**: For each pair of lines, the area is calculated using the width (distance between `i` and `j`) and the height (minimum of the two lines).
3. **Max Area Tracking**: The maximum area found is updated in each iteration.

### Time Complexity
- **Time Complexity**: O(n²) due to the nested loops, where each pair of lines is evaluated.
- **Space Complexity**: O(1), as no additional space is used that scales with input size. 

This approach is straightforward but less efficient for large inputs compared to the O(n) solution.

這裡是“容器能裝最多的水”問題的 **O(n²)** 時間複雜度解決方案。這個解決方案使用嵌套迴圈來評估所有可能的線對。

### 代碼解決方案

```python
def maxArea(height):
    max_water = 0
    n = len(height)
    
    for i in range(n):
        for j in range(i + 1, n):
            width = j - i
            current_height = min(height[i], height[j])
            current_area = width * current_height
            max_water = max(max_water, current_area)
    
    return max_water
```

### 解釋

1. **嵌套迴圈**：外層迴圈遍歷每一條線（由 `i` 指定），內層迴圈遍歷所有隨後的線（由 `j` 指定）。
2. **面積計算**：對於每對線，根據寬度（`i` 和 `j` 之間的距離）和高度（兩條線中較短的那一條）計算面積。
3. **最大面積跟踪**：在每次迭代中更新找到的最大面積。

### 時間複雜度
- **時間複雜度**：O(n²)，因為使用了嵌套迴圈，評估了每對線。
- **空間複雜度**：O(1)，因為沒有使用隨輸入大小增長的額外空間。

這種方法直觀但對於大輸入來說效率較低，相較於 O(n) 解決方案。
