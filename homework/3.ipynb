### **Homework 1: Sum of Perfect Squares (Brute Force Solution)**  
**Task:** Write a program that calculates the sum of all perfect squares between two integers `a` and `b` (inclusive) using a **brute force approach**.  

#### **Input/Output Example:**  
```  
Input: 3 25  
Output: 54  
(Explanation: 4 + 9 + 16 + 25 = 54)  
```  

#### **Requirements:**  
1. Input two integers `a` and `b` (1 ≤ `a` ≤ `b` ≤ 1,000,000).  
2. Check **every number** in `[a, b]` to see if it's a perfect square.  
3. Sum all perfect squares found.  

#### **Skills Tested:**  
- Basic looping and conditionals  
- Mathematical checks for perfect squares  
- Simple but less efficient problem-solving

### **Homework 3: Password Decoding with Digit Sum Rule**  
**Task:** Write a program that decodes a 6-digit password from 7 uppercase letters, where each digit is derived from the distance between adjacent letters. If the raw distance is ≥10, sum its digits until it becomes a single digit.  

#### **Input/Output Example:**  
```  
Input: "POKEMON"  
Output: "146821"  
Explanation:  
P→O=1, O→K=4, K→E=6, E→M=8, M→O=2, O→N=1  

Input: "HACKERR"  
Output: "728640"  
Explanation:  
H→A=7, A→C=2, C→K=8, K→E=6, E→R=13→1+3=4, R→R=0  
```  

#### **Requirements:**  
1. Input a string of **7 uppercase letters** (e.g., `"POKEMON"`).  
2. Calculate absolute ASCII distance between adjacent letters.  
3. If distance ≥10, reduce it by summing digits until single-digit (e.g., 13→4).  
4. Output the 6-digit password.  

#### **Skills Tested:**  
- ASCII manipulation (`ord()`)  
- Digit sum calculation  
- String iteration and concatenation

### **作業三：字母距離密碼解碼（數字和規則）**  
**任務：** 寫一個程式，將7個大寫字母轉換為6位數密碼，其中每位數是相鄰字母的距離。若距離≥10，則將數字各位相加直到成為個位數。  

#### **輸入/輸出範例：**  
```  
輸入: "POKEMON"  
輸出: "146821"  
說明:  
P→O=1, O→K=4, K→E=6, E→M=8, M→O=2, O→N=1  

輸入: "HACKERR"  
輸出: "728640"  
說明:  
H→A=7, A→C=2, C→K=8, K→E=6, E→R=13→1+3=4, R→R=0  
```  

#### **要求：**  
1. 輸入7個大寫字母。  
2. 計算相鄰字母的ASCII絕對距離。  
3. 若距離≥10，則將數字各位相加至個位數（例如: 13 -> 4, 19 -> 10 -> 1）。  
4. 輸出6位數密碼。  

#### **訓練技能：**  
- ASCII碼轉換（`ord()`）  
- 數字位數處理  
- 字串操作
