### **Homework 1: Two Sum (Brute Force Solution)**  
**Task:** Given an array of integers `nums` and an integer `target`, return the indices of the two numbers that add up to `target`.  

#### **Input/Output Example:**  
```  
Input: nums = [2,7,11,15], target = 9  
Output: [0,1]  
(Explanation: nums[0] + nums[1] = 2 + 7 = 9)  
```  

#### **Requirements:**  
1. Each input has **exactly one solution**.  
2. You **cannot use the same element twice**.  
3. Return the answer in any order.  

#### **Skills Tested:**  
- Array traversal and indexing  
- Nested loops for pairwise checks  
- Handling edge cases (e.g., duplicate values)

### **作業一：兩數之和（暴力法）**  
**任務：** 給定一個整數陣列 `nums` 和目標值 `target`，返回兩數之和等於 `target` 的索引。  

#### **輸入/輸出範例：**  
```  
輸入: nums = [3,2,4], target = 6  
輸出: [1,2]  
(說明: nums[1] + nums[2] = 2 + 4 = 6)  
```  

#### **要求：**  
1. 每組輸入保證**有且只有一個解**。  
2. **不可重複使用同一元素**。  
3. 答案順序不限。  

#### **訓練技能：**  
- 陣列遍歷與索引操作  
- 雙重迴圈檢查所有組合  
- 邊界條件處理（如重複值）  

---  


### **Solution Code (Brute Force Approach)**  
```python  
def two_sum(nums, target):  
    for i in range(len(nums)):  
        for j in range(i + 1, len(nums)):  
            if nums[i] + nums[j] == target:  
                return [i, j]  
    return []  # As problem states one solution exists, this line is a fallback  

# Example Usage:  
nums = [3, 3]  
target = 6  
print(two_sum(nums, target))  # Output: [0, 1]  
```  

#### **Explanation:**  
1. **Nested Loops:** The outer loop picks each element `nums[i]`, and the inner loop checks all subsequent elements `nums[j]` for a pair summing to `target`.  
2. **Early Termination:** Returns the first valid pair found (problem guarantees exactly one solution).  
3. **Time Complexity:** O(n²), where n is the length of `nums`.

### **作業二：最長共同前綴**  
**任務：** 撰寫一個函式，找出字串陣列中的最長共同前綴。  

#### **輸入/輸出範例：**  
```  
輸入: strs = ["flower","flow","flight"]  
輸出: "fl"  
(說明: 最長的共同前綴為 "fl") 

輸入: strs = ["dog","racecar","car"]  
輸出: ""  
(說明: 無共同前綴，故回傳空字串)  
```  

#### **要求：**  
1. 若無共同前綴，回傳 `""`。  
2. 輸入陣列 `strs` 僅包含小寫英文字母。  
3. 需有效率處理邊界情況（如空陣列或單一字串）。  

#### **訓練技能：**  
- 字串操作與比對  
- 垂直掃描的迭代實作  
- 邊界條件與提前終止  


### **Solution Code **  
```python  
def longest_common_prefix(strs):  
    if not strs:  
        return ""  
    for i in range(len(strs[0])):  
        char = strs[0][i]  
        for s in strs[1:]:  
            if i >= len(s) or s[i] != char:  
                return strs[0][:i]  
    return strs[0]  

# Example Usage:  
strs = ["flower","flow","flight"]  
print(longest_common_prefix(strs))  # Output: "fl"  
```  

#### **Explanation:**  
1. **Edge Handling:** Returns `""` if the input array is empty.  
2. **Vertical Scanning:**  
   - Iterates through each character position `i` of the first string.  
   - Checks if all other strings have the same character at position `i`.  
   - Stops and returns the prefix up to `i` on the first mismatch.  
3. **Time Complexity:** O(S), where S is the total number of characters in all strings (optimal for this problem).  
 

````python
def longestCommonPrefix(strs):
    """
    :type strs: List[str]
    :rtype: str
    """
    result=[]
    for i in zip(*strs):
        # print(i)
        x = i[0]
        if all([y==x for y in i]):
            result.append(x)
        else:
            break
    return "".join(result) 
strs = ["flower","flow","flight"]
print(f'Passed? {longestCommonPrefix(strs)=='fl'}')
strs = ["dog","racecar","car"]
print(f'Passed? {longestCommonPrefix(strs)==''}')
````
