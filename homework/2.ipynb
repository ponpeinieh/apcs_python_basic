### **Homework 1: Greatest Common Divisor (GCD)**  
**Task:** Write a program that calculates the GCD of two integers using the **Euclidean algorithm** and prints the result.  

#### **Input/Output Example:**  
```  
Input: 24 36  
Output: 12  
```  

#### **Requirements:**  
1. Input two integers `a` and `b` (1 ≤ `a`, `b` ≤ 1,000,000).  
2. Output their GCD.  
3. Use the **Euclidean algorithm** (while loop).  

#### **Skills Tested:**  
- Understanding of GCD and the Euclidean algorithm  
- Looping
- Handling edge cases (e.g., one input is zero)

### **Homework 4: Armstrong Numbers**  
**Task:** Write a program that finds all Armstrong numbers(see definition below) within a given range.  

#### **Input/Output Example:**  
```  
Input start: 100  
Input end: 10000  
Output: 153, 370, 371, 407, 1634, 8208, 9474  
```  

#### **Requirements:**  
1. Input two integers `start` and `end` (1 ≤ `start` ≤ `end` ≤ 1,000,000).  
2. Output all Armstrong numbers in the range `[start, end]`, separated by commas.  
3. An **Armstrong number** is an *n*-digit number where the sum of each digit raised to the power *n* equals the number itself.  

#### **Skills Tested:**  
- Looping through a range  
- Mathematical operations (exponents and sums)  
- Digit extraction and manipulation

### **作業四：阿姆斯壯數**  
**任務：** 撰寫一個程式，找出指定範圍內的所有阿姆斯壯數(定義見後面説明)。  

#### **輸入/輸出範例：**  
```  
輸入起始值: 100  
輸入結束值: 10000  
輸出: 153, 370, 371, 407, 1634, 8208, 9474  
```  

#### **要求：**  
1. 輸入兩個整數 `start` 和 `end`（1 ≤ `start` ≤ `end` ≤ 1,000,000）。  
2. 輸出範圍 `[start, end]` 內的所有阿姆斯壯數，以逗號分隔。  
3. **阿姆斯壯數**是一個 *n* 位數，其每個位數的 *n* 次方和等於該數本身。  

#### **訓練技能：**  
- 範圍迴圈處理  
- 數學運算（冪次與求和）  
- 數字位數操作
