### **Homework 1: Two Sum (Brute Force Solution)**  
**Task:** Given an array of integers `nums` and an integer `target`, return the indices of the two numbers that add up to `target`.  

#### **Input/Output Example:**  
```  
Input: nums = [2,7,11,15], target = 9  
Output: [0,1]  
(Explanation: nums[0] + nums[1] = 2 + 7 = 9)  
```  

#### **Requirements:**  
1. Each input has **exactly one solution**.  
2. You **cannot use the same element twice**.  
3. Return the answer in any order.  

#### **Skills Tested:**  
- Array traversal and indexing  
- Nested loops for pairwise checks  
- Handling edge cases (e.g., duplicate values)

### **作業一：兩數之和（暴力法）**  
**任務：** 給定一個整數陣列 `nums` 和目標值 `target`，返回兩數之和等於 `target` 的索引。  

#### **輸入/輸出範例：**  
```  
輸入: nums = [3,2,4], target = 6  
輸出: [1,2]  
(說明: nums[1] + nums[2] = 2 + 4 = 6)  
```  

#### **要求：**  
1. 每組輸入保證**有且只有一個解**。  
2. **不可重複使用同一元素**。  
3. 答案順序不限。  

#### **訓練技能：**  
- 陣列遍歷與索引操作  
- 雙重迴圈檢查所有組合  
- 邊界條件處理（如重複值）

### **Homework 2: Longest Common Prefix**  
**Task:** Write a function to find the longest common prefix string among an array of strings.  

#### **Input/Output Example:**  
```  
Input: strs = ["flower","flow","flight"]  
Output: "fl"  
(Explanation: The longest common prefix is "fl")  
```  

#### **Requirements:**  
1. If no common prefix exists, return an empty string `""`.  
2. The input array `strs` contains only lowercase English letters.  
3. The solution should efficiently handle edge cases (e.g., empty array or single string).  

#### **Skills Tested:**  
- String manipulation and comparison  
- Iterative vertical scanning  
- Handling edge cases and early termination

### **作業二：最長共同前綴**  
**任務：** 撰寫一個函式，找出字串陣列中的最長共同前綴。  

#### **輸入/輸出範例：**  
```  
輸入: strs = ["flower","flow","flight"]  
輸出: "fl"  
(說明: 最長的共同前綴為 "fl") 

輸入: strs = ["dog","racecar","car"]  
輸出: ""  
(說明: 無共同前綴，故回傳空字串)  
```  

#### **要求：**  
1. 若無共同前綴，回傳 `""`。  
2. 輸入陣列 `strs` 僅包含小寫英文字母。  
3. 需有效率處理邊界情況（如空陣列或單一字串）。  

#### **訓練技能：**  
- 字串操作與比對  
- 垂直掃描的迭代實作  
- 邊界條件與提前終止
