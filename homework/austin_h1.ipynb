### **作業一：身分證字號驗證**

#### **作業目標**
請撰寫一個程式來驗證台灣身分證字號的有效性。

#### **驗證規則**
1. **字母轉換**：將第一個英文字母轉換為2位數代碼
   ```
   A=10 台北市     J=18 新竹縣     S=26 高雄縣
   B=11 台中市     K=19 苗栗縣     T=27 屏東縣
   C=12 基隆市     L=20 台中縣     U=28 花蓮縣
   D=13 台南市     M=21 南投縣     V=29 台東縣
   E=14 高雄市     N=22 彰化縣     W=32 金門縣
   F=15 台北縣     O=35 新竹市     X=30 澎湖縣
   G=16 宜蘭縣     P=23 雲林縣     Y=31 陽明山
   H=17 桃園縣     Q=24 嘉義縣     Z=33 連江縣
   I=34 嘉義市     R=25 台南縣
   ```

2. **步驟2計算**：
   - 轉換後的數字：十位數 + 個位數 × 9

3. **步驟3計算**：
   - 其餘9個數字依序乘上權重 `[8,7,6,5,4,3,2,1,1]` 後相加

4. **最終驗證**：
   - 總和 = 步驟2結果 + 步驟3結果
   - 如果 `總和 % 10 == 0`：輸出 **real**，否則輸出 **fake**

#### **範例**
```
輸入：T112663836

1. 字母轉換：T → 27
2. 步驟2計算：2 + 7×9 = 2 + 63 = 65
3. 步驟3計算：1×8 + 1×7 + 2×6 + 6×5 + 6×4 + 3×3 + 8×2 + 3×1 + 6×1 = 115
4. 總和驗證：65 + 115 = 180 → 180 ÷ 10 = 18...0 → real
```

### **作業二：阿姆斯壯數**

#### **作業目標**
請撰寫一個程式，找出指定範圍內的所有阿姆斯壯數（Armstrong numbers）。

#### **阿姆斯壯數定義**
一個 **n 位數** 如果滿足以下條件，即為阿姆斯壯數：
- 每個位數的 **n 次方** 總和等於該數本身

#### **範例說明**
```
153 是 3 位數 → 1³ + 5³ + 3³ = 1 + 125 + 27 = 153
370 是 3 位數 → 3³ + 7³ + 0³ = 27 + 343 + 0 = 370
1634 是 4 位數 → 1⁴ + 6⁴ + 3⁴ + 4⁴ = 1 + 1296 + 81 + 256 = 1634
```

#### **輸入/輸出範例**
```
輸入起始值: 100
輸入結束值: 10000
輸出: 153, 370, 371, 407, 1634, 8208, 9474
```

#### **要求規範**
1. 輸入兩個整數 `start` 和 `end`，範圍為 1 ≤ `start` ≤ `end` ≤ 1,000,000
2. 輸出範圍內的所有阿姆斯壯數，以逗號分隔
3. 需要正確判斷每個數字是否為阿姆斯壯數

#### **訓練技能**
- 範圍迴圈的處理與優化
- 數學冪次運算的應用
- 數字位數的提取與操作
- 條件判斷與邏輯處理

#### **步驟2：分析問題需求**

#### **步驟4：實作位數計算**
- 找出計算數字位數的方法
- 考慮使用數學運算或字串轉換來取得位數

#### **步驟5：實作位數提取**
- 設計方法來取得數字的每個位數
- 考慮使用除法取餘數或字串轉換的方式
- 確保能正確處理所有位數

#### **步驟6：實作冪次和計算**
- 對每個位數計算 n 次方

- 使用迴圈遍歷從 start 到 end 的所有數字
- 對每個數字呼叫判斷函數
- 收集所有符合條件的阿姆斯壯數

#### **步驟8：格式化輸出**
- 將找到的阿姆斯壯數儲存在列表中
- 按照要求以逗號分隔輸出
