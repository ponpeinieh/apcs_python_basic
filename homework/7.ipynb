### **Homework 1: Two Sum**  
**Task:** Given an array of integers `nums` and an integer `target`, return the indices of the two numbers that add up to `target`.  

#### **Input/Output Examples:**  
1. **Example 1:**  
   - Input: `nums = [2, 7, 11, 15]`, target = `9`  
   - Output: `[0, 1]`  
   - Explanation: Because `nums[0] + nums[1] == 9`, we return `[0, 1]`.

2. **Example 2:**  
   - Input: `nums = [3, 2, 4]`, target = `6`  
   - Output: `[1, 2]`  

3. **Example 3:**  
   - Input: `nums = [3, 3]`, target = `6`  
   - Output: `[0, 1]`  

#### **Requirements:**  
1. Each input has **exactly one solution**.  
2. You **cannot use the same element twice**.  
3. Return the answer in any order.  

#### **Hint:**  
- Consider using a dictionary to store the indices of the numbers for quick lookups.

#### **Skills Tested:**  
- Efficient array traversal and indexing  
- Utilizing a dictionary for quick lookups  
- Handling edge cases (e.g., duplicate values)

### **Homework 2: Remove Element**  
**Task:** Given an integer array `nums` and an integer `val`, remove all occurrences of `val` in `nums` in-place. The order of the elements may be changed. Then return the number of elements in `nums` which are not equal to `val`.

Consider the number of elements in `nums` which are not equal to `val` to be `k`. To get accepted, you need to do the following:

1. Change the array `nums` such that the first `k` elements of `nums` contain the elements which are not equal to `val`. The remaining elements of `nums` are not important as well as the size of `nums`.
2. Return `k`.

#### **Input/Output Examples:**  
1. **Example 1:**  
   - Input: `nums = [3, 2, 2, 3]`, `val = 3`  
   - Output: `2`, `nums = [2, 2, _, _]`  
   - Explanation: Your function should return `k = 2`, with the first two elements of `nums` being `2`. The underscores indicate that the values beyond `k` do not matter.

2. **Example 2:**  
   - Input: `nums = [0, 1, 2, 2, 3, 0, 4, 2]`, `val = 2`  
   - Output: `5`, `nums = [0, 1, 4, 0, 3, _, _, _]`  
   - Explanation: Your function should return `k = 5`, with the first five elements of `nums` containing `0, 1, 4, 0, 3`. The order may vary.

3. **Example 3:**  
   - Input: `nums = [1]`, `val = 1`  
   - Output: `0`, `nums = [_]`  
   - Explanation: Your function should return `k = 0`, indicating that there are no elements left in `nums` that are not equal to `1`.

#### **Requirements:**  
1. You need to modify the array in-place.
2. The order of the remaining elements can be changed.
3. Return the count of elements not equal to `val`.

#### **Hint:**  
- Consider using a two-pointer technique to efficiently remove elements.

### **作業二：移除元素**  
**任務：** 給定一個整數陣列 `nums` 和整數 `val`，在原地移除 `nums` 中所有出現的 `val`。元素的順序可以改變。然後返回 `nums` 中不等於 `val` 的元素數量。

考慮 `nums` 中不等於 `val` 的元素數量為 `k`。要獲得通過，你需要做以下事情：

1. 更改陣列 `nums`，使得前 `k` 個元素包含不等於 `val` 的元素。`nums` 中剩餘的元素不重要，`nums` 的大小也不重要。
2. 返回 `k`。

#### **輸入/輸出範例：**  
1. **範例 1：**  
   - 輸入: `nums = [3, 2, 2, 3]`，`val = 3`  
   - 輸出: `2`，`nums = [2, 2, _, _]`  
   - 說明: 你的函數應該返回 `k = 2`，前兩個元素為 `2`。下劃線表示超出 `k` 的值不重要。

2. **範例 2：**  
   - 輸入: `nums = [0, 1, 2, 2, 3, 0, 4, 2]`，`val = 2`  
   - 輸出: `5`，`nums = [0, 1, 4, 0, 3, _, _, _]`  
   - 說明: 你的函數應該返回 `k = 5`，前五個元素包含 `0, 1, 4, 0, 3`。順序可以改變。

3. **範例 3：**  
   - 輸入: `nums = [1]`，`val = 1`  
   - 輸出: `0`，`nums = [_]`  
   - 說明: 你的函數應該返回 `k = 0`，表示 `nums` 中沒有不等於 `1` 的元素。

#### **要求：**  
1. 你需要在原地修改陣列。
2. 剩餘元素的順序可以改變。
3. 返回不等於 `val` 的元素數量。

#### **提示：**  
- 考慮使用雙指針技術來高效移除元素。
