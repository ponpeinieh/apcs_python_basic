### **Homework 1: Two Sum**  
**Task:** Given an array of integers `nums` and an integer `target`, return the indices of the two numbers that add up to `target`.  

#### **Input/Output Examples:**  
1. **Example 1:**  
   - Input: `nums = [2, 7, 11, 15]`, target = `9`  
   - Output: `[0, 1]`  
   - Explanation: Because `nums[0] + nums[1] == 9`, we return `[0, 1]`.

2. **Example 2:**  
   - Input: `nums = [3, 2, 4]`, target = `6`  
   - Output: `[1, 2]`  

3. **Example 3:**  
   - Input: `nums = [3, 3]`, target = `6`  
   - Output: `[0, 1]`  

#### **Requirements:**  
1. Each input has **exactly one solution**.  
2. You **cannot use the same element twice**.  
3. Return the answer in any order.  

#### **Hint:**  
- Consider using a dictionary to store the indices of the numbers for quick lookups.

#### **Skills Tested:**

### **作業一：兩數之和**  
**任務：** 給定一個整數陣列 `nums` 和目標值 `target`，返回兩數之和等於 `target` 的索引。  

#### **輸入/輸出範例：**  
1. **範例 1：**  
   - 輸入: `nums = [2, 7, 11, 15]`，目標 = `9`  
   - 輸出: `[0, 1]`  
   - 說明: 因為 `nums[0] + nums[1] == 9`，所以返回 `[0, 1]`。

2. **範例 2：**  
   - 輸入: `nums = [3, 2, 4]`，目標 = `6`  
   - 輸出: `[1, 2]`  

3. **範例 3：**  
   - 輸入: `nums = [3, 3]`，目標 = `6`  
   - 輸出: `[0, 1]`  

#### **要求：**  
1. 每組輸入保證**有且只有一個解**。  
2. **不可重複使用同一元素**。  
3. 答案順序不限。  

#### **提示：**  
- 考慮使用字典來存儲數字的索引以便快速查找。

#### **訓練技能：**  
- 高效的陣列遍歷與索引操作  
- 使用字典進行快速查找  
- 邊界條件處理（如重複值）  

### **Code Solution**

```python
def two_sum(nums, target):
    # Create a dictionary to store the indices of the numbers
    num_to_index = {}
    
    # Iterate through the list of numbers
    for index, num in enumerate(nums):
        # Calculate the complement needed to reach the target
        complement = target - num
        
        # Check if the complement exists in the dictionary
        if complement in num_to_index:
            return [num_to_index[complement], index]  # Return indices of the two numbers
        
        # Store the index of the current number in the dictionary
        num_to_index[num] = index

# Example usage
nums = [2, 7, 11, 15]
target = 9
result = two_sum(nums, target)
print(result)  # Output: [0, 1]
```

### **作業二：移除元素**  
**任務：** 給定一個整數陣列 `nums` 和整數 `val`，在原地移除 `nums` 中所有出現的 `val`。元素的順序可以改變。然後返回 `nums` 中不等於 `val` 的元素數量。

考慮 `nums` 中不等於 `val` 的元素數量為 `k`。要獲得通過，你需要做以下事情：

1. 更改陣列 `nums`，使得前 `k` 個元素包含不等於 `val` 的元素。`nums` 中剩餘的元素不重要，`nums` 的大小也不重要。
2. 返回 `k`。

#### **輸入/輸出範例：**  
1. **範例 1：**  
   - 輸入: `nums = [3, 2, 2, 3]`，`val = 3`  
   - 輸出: `2`，`nums = [2, 2, _, _]`  
   - 說明: 你的函數應該返回 `k = 2`，前兩個元素為 `2`。下劃線表示超出 `k` 的值不重要。

2. **範例 2：**  
   - 輸入: `nums = [0, 1, 2, 2, 3, 0, 4, 2]`，`val = 2`  
   - 輸出: `5`，`nums = [0, 1, 4, 0, 3, _, _, _]`  
   - 說明: 你的函數應該返回 `k = 5`，前五個元素包含 `0, 1, 4, 0, 3`。順序可以改變。

3. **範例 3：**  
   - 輸入: `nums = [1]`，`val = 1`  
   - 輸出: `0`，`nums = [_]`  
   - 說明: 你的函數應該返回 `k = 0`，表示 `nums` 中沒有不等於 `1` 的元素。

#### **要求：**  
1. 你需要在原地修改陣列。
2. 剩餘元素的順序可以改變。
3. 返回不等於 `val` 的元素數量。

#### **提示：**  
- 考慮使用雙指針技術來高效移除元素。

---

### **Code Solution**

```python
def remove_element(nums, val):
    k = 0  # Initialize a counter for elements not equal to val
    for i in range(len(nums)):
        if nums[i] != val:
            nums[k] = nums[i]  # Move the element to the front
            k += 1  # Increment the counter
    return k

# Example usage
nums1 = [3, 2, 2, 3]
val1 = 3
k1 = remove_element(nums1, val1)
print(k1)  # Output: 2
print(nums1[:k1])  # Output: [2, 2]

nums2 = [0, 1, 2, 2, 3, 0, 4, 2]
val2 = 2
k2 = remove_element(nums2, val2)
print(k2)  # Output: 5
print(nums2[:k2])  # Output: [0, 1, 3, 0, 4]

nums3 = [1]
val3 = 1
k3 = remove_element(nums3, val3)
print(k3)  # Output: 0
print(nums3[:k3])  # Output: []
```


````python
from typing import List

class Solution:
    def removeElement(self, nums: List[int], val: int) -> int:
        # 從前面開始找第一個等於 val 的元素, 從後面開始找第一個不等於 val 的元素, 然後做交換
        count = 0  # 計數器，用來記錄不等於 val 的元素數量
        left = 0  # 左邊開始指針
        right = len(nums) - 1  # 右邊開始指針
        
        while left <= right and left >= 0 and right < len(nums):
            # 左邊尋找等於 val 的元素
            while left < len(nums) and nums[left] != val:
                left += 1  # 移動左邊指針
                count += 1  # 增加計數
            
            # 右邊尋找不等於 val 的元素
            while right >= 0 and nums[right] == val:
                right -= 1  # 移動右邊指針
            
            # 如果開始指針小於結束指針，則進行交換
            if left < right:
                nums[left], nums[right] = nums[right], nums[left]
                # print(f'swap {left} , {right}')
                # print(nums)
                count += 1  # 增加計數
                left += 1  # 移動左邊指針
                right -= 1  # 移動右邊指針
        
        return count  # 返回不等於 val 的元素數量

# 測試示例
sol = Solution()

# Example usage
nums1 = [3, 2, 2, 3]
val1 = 3
k1 = sol.removeElement(nums1, val1)
print(k1)  # Output: 2
print(nums1[:k1])  # Output: [2, 2]

nums2 = [0, 1, 2, 2, 3, 0, 4, 2]
val2 = 2
k2 = sol.removeElement(nums2, val2)
print(k2)  # Output: 5
print(nums2[:k2])  # Output: [0, 1, 3, 0, 4]

nums3 = [1]
val3 = 1
k3 = sol.removeElement(nums3, val3)
print(k3)  # Output: 0
print(nums3[:k3])  # Output: []
````
