### **Homework 1: Integer to Roman**  
**Task:** Convert an integer to a Roman numeral according to the following rules:  

#### **Roman Numeral Rules:**  
| Symbol | Value |
|--------|-------|
| I      | 1     |
| V      | 5     |
| X      | 10    |
| L      | 50    |
| C      | 100   |
| D      | 500   |
| M      | 1000  |

1. **Subtractive Notation:**  
   - `4` = IV (1 before 5), `9` = IX (1 before 10)  
   - `40` = XL (10 before 50), `90` = XC (10 before 100)  
   - `400` = CD (100 before 500), `900` = CM (100 before 1000).  
2. **Additive Notation:**  
   - Symbols are appended left-to-right from largest to smallest (e.g., `3` = III).  
   - No symbol is used more than 3 times consecutively (use subtractive form instead).  

#### **Input/Output Examples:**  
```  
Input: num = 3749  
Output: "MMMDCCXLIX"  
Explanation:  
3000 = MMM  
700  = DCC  
40   = XL  
9    = IX  

Input: num = 58  
Output: "LVIII"  
Explanation:  
50 = L  
8  = VIII  

Input: num = 1994  
Output: "MCMXCIV"  
Explanation:  
1000 = M  
900  = CM  
90   = XC  
4    = IV  
```  

#### **Constraints:**  
- `1 <= num <= 3999` (Romans had no symbol for 0 or numbers ≥4000).  

#### **Skills Tested:**  
- Greedy algorithm (breaking down the number into largest possible values).  
- Hash table/dictionary for symbol-value mapping.  
- String concatenation.

### **作業一：整數轉羅馬數字**  
**任務：** 將給定的整數轉換為羅馬數字，需滿足以下規則：  

#### **羅馬數字規則：**  
| 符號 | 值  |
|------|-----|
| I    | 1   |
| V    | 5   |
| X    | 10  |
| L    | 50  |
| C    | 100 |
| D    | 500 |
| M    | 1000|

1. **減法表示法：**  
   - `4` = IV (5前放1)，`9` = IX (10前放1)  
   - `40` = XL (50前放10)，`90` = XC (100前放10)  
   - `400` = CD (500前放100)，`900` = CM (1000前放100)。  
2. **加法表示法：**  
   - 符號從大到小由左到右疊加（如`3` = III）。  
   - 同一符號最多連續使用3次（超過時改用減法形式）。  

#### **輸入/輸出範例：**  
```  
輸入: num = 3749  
輸出: "MMMDCCXLIX"  
解釋:  
3000 = MMM  
700  = DCC  
40   = XL  
9    = IX  

輸入: num = 58  
輸出: "LVIII"  
解釋:  
50 = L  
8  = VIII  

輸入: num = 1994  
輸出: "MCMXCIV"  
解釋:  
1000 = M  
900  = CM  
90   = XC  
4    = IV  
```  

#### **限制條件：**  
- `1 <= num <= 3999`（羅馬數字無表示0或≥4000的符號）。  

#### **訓練技能：**  
- 貪婪演算法（分解數字為最大可能值）。  
- 哈希表/字典用於符號映射。  
- 字串拼接。  


### **Homework 2: Reconstruct Original Digits from English**  
**Task:** Given a string containing an out-of-order English representation of digits 0-9, return the digits in ascending order.

#### **Unique Digit Representation:**  
| Digit | English Representation |
|-------|-----------------------|
| 0     | zero                  |
| 1     | one                   |
| 2     | two                   |
| 3     | three                 |
| 4     | four                  |
| 5     | five                  |
| 6     | six                   |
| 7     | seven                 |
| 8     | eight                 |
| 9     | nine                  |

1. **Unique Letters for Identification:**  
   - Use unique letters to identify certain digits:
     - `z` for 0 (zero)
     - `w` for 2 (two)
     - `u` for 4 (four)
     - `x` for 6 (six)
     - `g` for 8 (eight)

2. **Determining Remaining Digits:**  
   - After identifying unique digits, deduce the others based on the counts of letters:
     - Use counts of `o`, `t`, `f`, `s`, and `i` to derive 1, 3, 5, 7, and 9 respectively.

#### **Input/Output Examples:**  
```  
Input: s = "owoztneoer"  
Output: "012"  
Explanation:  
- 'z' → 0  
- 'w' → 2  
- 'o' → 1 (after accounting for 0 and 2)  

Input: s = "fviefuro"  
Output: "45"  
Explanation:  
- 'u' → 4  
- 'f' → 5 (after accounting for 4)  
```  

#### **Constraints:**  
- `1 <= s.length <= 10^5`  
- s consists of characters from the set ["e","g","f","i","h","o","n","s","r","u","t","w","v","x","z"].  
- s is guaranteed to be valid.

#### **Skills Tested:**  
- String manipulation and counting.  
- Logical deduction to derive counts of remaining digits.  
- Efficiently constructing and sorting the output.

### **作業二：從英文重建原始數字**  
**任務：** 給定一個字符串，包含無序的數字0-9的英文表示，返回按升序排列的數字。

#### **獨特的數字表示：**  
| 數字 | 英文表示 |
|------|----------|
| 0    | zero     |
| 1    | one      |
| 2    | two      |
| 3    | three    |
| 4    | four     |
| 5    | five     |
| 6    | six      |
| 7    | seven    |
| 8    | eight    |
| 9    | nine     |

1. **用於識別的獨特字母：**  
   - 使用獨特字母來識別某些數字：
     - `z` 代表 0（zero）
     - `w` 代表 2（two）
     - `u` 代表 4（four）
     - `x` 代表 6（six）
     - `g` 代表 8（eight）

2. **確定剩餘的數字：**  
   - 在識別獨特數字後，根據字母的計數推導出其他數字：
     - 使用 `o`、`t`、`f`、`s` 和 `i` 的計數來推導 1、3、5、7 和 9 分別。

#### **輸入/輸出範例：**  
```  
輸入: s = "owoztneoer"  
輸出: "012"  
解釋：  
- 'z' → 0  
- 'w' → 2  
- 'o' → 1 （扣除0和2的計數）  

輸入: s = "fviefuro"  
輸出: "45"  
解釋：  
- 'u' → 4  
- 'f' → 5 （扣除4的計數）  
```  

#### **限制條件：**  
- `1 <= s.length <= 10^5`  
- s 的字符來自集合 ["e","g","f","i","h","o","n","s","r","u","t","w","v","x","z"]。  
- s 保證是有效的。

#### **訓練技能：**  
- 字符串操作和計數。  
- 邏輯推理以推導剩餘數字的計數。  
- 高效構建和排序輸出。
