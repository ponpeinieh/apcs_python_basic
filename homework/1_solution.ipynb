### **Homework 1: ID Number Validation**  
**Task:** Validate whether the input is valid based on Taiwan's ID number rules.  

#### **Rules:**  
1. **Letter Conversion:** Convert the first letter to a 2-digit code.

```text
      A=10 Taipei City     J=18 Hsinchu County     S=26 Kaohsiung County
      B=11 Taichung City    K=19 Miaoli County      T=27 Pingtung County
      C=12 Keelung City      L=20 Taichung County    U=28 Hualien County
      D=13 Tainan City       M=21 Nantou County      V=29 Taitung County
      E=14 Kaohsiung City    N=22 Changhua County    W=32 Kinmen County
      F=15 Taipei County      O=35 Hsinchu City      X=30 Penghu County
      G=16 Yilan County      P=23 Yunlin County      Y=31 Yangmingshan
      H=17 Taoyuan County     Q=24 Chiayi County      Z=33 Lienchiang County
      I=34 Chiayi City       R=25 Tainan County
```

2. **Step 2 Calculation:**  
   - Converted number: Tens digit + Units digit × 9  
3. **Step 3 Calculation:**  
   - The remaining digits are multiplied sequentially by `[8,7,6,5,4,3,2,1,1]` and summed.  
4. **Validation:**  
   - Total = Step 2 + Step 3  
   - If `Total % 10 == 0`: **real**, otherwise **fake**.  

#### **Example:**  
```  
For example: T112663836 

1. **Letter Conversion:** T -> 27
2. **Step 2 Calculation:** 2 + 7*9
3. **Step 3 Calculation:** 1*8 + 1*7 + 2*6 + 6*5 + 6*4 + 3*3 + 8*2 + 3*1 + 6*1 
4. **Validation:** Total = Step 2 + Step 3 = 180 -> Divisible by 10, therefore it is real   
```
---

### **作業1：身分證字號驗證**  
**任務：** 根據台灣身分證字號規則驗證輸入是否有效。  

#### **規則：**  
1. **字母轉換**：首字母轉為2位數代碼。

```text
      A=10 台北市     J=18 新竹縣     S=26 高雄縣
      B=11 台中市     K=19 苗栗縣     T=27 屏東縣
      C=12 基隆市     L=20 台中縣     U=28 花蓮縣
      D=13 台南市     M=21 南投縣     V=29 台東縣
      E=14 高雄市     N=22 彰化縣     W=32 金門縣
      F=15 台北縣     O=35 新竹市     X=30 澎湖縣
      G=16 宜蘭縣     P=23 雲林縣     Y=31 陽明山
      H=17 桃園縣     Q=24 嘉義縣     Z=33 連江縣
      I=34 嘉義市     R=25 台南縣
```

2. **步驟2計算**：  
   - 步驟一中英文轉成的數字: 十位數 + 個位數 × 9  
3. **步驟3計算**：  
   - 其餘數字依序乘 `[8,7,6,5,4,3,2,1,1]` 後相加。  
4. **驗證**：  
   - 總和 = 步驟2 + 步驟3  
   - 若 `總和 % 10 == 0`：**real**，否則 **fake**。  

#### **範例：**  
```  
例如: T112663836 

1. **字母轉換**：T -> 27
2. **步驟2計算** : 2 + 7*9
3. **步驟3計算**：1*8 + 1*7 + 2*6 + 6*5 + 6*4 + 3*3 + 8*2 + 3*1 + 6*1 
4. **驗證**：總和 = 步驟2 + 步驟3 = 180 -> 除以 10 整除，因此為 real   
```  

---

### **Sample Solution Code (解答範例):**  
```python  
id_code = input().strip().upper()  
letter_map = {  
    'A': 10, 'B': 11, 'C': 12, 'D': 13, 'E': 14, 'F': 15, 'G': 16, 'H': 17,  
    'I': 34, 'J': 18, 'K': 19, 'L': 20, 'M': 21, 'N': 22, 'O': 35, 'P': 23,  
    'Q': 24, 'R': 25, 'S': 26, 'T': 27, 'U': 28, 'V': 29, 'W': 32, 'X': 30,  
    'Y': 31, 'Z': 33  
}  

# Step 1: Letter to code  
code = letter_map[id_code[0]]  
step2 = (code // 10) + (code % 10) * 9  

# Step 2: Weights [8,7,...,1,1]  
weights = [8, 7, 6, 5, 4, 3, 2, 1, 1]  
step3 = sum(int(digit) * weight for digit, weight in zip(id_code[1:], weights))  

# Validation  
total = step2 + step3   
print("real" if total % 10 == 0 else "fake")  
```  

### **教學重點：**  
1. **字典映射**：用字典快速查詢字母對應的數字。  
2. **權重計算**：  
   - `zip(id_code[1:], weights)` 配對數字與權重。  
3. **模運算**：`% 10` 檢查是否整除。

### **作業2：質因數分解**  
**任務：** 撰寫一個程式，將輸入的整數分解為質因數的乘積，並以 `2^3 * 3^2` 的格式輸出結果。  

#### **輸入/輸出範例：**  
```  
輸入: 20  
輸出: 2^2 * 5  
```  

#### **要求：**  
1. 輸入一個整數 `n`（1 < `n` ≤ 100,000,000）。  
2. 輸出質因數分解字串（用 `^` 表示次方）。  
3. 質因數按從小到大排列，以 ` * ` 分隔。  

#### **訓練技能：**  
- 質數判斷  
- 迴圈與除法運算  
- 字串格式化  

---  

### **Sample Solution Code (解答範例):**  
```python  
n = int(input())  
if n == 1:  
    print("1")  
    exit()  

factors = {}  
current = 2  
remaining = n  

while current * current <= remaining:  
    if remaining % current == 0:  
        factors[current] = factors.get(current, 0) + 1  
        remaining = remaining // current  
    else:  
        current += 1  

if remaining > 1:  
    factors[remaining] = factors.get(remaining, 0) + 1  

result = []  
for prime in sorted(factors):  
    if factors[prime] > 1:  
        result.append(f"{prime}^{factors[prime]}")  
    else:  
        result.append(f"{prime}")  

print(" * ".join(result))  
```  

### **教學重點：**  
1. **質因數分解邏輯**：  
   - 從最小的質數（2）開始，逐步測試能否整除剩餘的數。  
   - 如果能整除，記錄該質數及其次方；否則測試下一個數。  
2. **字典記錄次方**：用字典 `factors` 儲存質因數及其出現次數。  
3. **輸出格式化**：  
   - 次方大於 1 時，表示為 `p^e`（如 `2^2`）。  
   - 次方等於 1 時，直接輸出質數（如 `5`）。

### **如何在 Python 中使用 Unicode 上標符號取代 `^`**  
要將質因數分解的輸出格式從 `2^2` 改為 `2²`，我們可以用 **Unicode 上標數字** 替代 `^` 和指數。  

---

### **1. Unicode 上標數字對照表**  
| 數字 | 上標符號 | Unicode 碼點 |
|------|----------|--------------|
| 0    | ⁰        | `\u2070`     |
| 1    | ¹        | `\u00B9`     |
| 2    | ²        | `\u00B2`     |
| 3    | ³        | `\u00B3`     |
| 4-9  | ⁴-⁹      | `\u2074`-`\u2079` |

---

### **2. 實現代碼**  
在質因數分解的程式中，修改輸出部分的程式碼：  
```python  
def get_superscript(num):  
    # 將數字轉換為對應的 Unicode 上標符號  
    superscript_map = {  
        '0': '\u2070',  
        '1': '\u00B9',  
        '2': '\u00B2',  
        '3': '\u00B3',  
        '4': '\u2074',  
        '5': '\u2075',  
        '6': '\u2076',  
        '7': '\u2077',  
        '8': '\u2078',  
        '9': '\u2079'  
    }  
    return ''.join(superscript_map[digit] for digit in str(num))  

# 修改輸出的部分  
result = []  
for prime in sorted(factors):  
    exponent = factors[prime]  
    if exponent > 1:  
        # 使用上標符號（例如 2²）  
        result.append(f"{prime}{get_superscript(exponent)}")  
    else:  
        result.append(f"{prime}")  

print(" × ".join(result))  # 用 × 替代 * 更美觀  
```

### **4. 關鍵說明**  
1. **`get_superscript()` 函數**  
   - 將數字（如 `2`）轉換為 Unicode 上標（如 `²`）。  
2. **符號替換**  
   - 用 `×`（`\u00D7`）替代 `*`，更符合數學顯示習慣。  
3. **適用性**  
   - 適用於任何需要上標的場景（如科學記數法 `x³ + y²`）。  

---

### **5. 進階應用**  
如果想支援 **負數次方**（如 `2⁻³`），可以擴展 `superscript_map`：  
```python  
superscript_map['-'] = '\u207B'  # 負號上標  
```  

這樣就能顯示 `2⁻³` 這樣的格式！
