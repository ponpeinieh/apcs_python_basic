### **Homework 1: Integer to Roman**  
**Task:** Convert an integer to a Roman numeral according to the following rules:  

#### **Roman Numeral Rules:**  
| Symbol | Value |
|--------|-------|
| I      | 1     |
| V      | 5     |
| X      | 10    |
| L      | 50    |
| C      | 100   |
| D      | 500   |
| M      | 1000  |

1. **Subtractive Notation:**  
   - `4` = IV (1 before 5), `9` = IX (1 before 10)  
   - `40` = XL (10 before 50), `90` = XC (10 before 100)  
   - `400` = CD (100 before 500), `900` = CM (100 before 1000).  
2. **Additive Notation:**  
   - Symbols are appended left-to-right from largest to smallest (e.g., `3` = III).  
   - No symbol is used more than 3 times consecutively (use subtractive form instead).  

#### **Input/Output Examples:**  
```  
Input: num = 3749  
Output: "MMMDCCXLIX"  
Explanation:  
3000 = MMM  
700  = DCC  
40   = XL  
9    = IX  

Input: num = 58  
Output: "LVIII"  
Explanation:  
50 = L  
8  = VIII  

Input: num = 1994  
Output: "MCMXCIV"  
Explanation:  
1000 = M  
900  = CM  
90   = XC  
4    = IV  
```  

#### **Constraints:**  
- `1 <= num <= 3999` (Romans had no symbol for 0 or numbers ≥4000).  

#### **Skills Tested:**  
- Greedy algorithm (breaking down the number into largest possible values).  
- Hash table/dictionary for symbol-value mapping.  
- String concatenation.

### **作業一：整數轉羅馬數字**  
**任務：** 將給定的整數轉換為羅馬數字，需滿足以下規則：  

#### **羅馬數字規則：**  
| 符號 | 值  |
|------|-----|
| I    | 1   |
| V    | 5   |
| X    | 10  |
| L    | 50  |
| C    | 100 |
| D    | 500 |
| M    | 1000|

1. **減法表示法：**  
   - `4` = IV (5前放1)，`9` = IX (10前放1)  
   - `40` = XL (50前放10)，`90` = XC (100前放10)  
   - `400` = CD (500前放100)，`900` = CM (1000前放100)。  
2. **加法表示法：**  
   - 符號從大到小由左到右疊加（如`3` = III）。  
   - 同一符號最多連續使用3次（超過時改用減法形式）。  

#### **輸入/輸出範例：**  
```  
輸入: num = 3749  
輸出: "MMMDCCXLIX"  
解釋:  
3000 = MMM  
700  = DCC  
40   = XL  
9    = IX  

輸入: num = 58  
輸出: "LVIII"  
解釋:  
50 = L

輸入: num = 1994  
輸出: "MCMXCIV"  
解釋:  
1000 = M  
900  = CM  
90   = XC  
4    = IV  
```  

#### **限制條件：**  
- `1 <= num <= 3999`（羅馬數字無表示0或≥4000的符號）。  

#### **訓練技能：**  
- 貪婪演算法（分解數字為最大可能值）。  
- 哈希表/字典用於符號映射。  
- 字串拼接。

### **Solution Code :**  
```python
def intToRoman(num: int) -> str:
    val_symbols = [
        (1000, "M"), (900, "CM"), (500, "D"), (400, "CD"),
        (100, "C"), (90, "XC"), (50, "L"), (40, "XL"),
        (10, "X"), (9, "IX"), (5, "V"), (4, "IV"), (1, "I")
    ]
    roman = []
    for value, symbol in val_symbols:
        while num >= value:
            roman.append(symbol)
            num -= value
    return ''.join(roman)
```

### **Explanation:**  
1. **Greedy Approach:**  
   - The algorithm always picks the largest possible Roman symbol at each step (e.g., `3749` → `1000 (M)` first).  
2. **Predefined Mappings:**  
   - The `val_symbols` list includes subtractive forms (like `CM`, `XL`) to handle edge cases directly.  
3. **Efficiency:**  
   - O(1) time complexity (fixed iterations for values ≤3999) and O(1) space.  
 


### 將阿拉伯數字轉換為羅馬數字的解決方案
#### Another solution

```python

nums = 2597  # 要轉換的阿拉伯數字
data = []    # 儲存轉換後的羅馬數字字符

# 將數字分解為各個位數的列表
digits = [int(d) for d in str(nums)]

# 計算初始偏移量，用於確定羅馬數字字符的級別（個位、十位、百位、千位）
counter = len(digits) * 2 - 2  

# 定義基本數字對應的羅馬數字模式
mydict = {
  1: ['I'],          # 1 對應 I
  2: ['I', 'I'],     # 2 對應 II
  3: ['I', 'I', 'I'],# 3 對應 III
  4: ['I', 'V'],     # 4 對應 IV
  5: ['V'],          # 5 對應 V
  6: ['V', 'I'],     # 6 對應 VI
  7: ['V', 'I', 'I'],# 7 對應 VII
  8: ['V', 'I', 'I', 'I'],  # 8 對應 VIII
  9: ['I', 'X']      # 9 對應 IX
}

# 羅馬數字的基本字符，按順序排列：I, V, X, L, C, D, M
roman_values = ['I', 'V', 'X', 'L', 'C', 'D', 'M']

# 遍歷每個位數數字
for i in digits:
    # 獲取該數字對應的基本羅馬數字模式
    v = mydict[i]
    
    # 處理模式中的每個字符
    for z in v:
        # 根據當前偏移量計算正確的羅馬數字字符
        idx = roman_values.index(z) + counter
        data.append(roman_values[idx])
    
    # 減少偏移量，處理下一個較低的位數（從千位到百位到十位到個位）
    counter -= 2

# 輸出轉換結果
print(data)
```

這個程式的工作原理：
1. 將輸入數字分解為各個位數（如 2597 → [2, 5, 9, 7]）
2. 使用 counter 變數來追蹤當前處理的是哪個位數（千位、百位、十位或個位）
3. 對於每個位數數字，查找對應的基本羅馬數字模式
4. 根據當前位數的級別，將基本字符轉換為正確的羅馬數字字符
5. 將所有字符收集到 data 列表中並輸出

轉換邏輯：
- 個位數：I, V, X
- 十位數：X, L, C  
- 百位數：C, D, M
- 千位數：M（以及更大的數字）

### **Solution Code (Python):**  
```python
from collections import Counter

def originalDigits(s: str) -> str:
    count = Counter(s)
    
    # Initialize the count of digits
    digit_count = [0] * 10
    
    # Unique character mapping to digits
    digit_count[0] = count['z']  # 0
    digit_count[2] = count['w']  # 2
    digit_count[4] = count['u']  # 4
    digit_count[6] = count['x']  # 6
    digit_count[8] = count['g']  # 8
    
    # For the remaining digits
    digit_count[1] = count['o'] - digit_count[0] - digit_count[2] - digit_count[4]  # 1
    digit_count[3] = count['h'] - digit_count[8]  # 3
    digit_count[5] = count['f'] - digit_count[4]  # 5
    digit_count[7] = count['s'] - digit_count[6]  # 7
    digit_count[9] = count['i'] - digit_count[5] - digit_count[8]  # 9
    
    # Construct the output
    result = ''.join(str(i) * digit_count[i] for i in range(10))
    return result
```
