### **作業一：最大公因數 (GCD)**  
**任務：** 撰寫一個程式，用**輾轉相除法（歐幾里得算法）**計算兩整數的最大公因數並輸出結果。  

#### **輸入/輸出範例：**  
```  
輸入: 24 36  
輸出: 12  
```  

#### **要求：**  
1. 輸入兩個整數 `a` 和 `b`（1 ≤ `a`, `b` ≤ 1,000,000）。  
2. 輸出它們的最大公因數。  
3. 必須使用**輾轉相除法**（迭代或遞迴均可）。  

#### **訓練技能：**  
- 最大公因數與輾轉相除法的理解  
- 迴圈或遞迴運用  
- 邊界條件處理（例如輸入為零）  

### **Code Solution (Python):**  
```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

# Input
a, b = map(int, input().split())

# Output
print(gcd(a, b))
```

#### **Explanation:**  
1. The Euclidean algorithm repeatedly replaces `(a, b)` with `(b, a % b)` until `b` becomes zero. The GCD is the last non-zero remainder (`a`).  
2. Example: For `a = 24`, `b = 36`:  
   - Step 1: `36 % 24 = 12` → `(24, 12)`  
   - Step 2: `24 % 12 = 0` → GCD is `12`.  

This solution is efficient with a time complexity of **O(log(min(a, b)))**.

### **作業二：十進位轉二進位**  
**任務：** 撰寫一個程式，使用 **`while` 迴圈** 將十進位數字轉換為二進位並輸出結果。  

#### **輸入/輸出範例：**  
```  
輸入: 13  
輸出: 1101  
```  

#### **要求：**  
1. 輸入一個整數 `n`（0 ≤ `n` ≤ 1,000,000）。  
2. 輸出其二進位表示法（字串形式）。  
3. 必須使用 **`while` 迴圈**（不可用內建函數如 `bin()`）。  
4. 需特別處理 `n = 0` 的情況。  

#### **訓練技能：**  
- 二進位制理解  
- 迴圈與餘數運算  
- 字串拼接  

### **Code Solution:**  
```python
n = int(input("Input: "))

if n == 0:
    print("0")
else:
    binary = ""
    while n > 0:
        binary = str(n % 2) + binary  # Prepend remainder
        n = n // 2
    print(binary)
```

#### **Explanation:**  
1. **Special Case (`n = 0`):** Directly outputs `"0"`.  
2. **While Loop Logic:**  
   - Repeatedly divides `n` by 2 and prepends the remainder (`n % 2`) to the `binary` string.  
   - Updates `n` to `n // 2` until `n` becomes `0`.  
3. **Example (`n = 13`):**  
   - Iteration 1: `13 % 2 = 1` → `binary = "1"`, `n = 6`  
   - Iteration 2: `6 % 2 = 0` → `binary = "01"`, `n = 3`  
   - Iteration 3: `3 % 2 = 1` → `binary = "101"`, `n = 1`  
   - Iteration 4: `1 % 2 = 1` → `binary = "1101"`, `n = 0` (loop ends).  

This method ensures correct binary conversion without built-in functions.

### **作業三：數字倒轉（正數）**  
**任務：** 撰寫一個程式，將輸入的**正整數**數字順序倒轉。  

#### **輸入/輸出範例：**  
```  
輸入: 12345  
輸出: 54321  

輸入: 1000  
輸出: 1  （注意：去除開頭的零）  
```  

#### **要求：**  
1. 輸入正整數 `n`（0 ≤ `n` ≤ 1,000,000）。  
2. 輸出倒轉後的數字（去除開頭的零）。  
3. 使用**迴圈或數學運算**（不可用字串倒轉技巧）。  

#### **訓練技能：**  
- 迴圈控制與終止條件  
- 數學運算（取餘和除法）  
- 處理末尾零  

### **Code Solution (Python):**  
```python
n = int(input("Input: "))  

reversed_num = 0  
while n > 0:  
    reversed_num = reversed_num * 10 + n % 10  
    n = n // 10  

print(reversed_num)  
```

#### **Explanation:**  
1. **Loop Logic:**  
   - `n % 10` extracts the last digit (e.g., `12345 % 10 = 5`).  
   - `reversed_num * 10 + digit` appends the digit (e.g., `0 * 10 + 5 = 5`).  
   - `n // 10` removes the last digit (e.g., `12345 // 10 = 1234`).  

2. **Edge Cases:**  
   - Input `1000` → Reversed digits: `0001` → Output: `1` (leading zeros dropped).  
   - Input `0` → Output: `0` (loop skips, `reversed_num` remains `0`).  

---

### **Key Concepts:**  
- **Modulo (`% 10`)** and **division (`// 10`)** are the core operations.  
- **Loop terminates** when `n` becomes `0`.

### **作業四：阿姆斯壯數**  
**任務：** 撰寫一個程式，找出指定範圍內的所有阿姆斯壯數。  

#### **輸入/輸出範例：**  
```  
輸入起始值: 100  
輸入結束值: 10000  
輸出: 153, 370, 371, 407, 1634, 8208, 9474  
```  

#### **要求：**  
1. 輸入兩個整數 `start` 和 `end`（1 ≤ `start` ≤ `end` ≤ 1,000,000）。  
2. 輸出範圍 `[start, end]` 內的所有阿姆斯壯數，以逗號分隔。  
3. **阿姆斯壯數**是一個 *n* 位數，其每個位數的 *n* 次方和等於該數本身。  

#### **訓練技能：**  
- 範圍迴圈處理  
- 數學運算（冪次與求和）  
- 數字位數操作  

#### **Code Solution:**  
```python
start = int(input("Input start: "))
end = int(input("Input end: "))

armstrong_numbers = []

for num in range(start, end + 1):
    # Count digits (n) and extract each digit
    temp = num
    digits = []
    while temp > 0:
        digits.append(temp % 10)
        temp = temp // 10
    n = len(digits)
    
    # Calculate sum of digits^n
    sum_of_powers = 0
    for d in digits:
        sum_of_powers += d ** n
    
    # Check if Armstrong number
    if sum_of_powers == num:
        armstrong_numbers.append(num)

# Output results separated by commas
if armstrong_numbers:
    print(*armstrong_numbers, sep=", ")
else:
    print("No Armstrong numbers in this range.")
```

### **Code Solution (using list comprehension):**  
```python
start = int(input("Input start: "))  
end = int(input("Input end: "))  

armstrong_numbers = []  

for num in range(start, end + 1):  
    # Convert number to string to count digits and get each digit  
    digits = [int(d) for d in str(num)]  
    n = len(digits)  
    # Calculate sum of digits raised to the power n  
    sum_of_powers = sum(d ** n for d in digits)  
    if sum_of_powers == num:  
        armstrong_numbers.append(num)  

# Output results separated by commas  
print(*armstrong_numbers, sep=", ")  
```

#### **Explanation:**  
1. **Digit Handling:**  
   - Convert the number to a string to easily iterate over each digit.  
   - Count digits (`n`) to determine the power for each digit.  

2. **Armstrong Check:**  
   - For each digit in the number, raise it to the power `n` and sum all results.  
   - If the sum equals the original number, it’s an Armstrong number.  

3. **Edge Cases:**  
   - Single-digit numbers (1–9) are Armstrong numbers by definition (`1^1 = 1`, `9^1 = 9`).  
   - The range `[100, 10000]` includes the examples `153` (`1³ + 5³ + 3³ = 153`) and `1634` (`1⁴ + 6⁴ + 3⁴ + 4⁴ = 1634`).  

---

### **Key Concepts:**  
- **Armstrong Definition:**  
  - A 3-digit number: \( abc = a^3 + b^3 + c^3 \).  
  - A 4-digit number: \( abcd = a^4 + b^4 + c^4 + d^4 \).  
- **Efficiency:** The algorithm checks each number in the range linearly, which is feasible for `end ≤ 1,000,000`.  
 

### **Optimized Armstrong Number Finder (with Precomputed Powers)**
```python
start = int(input("Input start: "))
end = int(input("Input end: "))

armstrong_numbers = []

# Precompute powers for digits 0-9 up to max possible exponent (max_digits)
max_digits = len(str(end))
power_table = [[d ** n for n in range(1, max_digits + 1)] for d in range(10)]

for num in range(start, end + 1):
    # Count digits and extract each digit
    temp = num
    digits = []
    while temp > 0:
        digits.append(temp % 10)
        temp = temp // 10
    n = len(digits)
    
    # Calculate sum using precomputed powers
    sum_of_powers = 0
    for d in digits:
        sum_of_powers += power_table[d][n-1]  # d^n from precomputed table
    
    if sum_of_powers == num:
        armstrong_numbers.append(num)

# Output results
if armstrong_numbers:
    print(*armstrong_numbers, sep=", ")
else:
    print("No Armstrong numbers in this range.")
```

### **Key Optimizations:**
1. **Power Table Precomputation**:
   - Builds a 10xN table (`power_table`) where `power_table[d][n-1] = dⁿ`.
   - Eliminates redundant `d**n` calculations during the main loop.

2. **Digit Counting Optimization**:
   - Uses `len(str(end))` to determine maximum needed exponent (avoids over-calculation).

3. **Efficient Digit Extraction**:
   - Still uses modulo/division loop (faster than string conversion for very large numbers).

### **作業五：十進位轉二進位（含負數）**  
**任務：** 撰寫一個程式，使用 **`while` 迴圈** 將十進位數字（正數、零或負數）轉換為 **8位元二補數** 的二進位表示法。  

#### **輸入/輸出範例：**  
```  
輸入: 13  
輸出: 00001101  

輸入: -13  
輸出: 11110011  
```  

#### **要求：**  
1. 輸入整數 `n`（-128 ≤ `n` ≤ 127）。  
2. 輸出 **8位元二補數** 的二進位字串（固定長度）。  
3. 必須使用 **`while` 迴圈**（不可用內建函數如 `bin()`）。  
4. 正確處理正數和負數。  

#### **訓練技能：**  
- 二補數表示法  
- 位元運算（負數處理）  
- 固定長度字串格式化  


### **Code Solution:**  
```python
n = int(input("Input: "))

def to_binary(n, bits=8):
    if n >= 0:
        # Positive number: Standard binary conversion
        binary = ""
        for _ in range(bits):
            binary = str(n % 2) + binary
            n = n // 2
        return binary
    else:
        # Negative number: Two's complement
        n = (1 << bits) + n  # Convert to positive equivalent
        binary = ""
        for _ in range(bits):
            binary = str(n % 2) + binary
            n = n // 2
        return binary

print(to_binary(n))
```

#### **Explanation:**  
1. **Positive Numbers:**  
   - Standard division-by-2 method, padded to 8 bits.  
   - Example: `13 → 00001101`.  

2. **Negative Numbers (Two's Complement):**  
   - Step 1: Compute `n + 2^8` (e.g., `-13 + 256 = 243`).  
   - Step 2: Convert the result to binary (`243 → 11110011`).  
   - Why? The 8-bit two's complement of `-13` is equivalent to the binary representation of `243`.  

3. **Edge Cases:**  
   - `n = 0` → `00000000`  
   - `n = -128` → `10000000` (smallest 8-bit negative).  

---

### **Key Concepts:**  
- **Two's Complement:** Flips bits and adds 1 to represent negatives.  
- **Fixed 8-bit Length:** Ensures consistency for both positives and negatives.
