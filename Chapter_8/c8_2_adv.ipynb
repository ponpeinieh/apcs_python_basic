CSV 檔案開頭出現 **BOM (Byte Order Mark)** ：  
 
### **CSV 檔案開頭出現 BOM 是正常的嗎？**  
是的，這在某些情況下是正常的，尤其是當 CSV 檔案採用 **UTF-8-BOM** 編碼時。  

### **為什麼 CSV 會有 BOM？**  
1. **UTF-8-BOM 編碼**：  
   - 部分程式（如 Microsoft Excel）在儲存 CSV 時，會自動加入 BOM（`\xEF\xBB\xBF`），以明確標示檔案為 UTF-8 編碼。  
2. **工具或編輯器的預設行為**：  
   - 某些文字編輯器（如 Windows 的記事本、Excel）在儲存 UTF-8 格式時，預設會加入 BOM。  
3. **相容性考量**：  
   - BOM 能幫助舊版軟體正確識別 Unicode 編碼。  

### **BOM 會造成問題嗎？**  
- **✅ 正常情況**：如果檔案本身就是 UTF-8-BOM 格式，BOM 是預期的，不會影響內容。  
- **⚠️ 潛在問題**：  
  - 部分程式（如 Python 的 `pandas.read_csv()` 或 Unix 工具）可能無法正確解析 BOM，導致：  
    - 第一列欄位名稱出現亂碼（如 `"ï»¿Column1"` 而非 `"Column1"`）。  
    - 腳本讀取時發生錯誤。  

### **如何處理 BOM？**  
1. **保留 BOM**：  
   - 如果需要與 Excel 或其他依賴 BOM 的工具相容，可以保留它。  

2. **移除 BOM**（若不需要）：  
   - **Python**：使用 `encoding='utf-8-sig'` 自動處理 BOM。  
     ```python
     import pandas as pd
     df = pd.read_csv("file.csv", encoding="utf-8-sig")
     ```  
   - **命令列（Linux/Mac）**：  
     ```sh
     sed -i '1s/^\xEF\xBB\xBF//' file.csv  # 移除 BOM
     ```  
   - **文字編輯器**：  
     - 用 Notepad++ 開啟檔案，選擇「編碼 > 轉換至 UTF-8（無 BOM）」後儲存。
