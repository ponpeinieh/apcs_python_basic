### 練習一：使用者輸入與三種字串格式化語法

**任務**：撰寫一個程式，要求使用者輸入他們的姓名、年齡和出生城市，然後使用三種不同的字串格式化語法（舊式 `%` 語法、`format()` 方法和 f-string），列印出以下的訊息：

"嗨，我的名字是 {姓名} 。我今年 {年齡} 歲，我出生在 {出生城市} 城市。"

**要求**：
1. 使用 `input()` 函數獲取使用者的姓名、年齡和城市。
2. 使用 `%` 語法格式化字串。
3. 使用 `format()` 方法格式化字串。
4. 使用 f-string 格式化字串。

### 練習二：使用 f-string 格式化浮點數和對齊文字

**任務**：使用變數儲存一個產品的名稱、價格（浮點數）和庫存數量。然後使用 f-string 格式化來列印產品資訊。

**要求**：
1. 產品名稱必須在 6 個字元以內，並且列印時佔用 12 個字元的寬度，左對齊。
2. 價格以浮點數格式顯示，顯示兩位小數。
3. 庫存數量向右對齊，佔用 5 個字元的寬度。

### 練習三：驗證正弦平方與餘弦平方的和等於1的數學定理

**任務**：撰寫一個程式，要求使用者輸入一個角度（以度為單位），然後計算該角度的正弦(sin)平方和餘弦(cos)平方，並驗證它們的和是否等於 1。

**要求**：
1. 使用 `input()` 函數獲取使用者輸入的角度。
2. 將輸入轉換為浮點數（`float`）。
3. 使用 `math` 模組計算該角度的正弦和餘弦，並計算它們的平方和。
4. 使用 f-string 格式化輸出結果，顯示以下訊息：
   - "角度 {角度} 的正弦平方加上餘弦平方等於 {和}。"
