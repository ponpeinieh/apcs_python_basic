**題目陳述：**
編寫一個程式，根據給定的半徑計算圓的面積和周長。半徑應作為變數定義。使用 `math` 模組進行計算。

### 練習 2：直角三角形的斜邊

**題目陳述：**
編寫一個程式，根據直角三角形的兩條邊的長度計算斜邊的長度。兩條邊的長度應作為變數定義。使用 `math` 模組進行計算。

### 練習 3：統計計算

**題目陳述：**
編寫一個程式，計算五個數字的平均值和標準差。五個數字應作為獨立的變數定義。使用 `math` 模組進行計算。


### 練習 1：圓的面積和周長

**代碼解決方案：**
```python
import math

# 定義半徑
radius = 5

# 計算面積
area = math.pi * (radius ** 2)

# 計算周長
circumference = 2 * math.pi * radius

# 輸出結果
print("圓的面積：", area)
print("圓的周長：", circumference)
```

### 練習 2：直角三角形的斜邊

```python
import math

# 定義兩條邊的長度
side_a = 3
side_b = 4

# 計算斜邊
hypotenuse = math.sqrt(side_a ** 2 + side_b ** 2)

# 輸出結果
print("斜邊長度：", hypotenuse)
```

### 練習 3：統計計算

**代碼解決方案：**
```python
import math

# 定義五個數字
num1 = 10
num2 = 20
num3 = 30
num4 = 40
num5 = 50

# 計算平均值
mean = (num1 + num2 + num3 + num4 + num5) / 5

# 計算標準差
variance = ((num1 - mean) ** 2 + (num2 - mean) ** 2 + (num3 - mean) ** 2 +
            (num4 - mean) ** 2 + (num5 - mean) ** 2) / 5
std_dev = math.sqrt(variance)

# 輸出結果
print("平均值：", mean)
print("標準差：", std_dev)
```

### 練習 4：字串處理

**題目陳述：**
編寫一個程式，將字串 "  Welcome to Python programming.   " 進行以下的連續操作：
1. 移除字串前後的空白字元。
2. 將字串轉換為大寫。
3. 分割字串並顯示每個單字。

**代碼解決方案：**
```python
# 預定義字串
input_string = "  Welcome to Python programming.   "

# 移除前後空白字符
stripped_string = input_string.strip()

# 將字符串轉換為大寫
upper_case_string = stripped_string.upper()

# 分割字符串
split_string = upper_case_string.split()

# 輸出結果
print("原始字符串：", input_string)
print("移除空白字符後的字符串：", stripped_string)
print("大寫字符串：", upper_case_string)
print("分割後的字符串：", split_string)
```
