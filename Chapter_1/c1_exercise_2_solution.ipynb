### 練習二：使用 f-string 格式化浮點數和對齊文字

**任務**：使用變數儲存一個產品的名稱、價格（浮點數）和庫存數量。然後使用 f-string 格式化來列印產品資訊。

**要求**：
1. 產品名稱必須在 6 個字元以內，並且列印時佔用 12 個字元的寬度，左對齊。

3. 庫存數量向右對齊，佔用 5 個字元的寬度。

### 練習三：驗證正弦平方與餘弦平方的和等於1的數學定理

**任務**：撰寫一個程式，要求使用者輸入一個角度（以度為單位），然後計算該角度的正弦(sin)平方和餘弦(cos)平方，並驗證它們的和是否等於 1。

**要求**：
1. 使用 `input()` 函數獲取使用者輸入的角度。
2. 將輸入轉換為浮點數（`float`）。
3. 使用 `math` 模組計算該角度的正弦和餘弦，並計算它們的平方和。
4. 使用 f-string 格式化輸出結果，顯示以下訊息：
   - "角度 {角度} 的正弦平方加上餘弦平方等於 {和}。"

### 練習一：使用三種字串格式化語法

### 代碼解決方案：

```python
# 獲取使用者輸入
name = input("請輸入您的姓名：")  # 使用者的姓名
age = input("請輸入您的年齡：")    # 使用者的年齡
city = input("請輸入您出生的城市：")  # 使用者的出生城市

# 使用 % 語法格式化字串
formatted_string_percent = "嗨，我的名字是 %s。我今年 %s 歲，我出生在 %s 城市。" % (name, age, city)
print(formatted_string_percent)

# 使用 format() 方法格式化字串
formatted_string_format = "嗨，我的名字是 {}。我今年 {} 歲，我出生在 {} 城市。".format(name, age, city)
print(formatted_string_format)

# 使用 f-string 格式化字串
formatted_string_fstring = f"嗨，我的名字是 {name}。我今年 {age} 歲，我出生在 {city} 城市。"
print(formatted_string_fstring)
```

### 練習二：使用 f-string 格式化浮點數和對齊文字

### 代碼解決方案：

```python
# 定義變數
product_name = "蘋果"  # 產品名稱（最多 6 個字元）
price = 1.2345
stock = 50

# 使用 f-string 格式化字串，指定寬度和浮點數格式
formatted_string = f"{product_name:<12} | 價格: {price:.2f} | 庫存: {stock:>5}"
print(formatted_string)  # 預期輸出："蘋果          | 價格: 1.23 | 庫存:   50"
```

### 說明：
- `:<12` 表示將產品名稱左對齊並佔用 12 個字元的寬度。
- `:.2f` 表示將價格格式化為浮點數，顯示兩位小數。
- `:>5` 表示將庫存數量右對齊，佔用 5 個字元的寬度。

### 練習三：驗證正弦平方與餘弦平方的和等於1的數學定理

### 代碼解決方案：

```python
import math  # 導入 math 模組

# 獲取使用者輸入
angle_input = input("請輸入一個角度（度）：")

# 將輸入轉換為浮點數
angle_degrees = float(angle_input)

# 將角度轉換為弧度
angle_radians = math.radians(angle_degrees)

# 計算正弦和餘弦
sin_value = math.sin(angle_radians)  # 計算正弦
cos_value = math.cos(angle_radians)  # 計算餘弦

# 計算正弦平方和餘弦平方的和
sum_of_squares = sin_value**2 + cos_value**2

# 使用 f-string 格式化字串輸出結果
formatted_message = f"角度 {angle_degrees} 的正弦平方加上餘弦平方等於 {sum_of_squares:.2f}。"
print(formatted_message)
```

### 說明：
- 程式首先要求使用者輸入一個角度（以度為單位）。
- 使用 `float()` 函數將輸入轉換為浮點數，然後使用 `math.radians()` 將角度轉換為弧度，以便計算。
- 使用 `math.sin()` 和 `math.cos()` 計算正弦和餘弦值，並計算它們的平方和。
- 最後，使用 f-string 格式化輸出結果，顯示正弦平方加上餘弦平方的和，並確保結果顯示兩位小數。
