### 練習三：河流與國家管理

**任務**：建立一個字典，該字典包含河流及其所經國家的對應關係。然後使用迴圈執行以下操作：

1. 列印每條河流及其所經的國家，格式為 "xxx河流經xxx國家"。
2. 列印所有的河流名稱。
3. 列印所有的國家名稱。

### 要求：
- 使用字典（`dict`）來儲存五條河流和國家的對應關係。
- 使用 `for` 迴圈來遍歷字典並執行上述操作。

### 練習一：字串處理與反轉計數
 
```python
# 獲取使用者輸入的字串
quote = input("請輸入一個字串：")

# 使用單一索引值提取第一個和最後一個字元，並設定預設值
first_char = quote[0]                       # 提取第一個字元
last_char = quote[-1]                       # 提取最後一個字元
substring = quote[2:8]                      # 提取第 3 到第 8 個字元
char_count = len(quote)                     # 計算字元數量
reversed_quote = quote[::-1]                # 反轉字串

# 格式化輸出
print(f"您輸入的字串為：{quote}，第一個字元為：{first_char}，最後一個字元為：{last_char}，"
      f"子字串為：{substring}，字元數量為：{char_count}，反轉後的字串為：{reversed_quote}。")
```

### 練習二：邀請晚宴的客人管理

### 代碼解答：
```python
# 初始客人清單
guest_list = ["客人A", "客人B", "客人C"]

# 添加一位新客人在清單第一順位的位置
new_guest = input("請輸入要添加的新客人名稱：")
guest_list.insert(0, new_guest)
print(f"已添加新客人：{new_guest}，現在的清單為：{guest_list}")

# 替換清單中第三個位置的客人
replace_guest_name = input("請輸入新的客人名稱以替換第三個位置的客人：")
guest_list[2] = replace_guest_name
print(f"已將第三個位置的客人替換為：{replace_guest_name}，現在的清單為：{guest_list}")

# 刪除清單中最後位置的客人
removed_guest = guest_list.pop()  # 刪除最後一位客人
print(f"已刪除客人：{removed_guest}，現在的清單為：{guest_list}")

# 將剛剛刪除的客人再加回到清單中
guest_list.append(removed_guest)
print(f"已將客人 {removed_guest} 加回清單，現在的清單為：{guest_list}")

# 使用 for 迴圈列印邀請訊息
print("\n邀請信發送給以下客人：")
for guest in guest_list:
    print(f"親愛的 {guest}，我們誠摯地邀請您參加晚宴！")
```

### 練習三：河流與國家管理

### 代碼解答：
```python
# 建立河流與國家的字典
rivers = {
    '尼羅河': '埃及',
    '恆河': '印度',
    '亞馬遜河': '巴西',
    '長江': '中國',
    '密西西比河': '美國'
}

# 列印每條河流及其所經的國家
for river, country in rivers.items():
    print(f"{river} 河流經 {country} 國家")

# 列印所有的河流名稱
print("\n所有河流名稱：")
for river in rivers.keys():
    print(river)

# 列印所有的國家名稱
print("\n所有國家名稱：")
for country in rivers.values():
    print(country)
```
