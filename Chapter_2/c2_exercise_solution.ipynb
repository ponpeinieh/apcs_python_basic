### 練習一：進位轉換與加法運算

**任務**：撰寫一個程式，要求使用者輸入兩個數字，這些數字可以是二進位、八進位、十進位或十六進位的表示。程式需要轉換這些數字為十進位，然後計算它們的和，並將結果以二進位、八進位、十進位和十六進位表示。

**要求**：
1. 使用 `input()` 函數獲取使用者輸入的兩個數字，並指定進位（如 `0b`、`0o`、`0x`、或是無前綴符號的十進位）。
2. 將這些數字轉換為十進位整數。(請見以下範例)
3. 計算這兩個數字的和。
4. 將結果轉換為二進位(bin)、八進位(oct)和十六進位(hex)表示。
5. 使用 f-string 格式化輸出，顯示以下訊息：
   - "數字 {數字1} 和數字 {數字2} 的和為 {和}，以二進位表示為 {二進位}，八進位表示為 {八進位}，十六進位表示為 {十六進位}。"

**補充説明**: 將數字字串轉換為十進位整數的方法
- 使用`int(數字字串, 基數)`函數
- 對於基數 0，字串的解釋方式會根據前綴符號自動判斷基數為 2、8、10 或 16。

```python
x_str = '0b1010'
y_str = '0o117'
z_str = '0xabc'
w_str = '123'
x = int(x_str, 0) # 基數爲0
y = int(y_str, 0)
z = int(z_str, 0)
w = int(w_str, 0)
print(x,y,z,w)
```

### 練習二：邏輯運算符與關係運算符

**任務**：撰寫一個程式，要求使用者輸入三個數字，然後判斷這些數字是否滿足以下條件：

1. 至少有一個數字大於 10。
2. 所有數字必須大於或等於 0。
3. 至少有兩個數字相等。

如果以上這些條件中的任何一個滿足，則輸出「條件滿足」，否則輸出「條件不滿足」。

**要求**：
1. 使用 `input()` 函數獲取三個數字。
2. 將輸入轉換為整數（`int`）。
3. 使用關係運算符（例如 `>=`）和邏輯運算符（例如 `and`, `or`, `not`）來檢查條件。
4. 使用 `if` 語句輸出相應的結果。

### 練習一：進位轉換與加法運算

### 代碼解決方案：

```python
# 獲取使用者輸入
num1_input = input("請輸入第一個數字（以二進位、八進位、十進位或十六進位表示，例如 0b1010、0o12、123、0x1A）：")
num2_input = input("請輸入第二個數字（以二進位、八進位、十進位或十六進位表示）：")

# 將輸入轉換為十進位整數
num1_decimal = int(num1_input, 0)  # 基於前綴自動判斷進位
num2_decimal = int(num2_input, 0)

# 計算兩個數字的和
sum_decimal = num1_decimal + num2_decimal

# 將結果轉換為二進位、八進位和十六進位表示
binary_representation = bin(sum_decimal)        # 二進位表示
octal_representation = oct(sum_decimal)         # 八進位表示
hexadecimal_representation = hex(sum_decimal)   # 十六進位表示

# 使用 f-string 格式化字串輸出結果
formatted_message = (
    f"數字 {num1_input} 和數字 {num2_input} 的和為 {sum_decimal}，"
    f"以二進位表示為 {binary_representation}，"
    f"八進位表示為 {octal_representation}，"
    f"十六進位表示為 {hexadecimal_representation}。"
)
print(formatted_message)
```

### 練習二：邏輯運算符與關係運算符

### 代碼解決方案：

```python
# 獲取使用者輸入
num1_input = input("請輸入第一個數字：")
num2_input = input("請輸入第二個數字：")
num3_input = input("請輸入第三個數字：")

# 將輸入轉換為整數
num1 = int(num1_input)
num2 = int(num2_input)
num3 = int(num3_input)

# 檢查條件
condition1 = (num1 > 10) or (num2 > 10) or (num3 > 10)  # 至少有一個數字大於 10
condition2 = (num1 >= 0) and (num2 >= 0) and (num3 >= 0)  # 所有數字必須大於或等於 0
condition3 = (num1 == num2) or (num1 == num3) or (num2 == num3)  # 至少有兩個數字相等

# 總結條件
if condition1 or condition2 or condition3:
    print("條件滿足")
else:
    print("條件不滿足")
```
