### 練習一：使用列表播放音樂音符

**描述**：編寫一個程序，使用列表來播放不同的音樂音符，包括高音和低音。

**要求**：
1. 使用 `music.play()` 方法來播放一系列音符。
2. 定義音符的格式，並能夠控制每個音符的音高和時長。
3. 使用 `sleep()` 來控制音樂之間的間隔。

**示例代碼**：
```python
from microbit import *
import music

# 定義音符的說明
'''
Do : c
Re : d
Mi : e
Fa : f
So : g
La : a
Si : b
rest : r

例子：
g4:8 -> 'So' 在第4個八度，音符長度：8
'''

# 播放不同音高的音符
sleep(2000) 
music.play(['c4:4', 'd', 'e', 'f', 'g', 'a', 'b', 'c5'])  # 第一組音符
music.play(['c3:4', 'd', 'e', 'f', 'g', 'a', 'b', 'c4'])  # 第二組音符
music.play(['c4:2', 'd', 'e', 'f', 'g', 'a', 'b', 'c5'])  # 第三組音符
```

### 練習二：使用列表播放音樂音符（正向和反向）

**描述**：編寫一個程序，使用列表來存儲音樂音符，並能夠播放這些音符的正向和反向版本。

**要求**：
1. 使用 `music.play()` 方法來播放一系列音符，並將其存儲在列表中。
2. 播放音符的正向序列。
3. 播放音符的反向序列。

**示例代碼**：
```python
from microbit import *
import music

# 定義第一組音符並存儲在列表中
notes = ['c4', 'd', 'e', 'f', 'g', 'a', 'b4', 'c5']

# 播放正向音符
sleep(2000) 
music.play(notes) 

# 播放反向音符 
sleep(2000)  # 暫停以便聽到音樂
music.play(notes[::-1])  # 使用切片反轉列表
```

### 程式說明：
- 使用列表 `notes` 來存儲音符。
- 使用 `music.play()` 方法播放音符列表的正向序列。
- 使用列表切片 `[::-1]` 來反轉音符的順序，並播放反向序列。

### 練習三：播放音符字符串

**描述**：編寫一個程序，使用列表來播放一系列音符，並能夠處理音符字符串的播放，特別是「兩隻老虎」的音樂。

**要求**：
1. 定義一個字符串，表示「兩隻老虎」的音符及其時長。
2. 使用 `split()` 方法將字符串轉換為音符列表。
3. 使用 `music.play()` 方法來播放這些音符。

**示例代碼**：
```python
from microbit import *
import music

sleep(2000)
# 播放 'Two Tigers'
notes = 'c4:4 d e c c d e c e f g:8 e:4 f g:8 g:2 a g f e:4 c g:2 a g f e:4 c c g3 c4:8 c:4 g3 c4:8'
music.play(notes.split(' '))  # 使用 split() 將字符串轉換為列表並播放
```

### 程式說明：
- 在 `notes` 中，定義了「兩隻老虎」的音符和時長的格式。
- 使用 `split(' ')` 方法將字符串轉換為列表，這樣每個音符就能被正確播放。

### 練習四：顯示預定義圖像

**描述**：編寫一個程序，使用 micro:bit 的預定義圖像，在列表中迭代顯示每個圖像。

**要求**：
1. 定義一個包含多個預定義圖像的列表。
2. 使用迴圈顯示每個圖像。

**示例代碼**：
```python
from microbit import *

# 定義預定義圖像列表
images = [
    Image.HAPPY,
    Image.SAD,
    Image.HEART,
    Image.SILLY,
    Image.SURPRISED
]

# 迭代列表並顯示每個圖像
for img in images:
    display.show(img)
    sleep(1000)  # 暫停1秒以便查看每個圖像
```

### 程式說明：
- **Image.HAPPY**、**Image.SAD** 等是 MicroPython 中的預定義圖像。
- 使用 `display.show(img)` 來顯示當前圖像。
- 每個圖像顯示後，暫停 1 秒，以便用戶可以看到圖像。


### 練習六：使用字典播放音樂

**描述**：編寫一個程序，使用字典來存儲音樂曲調並播放。

**要求**：
1. 定義一個字典，鍵是音樂名稱，值是音樂曲調。
2. 使用迴圈播放字典中的每個曲調。

**示例代碼**：
```python
from microbit import *
import music

# 定義音樂曲調字典
custom_songs = {
    'Ding': music.BA_DING,
    'Power Up': music.POWER_UP,
    'Power Down': music.POWER_DOWN
}

# 迭代字典:顯示每一個曲調名稱
for name in custom_songs:
    display.scroll(name)  # 顯示音樂名稱
    sleep(1000)          # 暫停1秒以便聽到每個音樂

# 迭代字典:播放每一個曲調
for _ in range(3):
    for song in custom_songs.values():
        music.play(song)
        sleep(1000)

# 迭代字典:顯示每一個曲調名稱並播放
for name, song in custom_songs.items():
    display.scroll(name)  # 顯示音樂名稱
    music.play(song)      # 播放音樂
    sleep(1000)          # 暫停1秒以便聽到每個音樂
    
```
