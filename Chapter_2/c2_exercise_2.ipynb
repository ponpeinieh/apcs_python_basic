### 練習一：字串處理與反轉計數

**任務**：撰寫一個程式，要求使用者輸入一個字串，然後提取並顯示該字串的第一個字元、最後一個字元，以及指定範圍內的子字串。此外，計算該字串中的字元數量，並反轉該字串。

**要求**：
1. 使用 `input()` 函數獲取使用者輸入的字串。
2. 使用索引提取並顯示字串的第一個字元和最後一個字元。
3. 使用切片(slice)提取並顯示字串的第 3 到第 8 個字元（根據使用者輸入的字串長度，需考慮邊界情況）。
4. 計算字串中的字元數量（包括空格和標點符號）。
5. 反轉該字串。
6. 使用 f-string 格式化輸出，顯示以下訊息：
   - "您輸入的字串為：{原始字串}，第一個字元為：{第一個字元}，最後一個字元為：{最後一個字元}，子字串為：{子字串}，字元數量為：{字元數量}，反轉後的字串為：{反轉字串}。"

### 練習二：邀請晚宴的客人管理

**任務**：撰寫一個程式，建立一個客人清單，然後執行添加、替換、刪除等操作，並為每位客人準備邀請訊息。

**要求**：
1. 將三名客人名字存儲在清單中。
2. 添加一位新客人在清單第一順位的位置。
3. 替換清單中第三個位置的客人。
4. 刪除清單中最後位置的客人。
5. 將剛剛刪除的客人再加回到清單中。
6. 使用 `for` 迴圈，給清單中每一位客人列印邀請訊息: "親愛的 {客人名字}，我們誠摯地邀請您參加晚宴！"

### 練習三：河流與國家管理

**任務**：建立一個字典，該字典包含河流及其所經國家的對應關係。然後使用迴圈執行以下操作：

1. 列印每條河流及其所經的國家，格式為 "xxx河流經xxx國家"。
2. 列印所有的河流名稱。
3. 列印所有的國家名稱。

### 要求：
- 使用字典（`dict`）來儲存五條河流和國家的對應關係。
- 使用 `for` 迴圈來遍歷字典並執行上述操作。
