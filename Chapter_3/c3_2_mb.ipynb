### 練習一：使用條件語句控制LED顯示

**描述**：編寫一個程序，根據按鈕的狀態控制微型顯示屏上的顯示內容。

**要求**：
1. 使用 `if`、`elif` 和 `else` 語句來檢查按鈕 A 和 B 的狀態。
2. 當按下按鈕 A 時，在顯示屏上顯示一顆心形。
3. 當按下按鈕 B 時，在顯示屏上顯示一個笑臉。
4. 如果兩個按鈕都沒有被按下，顯示一個悲傷的臉。

**示例代碼**：
```python
from microbit import *

while True:
    if button_a.is_pressed() and button_b.is_pressed():
        display.show(Image.SAD)  # 如果兩個按鈕都被按下，顯示悲傷的臉
    elif button_a.is_pressed():
        display.show(Image.HEART)  # 如果按下按鈕 A，顯示心形
    elif button_b.is_pressed():
        display.show(Image.HAPPY)  # 如果按下按鈕 B，顯示笑臉
    else:
        display.clear()  # 如果沒有按下任何按鈕，清除顯示
```
