### 練習題 1：計算第 n 個斐波那契數

**描述**：編寫一個程式，計算並輸出第 n 個斐波那契數。請注意，斐波那契數列的定義如下：

- 第 1 個數是 0
- 第 2 個數是 1
- 第 3 個數是 1
- 第 4 個數是 2
- 第 5 個數是 3
- 第 n 個數是前兩個數的和，即 `F(n) = F(n-1) + F(n-2)`

**要求**：
1. 使用 `input` 讀取一個整數 n。
2. 使用 `for` 迴圈計算第 n 個斐波那契數。
3. 輸出第 n 個斐波那契數。

### 練習題 2：猜數字遊戲

**描述**：建立一個互動的猜數字遊戲，玩家需要在最多 5 次的猜測中找出隨機生成的數字，並根據猜測結果提供最小和最大值範圍的提示。

**要求**：
1. 隨機生成一個介於 1 到 100 的整數作為答案。
2. 在每次猜測後，根據玩家所猜的數字，提示更新最小和最大值範圍，格式為：
   - 如果猜的數字太低，顯示訊息：`"請猜一個介於 {min} 和 {max} 的數字，因為 {guess} 太低了！"`
   - 如果猜的數字太高，顯示訊息：`"請猜一個介於 {min} 和 {max} 的數字，因為 {guess} 太高了！"`
3. 當玩家的猜測次數達到 5 次時，顯示所有猜測過的數字及正確答案，格式為：`"你所有的猜測：{guesses}，正確答案是 {answer}！"`

### 範例案例
- 假設隨機生成的正確答案為 `77`。
- 玩家第一次猜測 `89`。

**程式輸出**：
```plaintext
請猜一個介於 1 和 88 的數字，因為 89 太高了！
```

### 練習題 3：擴展猜數字遊戲

**描述**：在猜數字遊戲中，增加難度選項，讓玩家選擇「簡單」、「一般」或「困難」三個難度級別，並根據選擇的級別生成不同範圍的隨機數字和限制猜測次數。

**要求**：
1. 提供三個難度選項：「簡單」（1-100，最多 6 次）、 「一般」（1-1000，最多 8 次）、 「困難」（1-10000，最多 11 次）。
2. 根據玩家選擇的難度級別生成相應範圍的隨機數字。
3. 在每次猜測後顯示當前的最小和最大範圍，以及提示訊息，格式為：
   - 如果猜的數字太低，顯示訊息：`"請猜一個介於 {min} 和 {max} 的數字，因為 {guess} 太低了！"`
   - 如果猜的數字太高，顯示訊息：`"請猜一個介於 {min} 和 {max} 的數字，因為 {guess} 太高了！"`


### 練習題 4：尋找第一個偶數

**描述**：編寫一個程式，從一組整數中尋找第一個偶數。如果找到偶數，則輸出該偶數；如果所有數字都是奇數，則輸出「沒有找到偶數」。

**要求**：
1. 使用 `input` 讀取一行數字，數字之間用逗號分隔。
2. 將讀取的字串分割並轉換為整數清單。
3. 使用 `for` 或 `while` 迴圈遍歷數字清單。
4. 使用 `else` 語句來處理當迴圈正常結束（即沒有 `break`）的情況。
