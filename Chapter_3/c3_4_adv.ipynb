# Python Collections 模組中的 Counter 類別

## 簡介
`Counter` 是 `dict` 的一種子類別，專門用來計算可雜湊（Hashable）物件的數量。它將物件和它們的計數以字典的形式儲存，簡單來說，就是一個計數容器。計數可以是任何整數，包括 0 和負數，讓我們能靈活地處理不同的情況。

## 初始化
`Counter` 可以從多種方式來初始化，包括 iterable、其他 mapping、關鍵字參數，甚至是另一個 `Counter`：

```python
c = Counter()                           # 建立一個新的空 Counter
c = Counter('gallahad')                 # 從字串 iterable 初始化的 Counter
c = Counter({'red': 4, 'blue': 2})      # 從 mapping 初始化的 Counter
c = Counter(cats=4, dogs=8)             # 從關鍵字參數初始化的 Counter
```

## 使用介面
`Counter` 的操作方式與字典類似，但有一個特別之處：**如果你查詢的元素在 Counter 中不存在，它會返回 0，而不是引發 KeyError**，這樣使用起來更方便：

```python
cnt = Counter()
for word in ['red', 'blue', 'red', 'green', 'blue', 'blue']:
    cnt[word] += 1
print(cnt) # 輸出 Counter({'blue': 3, 'red': 2, 'green': 1})

c = Counter(['eggs', 'ham'])
c['bacon']                              # 查詢一個缺失的元素，返回的計數是零
```

## 刪除元素
如果將某個值的計數設為 0，這個元素不會自動從 Counter 中移除。要真正刪除它，請使用 `del` 語句：

```python
c['sausage'] = 0                        # 設定計數為零的 counter 條目
del c['sausage']                        # 使用 del 來實際移除這個條目
```

## 附加方法
除了字典本身的方法外，`Counter` 還支援幾個額外的便利方法：

### `elements()`
這個方法會回傳一個**迭代器**，包含每個元素重複出現的次數。**元素的返回順序依據首次出現的時間**。如果某個元素的計數小於 1，則會被忽略。

```python
c = Counter(a=2, b=4, c=0, d=-2)
print(list(c.elements()))
# 輸出: ['a', 'a', 'b', 'b', 'b', 'b']
```

### `most_common([n])`

返回一個包含 n 個最常見元素及其計數的清單，按從最常見到最不常見的順序。如果 n 被省略或為 None，`most_common()` 將返回計數器中的所有元素。計數相同的元素會按照首次出現的順序排列：

```python
Counter('abracadabra').most_common(3)
# 輸出: [('a', 5), ('b', 2), ('r', 2)]
```

### `subtract([iterable-or-mapping])`
這個方法可以從一個 iterable 或另一個映射（或 Counter）中減去計數。行為類似 `dict.update()`，但這裡是減去計數而不是取代值。輸入和輸出都可以是 0 或負數。

```python
c = Counter(a=4, b=2, c=0, d=-2)
d = Counter(a=1, b=2, c=3, d=4)
c.subtract(d)
# c 現在是: Counter({'a': 3, 'b': 0, 'c': -3, 'd': -6})
```

### `total()`
這個方法計算所有元素的總計數值。

```python
c = Counter(a=10, b=5, c=0)
c.total()
# 輸出: 15
```

## 注意事項
- 直接列印Counter物件, 其元素的順序會按照計數值由大到小排序. 
- 若使用for loop, 則會按照元素首次出現的順序排列
- 字典的方法也可以用於 Counter 物件，但有兩個例外：
    - `fromkeys(iterable)` 這個類別方法並不適用於 `Counter`。
    - `update([iterable-or-mapping])` 用來增加計數，行為類似 `dict.update()`，但這裡是增加計數而非取代值。注意，iterable 需要是一串將被計數的元素，而不是 (key, value) 形式的序列。


```python
c = Counter(a=2, b=2, c=0, d=-2)
d = Counter(a=1, b=2, c=3, d=4)
c.update(d) 
print(c) # 按照計數值由大到小排序
for i,j in c.items(): #按照元素首次出現的順序排列
    print(i,j)
```

- 輸出
```
Counter({'b': 4, 'a': 3, 'c': 3, 'd': 2})
a 3
b 4
c 3
d 2
```
