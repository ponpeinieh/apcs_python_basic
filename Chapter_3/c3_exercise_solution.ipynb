### 練習題 1：成績分類

**描述**：寫一個函數，接受一個數字成績（0-100），並根據以下標準返回字母等級：
- A: 90-100
- B: 80-89
- C: 70-79
- D: 60-69
- F: 0-59

**解答**：
```python
def classify_grade(grade):
    if grade >= 90:
        return "A"
    elif grade >= 80:
        return "B"
    elif grade >= 70:
        return "C"
    elif grade >= 60:
        return "D"
    else:
        return "F"

# 示例用法
grade = float(input("請輸入成績（0-100）："))
print(f"你的等級是: {classify_grade(grade)}")
```

### 練習題 2：星期幾

**描述**：寫一個函數，接受一個整數（1-7），表示星期幾，並返回對應的星期名稱。
1 = 星期一，2 = 星期二，...，7 = 星期日。

**解答**：
```python
def day_of_week(day):
    if day == 1:
        return "星期一"
    elif day == 2:
        return "星期二"
    elif day == 3:
        return "星期三"
    elif day == 4:
        return "星期四"
    elif day == 5:
        return "星期五"
    elif day == 6:
        return "星期六"
    elif day == 7:
        return "星期日"
    else:
        return "無效的輸入，請輸入1到7之間的數字"

# 示例用法
day = int(input("請輸入星期幾（1-7）："))
print(f"今天是: {day_of_week(day)}")
```

### 練習題 3：溫度轉換器

**描述**：寫一個函數，接受攝氏溫度並根據用戶選擇轉換為華氏或開爾文。用戶可以輸入 "F" 表示華氏或 "K" 表示開爾文。

**解答**：
```python
def convert_temperature(celsius, scale):
    if scale.upper() == "F":
        return (celsius * 9/5) + 32  # 攝氏轉華氏
    elif scale.upper() == "K":
        return celsius + 273.15  # 攝氏轉開爾文
    else:
        return "無效的單位，請輸入 'F' 或 'K'"

# 示例用法
temp_celsius = float(input("請輸入攝氏溫度："))
scale_choice = input("請輸入要轉換的單位（F/K）：")
converted_temp = convert_temperature(temp_celsius, scale_choice)

if isinstance(converted_temp, str):
    print(converted_temp)  # 輸出錯誤信息
else:
    print(f"轉換後的溫度是: {converted_temp:.2f}°{scale_choice.upper()}")
```
