### 練習題 2：多位學生成績分類 

**解答範例**：
```python
# 定義學生及其成績的字典
student_grades = {
    "Alice": 95,
    "Bob": 82,
    "Charlie": 76,
    "David": 58,
    "Eva": 66
}

# 根據成績顯示字母等級
for student, grade in student_grades.items():
    if grade >= 90:
        level = "A"
    elif grade >= 80:
        level = "B"
    elif grade >= 70:
        level = "C"
    elif grade >= 60:
        level = "D"
    else:
        level = "F"
    
    print(f"{student} 的等級是: {level}")
```
 
### 練習題 3：九宮格(3x3 網格座標)檢查 

**解答**：
```python
# 提示使用者輸入座標
x = int(input("請輸入 x 座標（0-2）："))
y = int(input("請輸入 y 座標（0-2）："))

# 根據座標顯示相應的訊息
if x < 0 or y < 0:
    print("座標不能為負數！")
elif 0 <= x < 3 and 0 <= y < 3:
    print("座標在網格內有效。")
    
    if (x, y) == (1, 1):
        print("您在網格的中心。")
    elif (x, y) in [(0, 0), (0, 2), (2, 0), (2, 2)]:
        print("您在網格的角落。")
    else:
        print("您在網格的邊緣。")
else:
    print("座標無效！請輸入 0 到 2 之間的值。")
``` 

### 練習題 4：偶數平方總和計算 

**解答範例**：
```python
# 初始化總和變數和公式字串
total_sum = 0
formula = ""

# 使用 for 迴圈計算從 2 到 20 的偶數的平方
for num in range(2, 21, 2):  # 每次增加 2，獲取偶數
    square = num ** 2
    total_sum += square  # 累加平方到總和
    
    # 構建公式字串
    if num == 20:
        formula += f"{num}^2"
    else:
        formula += f"{num}^2 + "

# 顯示偶數平方的總和和公式
print(f"平方公式: {formula} = {total_sum}")
```
### 練習題 5：使用 `for` 迴圈及 `zip()` 函數列印遊戲怪物資訊  

**解答範例**：
```python
# 定義三個清單
monsters = ["Spider", "Zombie", "Creeper"]
weapons = ["Sword", "Bow", "TNT"]
scores = [100, 50, 150]

# 使用 for 迴圈遍歷三個清單並顯示怪物資訊
for i in range(len(monsters)):
    print(f"怪物 {monsters[i]} 可以被 {weapons[i]} 所打敗，可以得到的分數為 {scores[i]}")

print()  # 空行分隔

# 使用 zip() 函數合併三個清單
combined = zip(monsters, weapons, scores)

# 使用 for 迴圈遍歷合併後的結果並顯示怪物資訊
for monster, weapon, score in combined:
    print(f"怪物 {monster} 可以被 {weapon} 所打敗，可以得到的分數為 {score}")
```
