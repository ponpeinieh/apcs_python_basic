### while 迴圈

#### 基本概念

`while` 迴圈是用來反覆執行某段程式碼的一種控制結構，直到指定的條件不再成立為止。這意味著只要條件為真，迴圈就會持續執行。`while` 迴圈非常適合用於當你不知道確切迴圈次數的情況，但可以根據某個條件來進行控制。

#### 語法結構

```python
while 條件:
    # 執行的程式碼
```

```mermaid
flowchart TD
    A[開始] --> B{條件為真?}
    B -- 是 --> C[執行程式碼]
    C --> D[更新狀態]
    D --> B
    B -- 否 --> E[結束]
```

````python
mm("""
flowchart TD
    A[開始] --> B{條件為真?}
    B -- 是 --> C[執行程式碼]
    C --> D[更新狀態]
    D --> B
    B -- 否 --> E[結束]""")
````

### 說明

1. **開始**：流程開始。
2. **條件為真?**：檢查條件是否為真。
   - 如果是，執行程式碼，然後更新狀態，並且重新判斷條件。
   - 如果否，結束迴圈。

#### 範例

```python
count = 0  # 初始化計數器

while count < 5:  # 當 count 小於 5 時執行
    print(f"目前計數: {count}")
    count += 1  # 更新計數器
```

### 無窮迴圈（Infinite Loop）

無窮迴圈是指一個迴圈將持續不斷地執行下去。這在程式設計中並不罕見，尤其是在需要持續運行的應用中，例如遊戲的主迴圈或伺服器的運行。

#### 特點

- **持續執行**：無窮迴圈會在沒有特定條件的情況下持續執行，直到程式被強制終止或遇到特定的終止條件。
- **常見用法**：經常用於需要持續監控或反覆檢查狀態的場景，如遊戲主迴圈、事件監聽、伺服器輪詢等。

#### 使用 `while True`

在 Python 中，無窮迴圈通常使用 `while True` 的語法來實現。這裡的 `True` 是一個永遠為真的條件，迴圈將不斷執行。

```python
while True:
    # 重複執行的動作
    print("這是一個無窮迴圈")
```

#### 結束無窮迴圈

要終止無窮迴圈，通常會使用 `break` 語句。這樣可以根據某些條件跳出迴圈，讓程式正常結束。以下是使用 `break` 的範例：

```python
while True:
    # 重複執行的動作
    user_input = input("請輸入 'exit' 來結束程式：")
    if user_input == 'exit':
        print("程式結束。")
        break  # 使用 break 跳出迴圈
```

### 使用場景

- **遊戲主迴圈**：在遊戲中，無窮迴圈通常用於主迴圈，持續檢查遊戲狀態、處理用戶輸入和更新畫面。
- **伺服器監控**：伺服器可以使用無窮迴圈來持續監聽請求，並根據請求執行相應的操作。
- **設備監控**：在嵌入式系統或設備中，無窮迴圈可用於持續監測傳感器的狀態。

### 遊戲主迴圈 (Game Main Loop)、事件監聽 (Event Listening) 與 伺服器輪詢 (Server Polling)

這三者都是無窮迴圈的實際應用場景，下面將逐一解釋它們的功能和用途。

#### 1. 遊戲主迴圈

**概念**：遊戲主迴圈是遊戲程式的核心部分，負責管理遊戲的運行流程。它持續運行，並不斷更新遊戲狀態和渲染畫面。

**流程**：

- **檢查輸入**：檢查玩家的鍵盤、滑鼠或其他控制器的輸入。
- **更新遊戲狀態**：根據玩家的輸入和遊戲邏輯更新角色位置、計分、敵人行為等。
- **渲染畫面**：根據最新的遊戲狀態繪製畫面，顯示給玩家。
- **控制幀率**：通常會有延遲或控制幀率的機制，以確保遊戲運行的平穩性。

**範例**：

```python
while True:
    handle_input()  # 處理輸入
    update_game_state()  # 更新遊戲狀態
    render()  # 渲染畫面
    check_exit()  # 檢查是否退出遊戲
```

#### 2. 事件監聽

**概念**：事件監聽是一種編程模式，用於等待和響應特定事件的發生。這些事件可以是用戶輸入、網絡請求或其他系統事件。

**流程**：

- **註冊事件**：將事件處理函數註冊到特定事件上。
- **監聽事件**：在無窮迴圈中持續檢查是否有事件發生。
- **觸發處理**：當事件發生時，自動呼叫預先註冊的事件處理函數來執行相應的操作。

**範例**：

```python
while True:
    event = check_for_event()  # 檢查是否有事件
    if event:
        handle_event(event)  # 處理事件
```

#### 3. 伺服器輪詢

**概念**：伺服器輪詢是指客户端持續向伺服器發送請求，以檢查是否有新的數據或事件發生。這是一種常見的網絡編程模式。

**流程**：

- **發送請求**：在無窮迴圈中定期向伺服器發送請求。
- **處理響應**：檢查伺服器的響應，並根據數據更新客戶端狀態。
- **控制請求頻率**：為了避免過於頻繁的請求，通常會添加延遲或計時器來控制輪詢的頻率。

**範例**：

```python
while True:
    response = poll_server()  # 向伺服器發送請求
    if response:
        update_client(response)  # 更新客戶端狀態
    time.sleep(1)  # 控制請求頻率
```

### 猜數字遊戲 (Guess the Number Game)

#### 遊戲規則

- 程式隨機生成一個 1 到 100 之間的數字。
- 玩家需要猜這個數字。
- 如果玩家的猜測比正確數字低，程式會提示 "太小了"；如果比正確數字高，程式會提示 "太大了"。
- 玩家可以在任何時候輸入 "exit" 結束遊戲。

#### 程式碼範例

```python
import random

# 隨機生成一個 1 到 100 之間的數字
target_number = random.randint(1, 100)
print("歡迎來到猜數字遊戲！我已經選擇了一個 1 到 100 之間的數字。")

while True:
    guess = input("請猜一個數字（或輸入 'exit' 來結束遊戲）：")
    
    # 檢查玩家是否希望結束遊戲
    if guess.lower() == 'exit':
        print("謝謝遊玩！再見！")
        break  # 使用 break 跳出迴圈
    
    # 檢查輸入是否為有效數字
    try:
        guess_number = int(guess)
    except ValueError:
        print("請輸入有效的數字！")
        continue  # 若輸入無效，重新開始迴圈

    # 檢查玩家猜測的數字
    if guess_number < target_number:
        print("太小了！再試一次。")
    elif guess_number > target_number:
        print("太大了！再試一次。")
    else:
        print("恭喜！你猜對了！正確的數字是", target_number)
        break  # 猜對了，跳出迴圈
```

### 程式碼解釋

1. **匯入隨機模組**：使用 `import random` 匯入隨機數生成的功能。
2. **隨機生成數字**：使用 `random.randint(1, 100)` 隨機生成一個 1 到 100 之間的數字。
3. **主迴圈**：使用 `while True` 開始無窮迴圈，持續接收玩家的輸入。
4. **結束遊戲選項**：檢查玩家是否輸入了 "exit"，如果是，則列印結束信息並使用 `break` 跳出迴圈。
5. **有效性檢查**：使用 `try` 和 `except` 來檢查玩家的輸入是否為有效的整數。如果無效，則提示並重新開始。
6. **猜測檢查**：
   - 如果玩家的猜測小於目標數字，提示 "太小了"。
   - 如果大於目標數字，提示 "太大了"。
   - 如果猜對了，祝賀玩家並使用 `break` 結束遊戲。

```python
def is_prime_simple(n):
    # 1 不是質數
    if n <= 1:
        return False
    # 2 和 3 是質數
    if n <= 3:
        return True
    
    i = 2
    # 檢查從 2 到 sqrt(n) 的所有整數
    while i * i <= n:
        # 如果 n 能被 i 整除，則 n 不是質數
        if n % i == 0:
            return False
        i += 1
    return True


def is_prime_simple2(n):
    answer = None
    
    # 1 不是質數
    if n <= 1:
        answer = False
    # 2 和 3 是質數
    elif n <= 3:
        answer = True
    else:
        i = 2
        answer = True
        # 檢查從 2 到 sqrt(n) 的所有整數
        while i * i <= n:
            # 如果 n 能被 i 整除，則 n 不是質數
            if n % i == 0:
                answer = False
                break
            i += 1
    return answer


def is_prime_smart(n):
    # 1 不是質數
    if n <= 1:
        return False
    # 2 和 3 是質數
    if n <= 3:
        return True
    # 排除能被 2 或 3 整除的數字
    if n % 2 == 0 or n % 3 == 0:
        return False
    
    # 檢查從 5 開始的奇數因數
    # 奇數可能的形式為 6n+1, 6n+3(排除, 因爲是3的倍數), 6n+5
    # 6n+5 可以視爲是 6n-1
    # 因此我們僅需判斷6n-1 與 6n+1兩種形式的奇數
    i = 5
    while i * i <= n:
        # 如果 n 能被 i 或 i+2 整除，則 n 不是質數
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6  # 每次跳過 6，檢查 5, 11, 17, ...
    
    return True
```

### 代碼說明

1. **`is_prime_simple(n)`**:
   - 使用簡單的方法檢查質數。
   - 檢查從 2 開始的整數，直到 `sqrt(n)`。

2. **`is_prime_simple2(n)`**:
   - 另一種簡單的方法，使用 `answer` 變數來儲存結果。
   - 同樣檢查從 2 開始的整數，直到 `sqrt(n)`。

3. **`is_prime_smart(n)`**:
   - 使用更高效的方法，首先排除能被 2 和 3 整除的數字。

### 請寫一個程式，列出輸入整數的所有質因數

- 例如輸入1100, 輸出 [2,2,5,5,11]

```python
def prime_factors(n):
    
    factors = []  # 用於儲存質因數的列表

    # 檢查 2 是否為質因數
    while n % 2 == 0:
        factors.append(2)  # 將 2 添加到質因數列表
        n //= 2  # 將 n 除以 2
    
    # 檢查所有奇數因數
    i = 3
    while i * i <= n:
        while n % i == 0:
            factors.append(i)  # 將質因數添加到列表
            n //= i  # 將 n 除以質因數
        i += 2
    # 如果 n 仍大於 2，則 n 本身是質數
    if n > 2:
        factors.append(n)
    
    return factors

# 主程式
if __name__ == "__main__":
    number = int(input("請輸入一個整數："))  # 輸入整數
    print(f"{number} 的質因數為：{prime_factors(number)}")
```

### 程式碼說明

1. **函數 `prime_factors(n)`**:
   - 接受一個整數 `n`，返回其質因數的列表。
   - 首先檢查 2 是否為質因數，並持續將 2 除去。
   - 接著檢查從 3 開始的每個奇數因數，直到 `sqrt(n)`。
   - 如果最終的 `n` 大於 2，則 `n` 本身是質數，也應添加到質因數列表中。

2. **主程式**:
   - 提示用戶輸入一個整數。
   - 呼叫 `prime_factors` 函數並列印結果。

### 使用範例

如果用戶輸入 `28`，程式會輸出：

```
28 的質因數為：[2, 2, 7]
```

這表示 28 的質因數是兩個 2 和一個 7。這段程式碼能夠有效地列出輸入整數的所有質因數。


````python
def prime_factors_with_powers(n):
    """
    列出輸入整數的所有質因數及其次方
    
    :param n: 要檢查的整數
    :return: 質因數及其次方的字典
    """
    factors = {}  # 用於儲存質因數及其次方的字典

    # 檢查 2 是否為質因數
    while n % 2 == 0:
        if 2 in factors:
            factors[2] += 1  # 增加次方
        else:
            factors[2] = 1  # 初始化次方為 1
        n //= 2  # 將 n 除以 2
    
    # 檢查所有奇數因數
    i = 3
    while i * i <= n:
        while n % i == 0:
            if i in factors:
                factors[i] += 1  # 增加次方
            else:
                factors[i] = 1  # 初始化次方為 1
            n //= i  # 將 n 除以質因數
        i += 2
    # 如果 n 仍大於 2，則 n 本身是質數
    if n > 2:
        factors[n] = 1  # 將質數添加到字典，次方為 1
    
    return factors

# 主程式
if __name__ == "__main__":
    number = int(input("請輸入一個整數："))  # 輸入整數
    factors = prime_factors_with_powers(number)
    
    # 輸出結果
    result = " * ".join([f"{base}^{exp}" for base, exp in factors.items()])
    print(f"{number} 的質因數為：{result}")
````

````python
from collections import Counter

def prime_factors_with_powers(n):
    
    factors = Counter()  # 用於儲存質因數及其次方的 Counter 物件

    # 檢查 2 是否為質因數
    while n % 2 == 0:
        factors[2] += 1  # 增加次方
        n //= 2  # 將 n 除以 2
    
    # 使用 while 迴圈檢查所有奇數因數
    i = 3
    while i * i <= n:  # 檢查到 sqrt(n)
        while n % i == 0:
            factors[i] += 1  # 增加次方
            n //= i  # 將 n 除以質因數
        i += 2  # 檢查下一個奇數
    
    # 如果 n 仍大於 2，則 n 本身是質數
    if n > 2:
        factors[n] += 1  # 將質數添加到 Counter，次方為 1
    
    return factors

# 主程式
if __name__ == "__main__":
    number = int(input("請輸入一個整數："))  # 輸入整數
    factors = prime_factors_with_powers(number)
    
    # 輸出結果
    result = " * ".join([f"{base}^{exp}" for base, exp in factors.items()])
    print(f"{number} 的質因數為：{result}")
````

````python
# 或是先計算所有的因數, 然後再利用Counter, 產生計數
def prime_factors(n):
    
    factors = []  # 用於儲存質因數的列表

    # 檢查 2 是否為質因數
    while n % 2 == 0:
        factors.append(2)  # 將 2 添加到質因數列表
        n //= 2  # 將 n 除以 2
    
    # 檢查所有奇數因數
    i = 3
    while i * i <= n:
        while n % i == 0:
            factors.append(i)  # 將質因數添加到列表
            n //= i  # 將 n 除以質因數
        i += 2
    # 如果 n 仍大於 2，則 n 本身是質數
    if n > 2:
        factors.append(n)
    
    return factors

# 主程式
if __name__ == "__main__":
    number = int(input("請輸入一個整數："))  # 輸入整數
    factors = prime_factors(number)
    print(f"{number} 的質因數為：{factors}")
    #
    from collections import Counter
    cnt = Counter(factors)
    print(cnt)
    result = " * ".join([f"{base}^{exp}" for base, exp in cnt.items()])
    print(f"{number} 的質因數為：{result}")
````

### `for-else` 語法搭配 `break`

在 Python 中，`for-else` 結構是一個獨特的語法，它可以用來在 `for` 迴圈結束時執行一些特定的邏輯，尤其是在迴圈因為正常結束而非因 `break` 中斷的情況下。以下是這種語法的詳細說明：

#### 基本概念

1. **`for-else` 語法**：
   - `for` 迴圈遍歷一個可迭代對象（如列表、字典、字符串等）。
   - `else` 部分會在 `for` 迴圈正常結束時執行，即當迴圈沒有被 `break` 中斷時。

2. **無需設置狀態變數**：
   - 使用 `for-else` 語法，可以不需要額外的狀態變數來檢查迴圈是否被 `break` 中斷。這種設計使得程式碼更簡潔。

3. **`break` 的作用**：
   - 在 `for` 迴圈中，當執行到 `break` 語句時，會立即跳出整個 `for-else` 結構，並不會執行 `else` 部分的代碼。

#### 範例: 鷄兔同籠問題

假設有一個籠子裡面有雞和兔子，已知總共有 `heads` 個頭和 `legs` 隻腿。請編寫程式來計算籠子裡面各有多少隻雞和兔子。

#### 1. 不使用 `else` 的版本

在這個版本中，我們使用一個標誌變數來判斷是否找到了解。

```python
def chicken_rabbit_without_else(heads, legs):
    found = False  # 標誌變數，初始為 False
    
    for chickens in range(heads + 1):  # 假設雞的數量從 0 到 heads
        rabbits = heads - chickens  # 兔子的數量
        if (2 * chickens + 4 * rabbits) == legs:  # 檢查腿的數量
            print(f"雞: {chickens}, 兔子: {rabbits}")
            found = True  # 找到解，更新標誌變數
            break  # 跳出迴圈

    if not found:
        print("沒有符合條件的解。")  # 只有在未找到解的情況下執行

# 測試
chicken_rabbit_without_else(10, 28)  # 有解的情況
chicken_rabbit_without_else(10, 25)  # 無解的情況
```

#### 2. 使用 `else` 的版本

在這個版本中，我們使用 `for-else` 結構來簡化邏輯。

```python
def chicken_rabbit_with_else(heads, legs):
    for chickens in range(heads + 1):  # 假設雞的數量從 0 到 heads
        rabbits = heads - chickens  # 兔子的數量
        if (2 * chickens + 4 * rabbits) == legs:  # 檢查腿的數量
            print(f"雞: {chickens}, 兔子: {rabbits}")
            break  # 找到解，跳出迴圈
    else:
        print("沒有符合條件的解。")  # 只有在正常結束時執行

# 測試
chicken_rabbit_with_else(10, 28)  # 有解的情況
chicken_rabbit_with_else(10, 25)  # 無解的情況
```

### 說明

- **不使用 `else` 的版本**：
  - 使用一個標誌變數 `found` 來跟蹤是否找到解。
  - 在找到解後，更新 `found` 並使用 `break` 跳出迴圈。
  - 如果迴圈結束時 `found` 為 `False`，則列印無解的消息。

- **使用 `else` 的版本**：
  - 直接在 `for` 循環後使用 `else`，這樣當沒有遇到 `break` 時，自動執行 `else` 部分。
  - 這種方式使程式碼更簡潔，無需額外的變數來跟蹤狀態。


````python
def chicken_rabbit(h,l):
    result = [(c,h-c) for c in range(h+1) if c*2+(h-c)*4==l]
    if result:
        print(f"雞: {result[0][0]}, 兔子: {result[0][1]}")
    else:
        print("沒有符合條件的解。") 
        
chicken_rabbit(10, 28)  # 有解的情況
chicken_rabbit(10, 25)  # 無解的情況
````

### `continue` 的用法

`continue` 語句用於迴圈中，當它被執行時，會跳過當前的迭代，並直接進入下一次迭代。這意味著當 `continue` 被觸發後，當前迴圈內的剩餘代碼將被跳過，迴圈將繼續進行下一次迭代。

### `continue` 的工作原理

1. **位置**：`continue` 可以放在 `for` 迴圈和 `while` 迴圈中。
2. **跳過**：當 `continue` 被執行時，迴圈不會終止；相反，它將跳過當前迭代中的任何後續代碼，並返回迴圈的開頭進行下一次迭代。
3. **條件**：通常，`continue` 是在條件語句內使用，以決定是否跳過當前迭代的剩餘代碼。

#### 範例 1：跳過偶數

在這個範例中，我們將遍歷一個數字範圍，只列印奇數，跳過偶數。

```python
def print_odd_numbers(limit):
    for number in range(limit):
        if number % 2 == 0:  # 檢查數字是否為偶數
            continue  # 跳過偶數的剩餘代碼
        print(number)  # 如果是奇數則列印

# 測試
print_odd_numbers(10)
```

#### 範例 2：在 `while` 迴圈中使用 `continue`

```python
def count_to_ten_skip_fives():
    count = 0
    while count < 10:
        count += 1
        if count == 5:  # 跳過數字 5
            continue
        print(count)

# 測試
count_to_ten_skip_fives()
```
