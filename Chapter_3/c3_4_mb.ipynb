### 練習一：使用 break 終止 while 迴圈

**描述**：編寫一個程序，使用 `while` 迴圈來持續顯示數字，直到按下按鈕 A 為止，然後終止迴圈。

**要求**：
1. 使用 `while` 迴圈來顯示從 1 開始的數字。
2. 當按下按鈕 A 時，使用 `break` 終止迴圈。
3. 每顯示一個數字後，暫停 1 秒鐘。

**示例代碼**：
```python
from microbit import *

number = 1  # 初始化數字

while True:
    display.show(number)  # 顯示當前數字
    sleep(1000)  # 暫停 1 秒
    if button_a.is_pressed():  # 檢查按鈕 A 是否被按下
        break  # 終止迴圈
    number += 1  # 數字加 1
    if number>9:
        number = 1

display.clear()  # 清除顯示
```

### 程式說明：
- 初始化變數 `number` 為 1。
- 使用 `while True` 創建一個不斷運行的循環，並顯示當前的數字。
- 每次顯示數字後，使用 `sleep(1000)` 暫停 1 秒鐘。
- 當檢查到按鈕 A 被按下時，使用 `break` 終止迴圈。
- 當迴圈終止後，使用 `display.clear()` 清除顯示。

備注: 爲了要能夠偵測到按鈕的被按下, 您可能需要持續的按下按鈕

### 練習二：使用 break 終止 LED 燈顯示

**描述**：編寫一個程序，使用 `while` 迴圈和嵌套的 `for` 迴圈來點亮微型顯示屏上的 LED 燈，直到按下按鈕 A，然後終止顯示。

**要求**：
1. 使用 `while` 迴圈持續運行。
2. 使用兩個 `for` 迴圈遍歷行（y）和列（x），點亮 LED 燈。
3. 當按下按鈕 A 時，使用 `break` 終止顯示。

**示例代碼**：
```python
from microbit import *

while True:
    for y in range(5):  # 遍歷行
        for x in range(5):  # 遍歷列
            if button_a.is_pressed():  # 檢查按鈕 A 是否被按下
                break  # 終止內層迴圈
            display.set_pixel(x, y, 9)  # 點亮 LED 燈
            sleep(50)  # 暫停 50 毫秒
        else:
            continue  # 如果內層迴圈正常結束，則繼續外層迴圈
        break  # 如果內層迴圈因按鈕 A 被終止，則也終止外層迴圈

display.clear()  # 清除顯示
```

### 程式說明：
- 使用 `while True` 創建一個不斷運行的循環。
- 內層 `for` 迴圈遍歷 0 到 4 的行（y）和列（x），點亮 LED 燈。
- 如果檢查到按鈕 A 被按下，使用 `break` 終止內層迴圈。
- 使用 `else` 子句來確保內層迴圈正常結束時，繼續外層迴圈。如果內層迴圈是因為按鈕 A 被按下而終止，則使用第二個 `break` 終止外層迴圈。
- 當迴圈終止後，使用 `display.clear()` 清除顯示。

### 練習三：使用 continue 跳過對角線 LED 燈

**描述**：編寫一個程序，使用 `for` 迴圈和 `continue` 語句在微型顯示屏上點亮所有 LED 燈，但跳過對角線上的燈。

**要求**：
1. 使用兩個 `for` 迴圈，第一個用於行（y），第二個用於列（x）。
2. 當 `x` 等於 `y` 時，或者 x+y 等於 4，使用 `continue` 跳過該 LED 燈的點亮。
3. 其他 LED 燈照常點亮。

**示例代碼**：
```python
from microbit import *

while True:
    for y in range(5):  # 遍歷行
        for x in range(5):  # 遍歷列
            if x == y or x+y == 4:  # 檢查是否為對角線
                continue  # 跳過對角線上的 LED 燈
            display.set_pixel(x, y, 9)  # 點亮 LED 燈
            sleep(50)  # 暫停 50 毫秒
    sleep(2000)  # 暫停 2000 毫秒
    display.clear()  # 清除顯示
```

### 程式說明：
- 使用外層 `for` 迴圈遍歷 0 到 4 的行數（y），內層 `for` 迴圈遍歷 0 到 4 的列數（x）。
- 當 `x` 等於 `y` 或者 `x+y` 等於 4 時，表示當前的 LED 燈位於對角線上，則使用 `continue` 跳過該燈的點亮。
- 其他的 LED 燈會正常點亮，並在每次點亮後暫停 50 毫秒。
- 當所有點亮操作完成後暫停 2000 毫秒，然後使用 `display.clear()` 清除顯示。
