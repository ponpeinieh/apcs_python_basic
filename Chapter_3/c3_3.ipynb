### 迴圈概述

迴圈是程式設計中一個基本的概念，允許程式在特定條件下重複執行某一段代碼。這種重複執行的能力使得迴圈非常適合用於處理需要多次執行的任務。

#### 迴圈的類型

1. **`for` 迴圈**：
   - `for` 迴圈用於在固定次數內執行某段代碼。它特別適合用於遍歷序列（如清單、元組或字典）或生成特定範圍的數字。
   - 語法示例：
     ```python
     for x in range(5):  # 將執行五次
         print(x)
     ```
   - 在這個例子中，`range(5)` 生成從 0 到 4 的數字，`for` 迴圈會依次將這些數字賦值給變數 `x` 並執行相關的代碼。

2. **`while` 迴圈**：
   - `while` 迴圈根據給定的條件反覆執行代碼。當條件為真時，迴圈繼續執行；當條件變為假時，迴圈停止。
   - 語法示例：
     ```python
     count = 0
     while count < 5:  # 當 count 小於 5 時執行
         print(count)
         count += 1  # 更新條件
     ```
   - 在這個例子中，`while` 迴圈會持續執行，直到 `count` 的值不再小於 5。


### Range (範圍)物件

`range` 是 Python 中一個內建的範圍物件，用於生成一系列的整數。這些整數形成一個不可變的數字序列，這意味著一旦創建後，這個序列的內容就無法被修改。`range` 範圍物件在循環、數據處理和迭代等多種情況下都非常有用。

#### 如何使用 `range` 函式？

`range` 函式可以接受一到三個參數來定義生成的數字序列。

1. **單一整數參數**：
   - 當只傳遞一個整數時，`range` 會生成從 0 到該整數減 1 的數字序列。
   - 語法示例：
     ```python
     r = range(5)
     print(r) # 輸出 range(0, 5)
     print(list(r))  # 輸出: [0, 1, 2, 3, 4]
     ```
   - 在這個例子中，創建了一個範圍物件 `r`，它包含了 0 到 4 的整數。

2. **兩個整數參數**：
   - 當傳遞兩個整數時，第一個整數作為序列的起始值，第二個整數作為結束值（不包括該值）。
   - 語法示例：
     ```python
     r = range(2, 6)
     print(list(r))  # 輸出: [2, 3, 4, 5]
     ```
   - 這裡，範圍物件 `r` 從 2 開始，到 5 結束（不包括 6）。

3. **三個整數參數**：
   - 可以使用三個整數參數，第三個參數指定**步長(step)**，表示每次遞增的數量。
   - 語法示例：
     ```python
     r = range(1, 10, 2)
     print(list(r))  # 輸出: [1, 3, 5, 7, 9]
     ```
   - 在這個例子中，範圍物件 `r` 從 1 開始，以 2 為步長，生成到 9 的序列。

#### 列出範圍物件的內容

要查看範圍物件中的具體數字，可以使用內建的 `list` 函式將範圍物件轉換為清單。這樣可以方便地查看範圍內的所有數字。

**示例**：
```python
r = range(5)  # 創建範圍物件
print(list(r))  # 將範圍物件轉換為清單並輸出
```

### for 迴圈

`for` 迴圈是 Python 中用來重複執行某段程式碼的結構。它通常用於遍歷集合（如清單、元組、字典、集合或字串），並對集合中的每個元素執行相同的操作。這種迴圈形式非常靈活，適合處理需要對多個元素執行的任務。

#### 迴圈的特點

- **重複執行**：`for` 迴圈允許你指定執行某段程式碼的次數，根據集合的大小自動決定迴圈的次數。
- **語法結構**：
  ```python
  for 變數 in 集合:
      # 執行的程式碼
  ```
  在這裡，`變數` 將儲存每次從集合中取出的值，並在每次迴圈中執行程式碼。

#### 集合物件

集合可以是多種形式，包括：

1. **範圍物件**

     ```python
     for i in range(3, 8):  # 從 3 到 7
         print(i)

     for i in range(0, 10, 2):  # 每次遞增2
         print(i)
     ```
     
2. **序列**：如清單或元組。

     ```python
     fruits = ["蘋果", "香蕉", "橙子"]

     for fruit in fruits:
         print(fruit)
     ```

5. **字串**：字串中的每個字元都可以被遍歷。

     ```python
     for char in "Hello":
         print(char)
     ```

#### 迴圈的執行流程

1. **宣告變數**：在每次迴圈開始時，`for` 迴圈會自動將集合中的下一個元素賦值給宣告的變數。
2. **執行程式碼**：執行與該變數相關的程式碼區塊。
3. **重複**：直到所有元素都被遍歷完畢，迴圈才會結束。

### Zip 打包

在 Python 中，`zip()` 函式是一個強大的工具，用於將多個可迭代對象（如清單、元組等）中的元素進行打包。這意味著你可以將不同集合中對應索引的元素組合成一個新的可迭代對象，通常是由元組組成的清單。

#### 基本概念

- `zip()` 函式接受多個可迭代對象作為參數，並將這些對象中對應位置的元素打包成一個元組。最終結果是一個可迭代的 `zip` 物件，這個物件可以用於 `for` 迴圈等。

#### 語法示例

```python
subjs = ['國語', '數學', '英文']
scores = [100, 80, 95]

zipped = zip(subjs, scores)
```

在這個例子中，`subjs` 和 `scores` 兩個清單被打包在一起，對應位置的元素將組成元組。

#### 如何使用 Zip 打包的結果？

打包後的 `zipped` 是一個可迭代(`iterable`)的物件，你可以使用 `for` 迴圈來遍歷這些元組，從而同時取出來自不同集合的元素。

```python
for subj, score in zipped:
    print(f"{subj} 的分數是: {score}")
```

### 輸出示例

這段程式碼的輸出會是：

```
國語 的分數是: 100
數學 的分數是: 80
英文 的分數是: 95
```
