````python
import base64
from IPython.display import Image, display
import matplotlib.pyplot as plt

def mm(graph):
    graphbytes = graph.encode("utf8")
    base64_bytes = base64.urlsafe_b64encode(graphbytes)
    base64_string = base64_bytes.decode("ascii")
    display(Image(url="https://mermaid.ink/img/" + base64_string))
````

### 1. `if` 語句

基本的 `if` 語句用於檢查一個條件，如果條件為 `True`，則執行一段代碼。

**流程圖：**

```mermaid
flowchart TD
    A[開始] --> B{檢查條件}
    B -- 是 --> C[執行代碼塊1]
    C --> D[結束]
    B -- 否 --> D[結束]
```


````python
mm("""
flowchart TD
    A[開始] --> B{檢查條件}
    B -- 是 --> C[執行代碼塊1]
    C --> D[結束]
    B -- 否 --> D[結束]""")
````


**語法：**

```python
if 條件:
    # 如果條件為 True，執行的代碼
```

**範例：**

```python
age = 18

if age >= 18:
    print("你是成年人。")  # 輸出: 你是成年人。
print("完成!")            # 輸出: 完成!
```

### 2. `if-else` 語句

`if-else` 語句提供了一個替代的代碼塊，如果條件為 `False`，則執行該代碼塊。

**流程圖：**

```mermaid
flowchart TD
    A[開始] --> B{檢查條件}
    B -- 是 --> C[執行代碼塊1]
    B -- 否 --> D[執行代碼塊2]
    C --> E[結束]
    D --> E
```

````python
mm("""
flowchart TD
    A[開始] --> B{檢查條件}
    B -- 是 --> C[執行代碼塊1]
    B -- 否 --> D[執行代碼塊2]
    C --> E[結束]
    D --> E""")
````

**語法：**

```python
if 條件:
    # 如果條件為 True，執行的代碼
else:
    # 如果條件為 False，執行的代碼
```

**範例：**

```python
age = 16

if age >= 18:
    print("你是成年人。")
else:
    print("你是未成年人。")  # 輸出: 你是未成年人。
print("完成!")              # 輸出: 完成!
```

### 3. `if-elif-else` 語句

`if-elif-else` 語句允許檢查多個條件。它代表 "if"、"else if" 和 "else"。你可以有多個 `elif` 塊來檢查額外的條件。

**流程圖：**

```mermaid
flowchart TD
    A[開始] --> B{檢查條件1}
    B -- 是 --> C[執行代碼塊1]
    B -- 否 --> D{檢查條件2}
    D -- 是 --> E[執行代碼塊2]
    D -- 否 --> F{檢查條件3}
    F -- 是 --> G[執行代碼塊3]
    F -- 否 --> H[執行代碼塊4]
    C --> I[結束]
    E --> I
    G --> I
    H --> I
```

````python
mm("""
flowchart TD
    A[開始] --> B{檢查條件1}
    B -- 是 --> C[執行代碼塊1]
    B -- 否 --> D{檢查條件2}
    D -- 是 --> E[執行代碼塊2]
    D -- 否 --> F{檢查條件3}
    F -- 是 --> G[執行代碼塊3]
    F -- 否 --> H[執行代碼塊4]
    C --> I[結束]
    E --> I
    G --> I
    H --> I""")
````



**語法：**

```python
if 條件1:
    # 如果條件1為 True，執行的代碼
elif 條件2:
    # 如果條件2為 True，執行的代碼
elif 條件3:
    # 如果條件3為 True，執行的代碼
else:
    # 如果以上條件都不為 True，執行的代碼
```

**範例：**

```python
score = 85

if score >= 90:
    print("成績優異！")  # 輸出: 成績優異！
elif score >= 75:
    print("成績良好。")  # 輸出: 成績良好。
elif score >= 60:
    print("成績及格。")
else:
    print("成績不及格。")
print("完成!")            # 輸出: 完成!
```

---

### 主要要點

- **縮排**：Python 使用縮排來定義每個條件語句所屬的代碼塊。確保保持一致的縮排。

- **布林表達式**：條件可以包含各種比較（如 `==`、`!=`、`<`、`>` 等）並可以使用邏輯運算符（`and`、`or`、`not`）來組合多個條件。

- **單行語句**：對於簡單條件，可以將 `if` 語句寫成單行。

```python
if age >= 18: print("你是成年人。")
```

- **嵌套條件**：可以在其他 `if` 語句內部嵌套 `if` 語句，以創建更複雜的條件。


```mermaid

    C --> D{檢查 number % 2 == 0}

    D -- 否 --> F[數字是奇數。]
    E --> G[結束]
    F --> G
    B -- 否 --> H[數字是非正數。]
    H --> G
```

### 嵌套條件的範例

```python
number = 10

if number > 0:
    print("數字是正數。")
    if number % 2 == 0:
        print("數字是偶數。")
    else:
        print("數字是奇數。")
else:
    print("數字是非正數。")
print("完成!") 
```

### BMI（身體質量指數）範例

```python
def calculate_bmi(weight, height):
    # 計算BMI
    bmi = weight / (height ** 2)
    return bmi

def categorize_bmi(bmi):
    # 使用if、elif和else來分類BMI
    if bmi < 18.5:
        return "體重過輕"  # Underweight
    elif 18.5 <= bmi < 25:
        return "正常範圍"  # Normal weight
    elif 25 <= bmi < 30:
        return "過重"  # Overweight
    else:
        return "肥胖"  # Obesity

# 示例用法
weight = float(input("請輸入體重（公斤）："))
height = float(input("請輸入身高（米）："))

bmi = calculate_bmi(weight, height)
category = categorize_bmi(bmi)

print(f"您的BMI是: {bmi:.2f}，分類為: {category}")
```

### 代碼解釋

1. **函數 `calculate_bmi`**:
   - 此函數接受 `weight`（體重，單位：公斤）和 `height`（身高，單位：米）作為輸入，並使用公式計算BMI：
     $$
     \text{BMI} = \frac{\text{weight}}{\text{height}^2}
     $$

2. **函數 `categorize_bmi`**:
   - 此函數接受計算出的BMI，並使用 `if`、`elif` 和 `else` 語句來對BMI進行分類：
     - **體重過輕**: BMI < 18.5
     - **正常範圍**: 18.5 ≤ BMI < 25
     - **超重**: 25 ≤ BMI < 30
     - **肥胖**: BMI ≥ 30

3. **示例用法**:
   - 提示用戶輸入體重和身高。
   - 計算BMI並進行分類。
   - 最後，將結果以格式化字符串的形式輸出。
