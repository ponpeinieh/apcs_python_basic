### Python 的分支結構

在 Python 中，分支結構通常使用 `if`、`elif` 和 `else` 來控制程式的執行流程。其語法如下：

```python
if condition:
    # 當 condition 為 True 時執行的程式碼
elif another_condition: # else if
    # 當 another_condition 為 True 時執行的程式碼
else:
    # 當以上條件都不符合時執行的程式碼
```

#### 縮排的重要性

- **不使用大括號**：與許多其他程式語言不同，Python 不使用 `{}` 來定義區塊範圍。相反，Python 使用縮排來表示程式碼的區塊。

- **顯示區塊範圍**：每個 `if`、`elif` 和 `else` 區塊內的程式碼必須使用相同的縮排，這樣 Python 才能識別哪些程式碼屬於同一個區塊。

### Python 的迴圈結構

Python 支援兩種主要的迴圈結構：`for` 迴圈和 `while` 迴圈。

#### `for` 迴圈

```python
for item in iterable:
    # 對 iterable 中的每個 item 執行的程式碼
```

#### `while` 迴圈

```python
while condition:
    # 當 condition 為 True 時執行的程式碼
```

### 縮排的一致性

- **縮排方式必須一致**：在 Python 中，所有相同區塊範圍的程式碼必須使用相同的縮排方式。例如，如果你在一個區塊中使用了四個空白字元來縮排，則在相同區塊的其他程式碼行中也必須使用四個空白字元。

- **空白與 Tab 的區別**：使用四個空白字元的縮排和使用一個 Tab 鍵的縮排是不同的。這可能導致 `IndentationError`，因為 Python 對縮排的要求非常嚴格。

### 程式中的條件控制

在程式中，根據不同的資料或條件，常常需要採取不同的行動。這通常是通過條件語句來實現的，最常見的就是 `if` 語句。

### 布林值 (Boolean)

- **布林值** 是一種資料型態，只有兩個可能的值：
  - `True`：表示真。
  - `False`：表示假。

在 Python 中，你可以這樣使用布林值：

```python
is_active = True
is_completed = False
```

### 關係運算

關係運算用於比較兩個值，並返回一個布林值。以下是一些常用的關係運算子：

| 運算子 | 意義          | 範例             |
|--------|---------------|------------------|
| `<`    | 小於         | `5 < 10`  (True) |
| `<=`   | 小於等於     | `5 <= 5` (True)  |
| `>`    | 大於         | `10 > 5` (True)  |
| `>=`   | 大於等於     | `5 >= 10` (False)|
| `==`   | 兩邊是否等於 | `5 == 5` (True)  |
| `!=`   | 兩邊是否不等於 | `5 != 10` (True)  |
| `in`     | 包含        | `3 in [1, 2, 3]` (True) |
| `not in` | 不包含      | `4 not in [1, 2, 3]` (True) |

例如：

```python
a = 10
b = 5
print(a > b)  # 輸出: True
print(a == b)  # 輸出: False
print(3 in [1, 2, 3])  # 輸出: True
```

### 邏輯運算

邏輯運算用於將多個布林值結合在一起，返回一個新的布林值。常見的邏輯運算子包括：

- `and`：當所有條件都為 `True` 時，結果為 `True`。
- `or`：只要有一個條件為 `True`，結果就為 `True`。
- `not`：對布林值取反，`True` 變成 `False`，反之亦然。

#### 使用邏輯運算的範例

以下是使用關係運算子和邏輯運算子的範例，檢查一個數字是否在兩個數字之間：

```python
number = 49
lower_bound = 1
upper_bound = 100

# 注意：這裡使用了較常見的語法，但是 Python 支援更簡單的寫法 lower_bound < number < upper_bound
if lower_bound < number and number < upper_bound:
    print("數字在範圍內")  # 輸出: 數字在範圍內
else:
    print("數字不在範圍內")
```

類似地, 我們可以檢查一個數字是否**不在**兩個數字之間：

```python
number = -10
lower_bound = 1
upper_bound = 100

if number < lower_bound or number > upper_bound: # or use 'not' -> not (lower_bound < number < upper_bound)
    print("數字不在範圍內")  
else:
    print("數字在範圍內")
```

### 物件類別關係運算子

這些運算子用於檢查兩個物件是否為同一個物件，或是否不是同一個物件。

| 運算子   | 意義                 | 範例                     |
|----------|----------------------|--------------------------|
| `is`     | 是否為同一個物件    | `a is b` (True 如果 a 和 b 指向同一個物件) |
| `is not` | 是否非同一個物件    | `a is not b` (True 如果 a 和 b 不指向同一個物件) |

#### 使用 `is` 和 `is not` 的範例

```python
a = [1, 2, 3]
b = a  # b 是 a 的參考
c = a[:]  # c 是 a 的淺拷貝

print(a is b)      # 輸出: True
print(a is c)      # 輸出: False
print(a is not c)  # 輸出: True

# 使用 id() 函數檢查物件的 ID
print("a 的 ID:", id(a))
print("b 的 ID:", id(b))
print("c 的 ID:", id(c))

```
