Python 不支持像 Java 的 `標籤迴圈`。在 Python 中，你不能標記一個迴圈並直接從內部迴圈跳出外部迴圈。相反，通常會使用 `標誌變數` 或重構代碼來實現類似的功能。

### Python 示例：

這裡是如何使用標誌變數來模擬跳出外部迴圈：

```python
from microbit import *

break_outer_loop = False

while not break_outer_loop:
    for y in range(5):  # 遍歷行
        for x in range(5):  # 遍歷列
            if button_a.is_pressed():  # 檢查按鈕 A 是否被按下
                break_outer_loop = True  # 設置標誌以終止外層迴圈
                break  # 終止內層迴圈
            display.set_pixel(x, y, 9)  # 點亮 LED 燈
            sleep(50)  # 暫停 50 毫秒
        if break_outer_loop:
            break  # 也終止外層迴圈

display.clear()  # 清除顯示
```

### 程式說明：
- 使用 `break_outer_loop` 標誌來控制外層迴圈的運行。
- 當按鈕 A 被按下時，設置 `break_outer_loop` 為 `True`，並使用 `break` 來終止內層迴圈。
- 在外層迴圈中，檢查這個標誌，並根據需要終止外層迴圈。

這種方法可以有效地模擬 Java 的標籤迴圈行為，但需要額外的邏輯。 

你可以在 Python 中使用 `else` 子句與迴圈搭配，以達成類似的效果，而不需要使用標誌變數。`else` 區塊僅在迴圈正常完成時執行（即未被 `break` 語句終止）。

### 使用 `for` 迴圈和 `else` 的 Python 示例：

以下是實現的方法：

```python
from microbit import *

while True:
    for y in range(5):  # 遍歷行
        for x in range(5):  # 遍歷列
            if button_a.is_pressed():  # 檢查按鈕 A 是否被按下
                break  # 終止內層迴圈
            display.set_pixel(x, y, 9)  # 點亮 LED 燈
            sleep(50)  # 暫停 50 毫秒
        else:
            continue  # 如果內層迴圈正常結束，繼續外層迴圈
        break  # 如果內層迴圈是因為按鈕 A 被按下而終止，則終止外層迴圈

display.clear()  # 清除顯示
```

### 程式說明：
- 使用 `while True` 創建一個無限循環。
- 內層 `for` 迴圈遍歷 0 到 4 的行（y）和列（x），點亮 LED 燈。
- 當檢查到按鈕 A 被按下時，使用 `break` 終止內層迴圈。
- `else` 子句將在內層 `for` 迴圈正常完成後執行，如果內層迴圈被 `break` 終止，則不會執行 `else`。
- 在外層迴圈中，檢查內層迴圈是否正常結束。如果是，則使用 `continue` 繼續外層迴圈；如果不是，則 `break` 終止外層迴圈。

這種方式可以有效地避免使用標誌變數，並利用 `else` 結構來控制流程。
