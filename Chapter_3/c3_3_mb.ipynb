### 練習一：使用 for 迴圈和 range() 控制 LED 顯示

**描述**：編寫一個程序，使用 `for` 迴圈和 `range()` 來控制微型顯示屏上的 LED 燈。顯示 1 到 5 的數字，每次顯示後暫停一段時間。

**要求**：
1. 使用 `for` 迴圈遍歷從 1 到 5 的數字。
2. 在顯示屏上顯示每個數字。
3. 每顯示一個數字後，暫停 1 秒鐘。

**示例代碼**：
```python
from microbit import *

while True:
    for number in range(1, 6):  # 從 1 到 5
        display.show(number)  # 顯示數字
        sleep(1000)  # 暫停 1 秒
``` 

### 練習二：使用嵌套 for 迴圈點亮 LED 燈

**描述**：編寫一個程序，使用嵌套的 `for` 迴圈在微型顯示屏上點亮所有的 LED 燈，形成一個亮點的效果。

**要求**：
1. 使用兩個 `for` 迴圈，第一個用於行（y），第二個用於列（x）。
2. 使用 `display.set_pixel(x, y, 9)` 來點亮每個 LED 燈。
3. 每點亮一個燈後，暫停 50 毫秒。

**示例代碼**：
```python
from microbit import *

while True:
    for y in range(5):  # 遍歷行
        for x in range(5):  # 遍歷列
            display.set_pixel(x, y, 9)  # 點亮 LED 燈
            sleep(50)  # 暫停 50 毫秒
    display.clear()  # 清除顯示
```

### 程式說明：
- 使用外層 `for` 迴圈遍歷 0 到 4 的行數（y），內層 `for` 迴圈遍歷 0 到 4 的列數（x）。
- `display.set_pixel(x, y, 9)` 用於將指定位置的 LED 燈點亮，亮度為 9。
- 在每次點亮一個燈後，使用 `sleep(50)` 暫停 50 毫秒，以便形成逐個亮燈的效果。
- 使用 `display.clear()` 清除顯示，在每次循環結束後清空顯示屏。 

### 練習三：使用嵌套 for 迴圈點亮 LED 燈並清除顯示

**描述**：編寫一個程序，使用嵌套的 `for` 迴圈在微型顯示屏上點亮所有的 LED 燈，並在每次點亮後清除顯示。

**要求**：
1. 使用兩個 `for` 迴圈，第一個用於行（y），第二個用於列（x）。
2. 使用 `display.set_pixel(x, y, 9)` 來點亮每個 LED 燈。
3. 每點亮一個燈後，暫停 50 毫秒，然後清除顯示。

**示例代碼**：
```python
from microbit import *

while True:
    for y in range(5):  # 遍歷行
        for x in range(5):  # 遍歷列
            display.set_pixel(x, y, 9)  # 點亮 LED 燈
            sleep(50)  # 暫停 50 毫秒
            display.clear()  # 清除顯示
```

### 程式說明：
- 使用外層 `for` 迴圈遍歷 0 到 4 的行數（y），內層 `for` 迴圈遍歷 0 到 4 的列數（x）。
- `display.set_pixel(x, y, 9)` 用於將指定位置的 LED 燈點亮，亮度為 9。
- 在每次點亮一個燈後，使用 `sleep(50)` 暫停 50 毫秒，然後立即使用 `display.clear()` 清除顯示。

### 與先前練習的區別：
- 在這個練習中，每次點亮一個 LED 燈後，會立即清除顯示，這樣 LED 燈的亮點效果不會持續，而是每次只顯示一個燈。
- 相較於之前的練習，LED 燈在點亮後會保持亮著一段時間，而這個練習則是每次只顯示一個燈，然後清空顯示，讓效果變得不同。
