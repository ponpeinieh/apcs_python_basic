### 練習題 1：計算第 n 個斐波那契數

**描述**：編寫一個程式，計算並輸出第 n 個斐波那契數。請注意，斐波那契數列的定義如下：

- 第 1 個數是 0
- 第 2 個數是 1
- 第 3 個數是 1
- 第 4 個數是 2
- 第 5 個數是 3
- 第 n 個數是前兩個數的和，即 `F(n) = F(n-1) + F(n-2)`

**要求**：
1. 使用 `input` 讀取一個整數 n。
2. 使用 `for` 迴圈計算第 n 個斐波那契數。
3. 輸出第 n 個斐波那契數。


**解答範例**： 
```python
# 計算第 n 個斐波那契數的程式
n = int(input("請輸入一個正整數 n："))

if n < 1:
    print("請輸入正整數")
elif n == 1:
    print("第 1 個斐波那契數是: 0")
elif n == 2:
    print("第 2 個斐波那契數是: 1")
else:
    a, b = 0, 1
    for _ in range(2, n):
        a, b = b, a + b  # 更新斐波那契數
    print(f"第 {n} 個斐波那契數是: {b}")
```

### 練習題 2：猜數字遊戲

**描述**：建立一個互動的猜數字遊戲，玩家需要在最多 5 次的猜測中找出隨機生成的數字，並根據猜測結果提供最小和最大值範圍的提示。

**要求**：
1. 隨機生成一個介於 1 到 100 的整數作為答案。
2. 在每次猜測後，根據玩家所猜的數字，提示更新最小和最大值範圍，格式為：
   - 如果猜的數字太低，顯示訊息：`"請猜一個介於 {min} 和 {max} 的數字，因為 {guess} 太低了！"`
   - 如果猜的數字太高，顯示訊息：`"請猜一個介於 {min} 和 {max} 的數字，因為 {guess} 太高了！"`
3. 當玩家的猜測次數達到 5 次時，顯示所有猜測過的數字及正確答案，格式為：`"你所有的猜測：{guesses}，正確答案是 {answer}！"`

### 範例案例
- 假設隨機生成的正確答案為 `77`。
- 玩家第一次猜測 `89`。

**程式輸出**：
```plaintext
請猜一個介於 1 和 88 的數字，因為 89 太高了！
```

### 完整的程式碼範例
```python
import random

# 隨機生成答案
answer = random.randint(1, 100)  # 假設答案是77
min_value = 1
max_value = 100
guesses = []

print("歡迎來到猜數字遊戲！你有 5 次機會來猜出隨機生成的數字。")

for attempt in range(5):
    guess = int(input(f"請猜一個介於 {min_value} 和 {max_value} 的數字："))
    guesses.append(guess)

    if guess < answer:
        min_value = guess + 1
        print(f"請猜一個介於 {min_value} 和 {max_value} 的數字，因為 {guess} 太低了！")
    elif guess > answer:
        max_value = guess - 1
        print(f"請猜一個介於 {min_value} 和 {max_value} 的數字，因為 {guess} 太高了！")
    else:
        print(f"恭喜！你猜對了，數字是 {answer}！")
        break
else:
    print(f"你所有的猜測：{guesses}，正確答案是 {answer}！")
```

### 說明：
1. 當玩家猜測 `89` 時，因為 `89` 大於 `77`，程式會更新最大值為 `88`，並提示玩家下一次猜測的範圍。
2. 隨著每次猜測，程序將根據玩家的輸入不斷更新範圍，直到猜對或用完次數。

這樣的設計能夠引導玩家逐步接近正確答案。如果有其他需求或修改，請告訴我！

### 練習題 3：擴展猜數字遊戲

**描述**：在猜數字遊戲中，增加難度選項，讓玩家選擇「簡單」、「一般」或「困難」三個難度級別，並根據選擇的級別生成不同範圍的隨機數字和限制猜測次數。

**要求**：
1. 提供三個難度選項：「簡單」（1-100，最多 6 次）、 「一般」（1-1000，最多 8 次）、 「困難」（1-10000，最多 11 次）。
2. 根據玩家選擇的難度級別生成相應範圍的隨機數字。
3. 在每次猜測後顯示當前的最小和最大範圍，以及提示訊息，格式為：
   - 如果猜的數字太低，顯示訊息：`"請猜一個介於 {min} 和 {max} 的數字，因為 {guess} 太低了！"`
   - 如果猜的數字太高，顯示訊息：`"請猜一個介於 {min} 和 {max} 的數字，因為 {guess} 太高了！"`

### 解答範例：
```python
import random

# 選擇難度
level = int(input("請選擇難度 - 1. 簡單, 2. 一般, 3. 困難: "))
if level == 1:
    min_value, max_value = 1, 100
    max_attempts = 6
elif level == 2:
    min_value, max_value = 1, 1000
    max_attempts = 8
elif level == 3:
    min_value, max_value = 1, 10000
    max_attempts = 11
else:
    print("無效的難度選擇，預設為簡單模式。")
    min_value, max_value = 1, 100
    max_attempts = 6

answer = random.randint(min_value, max_value)
print(f'歡迎來到猜數字遊戲, 答案介於 {min_value} 和 {max_value} 之間，最多可猜 {max_attempts} 次。')

count = 0
guessed_numbers = []  # 用來儲存猜的數字

while count < max_attempts:
    guess = int(input(f'請猜一個介於 {min_value} 和 {max_value} 的整數: '))
    guessed_numbers.append(guess)  # 儲存猜的數字
    count += 1

    if guess == answer:
        print('恭喜您猜對了!')
        break
    elif guess < answer:
        min_value = guess + 1
        print(f"請猜一個介於 {min_value} 和 {max_value} 的數字，因為 {guess} 太低了！")
    elif guess > answer:
        max_value = guess - 1
        print(f"請猜一個介於 {min_value} 和 {max_value} 的數字，因為 {guess} 太高了！")
else:
    print(f'您已經猜了 {max_attempts} 次了, 還沒猜對! 正確答案為 {answer}')
    print(f'您猜的數字是: {guessed_numbers}')
```

### 練習題 4：尋找第一個偶數

**描述**：編寫一個程式，從一組整數中尋找第一個偶數。如果找到偶數，則輸出該偶數；如果所有數字都是奇數，則輸出「沒有找到偶數」。

**要求**：
1. 使用 `input` 讀取一行數字，數字之間用逗號分隔。
2. 將讀取的字串分割並轉換為整數清單。
3. 使用 `for` 或 `while` 迴圈遍歷數字清單。
4. 使用 `else` 語句來處理當迴圈正常結束（即沒有 `break`）的情況。

**解答範例**：
```python
# 尋找第一個偶數的程式
input_numbers = input("請輸入一組整數（用逗號分隔）: ")
# 將輸入的字符串分割並轉換為整數清單
numbers = [int(num.strip()) for num in input_numbers.split(',')]

for num in numbers:
    if num % 2 == 0:
        print(f"找到的第一個偶數是: {num}")
        break
else:
    # 只有當沒有執行 break 時才會執行這裡
    print("沒有找到偶數")
```
