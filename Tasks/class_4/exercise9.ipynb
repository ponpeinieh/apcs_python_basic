### **Exercise 1: Basic Lambda Syntax**  
**Task:** Create simple lambda functions to understand the syntax.

#### **Instructions:**  
1. Convert these regular functions to lambda expressions:  
   - A function that adds two numbers  
   - A function that squares a number  
   - A function that checks if a number is even  

```python
# Regular function
def add(x, y):
    return x + y

# Lambda version
add_lambda = lambda x, y: x + y

print(add(3, 5))         # 8
print(add_lambda(3, 5))  # 8

# Now create these lambdas:
# 1. Square lambda (takes n, returns n²)
# 2. Even-check lambda (takes n, returns True/False)
```
