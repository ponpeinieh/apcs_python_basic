### **練習一：使用 `*args` 的數學計算器**  
**任務：** 創建能處理任意數量輸入的計算函式。  

#### **要求：**  
1. 定義 `calculate_stats()`：  
   - 使用 `*args` 接受任意數量的值  
   - 返回包含以下值的元組：  
     - 總和  
     - 平均值  
     - 最大值  
     - 最小值  
2. 示範：  
   - 3個數字  
   - 5個數字  

#### **範例輸出：**  
```
輸入: (5, 10, 15)  
總和: 30, 平均: 10.0, 最大: 15, 最小: 5  

輸入: (2, 4, 6, 8, 10)  
總和: 30, 平均: 6.0, 最大: 10, 最小: 2  
```

---

### **Sample Solution Code**  
```python
# English Version
def calculate_stats(*args):
    total = sum(args)
    average = total / len(args)
    return total, average, max(args), min(args)

# Demonstrations
print("Input: (5, 10, 15)")
sum_result, avg_result, max_result, min_result = calculate_stats(5, 10, 15)
print(f"Sum: {sum_result}, Avg: {avg_result}, Max: {max_result}, Min: {min_result}\n")

print("Input: (2, 4, 6, 8, 10)")
stats = calculate_stats(2, 4, 6, 8, 10)
print(f"Sum: {stats[0]}, Avg: {stats[1]}, Max: {stats[2]}, Min: {stats[3]}")

# 繁體中文版

    total = sum(args)
    average = total / len(args)
    return total, average, max(args), min(args)

# 示範

```

#### **要求：**  
1. 定義 `build_sentence(*words)` 功能：  
   - 用空格連接詞語  
   - 首字母大寫  
   - 添加句號（如果末尾沒有標點）  
2. 用2-4個詞語測試  

#### **範例輸出：**  
```
輸入: ("你好", "世界") → "你好 世界。"  
輸入: ("今天", "天氣", "很好") → "今天 天氣 很好。"  
```

#### **Exercise Solution**  
```python 

def build_sentence(*words):
    sentence = " ".join(words).capitalize()
    return sentence + "." if not sentence.endswith(".") else sentence

# Demo
print(f'''Input: ("hello", "world") → "{build_sentence("hello", "world")}"''')
 
```

```
輸入: ("只有一個 {}", "佔位符")

```

### **Sample Solution Code (without using exception handling)**  
```python
# English Version
def dynamic_format(fmt_str, *args):
    required = fmt_str.count('{}') # calculates count of placeholders
    if len(args) == required:
        return fmt_str.format(*args)
    return f"Error: Needs {required} arguments, got {len(args)}"

# Demonstrations
print('Input: ("Hello {}!", "Alice")')
print(f'Output: "{dynamic_format("Hello {}!", "Alice")}"')  # Success

print('\nInput: ("Hello {}!", )')
print(f'Output: "{dynamic_format("Hello {}!", )}"')  # Error

# 繁體中文版
def dynamic_format(fmt_str, *args):
    required = fmt_str.count('{}')
    if len(args) == required:
        return fmt_str.format(*args)
    return f"錯誤: 需要 {required} 個參數，但得到 {len(args)} 個"

print('\n輸入: ("你好 {}！", "小明")')
print(f'輸出: "{dynamic_format("你好 {}！", "小明")}"')  # 成功

print('\n輸入: ("你好 {}！", )')
print(f'輸出: "{dynamic_format("你好 {}！", )}"')  # 錯誤
```

```python
def dynamic_format(fmt_str, *args):

# Demonstrations
print('Input: ("Hello {}! Today is {}", "Alice", "Monday")')
print(f'Output: "{dynamic_format("Hello {}! Today is {}", "Alice", "Monday")}"\n')

print('Input: ("Only one {}", "placeholder")')

print('Input: ("Missing {}", )')

print('\n輸入: ("你好 {}！今天是{}", "小明", "星期一")')
```

### Python 函數參數：帶預設值的參數與 `*args` 的位置關係

在 Python 函數定義中，帶有預設值的參數（keyword arguments）和 `*args`（可變位置參數）的位置關係確實比較靈活，但有一些規則需要注意。

#### 基本規則

1. **帶預設值的參數可以放在 `*args` 前面**：
   ```python
   def func(a, b=2, *args):
       print(a, b, args)
   
   func(1)            # 輸出: 1 2 ()
   func(1, 3)         # 輸出: 1 3 ()
   func(1, 3, 4, 5)   # 輸出: 1 3 (4, 5)
   ```

2. **帶預設值的參數也可以放在 `*args` 後面**：
   ```python
   def func(a, *args, b=2):
       print(a, args, b)
   
   func(1)            # 輸出: 1 () 2
   func(1, 3, 4)      # 輸出: 1 (3, 4) 2
   func(1, 3, 4, b=5) # 輸出: 1 (3, 4) 5
   ```

#### 重要區別

這兩種方式的關鍵區別在於**如何傳遞帶預設值的參數**：

- 當預設參數在 `*args` 前時，可以透過位置或關鍵字傳遞
- 當預設參數在 `*args` 後時，**必須**使用關鍵字參數傳遞 

#### 實際應用建議

1. **推薦將帶預設值的參數放在 `*args` 後面**：
   - 這樣更明確，呼叫者必須使用關鍵字參數指定這些參數
   - 避免參數傳遞時的歧義

2. **混合使用時的完整參數順序**：
   ```
   def func(positional_args, *args, keyword_args=default, **kwargs)
   ```

#### 範例對比

```python
# 預設參數在前
def func1(a, b=2, *args):
    print(a, b, args)

# 預設參數在後
def func2(a, *args, b=2):
    print(a, args, b)

# 呼叫對比
func1(1, 3, 4, 5)    # b透過位置參數設置: 1 3 (4, 5)
func2(1, 3, 4, 5)    # b使用預設值: 1 (3, 4, 5) 2
func2(1, 3, 4, b=5)  # b必須用關鍵字: 1 (3, 4) 5
```
