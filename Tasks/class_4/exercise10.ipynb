### **Exercise 1: Call Stack Demonstration**  
**Task:** Visualize the call stack behavior in function execution.

#### **Part 1: Nested Function Calls**
```python
def function_a():
    print("Entering A")
    function_b()
    print("Leaving A")

def function_b():
    print("Entering B")
    function_c()
    print("Leaving B")

def function_c():
    print("Entering C")
    print("Leaving C")

print("Call Stack Growth:")
function_a()
```

**Expected Output:**
```
Call Stack Growth:
Entering A
Entering B
Entering C
Leaving C
Leaving B
Leaving A
```

**Visualization:**  
```
Call Stack:
[MAIN]
[MAIN, A]
[MAIN, A, B]
[MAIN, A, B, C]  ← Peak
[MAIN, A, B]
[MAIN, A]
[MAIN]
```

---

### **Part 2: Recursive Call Stack**
```python
def recursive_countdown(n):
    print(f"Pushing stack frame for n={n}")
    if n <= 0:  # Base case
        print("Base case reached!")
    else:
        recursive_countdown(n-1)
    print(f"Popping stack frame for n={n}")

print("\nRecursive Stack Behavior:")
recursive_countdown(3)
```

**Expected Output:**
```
Recursive Stack Behavior:
Pushing stack frame for n=3
Pushing stack frame for n=2
Pushing stack frame for n=1
Pushing stack frame for n=0
Base case reached!
Popping stack frame for n=0
Popping stack frame for n=1
Popping stack frame for n=2
Popping stack frame for n=3
```

**Stack Memory Illustration:**  
```
n=3 → n=2 → n=1 → n=0 (Base case)
Then unwinds: n=0 ← n=1 ← n=2 ← n=3
```

### **Stack Overflow Demonstration**
```python
def cause_overflow():
    cause_overflow()  # Infinite recursion

print("\nStack Overflow Example:")
# Uncomment to crash (runs until stack memory exhausted)
# cause_overflow()
```

**Key Observations:**  
1. Each call consumes stack memory  
2. Python defaults to ~1000 recursion depth

### **練習一：呼叫堆疊演示**  
**任務：** 觀察函數執行時的呼叫堆疊行為。

#### **第一部分：嵌套函數呼叫**
```python
def function_a():
    print("進入 A 層堆疊")
    function_b()
    print("離開 A 層堆疊")

def function_b():
    print("進入 B 層堆疊")
    function_c()
    print("離開 B 層堆疊")

def function_c():
    print("進入 C 層堆疊")
    print("離開 C 層堆疊")

print("呼叫堆疊增長過程:")
function_a()
```

**預期輸出：**
```
呼叫堆疊增長過程:
進入 A 層堆疊
進入 B 層堆疊
進入 C 層堆疊
離開 C 層堆疊
離開 B 層堆疊
離開 A 層堆疊
```

**堆疊可視化：**  
```
[主程式]
[主程式, A]
[主程式, A, B]
[主程式, A, B, C]  ← 最高點
[主程式, A, B]
[主程式, A]
[主程式]
```

---

### **第二部分：遞歸呼叫堆疊**
```python
def recursive_countdown(n):
    print(f"壓入堆疊框 n={n}")
    if n <= 0:  # 終止條件
        print("到達終止條件!")
    else:
        recursive_countdown(n-1)
    print(f"彈出堆疊框 n={n}")

print("\n遞歸堆疊行為:")
recursive_countdown(3)
```

**預期輸出：**
```
遞歸堆疊行為:
壓入堆疊框 n=3
壓入堆疊框 n=2
壓入堆疊框 n=1
壓入堆疊框 n=0
到達終止條件!
彈出堆疊框 n=0
彈出堆疊框 n=1
彈出堆疊框 n=2
彈出堆疊框 n=3
```

### **堆疊溢出演示**
```python
def cause_overflow():
    cause_overflow()  # 無限遞歸

print("\n堆疊溢出範例:")
# 取消註解會導致程式崩潰
# cause_overflow()
```

**重點觀察：**
