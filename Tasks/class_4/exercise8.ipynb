### **Sample Solution Code**  
```python
def circle_area(radius):
    """Calculate the area of a circle.
    
    Args:
        radius (float): Radius of the circle in meters
        
    Returns:
        float: Area in square meters
        
    Raises:
        ValueError: If radius is negative
        
    Examples:
        >>> circle_area(3)
        28.274333882308138
        >>> circle_area(-1)
        Traceback (most recent call last):
        ...
        ValueError: Radius cannot be negative
    """
    if radius < 0:
        raise ValueError("Radius cannot be negative")
    return 3.141592653589793 * radius ** 2

def factorial(n):
    """Calculate factorial of a non-negative integer.
    
    Args:
        n (int): Number to calculate
        
    Returns:
        int: Factorial value
        
    Raises:
        ValueError: If n is negative
        
    Examples:
        >>> factorial(5)
        120
        >>> factorial(0)
        1
    """
    if n < 0:
        raise ValueError("Factorial undefined for negative numbers")
    return 1 if n <= 1 else n * factorial(n-1)

# Display documentation
print("Circle Area Function:")
print(circle_area.__doc__)
print("\nFactorial Function:")
print(factorial.__doc__)
```

### **Key Features**  
1. **Complete Docstrings**:  
   ```python

       param (type): Description
       
   Returns:
       type: Description
   """
   ```

2. **Error Handling**:  
   ```python

       raise ValueError("Radius cannot be negative")
   ```

3. **Documentation Access**:  
   ```python

   help(func)          # Formatted help
   ```

### **練習二：變數作用域與 LEGB 規則**  
**任務：** 通過實際範例展示 Python 的作用域規則。  

#### **要求:**

#### **範例輸出:**  
```
內部函數 value: 30 (局部)  
中層函數 value: 20 (閉包)  
全局 value: 10  
內建 len(): <built-in function len>  

修改後:  
內部 value: 300  
中層 value: 200  
全局 value: 100  
```

### **Sample Solution Code**  
```python
# Global scope
value = 10
print(f"Global value: {value}")

def outer_function():
    # Enclosing scope
    value = 20
    
    def inner_function():
        # Local scope
        value = 30
        print(f"Inner function value: {value} (local)")
        
        # Uncomment to see the error:
        # print(unknown_var)  # NameError
        
    inner_function()
    print(f"Middle function value: {value} (enclosing)")

outer_function()
print(f"Built-in len(): {len}")

# Demonstrate scope modification
def modifier():
    global value
    nonlocal_value = value  # This will cause SyntaxError!
    
    def changer():
        nonlocal nonlocal_value  # Corrected version
        nonlocal_value = 200
        return nonlocal_value
    
    value = 100
    return changer()

# Corrected version
def proper_modifier():
    global value
    value = 100
    
    def inner():
        nonlocal_value = 200
        return nonlocal_value
    
    return inner()

print("\nAfter modification:")
print(f"Global value: {value}")
print(f"Nonlocal value: {proper_modifier()}")
```

### **Key Teaching Points**  
1. **LEGB Lookup Order**:  
   ```python
   def outer():
       x = 10
       def inner():
           print(x)  # Finds x in enclosing scope
   ```

2. **Scope Modification**:  
   ```python
   global_var = 1
   def func():
       global global_var  # Modify global variable
       nonlocal outer_var  # Modify enclosing scope
   ```

3. **Common Errors**:  
   - `UnboundLocalError`: When local variable referenced before assignment  
   - `SyntaxError`: Invalid `nonlocal` declaration  

4. **Best Practices**:  
   - Avoid excessive global variables  
   - Use clear variable names in different scopes
