### **練習一：函數文件字串**  
**任務：** 創建具有完整文件說明的數學函數。  

#### **要求:**  
1. 實作兩個函數並包含完整文件字串：  
   - `circle_area(radius)` 計算圓面積  
   - `factorial(n)` 計算階乘  
2. 文件字串必須包含：  
   - 一行簡述  
   - `Args:` 參數說明  
   - `Returns:` 回傳值說明  
   - `Raises:` 例外說明  
   - `Examples:` 呼叫範例  
3. 展示如何讀取文件字串  

---

### **Sample Solution Code**  
```python
def circle_area(radius):
    """Calculate the area of a circle.
    
    Args:
        radius (float): Radius of the circle in meters
        
    Returns:
        float: Area in square meters
        
    Raises:
        ValueError: If radius is negative
        
    Examples:
        >>> circle_area(3)
        28.274333882308138
        >>> circle_area(-1)
        Traceback (most recent call last):
        ...
        ValueError: Radius cannot be negative
    """
    if radius < 0:
        raise ValueError("Radius cannot be negative")
    return 3.141592653589793 * radius ** 2

def factorial(n):
    """Calculate factorial of a non-negative integer.
    
    Args:
        n (int): Number to calculate
        
    Returns:
        int: Factorial value
        
    Raises:
        ValueError: If n is negative
        
    Examples:
        >>> factorial(5)
        120
        >>> factorial(0)
        1
    """
    if n < 0:
        raise ValueError("Factorial undefined for negative numbers")
    return 1 if n <= 1 else n * factorial(n-1)

# Display documentation
print("Circle Area Function:")
print(circle_area.__doc__)
print("\nFactorial Function:")
print(factorial.__doc__)
```

---

### **Key Features**  
1. **Complete Docstrings**:  
   ```python
   """One-line description.
   
   Args:
       param (type): Description
       
   Returns:
       type: Description
   """
   ```

2. **Error Handling**:  
   ```python
   if radius < 0:
       raise ValueError("Radius cannot be negative")
   ```

3. **Documentation Access**:  
   ```python
   print(func.__doc__)  # View docstring
   help(func)          # Formatted help
   ```
 

### **Exercise 2: Global Variable Access**  
**Task:** Observe how functions interact with global variables through three progressive examples.

#### **Instructions:**
1. Run each code block separately  
2. Observe the output and error messages  
3. Compare the behavior differences  

---

### **Part 1: Read Access (No Modification)**
```python
# Global variable
counter = 0

def increment():
    """Can read but not modify global variable"""
    print("Inside function (read access):", counter)
    
print("Before calling:", counter)
increment()
print("After calling:", counter)
```

**Expected Output:**
```
Before calling: 0
Inside function (read access): 0
After calling: 0
```

**Key Point:**  
Functions can read global variables without any special declaration.

---

### **Part 2: Failed Write Attempt**
```python
# Global variable
counter = 0

def increment():
    """Attempt to modify global variable (will fail)"""
    counter += 1
    print("Inside function:", counter)
    
print("Before calling:", counter)
increment()  # This will crash!
print("After calling:", counter)
```

**Expected Error:**
```
UnboundLocalError: local variable 'counter' referenced before assignment
```

**Key Point:**  
Python creates a new local variable when you try to assign to a global variable without declaration.

---

### **Part 3: Correct Global Modification**
```python
# Global variable
counter = 0

def increment():
    """Properly modify global variable"""
    global counter  # Required declaration
    counter += 1
    print("Inside function:", counter)
    
print("Before calling:", counter)
increment()
print("After calling:", counter)
```

**Expected Output:**
```
Before calling: 0
Inside function: 1
After calling: 1
```

**Key Point:**  
The `global` keyword is required to modify global variables.

### **練習二：全域變數存取**  
**任務：** 通過三個漸進範例觀察函數如何與全域變數互動。

#### **操作步驟:**
1. 分別執行每個代碼塊  
2. 觀察輸出和錯誤訊息  
3. 比較行為差異  

---

### **第一部分：讀取存取 (不可修改)**
```python
# 全域變數
counter = 0

def increment():
    """可讀取但不可修改全域變數"""
    print("函數內部 (讀取存取):", counter)
    
print("呼叫前:", counter)
increment()
print("呼叫後:", counter)
```

**預期輸出:**
```
呼叫前: 0
函數內部 (讀取存取): 0
呼叫後: 0
```

**重點說明:**  
函數不需特別宣告即可讀取全域變數。

---

### **第二部分：失敗的修改嘗試**
```python
# 全域變數
counter = 0

def increment():
    """嘗試修改全域變數 (將會失敗)"""
    counter += 1
    print("函數內部:", counter)
    
print("呼叫前:", counter)
increment()  # 這裡會出錯!
print("呼叫後:", counter)
```

**預期錯誤:**
```
UnboundLocalError: local variable 'counter' referenced before assignment
```

**重點說明:**  
當嘗試修改未宣告的全域變數時，Python會創建新的局部變數。

---

### **第三部分：正確的全域修改**
```python
# 全域變數
counter = 0

def increment():
    """正確修改全域變數的方式"""
    global counter  # 必要宣告
    counter += 1
    print("函數內部:", counter)
    
print("呼叫前:", counter)
increment()
print("呼叫後:", counter)
```

**預期輸出:**
```
呼叫前: 0
函數內部: 1
呼叫後: 1
```

**重點說明:**  
修改全域變數必須使用`global`關鍵字宣告。
 

### **Exercise 3: Nonlocal Variable Access**  
**Task:** Understand how nested functions modify enclosing scope variables using `nonlocal`.

#### **Instructions:**
1. Run all three parts sequentially  
2. Compare the behavior with the previous global scope exercise  
3. Note when `nonlocal` is required  

---

### **Part 1: Read Access Only**
```python
def outer():
    enclosing_var = 10
    
    def inner():
        print("Inner function reading:", enclosing_var)
    
    inner()
    print("Outer function value:", enclosing_var)

outer()
```

**Expected Output:**
```
Inner function reading: 10
Outer function value: 10
```

**Key Point:**  
Inner functions can read enclosing scope variables without declaration.

---

### **Part 2: Failed Modification Attempt**
```python
def outer():
    enclosing_var = 10
    
    def inner():
        enclosing_var += 1  # This will fail!
        print("Inner function:", enclosing_var)
    
    print("Before inner call:", enclosing_var)
    inner()  # Raises UnboundLocalError
    print("After inner call:", enclosing_var)

outer()
```

**Expected Error:**
```
UnboundLocalError: local variable 'enclosing_var' referenced before assignment
```

**Key Point:**  
Just like global scope, assignment creates a new local variable by default.

---

### **Part 3: Correct nonlocal Usage**
```python
def outer():
    enclosing_var = 10
    
    def inner():
        nonlocal enclosing_var  # Critical declaration
        enclosing_var += 1
        print("Modified in inner:", enclosing_var)
    
    print("Before inner call:", enclosing_var)
    inner()
    print("After inner call:", enclosing_var)

outer()
```

**Expected Output:**
```
Before inner call: 10
Modified in inner: 11
After inner call: 11
```

**Key Point:**  
`nonlocal` allows modifying variables from enclosing (but non-global) scopes.

### **練習三：nonlocal 變數存取**  
**任務：** 理解嵌套函數如何使用`nonlocal`修改閉包作用域變數。

#### **操作步驟:**
1. 依序執行所有三個部分  
2. 與先前全域作用域的練習比較行為差異  
3. 注意何時需要`nonlocal`宣告  

---

### **第一部分：僅讀取存取**
```python
def outer():
    enclosing_var = 10
    
    def inner():
        print("內部函數讀取:", enclosing_var)
    
    inner()
    print("外部函數值:", enclosing_var)

outer()
```

**預期輸出:**
```
內部函數讀取: 10
外部函數值: 10
```

**重點說明:**  
內部函數不需宣告即可讀取閉包作用域變數。

---

### **第二部分：失敗的修改嘗試**
```python
def outer():
    enclosing_var = 10
    
    def inner():
        enclosing_var += 1  # 這裡會出錯!
        print("內部函數:", enclosing_var)
    
    print("呼叫內部函數前:", enclosing_var)
    inner()  # 引發UnboundLocalError
    print("呼叫內部函數後:", enclosing_var)

outer()
```

**預期錯誤:**
```
UnboundLocalError: local variable 'enclosing_var' referenced before assignment
```

**重點說明:**  
與全域變數相同，賦值操作預設會創建新的局部變數。

---

### **第三部分：正確nonlocal用法**
```python
def outer():
    enclosing_var = 10
    
    def inner():
        nonlocal enclosing_var  # 關鍵宣告
        enclosing_var += 1
        print("在內部函數修改:", enclosing_var)
    
    print("呼叫內部函數前:", enclosing_var)
    inner()
    print("呼叫內部函數後:", enclosing_var)

outer()
```

**預期輸出:**
```
呼叫內部函數前: 10
在內部函數修改: 11
呼叫內部函數後: 11
```

**重點說明:**  
`nonlocal`允許修改閉包(非全域)作用域中的變數。
