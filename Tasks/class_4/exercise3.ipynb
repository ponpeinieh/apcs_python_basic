### **練習一：圓形計算器**  
**任務：** 創建返回圓形多種屬性的函式。  

#### **要求：**  
1. 定義 `circle_calc(radius)` 計算：  
   - 面積 (`πr²`)  
   - 周長 (`2πr`)  
   - 直徑 (`2r`)  
2. 以字典形式返回所有值：  
   ```python
   {'radius': 半徑, 'area': 面積, 'circumference': 周長, 'diameter': 直徑}
   ```  
3. 測試半徑5和7.5  

#### **範例輸出：**  
```
半徑5結果: {'radius': 5, 'area': 78.54, 'circumference': 31.42, 'diameter': 10}  
半徑7.5結果: {'radius': 7.5, 'area': 176.71, 'circumference': 47.12, 'diameter': 15}  
```

---

### **Sample Solution Code**  
```python
# English Version
import math

def circle_calc(radius):
    return {
        'radius': radius,
        'area': round(math.pi * radius ** 2, 2),
        'circumference': round(2 * math.pi * radius, 2),
        'diameter': 2 * radius
    }

# Demonstrations
print("Radius 5 results:", circle_calc(5))
print("Radius 7.5 results:", circle_calc(7.5))

# 繁體中文版
import math

def circle_calc(radius):
    return {
        'radius': radius,
        'area': round(math.pi * radius ** 2, 2),
        'circumference': round(2 * math.pi * radius, 2),
        'diameter': 2 * radius
    }

# 測試
print("半徑5結果:", circle_calc(5))
print("半徑7.5結果:", circle_calc(7.5))
```

### **練習二：購物車處理器**  
**任務：** 創建處理購物車列表的函式。  

#### **要求：**  
1. 定義 `calculate_total(items)`：  
   - 接收商品價格列表  
   - 返回總和與平均值(元組)  
2. 定義 `apply_discount(items, discount=10)`：  
   - 直接修改原列表(打折扣)  
   - 同時返回新列表  
3. 測試數據:  
   ```python
   prices = [100, 50, 75, 200]
   ```

#### **範例輸出：**  
```
原始價格: [100, 50, 75, 200]  
總計: 425, 平均: 106.25  
打9折後: [90.0, 45.0, 67.5, 180.0]  
```

---

### **Sample Solution Code**  
```python
# English Version
def calculate_total(items):
    total = sum(items)
    average = total / len(items)
    return total, average

def apply_discount(items, discount=10):
    for i in range(len(items)):
        items[i] *= (100 - discount) / 100
    return items

# Demonstration
prices = [100, 50, 75, 200]
print("Original:", prices)
total, avg = calculate_total(prices)
print(f"Total: {total}, Average: {avg}")
print(f"After 10% discount:", apply_discount(prices.copy()))

# 繁體中文版
def calculate_total(items):
    total = sum(items)
    average = total / len(items)
    return total, average

def apply_discount(items, discount=10):
    for i in range(len(items)):
        items[i] *= (100 - discount) / 100
    return items

# 測試
prices = [100, 50, 75, 200]
print("原始價格:", prices)
total, avg = calculate_total(prices)
print(f"總計: {total}, 平均: {avg}")
print(f"打9折後:", apply_discount(prices.copy()))
```

### **Key Teaching Points**  
1. **List as Parameter**  
   ```python
   def func(items):  # Receives original list reference
       items[0] = 99  # Modifies original list
   ```

2. **Modification Effects**  
   - `apply_discount()` changes the original list  
   - Used `copy()` to preserve original for demo

### **練習五：防止資料修改的複製技巧**  
**任務A: 切片複製**  
1. 創建會修改列表的 `add_item_with_slice()`  
2. 用切片 `[:]` 傳遞副本  
3. 展示原始列表不變  

**任務B: copy()**  
1. 創建 `add_item_with_copy()`  
2. 使用 `list.copy()` 方法  
3. 展示淺複製行為  

**任務C: deepcopy()**  
1. 創建 `modify_nested_list()`  
2. 使用 `copy.deepcopy()`  
3. 展示嵌套結構保持原樣  

#### **範例輸出:**  
```
切片: 原始 [1,2,3] -> 修改後 [1,2,3,100] (原始保持 [1,2,3])  
copy(): 原始 [1,2,3] -> 修改後 [1,2,3,100] (原始保持 [1,2,3])  
deepcopy(): 原始 [[1,2],[3,4]] -> 修改後 [[1,2,100],[3,4]] (原始保持 [[1,2],[3,4]])  
```

---

### **Sample Solution Code**  
```python
import copy

# Task A: Slice copy
def add_item_with_slice(items):
    items.append(100)
    print(f"Modified (slice): {items}")

original = [1, 2, 3]
print(f"Slice: Original {original}", end=" -> ")
add_item_with_slice(original[:])  # Pass slice copy
print(f"(Original stays {original})")

# Task B: copy()
def add_item_with_copy(items):
    items.append(100)
    print(f"Modified (copy()): {items}")

original = [1, 2, 3]
print(f"\ncopy(): Original {original}", end=" -> ")
add_item_with_copy(original.copy())  # Pass shallow copy
print(f"(Original stays {original})")

# Task C: deepcopy()
def modify_nested_list(items):
    items[0].append(100)
    print(f"Modified (deepcopy): {items}")

original = [[1, 2], [3, 4]]
print(f"\ndeepcopy(): Original {original}", end=" -> ")
modify_nested_list(copy.deepcopy(original))  # Pass deep copy
print(f"(Original stays {original})")
```

---

### **Key Learning Points**  
| Technique  | Level       | Import Needed | Nested Safety | Performance |  
|------------|-------------|---------------|---------------|-------------|  
| `[:]`      | Shallow     | No            | No            | Fastest     |  
| `.copy()`  | Shallow     | No            | No            | Fast        |  
| `deepcopy`| Deep        | `import copy` | Yes           | Slowest     |  

**When to Use:**  
- **Slicing**: Simple 1D lists when performance matters  
- **copy()**: Clear intent for shallow copies  
- **deepcopy()**: Nested structures with mutable elements
