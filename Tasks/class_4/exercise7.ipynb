### **練習一：同名函數行為**  
**任務：** 展示 Python 如何處理同名函數定義（對比 Java 的方法重載）。  

#### **要求：**  
1. 創建三個版本的 `calculate()` 函數：  
   - 第一個版本無參數，返回 100  
   - 第二個版本接受一個數字，返回其兩倍  
   - 第三個版本接受兩個數字，返回它們的和  
2. 展示每次定義後的呼叫行為  
3. 與 Java 的方法重載概念進行比較  

#### **範例輸出：**  
```
初始 calculate(): 100  
重新定義後 (1 參數): 20  
再次重新定義後 (2 參數): 15  

注意：每次重新定義都會完全取代前一個版本。  
```

---

### **Sample Solution Code**  
```python
# English Version
def calculate():
    return 100

print("Initial calculate():", calculate())  # Output: 100

def calculate(x):
    return x * 2

print("After redefinition (1 param):", calculate(10))  # Output: 20

def calculate(x, y):
    return x + y

print("After redefinition (2 params):", calculate(10, 5))  # Output: 15

# This would now cause an error:
# calculate() or calculate(1) would raise TypeError

# 繁體中文版
def calculate():
    return 100

print("初始 calculate():", calculate())  # 輸出: 100

def calculate(x):
    return x * 2

print("重新定義後 (1 參數):", calculate(10))  # 輸出: 20

def calculate(x, y):
    return x + y

print("再次重新定義後 (2 參數):", calculate(10, 5))  # 輸出: 15

# 現在這些呼叫會出錯:
# calculate() 或 calculate(1) 會引發 TypeError
```

---

### **Key Teaching Points**  
1. **Python's Behavior**:  
   ```python
   def func(): pass        # First definition
   def func(x): pass       # Replaces first
   def func(x, y): pass    # Replaces second
   ```
   - Each new definition completely overwrites the previous one  
   - Only the last definition remains callable  

2. **Java Contrast**:  
   ```java
   // Java would allow all three to coexist
   int calculate() {...}  
   int calculate(int x) {...}  
   int calculate(int x, int y) {...}  
   ```
   - Java selects version based on argument count/types  
   - Python requires different approaches (default args, `*args`, type checking)  

3. **Practical Implications**:  
   - Python developers must use unique names or parameter flexibility  
   - Common alternatives:  
     ```python
     def calculate(x=None, y=None):
         if x is None: return 100
         elif y is None: return x*2
         else: return x+y
     ```

### **Exercise 2: Type-Based Function Behavior**

#### **Requirements:**

### **練習二：基於類型的函數行為**  
**任務：** 創建根據輸入類型改變行為的靈活函數，展示 Python 處理不同參數類型的方法。  

#### **要求:**  
1. 定義 `process_data()` 函數：  
   - 接受一個必要參數(例如:`data`)和可選的 `action` 關鍵字  
   - 為不同類型實現不同行為：  
     - 字串：返回大寫版本  
     - 數字：返回平方值  
     - 列表：根據內容類型返回總和或拼接  
     - 字典：返回按字母排序的鍵  
   - 使用 `action` 參數修改行為：  
     - `"reverse"` 反轉字串/列表或取數字負值  
     - `"info"` 返回類型信息  

#### **範例輸出:**  
```
處理字串: '你好' → '你好'  
處理數字: 5 → 25  
處理列表: [1,2,3] → 6  
處理字典: {'b':2,'a':1} → ['a', 'b']  

action='reverse' 時:  
'你好' → '好你'  
[1,2,3] → [3,2,1]  
5 → -5  
```

---

### **Sample Solution Code**  
```python
# English Version
def process_data(data, action=None):
    if action == "info":
        return f"Type: {type(data).__name__}, Length: {len(data) if hasattr(data, '__len__') else 'N/A'}"
    
    if isinstance(data, str):
        return data[::-1] if action == "reverse" else data.upper()
    elif isinstance(data, (int, float)):
        return -data if action == "reverse" else data ** 2
    elif isinstance(data, list):
        if action == "reverse":
            return data[::-1]
        return sum(data) if all(isinstance(x, (int, float)) for x in data) else ''.join(map(str, data))
    elif isinstance(data, dict):
        return sorted(data.keys())
    else:
        return "Unsupported type"

# Demonstration
print(process_data("hello"))                  # HELLO
print(process_data(5))                       # 25
print(process_data([1, 2, 3]))               # 6
print(process_data({'b': 2, 'a': 1}))        # ['a', 'b']
print(process_data("hello", action="reverse"))  # olleh
print(process_data([1, 2, 3], action="info"))  # Type: list, Length: 3

# 繁體中文版
def process_data(data, action=None):
    if action == "info":
        return f"類型: {type(data).__name__}, 長度: {len(data) if hasattr(data, '__len__') else '不適用'}"
    
    if isinstance(data, str):
        return data[::-1] if action == "reverse" else data.upper()
    elif isinstance(data, (int, float)):
        return -data if action == "reverse" else data ** 2
    elif isinstance(data, list):
        if action == "reverse":
            return data[::-1]
        return sum(data) if all(isinstance(x, (int, float)) for x in data) else ''.join(map(str, data))
    elif isinstance(data, dict):
        return sorted(data.keys())
    else:
        return "不支援的類型"

# 示範
print(process_data("你好"))                   # 你好
print(process_data(5))                       # 25
print(process_data([1, 2, 3]))               # 6
print(process_data({'b': 2, 'a': 1}))        # ['a', 'b']
print(process_data("你好", action="reverse"))  # 好你
print(process_data([1, 2, 3], action="info"))  # 類型: list, 長度: 3
```
