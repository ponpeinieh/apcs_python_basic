### **練習一：溫度轉換**  
**任務：** 使用函式轉換攝氏溫度(°C)和華氏溫度(°F)。  

#### **要求：**  
1. 創建函式 `c_to_f(celsius)`：  
   - 接收攝氏溫度參數  
   - 返回華氏溫度（公式：`F = (9/5) × °C + 32`）  
2. 在主程式中：  
   - 詢問攝氏溫度輸入  
   - 呼叫轉換函式  
   - 以`[°C] = [°F]`格式印出結果  
   - 輸入'q'時退出  

#### **範例輸出：**  
```
請輸入攝氏溫度°C（或輸入'q'退出）：25  
25°C = 77.0°F  
請輸入攝氏溫度°C（或輸入'q'退出）：100  
100°C = 212.0°F  
請輸入攝氏溫度°C（或輸入'q'退出）：q  
再見！
```

---

### **Sample Solution Code**  
```python
# English Version
def c_to_f(celsius):
    return (9/5) * celsius + 32

while True:
    user_input = input("Enter temperature in \u00b0C (or 'q' to quit): ")
    if user_input.lower() == 'q':
        print("Goodbye!")
        break
    if user_input.replace('.', '').isdigit():  # Simple number check
        celsius = float(user_input)
        fahrenheit = c_to_f(celsius)
        print(f"{celsius}\u00b0C = {fahrenheit}\u00b0F")
    else:
        print("Please enter a number or 'q' to quit")

# 繁體中文版
def c_to_f(celsius):
    return (9/5) * celsius + 32

while True:
    user_input = input("請輸入攝氏溫度\u00b0C（或輸入'q'退出）：")
    if user_input.lower() == 'q':
        print("再見！")
        break
    if user_input.replace('.', '').isdigit():  # 簡單數字檢查
        celsius = float(user_input)
        fahrenheit = c_to_f(celsius)
        print(f"{celsius}\u00b0C = {fahrenheit}\u00b0F")
    else:
        print("請輸入數字或'q'退出")
```

### **Exercise 2: BMI Calculator with Multiple Return Values**

### **練習二：多返回值BMI計算器**  
**任務：** 創建能從函式返回數值和分類的BMI計算器。  

#### **要求：**  
1. 創建函式 `calculate_bmi`：  
   - 接收體重(kg)和身高(m)參數  
   - 返回兩個值：  
     - BMI數值 (浮點數)  
     - 分類字串 ("體重過輕"/"正常範圍"/"過重"/"肥胖")  
2. 在主程式中：  
   - 獲取用戶輸入  
   - 調用函式並解包返回值  
   - 打印格式化結果  

#### **範例輸出：**  
```
請輸入體重（公斤）：70  
請輸入身高（米）：1.75  
您的BMI是: 22.86  
分類: 正常範圍  
```

---

### **Sample Solution Code**  
```python
# English Version
def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    
    if bmi < 18.5:
        category = "Underweight"
    elif 18.5 <= bmi < 25:
        category = "Normal"
    elif 25 <= bmi < 30:
        category = "Overweight"
    else:
        category = "Obese"
    
    return bmi, category  # Return as a tuple

# Main program
weight = float(input("Enter weight (kg): "))
height = float(input("Enter height (m): "))

bmi_value, bmi_category = calculate_bmi(weight, height)  # Tuple unpacking

print(f"Your BMI is: {bmi_value:.2f}")
print(f"Category: {bmi_category}")

# 繁體中文版
def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    
    if bmi < 18.5:
        category = "體重過輕"
    elif 18.5 <= bmi < 25:
        category = "正常範圍"
    elif 25 <= bmi < 30:
        category = "過重"
    else:
        category = "肥胖"
    
    return bmi, category  # 以元組形式返回

# 主程式
weight = float(input("請輸入體重（公斤）："))
height = float(input("請輸入身高（米）："))

bmi_value, bmi_category = calculate_bmi(weight, height)  # 元組解包

print(f"您的BMI是: {bmi_value:.2f}")
print(f"分類: {bmi_category}")
```
