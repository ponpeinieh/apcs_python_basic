### **Exercise 1: Personal Profile with `**kwargs`**  
**Task:** Create a function that builds personal profiles using keyword arguments.  

#### **Requirements:**  
1. Define `build_profile()` that:  
   - Accepts arbitrary keyword arguments using `**kwargs`  
   - Prints each key-value pair as "Key: Value"  
   - Returns a dictionary of all valid entries  
2. Validate inputs:  
   - Only accept strings or numbers as values  
   - Skip invalid entries with warning: "Skipped [key]: Invalid type"  

#### **Example Output:**  
```
name: Alice  
age: 25  
city: New York  
Skipped hobbies: Invalid type  

Collected 3 valid entries.  
Returned: {'name': 'Alice', 'age': 25, 'city': 'New York'}  
```

#### **Skills Tested:**  
- Keyword arguments (`**kwargs`)  
- Dictionary operations  
- Type validation

### **練習一：使用 `**kwargs` 建立個人檔案**  
**任務：** 創建使用關鍵字參數建立個人檔案的函式。  

#### **要求：**  
1. 定義 `build_profile()`：  
   - 使用 `**kwargs` 接受任意關鍵字參數  
   - 打印每個鍵值對為 "Key: Value"  
   - 返回所有有效條目的字典  
2. 驗證輸入：  
   - 僅接受字串或數字作為值  
   - 跳過無效條目並提示："Skipped [key]: Invalid type"  

#### **範例輸出：**  
```
name: Alice  
age: 25  
city: New York  
Skipped hobbies: Invalid type  

收集到 3 個有效條目。  
返回: {'name': 'Alice', 'age': 25, 'city': 'New York'}  
```

---

### **Sample Solution Code**  
```python
# English Version
def build_profile(**kwargs):
    profile = {}
    for key, value in kwargs.items():
        if isinstance(value, (str, int, float)):
            print(f"{key}: {value}")
            profile[key] = value
        else:
            print(f"Skipped {key}: Invalid type")
    
    print(f"\nCollected {len(profile)} valid entries.")
    return profile

# Demonstration
profile_data = build_profile(
    name="Alice",
    age=25,
    city="New York",
    hobbies=["reading", "hiking"],  # Will be skipped
    height=165.5
)

print(f"Returned: {profile_data}")

# 繁體中文版
def build_profile(**kwargs):
    profile = {}
    for key, value in kwargs.items():
        if isinstance(value, (str, int, float)):
            print(f"{key}: {value}")
            profile[key] = value
        else:
            print(f"跳過 {key}: 無效類型")
    
    print(f"\n收集到 {len(profile)} 個有效條目。")
    return profile

# 示範
profile_data = build_profile(
    name="王小明",
    age=30,
    city="台北",
    hobbies=["閱讀", "旅行"],  # 將被跳過
    height=175.2
)

print(f"返回: {profile_data}")
```

---

### **Key Teaching Points**  
1. **`**kwargs` Behavior**:  
   ```python
   def func(**kwargs):  # kwargs becomes a dictionary
       print(kwargs)   # {'key1': val1, 'key2': val2}
   ```

2. **Type Validation**:  
   - `isinstance()` checks for multiple valid types  
   - Flexible error handling  


注意: 一般引數必須可以放在 `**kwargs` 前面

### **Exercise 2: Minecraft Inventory with `*args` and `**kwargs`**  
**Task:** Create a function that manages Minecraft inventories using both positional and keyword arguments.  

#### **Requirements:**  
1. Define `build_inventory()` that:  
   - Accepts blocks as positional arguments (`*args`)  
   - Accepts tools as keyword arguments (`**kwargs`)  
   - Prints each block with "Block: [name]"  
   - Prints each tool with "Tool: [type] = [material]"  
   - Returns a dictionary containing:  
     - `blocks`: List of all blocks  
     - `tools`: Dictionary of all tools  
     - `total_items`: Sum of blocks and tools  

2. Validate inputs:  
   - Skip empty block entries with warning: "Skipped: Empty block"  
   - Tools must have string values  

#### **Example Output:**  
```
Block: stone  
Block: oak_log  
Tool: pickaxe = diamond  
Tool: shovel = iron  

Collected 2 blocks and 2 tools.  
Returned: {
    'blocks': ['stone', 'oak_log'],
    'tools': {'pickaxe': 'diamond', 'shovel': 'iron'},
    'total_items': 4
}
```

#### **Skills Tested:**  
- Positional arguments (`*args`)  
- Keyword arguments (`**kwargs`)  
- Dictionary and list operations  
- Input validation

### **練習二：使用 `*args` 和 `**kwargs` 建立 Minecraft 背包**  
**任務：** 創建管理 Minecraft 背包的函式。  

#### **要求：**  
1. 定義 `build_inventory()`：  
   - 用 `*args` 接受方塊  
   - 用 `**kwargs` 接受工具  
   - 打印 "方塊: [名稱]" 格式  
   - 打印 "工具: [類型] = [材質]" 格式  
   - 返回包含以下內容的字典：  
     - `blocks`: 所有方塊的列表  
     - `tools`: 所有工具的字典  
     - `total_items`: 方塊和工具的總數  

2. 驗證輸入：  
   - 跳過空方塊並提示："跳過: 空方塊"  
   - 工具值必須為字串  

#### **範例輸出：**  
```
方塊: stone  
方塊: oak_log  
工具: pickaxe = diamond  
工具: shovel = iron  

收集到 2 個方塊和 2 個工具。  
返回: {
    'blocks': ['stone', 'oak_log'],
    'tools': {'pickaxe': 'diamond', 'shovel': 'iron'},
    'total_items': 4
}
```

---

### **Sample Solution Code**  
```python
# English Version
def build_inventory(*blocks, **tools):
    valid_blocks = []
    valid_tools = {}
    
    for block in blocks:
        if block and str(block).strip():
            print(f"Block: {block}")
            valid_blocks.append(block)
        else:
            print("Skipped: Empty block")
    
    for tool, material in tools.items():
        if isinstance(material, str):
            print(f"Tool: {tool} = {material}")
            valid_tools[tool] = material
        else:
            print(f"Skipped {tool}: Invalid material type")
    
    print(f"\nCollected {len(valid_blocks)} blocks and {len(valid_tools)} tools.")
    return {
        'blocks': valid_blocks,
        'tools': valid_tools,
        'total_items': len(valid_blocks) + len(valid_tools)
    }

# Demonstration
inventory = build_inventory(
    "stone", "", "oak_log",  # Blocks
    pickaxe="diamond",       # Tools
    shovel="iron",
    axe=123  # Will be skipped
)

print(f"Returned: {inventory}")

# 繁體中文版
def build_inventory(*blocks, **tools):
    valid_blocks = []
    valid_tools = {}
    
    for block in blocks:
        if block and str(block).strip():
            print(f"方塊: {block}")
            valid_blocks.append(block)
        else:
            print("跳過: 空方塊")
    
    for tool, material in tools.items():
        if isinstance(material, str):
            print(f"工具: {tool} = {material}")
            valid_tools[tool] = material
        else:
            print(f"跳過 {tool}: 無效材質類型")
    
    print(f"\n收集到 {len(valid_blocks)} 個方塊和 {len(valid_tools)} 個工具。")
    return {
        'blocks': valid_blocks,
        'tools': valid_tools,
        'total_items': len(valid_blocks) + len(valid_tools)
    }

# 示範
inventory = build_inventory(
    "stone", "", "oak_log",  # 方塊
    pickaxe="diamond",       # 工具
    shovel="iron",
    axe=123  # 將被跳過
)

print(f"返回: {inventory}")
```

