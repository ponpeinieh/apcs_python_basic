### **練習一：參數類型練習**  
**任務：** 示範位置參數與關鍵字參數。  

#### **要求：**  
1. 創建函式 `describe_pet()` 包含參數：  
   - `animal_type` (位置參數)  
   - `pet_name` (位置參數)  
2. 展示三種調用方式：  
   - 純位置參數  
   - 純關鍵字參數  
   - 混合方式  
3. 打印格式化描述  

#### **範例輸出：**  
```
位置參數: 我有一隻狗叫Spot。  
關鍵字參數: 我有一隻貓叫Mittens。  
混合調用: 我有一隻鸚鵡叫Polly。  
```

---
#### **Exercise Solution**  
```python
# English
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type} named {pet_name}.")

# Demonstrations
describe_pet("dog", "Spot")  # Positional
describe_pet(pet_name="Mittens", animal_type="cat")  # Keyword
describe_pet("parrot", pet_name="Polly")  # Mixed

# 繁體中文版
def describe_pet(animal_type, pet_name):
    print(f"我有一隻{animal_type}叫{pet_name}。")

# 示範
describe_pet("狗", "Spot")  # 位置參數
describe_pet(pet_name="Mittens", animal_type="貓")  # 關鍵字參數
describe_pet("鸚鵡", pet_name="Polly")  # 混合調用
```

### **Sample Solution Code**  
```python
# English
def describe_pet(pet_name, animal_type="dog"):
    print(f"I have a {animal_type} named {pet_name}.")

# Demonstrations
describe_pet("Rover")  # Uses default
describe_pet("Nibbles", "hamster")  # Overrides default

# 繁體中文版
def describe_pet(pet_name, animal_type="狗"):
    print(f"我有一隻{animal_type}叫{pet_name}。")

# 示範
describe_pet("Rover")  # 使用默認值
describe_pet("Nibbles", "倉鼠")  # 覆蓋默認值
```

### **練習三：T恤打印（位置與關鍵字參數）**  
**任務：** 創建T恤打印函式展示不同參數類型。  

#### **要求：**  
1. 定義 `make_shirt(size, text_to_print)`  
2. 打印格式:  
   `"製作{size}尺寸T恤，印有文字: '{text_to_print}'"`  
3. 示範:  
   - 位置參數調用: `("大號", "你好")`  
   - 關鍵字參數調用: `(size="中號", text_to_print="Python")`  

#### **範例輸出：**  
```
位置參數: 製作大號尺寸T恤，印有文字: '你好'  
關鍵字參數: 製作中號尺寸T恤，印有文字: 'Python'  
```

---

```python
# English
def make_shirt(size, text_to_print):
    print(f"Making a {size} shirt with text: '{text_to_print}'")

# Demonstrations
make_shirt("large", "Hello")  # Positional
make_shirt(size="medium", text_to_print="Python")  # Keyword

# 繁體中文版
def make_shirt(size, text_to_print):
    print(f"製作{size}尺寸T恤，印有文字: '{text_to_print}'")

# 示範
make_shirt("大號", "你好")  # 位置參數
make_shirt(size="中號", text_to_print="Python")  # 關鍵字參數
```

### **Exercise 4: T-Shirt Printer (Default Values)**  
**Task:** Modify the function to use default values.  

#### **New Requirements:**  
1. Update `make_shirt()` with defaults:  
   - `size="medium"`  
   - `text_to_print="I code Python!"`  
2. Demonstrate:  
   - Large shirt with default text  
   - Default size with custom text  
   - Small shirt with custom text  

#### **Example Output:**  
```
Large + default: Making a large shirt with text: 'I code Python!'  
Default size + custom: Making a medium shirt with text: 'Python Rocks!'  
Small + custom: Making a small shirt with text: 'I use Jupyter Notebook too!'  
```

---

### **Sample Solution Code**  
```python
# English
def make_shirt(size="medium", text_to_print="I code Python!"):
    print(f"Making a {size} shirt with text: '{text_to_print}'")

# Demonstrations
make_shirt(size="large")  # Large + default text
make_shirt(text_to_print="Python Rocks!")  # Default size + custom
make_shirt("small", "I use Jupyter Notebook too!")  # Full custom

# 繁體中文版
def make_shirt(size="中號", text_to_print="我寫Python!"):
    print(f"製作{size}尺寸T恤，印有文字: '{text_to_print}'")

# 示範
make_shirt(size="大號")  # 大號 + 默認文字
make_shirt(text_to_print="Python超棒!")  # 默認尺寸 + 自訂文字
make_shirt("小號", "我也用Jupyter Notebook!")  # 完全自訂
```
