### Python `nonlocal` Variable Examples

#### Example 1: Counter Factory Using `nonlocal`

```python
def make_counter(initial_value=0, step=1):
    """Factory function that creates counter functions with encapsulated state"""
    count = initial_value
    
    def counter():
        """Counter function that increments and returns the current count"""
        nonlocal count  # This is key to modifying the count variable
        current = count
        count += step
        return current
    
    return counter

# Create counters with different initial states
simple_counter = make_counter()      # Starts at 0, increments by 1
ten_step_counter = make_counter(100, 10)  # Starts at 100, increments by 10
decrementer = make_counter(50, -5)  # Starts at 50, decrements by 5

# Use the counters
print(simple_counter())  # 0
print(simple_counter())  # 1
print(simple_counter())  # 2

print(ten_step_counter())  # 100
print(ten_step_counter())  # 110
print(ten_step_counter())  # 120

print(decrementer())  # 50
print(decrementer())  # 45
print(decrementer())  # 40
```

#### Example 2: Alternative Solution Without `nonlocal` (Using Mutable Objects)

```python
def make_counter(initial_value=0, step=1):
    """Counter factory using a list as an alternative solution"""
    count = [initial_value]  # Store the value in a mutable list
    
    def counter():
        """Counter function that increments and returns the current count"""
        current = count[0]  # Access the element in the list
        count[0] += step    # Modify the element in the list
        return current
    
    return counter

# Create counter
my_counter = make_counter(5, 2)

# Use the counter
print(my_counter())  # 5
print(my_counter())  # 7
print(my_counter())  # 9
```

#### Key Differences Explained

1. **`nonlocal` Version**:
   - Directly modifies variables from the outer function
   - Cleaner and more intuitive code
   - Recommended approach for Python 3.x

2. **List Alternative**:
   - Changes state by modifying list elements
   - Suitable for environments that don't support `nonlocal`
   - Slightly less intuitive code but functionally equivalent
 

### Python `nonlocal` 變數範例

#### 案例 1：使用 `nonlocal` 的計數器工廠

```python
def make_counter(initial_value=0, step=1):
    """工廠函數，創建具有封裝狀態的計數器函數"""
    count = initial_value
    
    def counter():
        """計數器函數，增加並返回當前計數"""
        nonlocal count  # 這是修改 count 變數的關鍵
        current = count
        count += step
        return current
    
    return counter

# 創建不同初始狀態的計數器
simple_counter = make_counter()      # 從 0 開始，每次加 1
ten_step_counter = make_counter(100, 10)  # 從 100 開始，每次加 10
decrementer = make_counter(50, -5)  # 從 50 開始，每次減 5

# 使用計數器
print(simple_counter())  # 0
print(simple_counter())  # 1
print(simple_counter())  # 2

print(ten_step_counter())  # 100
print(ten_step_counter())  # 110
print(ten_step_counter())  # 120

print(decrementer())  # 50
print(decrementer())  # 45
print(decrementer())  # 40
```

#### 案例 2：不使用 `nonlocal` 的替代方案（使用可變物件）

```python
def make_counter(initial_value=0, step=1):
    """使用列表作為替代方案的計數器工廠"""
    count = [initial_value]  # 將值儲存在可變的列表中
    
    def counter():
        """計數器函數，增加並返回當前計數"""
        current = count[0]  # 存取列表中的元素
        count[0] += step    # 修改列表中的元素
        return current
    
    return counter

# 創建計數器
my_counter = make_counter(5, 2)

# 使用計數器
print(my_counter())  # 5
print(my_counter())  # 7
print(my_counter())  # 9
```

#### 關鍵差異說明

1. **`nonlocal` 版本**:
   - 直接修改外層函數的變數
   - 程式碼更清晰直觀
   - Python 3.x 推薦做法

2. **列表替代方案**:
   - 通過修改列表元素來改變狀態
   - 適用於不支援 `nonlocal` 的環境
   - 程式碼稍欠直觀但功能相同
