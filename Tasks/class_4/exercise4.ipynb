### **練習一：密碼強度驗證器**  
**任務：** 使用`all()`驗證密碼是否滿足多個條件。  

#### **要求：**  
1. 檢查密碼是否滿足以下所有條件：  
   - 至少8個字元  
   - 包含至少一個數字  
   - 包含至少一個大寫字母  
   - 包含至少一個特殊符號(`!@#$%`)  
2. 使用`all()`進行綜合驗證  
3. 全部符合才返回`True`  

#### **範例輸出：**  
```
輸入密碼: Hello123!
強密碼: True

輸入密碼: hello
強密碼: False
```

### **Simple solution**   

```python
def is_strong_password(password):
    has_digit = False
    has_upper = False
    has_special = False
    
    # Manual checks
    for char in password:
        if char.isdigit():
            has_digit = True
        if char.isupper():
            has_upper = True
        if char in '!@#$%':
            has_special = True
    
    conditions = [
        len(password) >= 8,
        has_digit,
        has_upper,
        has_special
    ]
    return all(conditions)
```

---

### **Using Generator Expressions and any()**  
```python
# English Version
def is_strong_password(password):
    # Create condition checks as a list
    conditions = [
        len(password) >= 8,
        any(char.isdigit() for char in password),
        any(char.isupper() for char in password),
        any(char in '!@#$%' for char in password)
    ]
    return all(conditions)

# Test
user_pw = input("Enter password: ")
print("Strong password:", is_strong_password(user_pw))

# 繁體中文版
def is_strong_password(password):
    # 建立條件檢查列表
    conditions = [
        len(password) >= 8,
        any(char.isdigit() for char in password),
        any(char.isupper() for char in password),
        any(char in '!@#$%' for char in password)
    ]
    return all(conditions)

# 測試
user_pw = input("輸入密碼: ")
print("強密碼:", is_strong_password(user_pw))
```

---

### **Key Teaching Points**  
1. **`all()` Behavior**:  
   - Returns `True` only if all iterable elements are truthy  
   - Short-circuits at first `False` (like `and` operations)  

2. **Real-World Analogy**:  
   Like a checklist where ALL items must be ticked:  
   ```
   [✓] Length ≥ 8  
   [✓] Has digit  
   [✓] Has uppercase  
   [✓] Has special char  
   → ALL ✓ = Strong password  
   ```  

3. **Comparison**:  
   | With `all()` | Without `all()` |  
   |-------------|----------------|  
   | Clean one-line check | Nested `if` statements |  
   | Readable rule set | Harder to add new rules |

### **練習二：使用 map() 的溫度轉換器**  
**任務：** 用 `map()` 和普通函式轉換攝氏溫度。  

#### **要求：**  
1. 建立攝氏溫度列表: `[0, 10, 20, 30, 40]`  
2. 定義轉換函式 `c_to_f()`  
3. 用 `map()` 轉換整個列表  
4. 印出原始和轉換後的溫度  

#### **範例輸出：**  
```
攝氏: [0, 10, 20, 30, 40]  
華氏: [32.0, 50.0, 68.0, 86.0, 104.0]  
```

---

### **Sample Solution Code**  
```python
# English Version
def c_to_f(celsius):
    """Convert Celsius to Fahrenheit"""
    return (9/5) * celsius + 32

# Original temperatures
celsius_temps = [0, 10, 20, 30, 40]

# Convert using map()
fahrenheit_temps = list(map(c_to_f, celsius_temps))

print("Celsius:", celsius_temps)
print("Fahrenheit:", fahrenheit_temps)

# 繁體中文版
def c_to_f(celsius):
    """將攝氏溫度轉為華氏"""
    return (9/5) * celsius + 32

# 原始溫度
celsius_temps = [0, 10, 20, 30, 40]

# 使用 map() 轉換
fahrenheit_temps = list(map(c_to_f, celsius_temps))

print("攝氏:", celsius_temps)
print("華氏:", fahrenheit_temps)
```

---

### **Key Teaching Points**  
1. **How `map()` Works**  
   - Takes a function and iterable  
   - Applies the function to each item  
   - Returns an iterator (convert to list to view)  

2. **Without Lambda**  
   ```python
   # Instead of:
   # map(lambda x: x*2, temps)
   
   # We use:
   def double(x):
       return x*2
   map(double, temps)
   ```

3. **Real-World Analogy**  
   Like a factory conveyor belt:  
   - `map()` = Assembly line applying the same operation  
   - Function = Worker doing one specific task

### **練習三：使用 map() 的姓名格式化**  
**任務：** 用 `map()` 將姓名轉為首字母大寫。  

#### **要求：**  
1. 定義函式 `capitalize_name(name)`  
2. 建立列表 `["alice", "bob", "charlie"]`  
3. 使用 `map()` 進行格式化  
4. 印出前後對比結果  

#### **範例輸出：**  
```
原始: ["alice", "bob", "charlie"]  
格式化後: ["Alice", "Bob", "Charlie"]  
```

---
#### **Exercise Solution**  
```python
# English Version
def capitalize_name(name):
    return name.title()

names = ["alice", "bob", "charlie"]
title_names = list(map(capitalize_name, names))

print("Original:", names)
print("Formatted:", title_names)

# 繁體中文版
def capitalize_name(name):
    return name.title()

names = ["alice", "bob", "charlie"]
title_names = list(map(capitalize_name, names))

print("原始:", names)
print("格式化後:", title_names)
```
