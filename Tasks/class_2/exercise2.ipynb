### **Exercise 4: Tuple Immutability Demonstration**  
**Task:** Observe how tuples cannot be modified after creation.  

#### **Requirements:**  
1. Create this tuple:  
   ```python
   colors = ("red", "green", "blue")
   ```
2. Attempt these operations one at a time and observe what happens:  
   - `colors[1] = "yellow"`  
   - `colors.append("black")`  
   - `del colors[0]`  

#### **Expected Outcomes:**  
- Each operation will cause Python to show an error message  
- The original tuple remains unchanged  

#### **Skills Tested:**  
- Understanding immutable sequences  
- Recognizing common Python error messages  

---

### **練習四：元組不可變性觀察**  
**任務：** 觀察元組創建後無法修改的特性。  

#### **要求：**  
1. 建立以下元組：  
   ```python
   colors = ("紅色", "綠色", "藍色")
   ```
2. 分別嘗試以下操作並觀察結果：  
   - `colors[1] = "黃色"`  
   - `colors.append("黑色")`  
   - `del colors[0]`  

#### **預期結果：**  
- 每個操作都會導致Python顯示錯誤訊息  
- 原始元組保持不變  

#### **訓練技能：**  
- 理解不可變序列  
- 認識常見的Python錯誤訊息  

---

### **Sample Solution Approach**  
```python
# English Version
colors = ("red", "green", "blue")
```

### **Key Teaching Points**  
1. **Learning Through Errors**: Students will see these error types:  
   - `TypeError` for item assignment/deletion  
   - `AttributeError` for append attempts  

2. **Visual Demonstration**:  
   - After each failed operation, have students print `colors` to see it's unchanged  

3. **Contrast Activity**:  
   - Repeat the same operations with a list to show mutability
