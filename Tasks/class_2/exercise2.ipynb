### **Exercise 1: Escape Character Fun**  
**Task:** Write a Python program that uses escape characters to create a simple text art and format a message.  

#### **Requirements:**  
1. **Text Art:** Use `\n` and `\t` to draw a simple cat:  

```
 /\_/\  
( o.o )  
 > ^ <
```  

2. **Message Formatting:** Print this sentence with proper spacing:  
   `"Python is awesome! Learn it today!"`  
   - Use `\t` to indent the first line  
   - Use `\n` to split into two lines  

#### **Example Output:**  
```  
    /\_/\  
   ( o.o )  
    > ^ <  

	Python is awesome!  
Learn it today!  
```  

#### **Skills Tested:**  
- Basic escape characters (`\n`, `\t`)  
- Simple text formatting  

---  

### **練習一：跳脫字元小遊戲**  
**任務：** 使用跳脫字元製作簡單的文字圖案和格式化訊息。  

#### **要求：**  
1. **文字圖案：** 用 `\n` 和 `\t` 畫一隻小貓：  
   ```  
    /\_/\  
   ( o.o )  
    > ^ <  
   ```  
2. **訊息格式化：** 印出以下句子並正確排版：  
   `"Python 很棒！今天開始學習吧！"`  
   - 用 `\t` 縮排第一行  
   - 用 `\n` 分成兩行  

#### **範例輸出：**  
```  
    /\_/\  
   ( o.o )  
    > ^ <  

	Python 很棒！  
今天開始學習吧！  
```  

#### **訓練技能：**  
- 基本跳脫字元 (`\n`, `\t`)  
- 簡單文字排版  


### **Exercise 2: Dinner Party Guest Management**  
**Task:** Create and manipulate a guest list for a dinner party, then generate invitation messages.  

**Requirements:**  
1. Start with 3 guest names in a list  
2. Add a new guest at the beginning  
3. Replace the guest at position 3  
4. Remove the last guest  
5. Re-add the removed guest  
6. Print invitations using a `for` loop:  
   `"Dear {guest}, you're cordially invited to our dinner!"`  

**Example Output:**  
```
Invitations:
Dear New Guest, you're cordially invited to our dinner!
Dear Alice, you're cordially invited to our dinner!
...
```

**Skills Tested:**  
- List operations (insert, replace, remove, append)  
- `for` loops  
- String formatting  

---

### **練習二：晚宴賓客管理**  
**任務：** 建立並操作賓客名單，生成邀請訊息。  

**要求：**  
1. 初始名單包含3位賓客  
2. 在名單開頭新增1位賓客  
3. 替換第3位的賓客  
4. 刪除最後1位賓客  
5. 將刪除的賓客加回  
6. 用`for`迴圈列印邀請：  
   `"親愛的 {賓客}，誠摯邀請您參加晚宴！"`  

**範例輸出：**  
```

親愛的新賓客，誠摯邀請您參加晚宴！
親愛的小明，誠摯邀請您參加晚宴！
...
```

**訓練技能：**  
- 清單操作（插入、替換、刪除、追加）  
- `for`迴圈  
- 字串格式化  

---

### **Exercise 3: River & Country Dictionary**  
**Task:** Manage river-country pairs using a dictionary and print specific information.  

**Requirements:**  
1. Store 5 river-country pairs (e.g., `{'Nile': 'Egypt'}`)  
2. Print each pair as:  
   `"The {river} runs through {country}."`  
3. Print all river names  
4. Print all country names  

**Example Output:**  
```
River-Country Pairs:
The Nile runs through Egypt.
The Amazon runs through Brazil.

All Rivers: Nile, Amazon...
All Countries: Egypt, Brazil...
```

**Skills Tested:**  
- Dictionary creation and traversal  
- `keys()` and `values()` methods  
- Looping through dictionaries  

---

### **練習三：河流與國家字典**  
**任務：** 使用字典管理河流-國家對應關係並列印資訊。  

**要求：**  
1. 儲存5組河流-國家對應（如`{'長江': '中國'}`）  
2. 列印每組對應：  
   `"{河流}流經{國家}"`  
3. 列印所有河流名稱  
4. 列印所有國家名稱  

**範例輸出：**  
```
河流-國家對應:
長江流經中國
亞馬遜河流經巴西

所有河流: 長江, 亞馬遜河...
所有國家: 中國, 巴西...
```

**訓練技能：**  
- 字典建立與遍歷  
- `keys()`與`values()`方法  
- 字典迴圈操作
