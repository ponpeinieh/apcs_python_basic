### **Exercise 1: Chicken-Rabbit Cage Problem (Basic Loop)**  
**Task:** Solve the classic chickens-and-rabbits problem without using `for-else`.  

#### **Requirements:**  
1. Given total `heads` and `legs` counts  
2. Try all possible chicken counts (0 to heads)  
3. Use a flag variable `found` to track solution status  
4. Print solution if found, else print "No solution"  

#### **Example Output:**  
```
Heads: 10, Legs: 28  
Chickens: 6, Rabbits: 4  

Heads: 10, Legs: 25  
No solution.
```

#### **Skills Tested:**  
- For loops  
- Flag variables  
- Conditional logic  

---

### **練習五：鷄兔同籠問題（基礎循環）**  
**任務：** 不使用`for-else`解決鷄兔同籠問題。  

#### **要求：**  
1. 給定總頭數和腿數  
2. 嘗試所有可能的雞的數量（0到頭數）  
3. 使用標誌變量`found`追蹤解決狀態  
4. 找到解則打印，否則打印「無解」  

#### **範例輸出：**  
```
頭數: 10, 腿數: 28  
雞: 6, 兔: 4  

頭數: 10, 腿數: 25  
無解。
```

---

#### **Exercise 5 (Basic Loop)**
```python

heads = int(input("Heads: "))
legs = int(input("Legs: "))
found = False

for chickens in range(heads + 1):
    rabbits = heads - chickens
    if 2*chickens + 4*rabbits == legs:
        print(f"Chickens: {chickens}, Rabbits: {rabbits}")
        found = True
        break

if not found:
    print("No solution.")

```

### **Exercise 6: Chicken-Rabbit Cage Problem (For-Else)**  
**Task:** Solve the same problem using `for-else` to eliminate the flag variable.  

#### **New Requirement:**  
1. Replace the flag with `for-else` structure  
2. The `else` block should handle "no solution" case  

#### **Key Difference:**  
- The `else` executes only if loop completes without `break`  

#### **Example Output:**  
```
Heads: 35, Legs: 94  
Chickens: 23, Rabbits: 12  

Heads: 7, Legs: 19  
No solution exists.
```

### **練習六：鷄兔同籠問題（For-Else結構）**  
**任務：** 使用`for-else`解決同樣問題。  

#### **新要求：**  
1. 用`for-else`替代標誌變量  
2. `else`區塊處理「無解」情況  

#### **關鍵區別：**  
- `else`僅在循環未觸發`break`時執行  

#### **範例輸出：**  
```
頭數: 35, 腿數: 94  
雞: 23, 兔: 12  

頭數: 7, 腿數: 19  
無解。
```

---
 
#### **Exercise 6 (For-Else)**
```python

heads = int(input("Heads: "))
legs = int(input("Legs: "))

for chickens in range(heads + 1):
    rabbits = heads - chickens
    if 2*chickens + 4*rabbits == legs:
        print(f"Chickens: {chickens}, Rabbits: {rabbits}")
        break
else:
    print("No solution exists.")

```
