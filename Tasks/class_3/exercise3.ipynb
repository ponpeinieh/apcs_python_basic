### **Exercise 1: Simple Prime Check**  
**Task:** Check if a number is prime using basic division.  

#### **Requirements:**  
1. Get user input for a number  
2. Handle special cases (≤1, 2, 3)  
3. Check divisibility from 2 to √n  
4. Print whether the number is prime  

#### **Example Output:**  
```
Enter a number: 17
17 is a prime number.
```

#### **Skills Tested:**  
- While loops  
- Mathematical operations  
- Conditional logic  

---

### **練習一：簡單質數檢查**  
**任務：** 用基本除法檢查數字是否為質數。  

#### **要求：**  
1. 獲取用戶輸入的數字  
2. 處理特殊情況（≤1, 2, 3）  
3. 檢查2到√n之間的整除性  
4. 打印結果  

#### **範例輸出：**  
```
請輸入數字: 17
17 是質數。
```

---

### **Solution Code (Simple)**  

```python 
n = int(input("Enter a number: "))

if n <= 1:
    print(f"{n} is not a prime number.")
elif n <= 3:
    print(f"{n} is a prime number.")
else:
    i = 2
    is_prime = True
    while i * i <= n:
        if n % i == 0:
            is_prime = False
            break
        i += 1
    
    print(f"{n} is {'a prime' if is_prime else 'not a prime'} number.")

```

---

### **Exercise 2: Optimized Prime Check**  
**Task:** Improve the prime check by skipping even divisors and multiples of 3.  

#### **Optimizations:**  
1. Immediately eliminate even numbers >2  
2. Skip checking multiples of 2 and 3  
3. Only check divisors of form 6k±1  

#### **Example Output:**  
```
Enter a number: 10000019
10000019 is a prime number.
```

#### **Skills Tested:**  
- Algorithm optimization  
- Mathematical patterns  

---

### **練習二：優化質數檢查**  
**任務：** 通過跳過偶數和3的倍數來優化質數檢查。  

#### **優化：**  
1. 直接排除>2的偶數  
2. 跳過2和3的倍數  
3. 只檢查6k±1形式的除數  

#### **範例輸出：**  
```
請輸入數字: 10000019
10000019 是質數。
```

---

### **Solution Code (Optimized)**  
```python 

n = int(input("Enter a number: "))

if n <= 1:
    print(f"{n} is not a prime number.")
elif n <= 3:
    print(f"{n} is a prime number.")
elif n % 2 == 0 or n % 3 == 0:
    print(f"{n} is not a prime number.")
else:
    i = 5
    is_prime = True
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            is_prime = False
            break
        i += 6
    
    print(f"{n} is {'a prime' if is_prime else 'not a prime'} number.")

```
