### **Exercise 1: Base Conversion and Addition**  
**Task:** Write a program that converts two numbers from any base (binary, octal, decimal, or hexadecimal) to decimal, calculates their sum, and displays the result in multiple bases.  

**Requirements:**  
1. Use `input()` to get two numbers with optional prefixes (`0b`, `0o`, `0x`, or none for decimal)  
2. Convert to decimal using `int(number_str, 0)`  
3. Calculate the sum  
4. Convert the result to binary (`bin()`), octal (`oct()`), and hexadecimal (`hex()`)  
5. Format output using f-strings  

**Example:**  
```
Enter first number: 0b1010  
Enter second number: 0o12  
Sum in decimal: 20  
Binary: 0b10100  
Octal: 0o24  
Hexadecimal: 0x14
```

**Skills Tested:**  
- Base conversion with `int()`  
- Number formatting with `bin()/oct()/hex()`  
- f-string formatting

### **練習一：進位轉換與加法運算**  
**任務：** 撰寫程式將兩個任意進位數字轉換為十進位後相加，並將結果以多種進位顯示。  

**要求：**  
1. 使用 `input()` 獲取兩個數字（可含 `0b`、`0o`、`0x` 前綴或無前綴）  
2. 用 `int(數字字串, 0)` 轉為十進位  
3. 計算兩數之和  
4. 將結果轉為二進位(`bin()`)、八進位(`oct()`)和十六進位(`hex()`)  
5. 使用 f-string 格式化輸出  

**範例：**  
```
輸入第一個數字: 0b1010  
輸入第二個數字: 0o12  
十進位和: 20  
二進位: 0b10100  
八進位: 0o24  
十六進位: 0x14
```

**訓練技能：**  
- 使用 `int()` 進行進位轉換  
- `bin()/oct()/hex()` 數字格式化  
- f-string 輸出

### **Exercise 3: Variable Inspection Lab**  
**Task:** Create a program to demonstrate variable properties using `type()` and `id()`  

**Requirements:**  
1. Create these variables:  
   - `a = 10`  
   - `b = 10.5`  
   - `c = "Hello"`  
   - `d = [1,2,3]`  
   - `e = a` (alias of `a`)  
2. For each variable:  
   - Print its value  
   - Print its type using `type()`  
   - Print its memory address using `id()`  
3. Special checks:  
   - Compare `id(a)` and `id(e)`  
   - Modify `d[0] = 99` and show ID before/after  

**Example Output:**  
```
Variable: a = 10
Type: <class 'int'> | ID: 1407088000
Variable: e = 10
Type: <class 'int'> | ID: 1407088000 (Same as a)

List before: [1, 2,3] ID: 246810121536
List after: [99,2,3] ID: 246810121536 (Same ID!)
```

**Skills Tested:**  
- Variable assignment  
- Using `type()` and `id()`  
- Understanding mutability vs immutability

### **Exercise 4: Boolean Conversion Lab**

**Requirements:**  
1. Test these values:  
   - **Integer**: `0`, `-5`, `100`

```
Value: 0 (Type: <class 'int'>) → False
Value: -5 (Type: <class 'int'>) → True
Value: "" (Type: <class 'str'>) → False
Value: [0] (Type: <class 'list'>) → True
```

**Skills Tested:**  
- Boolean conversion rules  
- Understanding truthy/falsy values
