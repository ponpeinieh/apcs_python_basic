### **練習一：BMI計算器**  
**任務：** 創建一個計算身體質量指數(BMI)並分類的程序。  

#### **要求：**  
1. 獲取用戶輸入：  
   - 體重（公斤）  
   - 身高（米）  
2. 用以下公式計算BMI：  
   ```python
   bmi = weight / (height ** 2)
   ```  
3. 分類標準：  
   - 體重過輕: BMI < 18.5  
   - 正常範圍: 18.5 ≤ BMI < 25  
   - 過重: 25 ≤ BMI < 30  
   - 肥胖: BMI ≥ 30  
4. 輸出BMI(保留2位小數)和分類結果  

#### **範例輸出：**  
```
請輸入體重（公斤）：70
請輸入身高（米）：1.75
您的BMI是: 22.86，分類為: 正常範圍
```

#### **訓練技能：**  
- 用戶輸入/輸出  
- 數學運算  
- 條件判斷 (`if/elif/else`)  

---

### **Sample Solution Code**  
```python
# English Version
weight = float(input("Enter weight (kg): "))
height = float(input("Enter height (m): "))

bmi = weight / (height ** 2)

if bmi < 18.5:
    category = "Underweight"
elif 18.5 <= bmi < 25:
    category = "Normal"
elif 25 <= bmi < 30:
    category = "Overweight"
else:
    category = "Obese"

print(f"Your BMI is: {bmi:.2f}, Category: {category}")
```

### **練習二：使用索引處理學生記錄**  
**任務：** 使用平行列表和索引循環處理學生成績。  

#### **要求：**  
1. 建立兩個列表：  
   ```python
   names = ["小明", "小花", "小華"]
   scores = [85, 92, 78]
   ```  
2. 使用`for`循環和`range()`：  
   - 印出每個學生姓名和分數  
   - 計算並印出平均分  
3. 格式化輸出結果  

#### **範例輸出：**  
```
學生: 小明, 分數: 85
學生: 小花, 分數: 92
學生: 小華, 分數: 78
平均分: 85.00
```

---

### **Sample Solution Code**  
#### **Exercise 2 Solution:**  
```python
# English
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]
total = 0

for i in range(len(names)):
    print(f"Student: {names[i]}, Score: {scores[i]}")
    total += scores[i]

print(f"Average Score: {total/len(scores):.2f}")

```

### **練習三：使用zip()處理學生記錄**  
**任務：** 用`zip()`改進學生成績處理。  

#### **要求：**  
1. 使用練習二的相同列表  
2. 用`zip()`替代索引循環：  
   - 印出姓名/分數組合  
   - 找出並印出最高分學生  
3. 增加條件判斷：  
   ```python
   if score > 90:
       print("優秀！")
   ```

#### **範例輸出：**  
```
學生: 小明 (85)
學生: 小花 (92) - 優秀！
學生: 小華 (78)
最高分: 小花 (92分)
```
---

#### **Exercise 3 Solution:**  

```python
# English
max_score = 0
top_student = ""

for name, score in zip(names, scores):
    print(f"Student: {name} ({score})", end="")
    if score > 90:
        print(" - Excellent!")
    else:
        print()
    
    if score > max_score:
        max_score = score
        top_student = name

print(f"Top Student: {top_student} with {max_score} points")
 
```
