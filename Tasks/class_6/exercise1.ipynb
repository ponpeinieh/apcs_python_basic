### **Key Note: Namespace Collision Risk**  

⚠️ **Danger of Direct Imports:**  
If you use simple `from ... import` for both modules:  
```python
from math import pi           # Imports 3.141592653589793
from math_operations import pi  # Imports 3.14159 (overwrites!)
```
The second import **silently overwrites** the first `pi` in your namespace.  
- You'll only access `math_operations.pi`  
- No error occurs, but calculations may be wrong  
- Hard to debug since Python won't warn you  

**Safe Alternatives:**  
1. Use module prefixes:  
   ```python 
   import math
   import math_operations
   math.pi          # 3.141592653589793
   math_operations.pi  # 3.14159
   ```  
2. Use aliases:  
   ```python
   from math import pi as math_pi
   from math_operations import pi as custom_pi
   ```

### **關鍵說明：命名空間衝突風險**  
⚠️ **直接導入的危險性:**  
若對兩個模組都使用簡單的 `from ... import`:  
```python
from math import pi           # 導入 3.141592653589793
from math_operations import pi  # 導入 3.14159 (會覆蓋!)
```  
第二次導入會**無聲地覆寫**第一個 `pi`  
- 你只能訪問到 `math_operations.pi`  
- 不會報錯，但計算結果可能錯誤  
- 難以除錯，因為 Python 不會警告  

**安全替代方案:**  
1. 使用模組前綴:  
   ```python
   import math
   import math_operations
   math.pi          # 3.141592653589793
   math_operations.pi  # 3.14159
   ```  
2. 使用別名:  
   ```python
   from math import pi as math_pi
   from math_operations import pi as custom_pi
   ```
