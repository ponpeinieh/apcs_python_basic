### **練習一：探索 `datetime` 模組**  
**任務：** 利用 `datetime` 模組的各種功能進行日期和時間操作  

#### **要求:**  
1. 使用 `datetime.now()` 獲取當前的日期和時間。
2. 使用 `date` 類別建立一個特定日期。
3. 使用 `time` 類別建立一個特定時間。
4. 計算今天與特定日期之間的天數差。
5. 使用 `timedelta` 增加一天到今天的日期。
6. 使用 `strftime()` 將當前日期和時間格式化為字串。
7. 使用 `strptime()` 將日期字串轉換為 `datetime` 物件。

#### **範例輸出:**  
```
當前日期和時間: 2023-07-18 14:30:00
特定日期: 2024-10-04
特定時間: 14:30:00
距離特定日期的天數:  78
明天的日期: 2023-07-19
格式化的日期和時間: 2023-07-18 14:30:00
轉換後的 datetime 物件: 2024-10-04 14:30:00
```

---

### **Sample Solution Code**  
```python
from datetime import datetime, date, time, timedelta

# 1. 獲取當前日期和時間
now = datetime.now()
print("當前日期和時間:", now)

# 2. 建立一個特定日期
specific_date = date(2024, 10, 4)
print("特定日期:", specific_date)

# 3. 建立一個特定時間
specific_time = time(14, 30)  # 14:30
print("特定時間:", specific_time)

# 4. 計算兩個日期之間的差異
today = date.today()
delta = specific_date - today
print("距離特定日期的天數:", delta.days)

# 5. 使用 timedelta 增加一天
tomorrow = today + timedelta(days=1)
print("明天的日期:", tomorrow)

# 6. 格式化日期
formatted_date = now.strftime("%Y-%m-%d %H:%M:%S")
print("格式化的日期和時間:", formatted_date)

# 7. 從字串轉換為 datetime
date_string = "2024-10-04 14:30:00"
converted_date = datetime.strptime(date_string, "%Y-%m-%d %H:%M:%S")
print("轉換後的 datetime 物件:", converted_date)
```

### **範例解答程式碼**  
```python
from datetime import datetime, date, time, timedelta

# 1. 獲取當前日期和時間
now = datetime.now()
print("當前日期和時間:", now)

# 2. 建立一個特定日期
specific_date = date(2024, 10, 4)
print("特定日期:", specific_date)

# 3. 建立一個特定時間
specific_time = time(14, 30)  # 14:30
print("特定時間:", specific_time)

# 4. 計算兩個日期之間的差異
today = date.today()
delta = specific_date - today
print("距離特定日期的天數:", delta.days)

# 5. 使用 timedelta 增加一天
tomorrow = today + timedelta(days=1)
print("明天的日期:", tomorrow)

# 6. 格式化日期
formatted_date = now.strftime("%Y-%m-%d %H:%M:%S")
print("格式化的日期和時間:", formatted_date)

# 7. 從字串轉換為 datetime
date_string = "2024-10-04 14:30:00"
converted_date = datetime.strptime(date_string, "%Y-%m-%d %H:%M:%S")
print("轉換後的 datetime 物件:", converted_date)
``` 

### **Exercise 2: Utilizing `timedelta` for Date Calculations**  
**Task:** Use the `timedelta` class to perform various date calculations.

#### **Requirements:**  
1. Calculate the date 30 days from today.
2. Calculate the date 15 days ago from today.
3. Determine how many weeks and days are left until a specific future date.
4. Create a function that takes a starting date and a number of days to add, returning the new date.

#### **Example Output:**  
```

從今天起 30 天後的日期: 2023-08-17
從今天起 15 天前的日期: 2023-07-03
距離下一個生日還有: 45 天（6 週 3 天）
添加天數後的新日期: 2023-09-01
```

---

### **Sample Solution Code**  
```python
from datetime import datetime, timedelta

# 1. 計算從今天起 30 天後的日期
today = datetime.now().date()
date_30_days_later = today + timedelta(days=30)
print("從今天起 30 天後的日期:", date_30_days_later)

# 2. 計算從今天起 15 天前的日期
date_15_days_ago = today - timedelta(days=15)
print("從今天起 15 天前的日期:", date_15_days_ago)

# 3. 確定距離特定未來日期還有多少週和天
next_birthday = datetime(2025, 12, 1)  # 假設下一個生日是 2025 年 12 月 1 日
days_until_birthday = (next_birthday.date() - today).days
weeks = days_until_birthday // 7
days = days_until_birthday % 7
print(f"距離下一個生日還有: {days_until_birthday} 天（{weeks} 週 {days} 天）")

# 4. 創建一個函數，返回添加天數後的新日期
def add_days(start_date, days_to_add):
    return start_date + timedelta(days=days_to_add)

new_date = add_days(today, 45)  # 添加 45 天
print("添加天數後的新日期:", new_date)
```
