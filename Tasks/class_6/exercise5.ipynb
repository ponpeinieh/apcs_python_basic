### **Sample Solution Code**  
```python
import time

# 1. Get current timestamp
current_time = time.time()
print(f"Current timestamp: {current_time}")

# 2. Sleep for 2 seconds
print("Sleeping for 2 seconds...")
time.sleep(2)
print("Awake!")

# 3. Measure execution time of a simple loop
start_time = time.time()
for _ in range(1000000):
    pass  # Simulated operation
end_time = time.time()
execution_time = end_time - start_time
print(f"Execution time: {execution_time:.3f} seconds")

# 4. Format current local time
local_time = time.localtime()
formatted_time = time.strftime("%Y-%m-%d %H:%M:%S", local_time)
print(f"Current local time: {formatted_time}")
```

### **範例解答程式碼**

```
當前時間戳: 1672531199.123456
UTC 時間: 2023-01-01 00:00:00
解析的時間元組: time.struct_time(tm_year=2024, tm_mon=10, tm_mday=4, tm_hour=12, tm_min=30, tm_sec=0, tm_wday=4, tm_yday=277, tm_isdst=-1)
時間差（秒）: 86400.00
```

---

### **Sample Solution Code**  
```python
import time

# 1. Get current timestamp
current_time = time.time()
print(f"Current timestamp: {current_time}")

# 2. Convert current timestamp to UTC time
utc_time = time.gmtime(current_time)
formatted_utc_time = time.strftime("%Y-%m-%d %H:%M:%S", utc_time)
print("UTC time:", formatted_utc_time)

# 3. Parse a predefined formatted time string
time_string = "2025-10-04 12:30:00"
time_tuple = time.strptime(time_string, "%Y-%m-%d %H:%M:%S")
print("Parsed time tuple:", time_tuple)

# 4. Calculate the difference in seconds between two timestamps
future_time = time.mktime(time_tuple)  # Convert time_tuple back to timestamp
time_difference = future_time - current_time
print(f"Difference in seconds: {time_difference:.2f}")
```

### **範例解答程式碼**  
```python
import time

# 1. 獲取當前時間戳
current_time = time.time()
print(f"當前時間戳: {current_time}")

# 2. 將當前時間戳轉換為 UTC 時間
utc_time = time.gmtime(current_time)
formatted_utc_time = time.strftime("%Y-%m-%d %H:%M:%S", utc_time)
print("UTC 時間:", formatted_utc_time)

# 3. 解析預定義的格式化時間字串
time_string = "2025-10-04 12:30:00"
time_tuple = time.strptime(time_string, "%Y-%m-%d %H:%M:%S")
print("解析的時間元組:", time_tuple)

# 4. 計算兩個時間戳之間的秒數差
future_time = time.mktime(time_tuple)  # 將 time_tuple 轉回時間戳
time_difference = future_time - current_time
print(f"時間差（秒）: {time_difference:.2f}")
``` 

 `time.mktime()` function 

### **Function Signature**
```python
time.mktime(t)
```

### **Parameters**
- **t**: This is a `struct_time` object or a tuple that represents local time. The tuple should contain 9 elements, which correspond to the following fields:
  1. **Year** (e.g., 2024)
  2. **Month** (1-12)
  3. **Day** (1-31)
  4. **Hour** (0-23)
  5. **Minute** (0-59)
  6. **Second** (0-60, where 60 accounts for leap seconds)
  7. **Weekday** (0-6, where Monday is 0)
  8. **Julian Day** (1-366)
  9. **DST flag** (0, 1, or -1, indicating whether Daylight Saving Time is in effect)

### **Return Value**
The function returns a floating-point number representing the number of seconds since the epoch. If the input values are out of range or invalid, it raises an `OverflowError` or `ValueError`.

### **Usage Example**
Here’s a simple example of how to use `time.mktime()`:

```python
import time

# Create a struct_time object representing a specific local time
local_time_tuple = (2024, 10, 4, 12, 30, 0, 4, 277, -1)  # (Year, Month, Day, Hour, Minute, Second, Weekday, Julian Day, DST flag)

# Convert local time to seconds since the epoch
timestamp = time.mktime(local_time_tuple)
print("Timestamp:", timestamp)

# Convert back to struct_time for verification
converted_time = time.localtime(timestamp)
print("Converted back to struct_time:", converted_time)
```

 `time.mktime()` 函數：

### **函數簽名**
```python
time.mktime(t)
```

### **參數**
- **t**: 這是一個 `struct_time` 對象或表示當地時間的元組。元組應包含 9 個元素，對應以下字段：
  1. **年份**（例如，2024）
  2. **月份**（1-12）
  3. **日期**（1-31）
  4. **小時**（0-23）
  5. **分鐘**（0-59）
  6. **秒數**（0-60，其中 60 用於閏秒）
  7. **星期幾**（0-6，其中星期一為 0）
  8. **儒略日**（1-366）
  9. **夏令時間標誌**（0、1 或 -1，指示是否適用夏令時間）

### **返回值**
該函數返回一個浮點數，表示自紀元以來的秒數（自 1970 年 1 月 1 日 00:00:00 UTC 起）。如果輸入值超出範圍或無效，則會引發 `OverflowError` 或 `ValueError`。

### **使用示例**
以下是一個簡單的 `time.mktime()` 使用示例：

```python
import time

# 創建一個表示特定當地時間的 struct_time 對象
local_time_tuple = (2024, 10, 4, 12, 30, 0, 4, 277, -1)  # （年份、月份、日期、小時、分鐘、秒數、星期幾、儒略日、夏令時間標誌）

# 將當地時間轉換為自紀元以來的秒數
timestamp = time.mktime(local_time_tuple)
print("時間戳:", timestamp)

# 轉換回 struct_time 以進行驗證
converted_time = time.localtime(timestamp)
print("轉換回 struct_time:", converted_time)
```
