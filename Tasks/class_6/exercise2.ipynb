### **Exercise 1: Parking Fee Calculator**  
**Task:** Calculate parking fees using time rounding rules  

#### **Requirements:**  
1. Parking charges 30 NTD per 15-minute block  
2. Round UP to next block for billing (`math.ceil`)  
3. Display actual parked time vs billed time  

#### **Test Cases:**  
```python
parking_times = [12, 15, 16, 30, 47]  # Minutes
```

#### **Example Output:**  
```
Parked 12 mins → Billed 15 mins (NTD 30)
Parked 15 mins → Billed 15 mins (NTD 30)  
Parked 16 mins → Billed 30 mins (NTD 60)
Parked 47 mins → Billed 60 mins (NTD 120)
```

---

### **練習一：停車費計算器**  
**任務：** 使用時間進位規則計算停車費  

#### **要求:**  
1. 每15分鐘收費30新台幣  
2. 計費單位無條件進位 (`math.ceil`)  
3. 顯示實際停車時間 vs 計費時間  

#### **測試案例:**  
```python
parking_times = [12, 15, 16, 30, 47]  # 分鐘
```

#### **範例輸出:**  
```
停車 12 分鐘 → 計費 15 分鐘 (30元)
停車 15 分鐘 → 計費 15 分鐘 (30元)  
停車 16 分鐘 → 計費 30 分鐘 (60元)
停車 47 分鐘 → 計費 60 分鐘 (120元)
```

---

### **Sample Solution Code**  
```python
# English Version
import math

RATE_PER_BLOCK = 30  # NTD
MINUTES_PER_BLOCK = 15

def calculate_fee(minutes):
    blocks = math.ceil(minutes / MINUTES_PER_BLOCK)
    return blocks * RATE_PER_BLOCK

for time in parking_times:
    fee = calculate_fee(time)
    billed_mins = math.ceil(time / MINUTES_PER_BLOCK) * MINUTES_PER_BLOCK
    print(f"Parked {time} mins → Billed {billed_mins} mins (${fee:.2f})")
```

### **練習二：隨機停車費計算器**

#### **要求:**  
1. 每15分鐘收費30新台幣  
2. 計費單位無條件進位 (`math.ceil`)  
3. 隨機生成停車時間，範圍從1到120分鐘  

#### **測試案例:**
