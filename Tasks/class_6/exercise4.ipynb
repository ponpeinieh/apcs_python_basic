### **Exercise 1: Importing Custom Modules**  
**Task:** Import a custom module from a specified path  

#### **Requirements:**  
1. Create a custom module named `my_module.py` in the path `/path/to/your/module`.
2. Use `sys.path.append()` to add the custom module path to the Python path.
3. Import a function (e.g., `my_function`) from `my_module`.
4. Call the imported function and display the result.

#### **Test Cases:**  
```python
# Sample function in my_module.py
def my_function():
    return "Hello from my_module!"
```

#### **Example Output:**  
```
Hello from my_module!
```

### **練習一：導入自定義模組**  
**任務：** 從指定路徑導入自定義模組  

#### **要求:**  
1. 在路徑 `/path/to/your/module` 中創建一個名為 `my_module.py` 的自定義模組。
2. 使用 `sys.path.append()` 將自定義模組路徑添加到 Python 路徑中。
3. 從 `my_module` 導入一個函數（例如 `my_function`）。
4. 調用導入的函數並顯示結果。

#### **測試案例:**  
```python
# my_module.py 中的範例函數
def my_function():
    return "來自 my_module 的問候！"
```

#### **範例輸出:**  
```
來自 my_module 的問候！
```

---

### **Sample Solution Code**  
```python
import sys

# Add custom module path
sys.path.append('/path/to/your/module')

# Import the custom module
from my_module import my_function

# Call the function and display the result
result = my_function()
print(result)
```

### **Exercise 2: Command Line Argument Processor**  
**Task:** Build a script that processes command line arguments  

#### **Requirements:**  
1. Use `sys.argv` to access command line arguments  
2. Handle different argument counts  
3. Demonstrate script termination methods  

#### **Test Cases:**  
```bash
$ python script.py Alice 25
$ python script.py Bob
$ python script.py
```

#### **Example Output:**  
```
# With 2 arguments

Name: Bob, Age: Unknown
```

### **練習二：命令列參數處理器**  
**任務：** 建立處理命令列參數的腳本  

#### **要求:**  
1. 使用 `sys.argv` 讀取命令列參數  
2. 處理不同數量的參數  
3. 示範腳本終止方法  

#### **測試案例:**  
```bash
$ python script.py Alice 25
$ python script.py Bob
$ python script.py
```

#### **範例輸出:**  
```
# 兩個參數時
姓名: Alice, 年齡: 25

# 一個參數時
姓名: Bob, 年齡: 未知

# 無參數時
錯誤: 請提供至少1個參數
退出代碼 1
```

---

### **Sample Solution Code**  
```python
# English Version
import sys

def process_arguments():
    if len(sys.argv) < 2:
        print("Error: Please provide at least 1 argument")
        sys.exit(1)
    
    name = sys.argv[1]
    age = sys.argv[2] if len(sys.argv) > 2 else "Unknown"
    
    print(f"Name: {name}, Age: {age}")

if __name__ == "__main__":
    process_arguments()

```

### **Key sys Module Features**  
| Function | Purpose | 用途 |  
|----------|---------|------|  
| `sys.argv` | Access command line arguments | 獲取命令列參數 |  
| `sys.exit()` | Terminate script with status code | 終止腳本並返回狀態碼 |  
| `sys.path` | Module search paths | 模組搜索路徑 |
