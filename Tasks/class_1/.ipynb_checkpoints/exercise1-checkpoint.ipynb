### **Exercise 1: Area and Circumference of a Circle**  
**Task:** Write a Python program that calculates the **area** and **circumference** of a circle given its radius. Use `math.pi` for the calculation.  

**Example:**  
- If radius `r = 5`, the output should be:  
  - Area = 78.53981633974483  
  - Circumference = 31.41592653589793  

**Skills Tested:**  
- Using `math.pi`  
- Basic arithmetic operations (`*` and `**`)  
- Variable definition and output formatting

### **Exercise 3: Statistical Calculations**  
**Task:** Write a Python program that calculates the **mean**, **variance**, and **standard deviation** of five numbers (defined as variables `num1` to `num5`). Use `math.sqrt()` for standard deviation.  

**Example:**  
- For `num1=2, num2=4, num3=4, num4=5, num5=5`, the output should be:  
  - Mean = 4.0  
  - Variance = 1.2  
  - Standard deviation = 1.0954451150103321  

**Skills Tested:**  
- Using `math.sqrt()` for standard deviation  
- Implementing statistical formulas (variance = mean of squared differences)  
- Handling multiple variables (optional: replace with lists for scalability)

### **Exercise 4: Verifying sin²θ + cos²θ = 1 Identity**  
**Task:** Write a Python program that verifies the trigonometric identity sin²θ + cos²θ = 1 for a user-input angle in degrees.  

**Requirements:**  
1. Use `input()` to get an angle in degrees  
2. Convert the input to `float`  
3. Use `math` module to compute sin²θ + cos²θ  
4. Print the result using f-strings:  
   - "For angle {θ}: sin²θ + cos²θ = {result}"  

**Example:**  
- Input: `30`  
  Output: `For angle 30.0: sin²θ + cos²θ = 1.0`  

**Skills Tested:**  
- Using `math.sin()`, `math.cos()`, and `math.radians()`  
- Unit conversion (degrees to radians)  
- Floating-point arithmetic  
- f-string formatting  

---

### **練習 4：驗證 sin²θ + cos²θ = 1 的三角恆等式**  
**任務：** 寫一個 Python 程式，驗證使用者輸入角度（度）是否滿足 sin²θ + cos²θ = 1 的恆等式。  

**要求：**  
1. 使用 `input()` 獲取角度輸入  
2. 將輸入轉換為 `float`  
3. 使用 `math` 模組計算 sin²θ + cos²θ  
4. 用 f-string 輸出結果：  
   - "角度 {θ}：sin²θ + cos²θ = {結果}"  

**範例：**  
- 輸入：`30`  
  輸出：`角度 30.0：sin²θ + cos²θ = 1.0`  

**訓練技能：**  
- 使用 `math.sin()`, `math.cos()` 和 `math.radians()`  
- 單位轉換（角度→弧度）  
- 浮點數運算  
- f-string 格式化輸出
