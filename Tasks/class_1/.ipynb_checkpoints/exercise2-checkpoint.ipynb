### **Exercise 3: f-string Formatting for Numbers and Alignment**  
**Task:** Use f-strings to format product information with specific alignment rules.  

**Requirements:**  
1. Store these variables:  
   - Product name (≤6 chars)  
   - Price (float)  
   - Stock quantity (integer)  
2. Format with:  
   - Name: 12-char width, left-aligned  
   - Price: 2 decimal places  
   - Stock: 5-char width, right-aligned  

**Example Output:**  
```
| Product  | Price   | Stock |
| Laptop   | 999.99  |    15 |
```

**Skills Tested:**  
- f-string formatting  
- Number precision control  
- Text alignment  

---

### **練習三：使用 f-string 格式化數字與對齊**  
**任務：** 使用 f-string 格式化產品資訊並符合對齊要求。  

**要求：**  
1. 儲存變數：  
   - 產品名稱（≤6 字元）  
   - 價格（浮點數）  
   - 庫存數量（整數）  
2. 格式化規則：  
   - 名稱：佔 12 字元寬度，左對齊  
   - 價格：顯示 2 位小數  
   - 庫存：佔 5 字元寬度，右對齊  

**範例輸出：**  
```
| 產品     | 價格    | 庫存 |
| 筆記本   | 999.99  |   15 |
```

**訓練技能：**  
- f-string 格式化  
- 數字精度控制  
- 文字對齊  

---

### **Sample Solution for Exercise 3:**
```python
# Variables
product = "Laptop"
price = 999.987
stock = 15

# Formatted output
header = f"| {'Product':<12} | {'Price':<8} | {'Stock':>5} |"
divider = "-" * len(header)
row = f"| {product:<12} | {price:.2f}  | {stock:>5} |"

print(header)
print(divider)
print(row)
```
