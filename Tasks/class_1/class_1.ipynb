### **Exercise 1: Escape Character Fun**  
**Task:** Write a Python program that uses escape characters to create a simple text art and format a message.  

#### **Requirements:**  
1. **Text Art:** Use `\n` and `\t` to draw a simple cat:  

   ```
   /\_/\  
  ( o.o )  
   > ^ <
   ```  

2. **Message Formatting:** Print this sentence with proper spacing:  
   `"Python is awesome! Learn it today!"`  
   - Use `\t` to indent the first line  
   - Use `\n` to split into two lines  

#### **Example Output:**  
```  
    /\_/\  
   ( o.o )  
    > ^ <  

	Python is awesome!  
Learn it today!  
```  

#### **Skills Tested:**  
- Basic escape characters (`\n`, `\t`)  
- Simple text formatting

### **練習一：跳脫字元小遊戲**  
**任務：** 使用跳脫字元製作簡單的文字圖案和格式化訊息。  

#### **要求：**  
1. **文字圖案：** 用 `\n` 和 `\t` 畫一隻小貓：  
   ```  
    /\_/\  
   ( o.o )  
    > ^ <  
   ```  
2. **訊息格式化：** 印出以下句子並正確排版：  
   `"Python 很棒！今天開始學習吧！"`  
   - 用 `\t` 縮排第一行  
   - 用 `\n` 分成兩行  

#### **範例輸出：**  
```  
    /\_/\  
   ( o.o )  
    > ^ <  

	Python 很棒！  
今天開始學習吧！  
```  

#### **訓練技能：**  
- 基本跳脫字元 (`\n`, `\t`)  
- 簡單文字排版  

---  

### **Sample Solution Code (解答範例):**  
```python  
# 1. 小貓文字圖案  
print("    /\\_/\\")  
print("   ( o.o )")  
print("    > ^ <\n")  

# 2. 格式化訊息  
print("\tPython 很棒！")  
print("今天開始學習吧！")  
```  

### **教學重點：**  
1. `\n` = 換行  
2. `\t` = 水平縮排 (Tab)  
3. `\\` = 顯示反斜線 (因為單獨的 `\` 是跳脫字元的開頭)
