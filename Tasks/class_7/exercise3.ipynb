### **練習一：使用 `raise` 進行異常處理**  
**任務：** 編寫一個程序，演示如何使用 `raise` 語句故意觸發異常。

#### **要求：**  
1. 創建一個函數 `check_positive(number)`，檢查一個數字是否為正數。
2. 如果該數字不是正數，使用 `raise` 觸發 `ValueError` 並附上自定義消息。
3. 實現一個 `main()` 函數，調用 `check_positive()` 並處理任何引發的異常。

#### **範例輸出：**  
```
捕獲到異常：ValueError: 數字必須是正數。
```
```
該數字是正數。
```

---

### **Sample Solution Code**  
```python
def check_positive(number):
    if number <= 0:
        raise ValueError("The number must be positive.")  # 觸發異常
    print("The number is positive.")

def main():
    try:
        check_positive(-5)  # 測試用例，這將引發 ValueError
    except ValueError as e:
        print(f"Caught an exception: {e}")  # 處理異常

    try:
        check_positive(10)  # 正常用例
    except ValueError as e:
        print(f"Caught an exception: {e}")

# 執行主程式
main()
```

### **練習二：引發自定義異常**  
**任務：** 創建一個自定義異常類，並使用它來處理程序中的特定錯誤條件。

#### **要求：**  
1. 定義一個自定義異常類 `NegativeNumberError`，繼承自內建的 `Exception` 類。
2. 創建一個函數 `check_positive(number)`，如果數字為負數則引發 `NegativeNumberError`。
3. 實現一個 `main()` 函數，調用 `check_positive()` 並處理自定義異常。

#### **範例輸出：**  
```
捕獲到異常：NegativeNumberError: 數字不能為負數。
```
```
該數字是正數。
```

---

### **Sample Solution Code**  
```python
class NegativeNumberError(Exception):
    """自定義異常類，表示負數錯誤"""
    pass

def check_positive(number):
    if number < 0:
        raise NegativeNumberError("The number cannot be negative.")  # 引發自定義異常
    print("The number is positive.")

def main():
    try:
        check_positive(-5)  # 測試用例，這將引發 NegativeNumberError
    except NegativeNumberError as e:
        print(f"Caught an exception: {e}")  # 處理自定義異常

    try:
        check_positive(10)  # 正常用例
    except NegativeNumberError as e:
        print(f"Caught an exception: {e}")

# 執行主程式
main()
```

### **練習三：BMI 計算程式中的例外處理**

#### **任務：** 修改一個 BMI 計算程式，以處理各種異常情況，確保在用戶輸入無效時程式不會崩潰。

#### **要求：**
1. 定義一個自定義異常類 `InvalidInputError`，用於處理無效的身高或體重輸入。
2. 實現一個函數 `calculate_bmi(weight, height)` 來計算 BMI。
3. 在 `main()` 函數中，提示用戶輸入體重和身高，並驗證輸入：
   - 捕獲 `ValueError`，以處理非數字輸入。
   - 對於負數或零的值，拋出 `InvalidInputError`。
4. 顯示適當的錯誤訊息，並提示用戶重新輸入有效值，直到成功輸入為止。

#### **範例輸出：**
```
請輸入體重 (公斤): abc
請輸入有效的數值！
請輸入體重 (公斤): -70
體重必須是正數！
請輸入體重 (公斤): 70
請輸入身高 (公尺): 0
身高必須是正數！
請輸入身高 (公尺): 1.75
您的 BMI 是: 22.86
```

---

### **Sample Solution Code**
```python
class InvalidInputError(Exception):
    """自定義異常類別，用於處理無效的身高或體重輸入"""
    pass

def calculate_bmi(weight, height):
    """計算 BMI"""
    return weight / (height ** 2)

def main():
    while True:
        try:
            weight = input("請輸入體重 (公斤): ")
            weight = float(weight)  # 嘗試將輸入轉換為浮點數
            if weight <= 0:
                raise InvalidInputError("體重必須是正數！")

            height = input("請輸入身高 (公尺): ")
            height = float(height)  # 嘗試將輸入轉換為浮點數
            if height <= 0:
                raise InvalidInputError("身高必須是正數！")

            bmi = calculate_bmi(weight, height)
            print(f"您的 BMI 是: {bmi:.2f}")
            break  # 成功計算 BMI 後跳出循環

        except ValueError:
            print("請輸入有效的數值！")
        except InvalidInputError as e:
            print(e)  # 列印自定義異常訊息

# 執行主程式
main()
```
