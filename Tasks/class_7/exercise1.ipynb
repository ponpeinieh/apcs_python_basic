### **練習三：異常處理練習**  
**任務：** 編寫代碼以處理特定異常：`ZeroDivisionError`、`TypeError` 和 `IndexError`。

#### **要求:**  
1. 創建一個函數，執行兩個數字的除法並處理 `ZeroDivisionError`。
2. 創建一個函數，嘗試串聯兩個輸入參數，並在發生 `TypeError` 時報告錯誤消息。
3. 創建一個函數，根據索引從列表中檢索元素，處理 `IndexError`。

每個函數在捕獲異常時應打印出適當的消息。

#### **範例輸出:**  
```
Cannot divide by zero!
Cannot concatenate the two input parameters.
IndexError: Index out of range.
```

---

### **Sample Solution Code**  
```python
# Function to handle ZeroDivisionError
def safe_division(a, b):
    try:
        result = a / b
        print(f"Result: {result}")
    except ZeroDivisionError:
        print("Cannot divide by zero!")

# Function to handle TypeError
def safe_concatenation(a, b):
    try:
        result = a + b
        print(f"Result: {result}")
    except TypeError:
        print("Cannot concatenate the two input parameters.")

# Function to handle IndexError
def safe_list_access(my_list, index):
    try:
        element = my_list[index]
        print(f"Element at index {index}: {element}")
    except IndexError:
        print("IndexError: Index out of range.")

# Test the functions
safe_division(10, 0)
safe_concatenation("Hello, ", 5)
safe_list_access([1, 2, 3], 5)
```

### **練習四：演示 LBYL 和 EAFP 的除法操作**  
**任務：** 編寫代碼以演示兩種錯誤處理概念：**先檢查再行動（LBYL）** 和 **寧可請求原諒而非請求許可（EAFP）**。

#### **要求:**  
1. **LBYL 範例**：創建一個函數，在執行除法之前檢查除數是否為零。
2. **EAFP 範例**：創建一個函數，嘗試執行除法，並在發生 `ZeroDivisionError` 時捕獲錯誤。

每個函數應根據結果打印出適當的消息。

#### **範例輸出:**  
```
Result: 5.0
Cannot divide by zero!
Result: 5.0
Cannot divide by zero!
```

---

### **Sample Solution Code**  
```python
# LBYL Example: Look Before You Leap
def safe_division_lbyl(a, b):
    if b == 0:
        print("Cannot divide by zero!")
    else:
        result = a / b
        print(f"Result: {result}")

# EAFP Example: Easier to Ask for Forgiveness than Permission
def safe_division_eafp(a, b):
    try:
        result = a / b
        print(f"Result: {result}")
    except ZeroDivisionError:
        print("Cannot divide by zero!")

# Test the functions
safe_division_lbyl(10, 2)  # Should print the result
safe_division_lbyl(10, 0)  # Should print error message

safe_division_eafp(10, 2)   # Should print the result
safe_division_eafp(10, 0)   # Should print error message
```
