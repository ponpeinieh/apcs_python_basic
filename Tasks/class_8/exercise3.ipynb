### **Exercise 1: Reading CSV Files with the `csv` Module**  
**Task:** Learn how to read CSV files in Python using the `csv` module.

#### **Concepts:**
- **CSV Format:**  
  CSV (Comma-Separated Values) files store tabular data in plain text. Each line represents a row, and values are separated by commas.

- **`csv.reader`:**  
  Provides functionality to read and parse CSV files.

#### **Important Code for Reading CSV:**
```python
import csv

filename = "data.csv"  # Example CSV file

with open(filename, mode="r", encoding="utf-8") as fh:  # Open the file in read mode
    csv_reader = csv.reader(fh)  # Create a CSV reader object
    header = next(csv_reader)  # Read the header (first row)
    print("Header:", header)  # Display the header

    for row in csv_reader:  # Iterate through the rows
        print("Row:", row)  # Display each row
```

### **Exercise 2: Reading CSV Files with `DictReader`**  
**Task:** Learn how to read CSV files in Python using `DictReader` for easier data handling.

#### **Concepts:**
- **CSV Format:**  
  CSV files store tabular data in plain text. Each line represents a row, and values are separated by commas.

- **`csv.DictReader`:**  
  Reads each row as a dictionary, where the keys are the column headers.

#### **Important Code for Reading CSV with `DictReader`:**
```python
# https://catalog.data.gov/dataset/adoptable-pets
import csv

filename = 'Adoptable_Pets.csv'  # Example CSV file

with open(filename) as fi:
    dictReader = csv.DictReader(fi)  # Create a DictReader object
    for record in dictReader:  # Iterate through the records
        # Pet name, Animal Type, Pet Age, Breed
        print(record['Pet name'], record['Animal Type'], record['Pet Age'], record['Breed'], sep=', ')
```

**Task:** Learn how to write data to a CSV file using the `csv.writer` in Python, while converting Gregorian years to Chinese ROC years.

#### **Concepts:**
- **CSV Format:**  
  CSV files store tabular data in plain text. Each line represents a row, and values are separated by commas.

- **`csv.writer`:**  
  Provides functionality to write data to a CSV file.

#### **Important Code for Writing CSV:**
```python
# DC year to Ming-Guo year
import csv

data = []
with open('data.csv', encoding='utf-8-sig') as f:
    csvCursor = csv.reader(f)  # Create a CSV reader object
    n = 0
    for line in csvCursor:  # Iterate through the lines
        if n >= 1:  # Skip the header
            y = line[0]  # Get the first column (year)
            mgn = str(int(y[:4]) - 1911)  # Convert to ROC year
            line[0] = line[0].replace(y[:4], mgn)  # Update the year
        data.append(line)  # Add the line to data
        n += 1

print(data)  # Display the modified data

with open('data_mgn.csv', mode='w', encoding='utf-8-sig', newline='') as f:
    csvCursor = csv.writer(f)  # Create a CSV writer object
    for line in data:  # Write each line to the new CSV
        csvCursor.writerow(line)
```

### **練習三：使用 `csv.writer` 寫入 CSV 檔案**  
**任務：** 學習如何使用 `csv.writer` 在 Python 中寫入 CSV 檔案，同時將公曆年份轉換為中華民國年份。

#### **概念：**
- **CSV 格式：**  
  CSV 檔案以純文本格式儲存表格數據。每行代表一行，值由逗號分隔。

- **`csv.writer`：**  
  提供將數據寫入 CSV 檔案的功能。

#### **重要代碼：**
```python
# DC year to Ming-Guo year
import csv

data = []
with open('data.csv', encoding='utf-8-sig') as f:
    csvCursor = csv.reader(f)  # 創建 CSV 讀取器對象
    n = 0
    for line in csvCursor:  # 遍歷每一行
        if n >= 1:  # 跳過標題
            y = line[0]  # 獲取第一列（年份）
            mgn = str(int(y[:4]) - 1911)  # 轉換為中華民國年份
            line[0] = line[0].replace(y[:4], mgn)  # 更新年份
        data.append(line)  # 將行添加到數據中
        n += 1

print(data)  # 顯示修改後的數據

with open('data_mgn.csv', mode='w', encoding='utf-8-sig', newline='') as f:
    csvCursor = csv.writer(f)  # 創建 CSV 寫入器對象
    for line in data:  # 將每一行寫入新的 CSV
        csvCursor.writerow(line)
```

#### **Note:**  
To avoid extra blank lines in the output CSV file, make sure to include the `newline=''` parameter when opening the file for writing. This prevents the `csv.writer` from adding additional newline characters between records. 

#### **注意：**  
為了避免在輸出的 CSV 檔案中出現額外的空行，請確保在打開檔案以寫入時包含 `newline=''` 參數。這可以防止 `csv.writer` 在記錄之間添加額外的換行符。


### **Exercise 4: Writing to CSV Files with `csv.DictWriter`**  
**Task:** Learn how to write data to a CSV file using `csv.DictWriter` in Python, filtering specific fields and modifying the pet names.

#### **Concepts:**
- **CSV Format:**  
  CSV files store tabular data in plain text. Each line represents a row, and values are separated by commas.

- **`csv.DictWriter`:**  
  Allows you to write dictionaries to a CSV file, making it easy to manage specific fields.

#### **Important Code for Writing CSV with `DictWriter`:**
```python
# Original fields: Animal ID, Intake Type, In Date, Pet name, Animal Type, Pet Age, Pet Size, Color, Breed, Sex, URL Link, Crossing
# Fields reduced to Pet name, Animal Type, Pet Age, Breed
# The Pet name is changed to the capitalized form
import csv

filename = 'Adoptable_Pets.csv'
data = []
with open(filename) as fi:
    dictReader = csv.DictReader(fi)  # Read the original CSV
    for record in dictReader:
        data.append(record)

filename = 'Reduced_Adoptable_Pets.csv'
field_names = ['Pet name', 'Animal Type', 'Pet Age', 'Breed']
with open(filename, mode='w', newline='') as fi:
    # Ignore the extra field values in data
    dictWriter = csv.DictWriter(fi, fieldnames=field_names, extrasaction='ignore')
    dictWriter.writeheader()  # Write the header
    for record in data:
        record['Pet name'] = record['Pet name'].lower().title()  # Capitalize the pet name
        dictWriter.writerow(record)  # Write the filtered record
```

### **練習四：使用 `csv.DictWriter` 寫入 CSV 檔案**  
**任務：** 學習如何使用 `csv.DictWriter` 在 Python 中寫入 CSV 檔案，過濾特定欄位並修改寵物名稱。

#### **概念：**
- **CSV 格式：**  
  CSV 檔案以純文本格式儲存表格數據。每行代表一行，值由逗號分隔。

- **`csv.DictWriter`：**  
  允許您將字典寫入 CSV 檔案，方便管理特定欄位。

#### **重要代碼：**
```python
# 原始欄位：Animal ID, Intake Type, In Date, Pet name, Animal Type, Pet Age, Pet Size, Color, Breed, Sex, URL Link, Crossing
# 欄位縮減為 Pet name, Animal Type, Pet Age, Breed
# 寵物名稱被修改為首字母大寫的形式
import csv

filename = 'Adoptable_Pets.csv'
data = []
with open(filename) as fi:
    dictReader = csv.DictReader(fi)  # 讀取原始 CSV
    for record in dictReader:
        data.append(record)

filename = 'Reduced_Adoptable_Pets.csv'
field_names = ['Pet name', 'Animal Type', 'Pet Age', 'Breed']
with open(filename, mode='w', newline='') as fi:
    # 忽略數據中的額外欄位值
    dictWriter = csv.DictWriter(fi, fieldnames=field_names, extrasaction='ignore')
    dictWriter.writeheader()  # 寫入標題
    for record in data:
        record['Pet name'] = record['Pet name'].lower().title()  # 將寵物名稱首字母大寫
        dictWriter.writerow(record)  # 寫入過濾後的記錄
```
