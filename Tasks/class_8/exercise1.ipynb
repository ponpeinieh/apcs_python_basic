### **Exercise 1: File Handling**  
**Task:** Learn how to open and read from files  

#### **Requirements:**  
1. **Create a text file**  
   Before running your Python code, create a text file named `example.txt` and add the following content:
   ```
   Hello, World!
   Welcome to file handling in Python.
   Enjoy learning!
   ```

2. **Open the file for reading**  
   ```python
   # Open the file in read mode
   file = open('example.txt', 'r')  # Open the file
   content = file.read()  # Read the entire content of the file
   file.close()  # Close the file
   ```

3. **Print the content**  
   ```python
   print(content)  # Display the content read from the file
   ```

4. **Count the number of lines**  
   ```python
   # Count the number of lines in the file
   file = open('example.txt', 'r')  # Open the file again
   lines = file.readlines()  # Read all lines into a list
   line_count = len(lines)    # Get the number of lines
   file.close()  # Close the file
   print(f'Total number of lines: {line_count}')  # Display line count
   ```

5. **Count the number of words**  
   ```python
   # Count the number of words in the file
   word_count = sum(len(line.split()) for line in lines)  # Split lines into words and count
   print(f'Total number of words: {word_count}')  # Display word count
   ```

### **Full Solution:**  
```python
# Step 1: Open the file and read content
file = open('example.txt', 'r')  # Open the file
content = file.read()  # Read the entire content of the file
file.close()  # Close the file

# Step 2: Print the content
print(content)

# Step 3: Count the number of lines
file = open('example.txt', 'r')  # Open the file again
lines = file.readlines()  # Read all lines into a list
line_count = len(lines)    # Get the number of lines
file.close()  # Close the file
print(f'Total number of lines: {line_count}')  # Display line count

# Step 4: Count the number of words
word_count = sum(len(line.split()) for line in lines)  # Split lines into words and count
print(f'Total number of words: {word_count}')  # Display word count
```

### **完整解答：**  
```python
# 步驟 1：打開檔案並讀取內容
file = open('example.txt', 'r')  # 打開檔案
content = file.read()  # 讀取檔案的全部內容
file.close()  # 關閉檔案

# 步驟 2：打印內容
print(content)

# 步驟 3：計算行數
file = open('example.txt', 'r')  # 再次打開檔案
lines = file.readlines()  # 將所有行讀入列表
line_count = len(lines)    # 獲取行數
file.close()  # 關閉檔案
print(f'總行數：{line_count}')  # 顯示行數

# 步驟 4：計算單詞數
word_count = sum(len(line.split()) for line in lines)  # 將行分割為單詞並計數
print(f'總單詞數：{word_count}')  # 顯示單詞數
```

### **Exercise 3: File Handling with `finally` and `with ... as`**  
**Task:** Learn to safely handle files using `finally` and the `with ... as` statement

2. **Open the file using try-finally**  
   Write code to open the file, read the first line using `readline()`, and ensure the file is closed using a `finally` clause.
   ```python
   file = None
   try:
       file = open('example.txt', 'r')
   ```

### **練習三：使用 `finally` 和 `with ... as` 的檔案處理**  
**任務：** 學習如何安全地處理檔案，使用 `finally` 和 `with ... as` 語句  

#### **要求：**  
1. **使用現有的文本檔案**  
   我們將使用之前練習中創建的文本檔案 `example.txt`，內容如下：

   ```
   Hello, World!
   Welcome to file handling in Python.
   Enjoy learning!
   ```

2. **使用 try-finally 打開檔案**  
   編寫代碼以打開檔案，使用 `readline()` 讀取第一行，並確保使用 `finally` 子句關閉檔案。
   ```python
   file = None
   try:
       file = open('example.txt', 'r')
       line = file.readline()
       print(line)  # 打印第一行
   finally:
       if file:
           file.close()  # 確保檔案被關閉
   ```

3. **使用 `with ... as` 打開檔案**  
   現在，使用 `with ... as` 語句重寫代碼以讀取第一行。這種方法自動處理檔案的關閉。
   ```python
   with open('example.txt', 'r') as f:
       line = f.readline()
       print(line)  # 打印第一行
   ```

---

#### **Full Solution:**  
```python
# Step 1: Open the file using try-finally
file = None
try:
    file = open('example.txt', 'r')
    line = file.readline()  # Read the first line
    print(line)  # Print the first line
finally:
    if file:
        file.close()  # Ensure the file is closed

# Step 2: Open the file using with ... as
with open('example.txt', 'r') as f:
    line = f.readline()  # Read the first line
    print(line)  # Print the first line
```

---

#### **完整解答：**  
```python
# 步驟 1：使用 try-finally 打開檔案
file = None
try:
    file = open('example.txt', 'r')
    line = file.readline()  # 讀取第一行
    print(line)  # 打印第一行
finally:
    if file:
        file.close()  # 確保檔案被關閉

# 步驟 2：使用 with ... as 打開檔案
with open('example.txt', 'r') as f:
    line = f.readline()  # 讀取第一行
    print(line)  # 打印第一行
```
