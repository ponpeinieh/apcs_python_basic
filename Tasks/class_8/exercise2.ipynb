### **Exercise 1: Reading Text Files with `tell()` and `seek()`**  
**Task:** Learn how to read text files in Python while managing the file pointer using `tell()` and `seek()`.

#### **Concepts:**
- **`tell()` Method:**  
  Returns the current position of the file pointer in bytes.

- **`seek(offset, whence=0)` Method:**  
  Changes the current position of the file pointer.  
  - `offset`: The number of bytes to move.  
  - `whence`: The reference point for the offset:
    - `os.SEEK_SET` or `0`: The beginning of the file.
    - `os.SEEK_CUR` or `1`: The current position in the file.
    - `os.SEEK_END` or `2`: The end of the file.

> **Note:** When accessing text files, offsets other than `0` can only set the reference point from the start of the file.

#### **Important Code for Reading:**
```python
from os import SEEK_SET  # Import SEEK_SET for reference point

filename = "input1.txt"

with open(filename, encoding="utf-8") as fh:
    print("目前存取位置(位元):", fh.tell())
    t = fh.read(17)  # Read the first 17 characters
    print(t)
    print("目前存取位置(位元):", fh.tell())  # Current byte position
    fh.seek(4, SEEK_SET)  # Move to the 4th byte from the start
    print("目前存取位置(位元):", fh.tell())
    t = fh.read(1)  # Read one character
    print(t)
```

### **練習一：使用 `tell()` 和 `seek()` 讀取文本檔案**  
**任務：** 學習如何在 Python 中讀取文本檔案，同時使用 `tell()` 和 `seek()` 管理檔案指標。

#### **概念：**
- **`tell()` 方法：**  
  返回檔案指標的當前位置（位元）。

- **`seek(offset, whence=0)` 方法：**  
  更改檔案指標的當前位置。  
  - `offset`：要移動的位元數。  
  - `whence`：偏移的參考點：
    - `os.SEEK_SET` 或 `0`：檔案的開頭。
    - `os.SEEK_CUR` 或 `1`：檔案的當前位置。
    - `os.SEEK_END` 或 `2`：檔案的結尾。

> **注意：** 在訪問文本檔案時，偏移量不等於 `0` 只能從檔案的開頭設置參考點。

#### **重要代碼：**
```python
from os import SEEK_SET  # 載入 SEEK_SET 以作為參考點

filename = "input1.txt"
# filename = "example.txt"
# filename = "example_utf8.txt"

with open(filename, encoding="utf-8") as fh:
    print("目前存取位置(位元):", fh.tell())
    t = fh.read(17)  # 讀取text文件, read()讀取字元, 17代表字元(char)數目
    print(t)
    print("目前存取位置(位元):", fh.tell())  # 注意tell()回傳所在的位元位置, utf-8中的每一個中文字佔3個位元寬, 換行符號為\r\n
    fh.seek(4, SEEK_SET)  # 從開頭移動到第 4 個位元 or fh.seek(4,0)
    # fh.seek(13, SEEK_SET)  # 下一行會有 UnicodeDecodeError
    # fh.seek(14, SEEK_SET)  # ok
    print("目前存取位置(位元):", fh.tell())
    t = fh.read(1)  # 讀取一個字元, 注意:如果目前位置不是一個unicode char開始的位元位置, 則會產生UnicodeDecodeError
    print(t)
```

### **Exercise 2: Reading Binary Files with `tell()`, `seek()`, and `read()`**  
**Task:** Learn how to read binary files in Python while managing the file pointer using `tell()`, `seek()`, and `read()`.

#### **Concepts:**
- **Binary Mode:**  
  When a file is opened in binary mode, data is read and written in bytes rather than characters.

- **`tell()` Method:**  
  Returns the current position of the file pointer in bytes.

- **`seek(offset, whence=0)` Method:**  
  Changes the current position of the file pointer in bytes.  
  - `offset`: The number of bytes to move.  
  - `whence`: The reference point for the offset:
    - `os.SEEK_SET` or `0`: The beginning of the file.
    - `os.SEEK_CUR` or `1`: The current position in the file.
    - `os.SEEK_END` or `2`: The end of the file.

#### **Important Code for Reading Binary:**
```python
from os import SEEK_SET, SEEK_CUR, SEEK_END  # Import for reference points

filename = "input1.txt"  # Example binary file

with open(filename, mode="rb") as fh:  # Open the file in binary mode
    print("目前存取位置(位元):", fh.tell())  # Current byte position
    t = fh.read(17)  # Read the first 17 bytes
    print(t)  # Display byte data
    text = t.decode('utf-8')  # Decode bytes to text
    print(text)  # Display the decoded text
    print("目前存取位置(位元):", fh.tell())  # Current byte position

    fh.seek(4, SEEK_SET)  # Move to the 4th byte from the start
    print("目前存取位置(位元):", fh.tell())
    t = fh.read(1)  # Read one byte
    print(t)  # Display byte data
    print("目前存取位置(位元):", fh.tell())

    fh.seek(4, SEEK_CUR)  # Move forward 4 bytes from the current position
    print("目前存取位置(位元):", fh.tell())
    t = fh.read(1)  # Read one byte
    print(t)  # Display byte data

    fh.seek(0, SEEK_END)  # Move to the end of the file
    print("檔案大小(位元數):", fh.tell())  # Display file size in bytes

    fh.seek(-4, SEEK_END)  # Move back 4 bytes from the end
    print("目前存取位置(位元):", fh.tell())
    t = fh.read(1)  # Read one byte
    print(t)  # Display byte data
```

### **練習二：使用 `tell()`、`seek()` 和 `read()` 讀取二進位檔案**  
**任務：** 學習如何在 Python 中讀取二進位檔案，同時使用 `tell()`、`seek()` 和 `read()` 管理檔案指標。

#### **概念：**
- **二進位模式：**  
  當以二進位模式打開檔案時，數據以位元組而非字元的形式被讀取和寫入。

- **`tell()` 方法：**  
  返回檔案指標的當前位置（位元組）。

- **`seek(offset, whence=0)` 方法：**  
  更改檔案指標的當前位置（位元組）。  
  - `offset`：要移動的位元組數。  
  - `whence`：偏移的參考點：
    - `os.SEEK_SET` 或 `0`：檔案的開頭。
    - `os.SEEK_CUR` 或 `1`：檔案的當前位置。
    - `os.SEEK_END` 或 `2`：檔案的結尾。

#### **重要代碼：**
```python
from os import SEEK_SET, SEEK_CUR, SEEK_END  # 載入參考點

filename = "input1.txt"  # 示例二進位檔案

with open(filename, mode="rb") as fh:  # 以二進位模式打開檔案
    print("目前存取位置(位元):", fh.tell())  # 當前位元組位置
    t = fh.read(17)  # 讀取前 17 個位元組
    print(t)  # 顯示位元組數據
    text = t.decode('utf-8')  # 將位元組解碼為文本
    print(text)  # 顯示解碼後的文本
    print("目前存取位置(位元):", fh.tell())  # 當前位元組位置

    fh.seek(4, SEEK_SET)  # 從開頭移動到第 4 個位元組
    print("目前存取位置(位元):", fh.tell())
    t = fh.read(1)  # 讀取一個位元組
    print(t)  # 顯示位元組數據
    print("目前存取位置(位元):", fh.tell())

    fh.seek(4, SEEK_CUR)  # 從當前位置向前移動 4 個位元組
    print("目前存取位置(位元):", fh.tell())
    t = fh.read(1)  # 讀取一個位元組
    print(t)  # 顯示位元組數據

    fh.seek(0, SEEK_END)  # 移到檔案的結尾
    print("檔案大小(位元數):", fh.tell())  # 顯示檔案大小（位元組）

    fh.seek(-4, SEEK_END)  # 從檔案結尾向前移動 4 個位元組
    print("目前存取位置(位元):", fh.tell())
    t = fh.read(1)  # 讀取一個位元組
    print(t)  # 顯示位元組數據
```
