### **Exercise 1: Using `__name__` in Modules and Classes**  
**Task:** Implement a module that demonstrates the usage of the `__name__` implicit variable both within the module and in classes.

#### **Requirements:**  
1. **Module Implementation:**  
   - Create a module named `my_module.py` that:
     - Contains a function named `greet` that prints a greeting message.
     - Contains a class named `Person` with an `__init__` method to initialize the name and age attributes.
     - In the main block, use `__name__` to check if the module is being run directly or imported.

2. **Function and Class Name Retrieval:**
   - Use `greet.__name__` to print the name of the function.
   - Use `Person.__name__` to print the name of the class.

3. **Creating a Main File:**
   - Create another file named `main.py` that imports `my_module` and calls the `greet` function and creates an instance of the `Person` class.
   - Print the value of `__name__` in this file.

#### **Emphasis on `__name__`:**
- The `__name__` variable helps distinguish whether the module is being executed directly or imported, which is useful for organizing code and preventing certain parts from running when the module is imported.

---

### **Example Code Snippet for `my_module.py`:**
```python
# my_module.py

def greet():
    """Print a greeting message."""
    print("Hello! Welcome to my module.")

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Print the value of __name__ to see how the module is being used
print(f"Module name: {__name__}")  # Prints __main__ when run directly, module name when imported

if __name__ == "__main__":
    print("This module is being run directly.")
else:
    print("This moudle is being imported.")

# Print the names of the function and class
print(f"Function name: {greet.__name__}")  # Output: greet
print(f"Class name: {Person.__name__}")     # Output: Person
```

### **Example Code Snippet for `main.py`:**
```python
# main.py
from my_module import greet, Person

# Print the name of the current module
print(__name__)  # Output: __main__

# Call the greet function
greet()

# Create an instance of Person
person1 = Person("Alice", 30)
print(f"Person created: {person1.name}, Age: {person1.age}")
```

### **練習 1：在模組和類別中使用 `__name__`**  
**任務：** 實現一個模組，演示 `__name__` 隱式變數在模組和類別中的用法。

#### **要求：**  
1. **模組實現：**  
   - 創建一個名為 `my_module.py` 的模組，該模組：
     - 包含一個名為 `greet` 的函數，打印問候消息。
     - 包含一個名為 `Person` 的類，具有 `__init__` 方法以初始化名稱和年齡屬性。
     - 在主區塊中，使用 `__name__` 來檢查模組是直接運行還是被匯入。

2. **函數和類名檢索：**
   - 使用 `greet.__name__` 打印函數的名稱。
   - 使用 `Person.__name__` 打印類的名稱。

3. **創建主文件：**
   - 創建另一個名為 `main.py` 的文件，該文件導入 `my_module`，並調用 `greet` 函數和創建 `Person` 類的實例。
   - 在此文件中打印 `__name__` 的值。

#### **強調 `__name__` 的重要性：**
- `__name__` 變數有助於區分模組是直接執行還是被匯入，這對於組織代碼和防止在匯入模組時運行某些部分非常有用。

---

### **`my_module.py` 的示例代碼片段：**
```python
# my_module.py

def greet():
    """打印問候消息。"""
    print("你好！歡迎來到我的模組。")

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# 打印 __name__ 的值，以查看模組的使用方式
print(f"模組名稱: {__name__}")  # 當直接運行時打印 __main__，被匯入時打印模組名稱

if __name__ == "__main__":
    print("這個模組是直接執行的。")
else:
    print("這個模組是被匯入的。")

# 打印函數和類的名稱
print(f"函數名稱: {greet.__name__}")  # 輸出: greet
print(f"類名稱: {Person.__name__}")     # 輸出: Person
```

### **`main.py` 的示例代碼片段：**
```python
# main.py
from my_module import greet, Person

# 打印當前模組的名稱
print(__name__)  # 輸出: __main__

# 調用 greet 函數
greet()

# 創建 Person 的實例
person1 = Person("Alice", 30)
print(f"創建的實例: {person1.name}, 年齡: {person1.age}")
```
