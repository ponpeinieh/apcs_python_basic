### **練習一：在 `Bookshelf` 類中實現 `__iter__()`**  
**任務：** 創建一個 `Bookshelf` 類，包含多個 `Book` 對象，並實現 `__iter__()` 方法，使 `Bookshelf` 對象**可迭代**(Iterable)。

#### **要求：**  
1. **類實現：**  
   - 創建一個名為 `Bookshelf` 的類，包括：
     - 一個屬性來保存 `Book` 對象的列表。
     - 一個 `__init__` 方法來初始化書籍列表。
     - 一個方法來向書架添加書籍。
     - 一個 `__iter__()` 方法，返回 `Book` 對象列表的迭代器。

2. **創建實例：**  
   - 創建 `Book` 類的實例，並將它們添加到 `Bookshelf` 類的實例中。

3. **演示功能：**  
   - 使用 `for` 循環遍歷 `Bookshelf` 實例，打印每本書的詳細信息。

#### **強調 `__iter__()` 的重要性：**
- `__iter__()` 方法使 `Bookshelf` 類可迭代，允許你循環遍歷其中包含的 `Book` 對象。

### **Example Code Snippet:**
```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def __str__(self):
        """User-friendly string representation."""
        return f"Title: {self.title}, Author: {self.author}, Year: {self.year}"

    def __repr__(self):
        """Formal string representation for recreating the object."""
        return f"Book('{self.title}', '{self.author}', {self.year})"

class Bookshelf:
    def __init__(self):
        """Initialize an empty bookshelf."""
        self.books = []

    def add_book(self, book):
        """Add a book to the bookshelf."""
        self.books.append(book)

    def __iter__(self):
        """Return an iterator for the books in the bookshelf."""
        return iter(self.books)

# Creating instances of the Book class
book1 = Book("1984", "George Orwell", 1949)
book2 = Book("To Kill a Mockingbird", "Harper Lee", 1960)

# Creating an instance of the Bookshelf class
my_shelf = Bookshelf()
my_shelf.add_book(book1)
my_shelf.add_book(book2)

# Demonstrating iteration over the Bookshelf instance
print("Books on the shelf:")
for book in my_shelf:
    print(book)  # Calls __str__()
```

### **示例代碼片段：**
```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def __str__(self):
        """用戶友好的字符串表示。"""
        return f"標題：{self.title}，作者：{self.author}，年份：{self.year}"

    def __repr__(self):
        """用於重建對象的正式字符串表示。"""
        return f"Book('{self.title}', '{self.author}', {self.year})"

class Bookshelf:
    def __init__(self):
        """初始化空書架。"""
        self.books = []

    def add_book(self, book):
        """將書籍添加到書架。"""
        self.books.append(book)

    def __iter__(self):
        """返回書架中書籍的迭代器。"""
        return iter(self.books)

# 創建 Book 類的實例
book1 = Book("1984", "喬治·奧威爾", 1949)
book2 = Book("殺死一隻知更鳥", "哈珀·李", 1960)

# 創建 Bookshelf 類的實例
my_shelf = Bookshelf()
my_shelf.add_book(book1)
my_shelf.add_book(book2)

# 演示遍歷 Bookshelf 實例
print("書架上的書籍：")
for book in my_shelf:
    print(book)  # 調用 __str__()
```

### 為什麼在 `__iter__()` 中使用 `iter(self.books)`？

`__iter__()` 方法對於使類可迭代至關重要，允許其在 `for` 循環和其他需要迭代的上下文中使用。以下是我們使用 `iter(self.books)` 的原因的詳細解釋：

#### 1. **返回一個迭代器：**
   - `iter()` 函數從可迭代對象（在這種情況下是 `self.books`，一個 `Book` 對象的列表）返回一個迭代器對象。這個迭代器將跟踪列表中的當前位置，允許你一次獲取一個項目。

#### 2. **簡單性和效率：**
   - 使用 `iter(self.books)` 是直接且高效的。Python 內置的 `list` 類型本身就是可迭代的，因此我們能利用其內置的迭代器功能，而無需實現自己的迭代邏輯。

#### 3. **關注點分離：**
   - 通過使用 `iter()`，我們將迭代邏輯與 `Bookshelf` 類分離。這意味著 `Bookshelf` 類不需要知道如何遍歷其書籍；它只需將該責任委託給列表本身。

#### 4. **兼容 Python 的迭代協議：**
   - `iter(self.books)` 返回的迭代器遵循 Python 的迭代協議，該協議由 `__iter__()` 和 `__next__()` 方法組成。這確保了 `Bookshelf` 類可以與任何期望可迭代對象的 Python 結構（如列表推導、`map()`、`filter()` 等）無縫配合使用。

#### 5. **維護狀態：**
   - 迭代器維護自己的狀態（即當前索引）。當你在該迭代器上調用 `next()` 時，它知道從哪裡繼續，這使得檢索列表中的下一本書變得簡單，而無需在 `Bookshelf` 類中實現額外的邏輯。

### 迭代示例
以下是一個簡短的示例以說明：

```python
# 假設我們有一個使用 iter(self.books) 的 Bookshelf 類
my_shelf = Bookshelf()
my_shelf.add_book(Book("1984", "喬治·奧威爾", 1949))
my_shelf.add_book(Book("殺死一隻知更鳥", "哈珀·李", 1960))

# 使用 for 循環遍歷書架
for book in my_shelf:
    print(book)  # 每本書通過迭代器訪問
```

在這個示例中，`for` 循環自動調用 `iter(my_shelf)`，然後反復調用迭代器返回的 `next()`，實現對 `Bookshelf` 中每個 `Book` 對象的平滑迭代。

### **練習二：使用 `Bookshelf` 與 `map()` 和 `zip()`**

**任務：** 演示 `Bookshelf` 類的可迭代能力，使用 `map()` 和 `zip()` 函數。

#### **要求：**

1. **類實現：**
   - 使用現有的 `Bookshelf` 和 `Book` 類。

2. **創建實例：**
   - 創建幾個 `Book` 類的實例並將它們添加到 `Bookshelf` 類的實例中。

3. **使用 `map()`：**
   - 使用 `map()` 創建一本書標題的列表，通過應用一個函數來提取 `Bookshelf` 中每本 `Book` 對象的標題。

4. **使用 `zip()`：**
   - 創建另一個作者列表，並使用 `zip()` 將每本書的標題與其對應的作者配對。

5. **演示功能：**
   - 打印 `map()` 和 `zip()` 的結果。

---

### **Example Code Snippet:**
```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def __str__(self):
        return f"Title: {self.title}, Author: {self.author}, Year: {self.year}"

    def __repr__(self):
        return f"Book('{self.title}', '{self.author}', {self.year})"

class Bookshelf:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def __iter__(self):
        return iter(self.books)

# Create Book instances
book1 = Book("1984", "George Orwell", 1949)
book2 = Book("To Kill a Mockingbird", "Harper Lee", 1960)
book3 = Book("Brave New World", "Aldous Huxley", 1932)

# Create a Bookshelf instance and add books
my_shelf = Bookshelf()
my_shelf.add_book(book1)
my_shelf.add_book(book2)
my_shelf.add_book(book3)

# Using map() to get book titles
titles = list(map(lambda book: book.title, my_shelf))
print("Book Titles:", titles)

# Using zip() to pair titles with authors
authors = list(map(lambda book: book.author, my_shelf))
title_author_pairs = list(zip(titles, authors))
print("Title and Author Pairs:", title_author_pairs)
```

### **示例代碼片段：**
```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def __str__(self):
        return f"標題：{self.title}，作者：{self.author}，年份：{self.year}"

    def __repr__(self):
        return f"Book('{self.title}', '{self.author}', {self.year})"

class Bookshelf:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def __iter__(self):
        return iter(self.books)

# 創建 Book 實例
book1 = Book("1984", "喬治·奧威爾", 1949)
book2 = Book("殺死一隻知更鳥", "哈珀·李", 1960)
book3 = Book("美麗新世界", "阿道司·赫胥黎", 1932)

# 創建 Bookshelf 實例並添加書籍
my_shelf = Bookshelf()
my_shelf.add_book(book1)
my_shelf.add_book(book2)
my_shelf.add_book(book3)

# 使用 map() 獲取書名
titles = list(map(lambda book: book.title, my_shelf))
print("書名：", titles)

# 使用 zip() 將書名與作者配對
authors = list(map(lambda book: book.author, my_shelf))
title_author_pairs = list(zip(titles, authors))
print("書名和作者配對：", title_author_pairs)
```
