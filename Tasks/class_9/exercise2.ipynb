#### **Example Code Snippet:**  
```python
class Person:
    def __new__(cls, *args, **kwargs):
        # Control the instance creation
        print("Creating a new instance")
        instance = super().__new__(cls)
        return instance

    def __init__(self, name, age):
        self.name = name
        self.age = age
        print(f"Initialized: {self.name}, Age: {self.age}")

# Using the example
person1 = Person("Alice", 30)
print(f"Person's Name: {person1.name}, Age: {person1.age}")
```

#### **Understanding `__new__`:**  
- The `__new__` method is responsible for creating a new instance of the class. It is called before `__init__`.
- This method allows customization of instance creation and can be useful when implementing **singleton patterns** or controlling how many instances are created.

---

#### **範例代碼片段：**  
```python
class Person:
    def __new__(cls, *args, **kwargs):
        # 控制實例創建
        print("創建新的實例")
        instance = super().__new__(cls)
        return instance

    def __init__(self, name, age):
        self.name = name
        self.age = age
        print(f"初始化：{self.name}, 年齡：{self.age}")

# 使用示例
person1 = Person("小明", 25)
print(f"人的名字：{person1.name}, 年齡：{person1.age}")
```

#### **理解 `__new__`：**  
- `__new__` 方法負責創建類別的新實例。它在 `__init__` 之前被調用。
- 此方法允許自定義實例創建，當實施**單例模式**或控制創建多少實例時特別有用。

### **練習二：實作單例模式**  
**任務：** 使用 `__new__` 方法實作一個單例類別，以確保只能創建一個類別的實例。

#### **要求：**  
1. **類別實作：**  
   - 創建一個名為 `Singleton` 的類別，使用 `__new__` 方法來控制實例創建：
     ```python
     class Singleton:
         _instance = None  # 類別變數用於保存唯一實例

         def __new__(cls, *args, **kwargs):
             if cls._instance is None:
                 cls._instance = super().__new__(cls)  # 如果實例不存在，則創建實例
             return cls._instance
     ```

2. **物件創建：**  
   - 創建多個 `Singleton` 類別的實例。
   - 通過打印它們的 ID 來驗證所有實例都引用同一個對象。

3. **理解單例模式：**  
   - 解釋單例模式的目的，以及如何使用 `__new__` 方法實現它。

---

#### **Example Code Snippet:**  
```python
class Singleton:
    _instance = None  # Class variable to hold the single instance

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)  # Create instance if it doesn't exist
        return cls._instance

# Using the example
singleton1 = Singleton()
singleton2 = Singleton()

# Print the IDs of the instances
print(id(singleton1))  # Output: ID of the first instance
print(id(singleton2))  # Output: ID of the second instance (should be the same as singleton1)
```

#### **Understanding the Singleton Pattern:**  
- The Singleton pattern ensures that a class has only one instance and provides a global point of access to that instance.
- By overriding the `__new__` method, we can control the instantiation process and ensure that only one object is created.

---

#### **範例代碼片段：**  
```python
class Singleton:
    _instance = None  # 類別變數用於保存唯一實例

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)  # 如果實例不存在，則創建實例
        return cls._instance

# 使用示例
singleton1 = Singleton()
singleton2 = Singleton()

# 打印實例的 ID
print(id(singleton1))  # 輸出：第一個實例的 ID
print(id(singleton2))  # 輸出：第二個實例的 ID（應與 singleton1 相同）
```

#### **理解單例模式：**  
- 單例模式確保一個類別只有一個實例，並提供對該實例的全局訪問點。
- 通過覆蓋 `__new__` 方法，我們可以控制實例化過程，確保僅創建一個對象。
