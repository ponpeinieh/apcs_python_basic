### **練習一：理解 `__call__` 方法**  
**任務：** 實作一個類別，以展示 `__call__` 方法的用法，允許類別的實例像函數一樣被調用。

#### **要求：**  
1. **類別實作：**  
   - 創建一個名為 `Counter` 的類別，使用 `__call__` 方法來跟踪實例被調用的次數：
     ```python
     class Counter:
         def __init__(self):
             self.count = 0  # 初始化計數器

         def __call__(self):
             self.count += 1  # 增加計數
             return self.count
     ```

2. **物件創建與方法調用：**  
   - 創建一個名為 `counter` 的 `Counter` 類別實例。
   - 多次調用 `counter` 實例並在每次調用後打印計數。

---

#### **Example Code Snippet:**  
```python
class Counter:
    def __init__(self):
        self.count = 0  # Initialize count

    def __call__(self):
        self.count += 1  # Increment count
        return self.count

# Using the example
counter = Counter()
print(counter())  # Output: 1
print(counter())  # Output: 2
print(counter())  # Output: 3
```

#### **Understanding `__call__`:**  
- The `__call__` method allows an instance of a class to be called as if it were a function.
- This can be useful for creating objects that maintain state and can perform actions when called.

---

#### **範例代碼片段：**  
```python
class Counter:
    def __init__(self):
        self.count = 0  # 初始化計數器

    def __call__(self):
        self.count += 1  # 增加計數
        return self.count

# 使用示例
counter = Counter()
print(counter())  # 輸出：1
print(counter())  # 輸出：2
print(counter())  # 輸出：3
```

#### **理解 `__call__`：**  
- `__call__` 方法允許類別的實例像函數一樣被調用。
- 這對於創建保持狀態並能在調用時執行操作的對象非常有用。
