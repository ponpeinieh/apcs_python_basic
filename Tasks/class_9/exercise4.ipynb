### **練習一：理解類別變數**  
**任務：** 實作一個類別，以展示類別變數的用法。

#### **要求：**  
1. **類別實作：**  
   - 創建一個名為 `Dog` 的類別，包含一個共享的類別變數 `species`：
     ```python
     class Dog:
         species = "Canis lupus familiaris"  # 類別變數

         def __init__(self, name, age):
             self.name = name  # 實例變數
             self.age = age    # 實例變數
     ```

2. **物件創建：**  
   - 創建多個 `Dog` 類別的實例，例如 `dog1` 和 `dog2`，並為它們設置不同的名字和年齡。
   - 打印每隻狗的物種，以展示共享的類別變數。
     - 我們可以使用物件參考 `dog1.species` 來訪問類別變數。
     - 但是最正規的用法是使用類別名 `Dog.species` 來訪問類別變數。
3. **修改類別變數：**  
   - 改變類別變數 `species` 的值，並顯示所有實例的變化。

#### **Example Code Snippet:**  
```python
class Dog:
    species = "Canis lupus familiaris"  # Class variable

    def __init__(self, name, age):
        self.name = name  # Instance variable
        self.age = age    # Instance variable

# Using the example
dog1 = Dog("Buddy", 3)
dog2 = Dog("Max", 5)

# Accessing class variable
print(f"{dog1.name} is a {dog1.species}.")  # Output: Buddy is a Canis lupus familiaris.
print(f"{dog2.name} is a {dog2.species}.")  # Output: Max is a Canis lupus familiaris.

# Modifying class variable
Dog.species = "Canis familiaris"

# Accessing modified class variable
print(f"{dog1.name} is now a {dog1.species}.")  # Output: Buddy is now a Canis familiaris.
print(f"{dog2.name} is now a {dog2.species}.")  # Output: Max is now a Canis familiaris.
```

#### **Understanding Class Variables:**  
- Class variables are shared across all instances of a class.
- Modifying a class variable through the class name affects all instances, demonstrating a common property among them.

---

#### **範例代碼片段：**  
```python
class Dog:
    species = "Canis lupus familiaris"  # 類別變數

    def __init__(self, name, age):
        self.name = name  # 實例變數
        self.age = age    # 實例變數

# 使用示例
dog1 = Dog("小白", 3)
dog2 = Dog("小黑", 5)

# 訪問類別變數
print(f"{dog1.name} 是 {dog1.species}。")  # 輸出：小白 是 Canis lupus familiaris。
print(f"{dog2.name} 是 {dog2.species}。")  # 輸出：小黑 是 Canis lupus familiaris。

# 修改類別變數
Dog.species = "Canis familiaris"

# 訪問修改後的類別變數
print(f"{dog1.name} 現在是 {dog1.species}。")  # 輸出：小白 現在是 Canis familiaris。
print(f"{dog2.name} 現在是 {dog2.species}。")  # 輸出：小黑 現在是 Canis familiaris。
```

#### **理解類別變數：**  
- 類別變數在所有實例之間共享。
- 通過類別名稱修改類別變數會影響所有實例，展示它們之間的共同屬性。

#### **Requirements:**

     ```python
     class Item:
         # Class variable

             self.name = name  # Instance variable for the item's name

         def get_item_count(cls):

         def reset_item_count(cls):

     ```

### **練習二：理解類別方法**  
**任務：** 實作一個類別，以展示類別方法的用法來跟踪和管理物品數量。

#### **要求：**  
1. **類別實作：**  
   - 創建一個名為 `Item` 的類別，包含類別方法來獲取和重置物品數量：
     ```python
     class Item:
         # 類別變數
         item_count = 0  # 所有 Item 實例共享的計數

         def __init__(self, name):
             self.name = name  # 實例變數，物品的名稱
             Item.item_count += 1  # 每次建立新實例時增加計數

         @classmethod
         def get_item_count(cls):
             return cls.item_count  # 返回目前的物品數量

         @classmethod
         def reset_item_count(cls):
             cls.item_count = 0  # 重置物品數量
     ```

2. **物件創建：**  
   - 建立幾個 `Item` 類別的實例，例如 `backpack`、`pencil` 和 `notebook`。

3. **顯示物品數量：**  
   - 使用類別方法 `get_item_count` 顯示創建的物品總數。

4. **重置物品計數：**  
   - 調用類別方法 `reset_item_count` 重置物品計數，並再次顯示計數。

---

#### **Example Code Snippet:**  
```python
class Item:
    # Class variable
    item_count = 0  # Count shared by all Item instances

    def __init__(self, name):
        self.name = name  # Instance variable for the item's name
        Item.item_count += 1  # Increment count each time a new instance is created

    @classmethod
    def get_item_count(cls):
        return cls.item_count  # Return the current number of items

    @classmethod
    def reset_item_count(cls):
        cls.item_count = 0  # Reset the item count

# Creating several Item instances
backpack = Item("Backpack")
pencil = Item("Pencil")
notebook = Item("Notebook")

# Getting and displaying the item count
print(f"Total number of items: {Item.get_item_count()}")  # Output: Total number of items: 3

# Resetting the item count
Item.reset_item_count()
print(f"Total number of items after reset: {Item.get_item_count()}")  # Output: Total number of items after reset: 0
```

#### **Understanding Class Methods:**  
- Class methods can operate on class variables and are called on the class itself rather than instances.
- They provide a way to manage class-level data, such as counting instances or resetting values.

---

#### **範例代碼片段：**  
```python
class Item:
    # 類別變數
    item_count = 0  # 所有 Item 實例共享的計數

    def __init__(self, name):
        self.name = name  # 實例變數，物品的名稱
        Item.item_count += 1  # 每次建立新實例時增加計數

    @classmethod
    def get_item_count(cls):
        return cls.item_count  # 返回目前的物品數量

    @classmethod
    def reset_item_count(cls):
        cls.item_count = 0  # 重置物品數量

# 建立幾個 Item 實例
backpack = Item("Backpack")
pencil = Item("Pencil")
notebook = Item("Notebook")

# 獲取和顯示物品數量
print(f"Total number of items: {Item.get_item_count()}")  # 輸出: Total number of items: 3

# 重置物品計數
Item.reset_item_count()
print(f"Total number of items after reset: {Item.get_item_count()}")  # 輸出: Total number of items after reset: 0
```

#### **理解類別方法：**  
- 類別方法可以操作類別變數，並且是直接在類別上調用，而非在實例上。
- 它們提供了一種管理類別級別數據的方法，例如計數實例或重置值。

備注: 我們仍然可以使用`實例變數`來呼叫類別方法, 例如 `notebook.get_item_count()`, 但這個不是正規的用法。

### **練習三：理解靜態方法**  
**任務：** 實作一個類別，以展示靜態方法的用法。

#### **要求：**  
1. **類別實作：**  
   - 創建一個名為 `Calculator` 的類別，包含靜態方法來進行基本的算術運算：加法、減法、乘法和除法。
   - **對這些方法使用 `@staticmethod` 裝飾器**，以表示它們不需要訪問實例或類別變數。

2. **靜態方法定義：**  
   - 定義以下靜態方法：
     - `add(a, b)`：返回 `a` 和 `b` 的和。
     - `subtract(a, b)`：返回 `a` 和 `b` 的差。
     - `multiply(a, b)`：返回 `a` 和 `b` 的積。
     - `divide(a, b)`：返回 `a` 和 `b` 的商，並對除以零的情況進行錯誤處理。

3. **使用靜態方法：**  
   - 通過執行一系列計算來演示使用靜態方法，並顯示結果。

---

#### **Example Code Snippet:**  
```python
class Calculator:
    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod
    def subtract(a, b):
        return a - b

    @staticmethod
    def multiply(a, b):
        return a * b

    @staticmethod
    def divide(a, b):
        if b == 0:
            return "Error: Division by zero!"
        return a / b

# Demonstrating the use of static methods
print("Addition: ", Calculator.add(5, 3))          # Output: 8
print("Subtraction: ", Calculator.subtract(5, 3))  # Output: 2
print("Multiplication: ", Calculator.multiply(5, 3))  # Output: 15
print("Division: ", Calculator.divide(5, 0))        # Output: Error: Division by zero!
print("Division: ", Calculator.divide(5, 2))        # Output: 2.5
```

### **Explanation:**
- The `Calculator` class contains static methods that perform arithmetic operations.
- These methods can be called directly on the class without creating an instance, making them useful for utility functions.
