### **Exercise 1: Understanding `__str__()` and `__repr__()`**  
**Task:** Implement a class that demonstrates the usage of the `__str__()` and `__repr__()` methods for string representation of objects.

#### **Requirements:**  
1. **Class Implementation:**  
   - Create a class named `Book` that includes:
     - Attributes for `title`, `author`, and `year`.
     - An `__init__` method to initialize these attributes.
     - A `__str__()` method that returns a user-friendly string representation of the book (e.g., "Title: Book Title, Author: Author Name, Year: Year").
     - A `__repr__()` method that returns a more formal string representation of the book, useful for debugging (e.g., "Book('Book Title', 'Author Name', Year)").

2. **Creating Instances:**  
   - Create several instances of the `Book` class and demonstrate the difference between using `print()` (which calls `__str__()`) and using `repr()` (which calls `__repr__()`).

#### **Emphasis on `__str__()` and `__repr__()`:**
- The `__str__()` method is intended for creating a readable string representation of an object, while `__repr__()` is aimed at providing an unambiguous representation that can ideally be used to recreate the object.

---

### **Example Code Snippet:**
```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def __str__(self):
        """User-friendly string representation."""
        return f"Title: {self.title}, Author: {self.author}, Year: {self.year}"

    def __repr__(self):
        """Formal string representation for debugging."""
        return f"Book('{self.title}', '{self.author}', {self.year})"

# Creating instances of the Book class
book1 = Book("1984", "George Orwell", 1949)
book2 = Book("To Kill a Mockingbird", "Harper Lee", 1960)

# Demonstrating __str__()
print(book1)  # Calls __str__()
print(book2)  # Calls __str__()

# Demonstrating __repr__()
print(repr(book1))  # Calls __repr__()
print(repr(book2))  # Calls __repr__()
```

     - 一個 `__repr__()` 方法，返回書籍的更正式字符串表示，便於調試（例如：“Book('書名', '作者姓名', 年份)”）。

2. **創建實例：**

#### **強調 `__str__()` 和 `__repr__()` 的重要性：**
- `__str__()` 方法旨在創建對象的可讀字符串表示，而 `__repr__()` 則旨在提供一種明確的表示，理想情況下可以用於重建該對象。

### **Exercise 2: Using `eval()` with `repr()` in the `Book` Class**  
**Task:** Demonstrate the usage of the `eval()` function in conjunction with the `__repr__()` method to recreate a `Book` object.

#### **Requirements:**  
1. **Class Implementation:**  
   - Use the existing `Book` class with attributes for `title`, `author`, and `year`.
   - Ensure that the `__repr__()` method returns a string representation that can be used with `eval()` to recreate the object (e.g., `Book('Book Title', 'Author Name', Year)`).

2. **Creating Instances:**  
   - Create an instance of the `Book` class.
   - Use `repr()` to get the string representation of the instance and then use `eval()` to recreate the object from that string.

3. **Demonstrate Functionality:**
   - Print the original object and the recreated object to show that they are the same.
   - Compare the identity of `book1` and `book2` using the `is` operator.

#### **Emphasis on `eval()` and `repr()`:**
- The `eval()` function executes a string expression, which can be useful for dynamic object creation. The `__repr__()` method should provide a valid Python expression for recreating the object.

---

### **Example Code Snippet:**
```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def __str__(self):
        """User-friendly string representation."""
        return f"Title: {self.title}, Author: {self.author}, Year: {self.year}"

    def __repr__(self):
        """Formal string representation for recreating the object."""
        return f"Book('{self.title}', '{self.author}', {self.year})"

# Creating an instance of the Book class
book1 = Book("1984", "George Orwell", 1949)

# Using repr() to get the string representation
book_repr = repr(book1)
print(f"String representation: {book_repr}")

# Using eval() to recreate the object
book2 = eval(book_repr)

# Demonstrating functionality
print(f"Original object: {book1}")  # Calls __str__()
print(f"Recreated object: {book2}")  # Calls __str__()
print(f"Are they equal? {book1.title == book2.title and book1.author == book2.author and book1.year == book2.year}")

# Comparing identity
print(f"Are book1 and book2 the same object? {book1 is book2}")
```

### **練習 2：在 `Book` 類中使用 `eval()` 和 `repr()`**  
**任務：** 演示 `eval()` 函數與 `__repr__()` 方法結合使用以重建 `Book` 對象。

#### **要求：**  
1. **類實現：**  
   - 使用現有的 `Book` 類，具有 `title`、`author` 和 `year` 的屬性。
   - 確保 `__repr__()` 方法返回一個可以與 `eval()` 一起使用的字符串表示（例如，`Book('書名', '作者姓名', 年份)`）。

2. **創建實例：**  
   - 創建一個 `Book` 類的實例。
   - 使用 `repr()` 獲取該實例的字符串表示，然後使用 `eval()` 從該字符串重建對象。

3. **演示功能：**
   - 打印原始對象和重建對象以顯示它們是相同的。
   - 使用 `is` 操作符比較 `book1` 和 `book2` 的身份。

#### **強調 `eval()` 和 `repr()` 的重要性：**
- `eval()` 函數執行字符串表達式，這對於動態對象創建非常有用。`__repr__()` 方法應提供一個有效的 Python 表達式，以重建對象。

---

### **示例代碼片段：**
```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def __str__(self):
        """用戶友好的字符串表示。"""
        return f"標題：{self.title}，作者：{self.author}，年份：{self.year}"

    def __repr__(self):
        """用於重建對象的正式字符串表示。"""
        return f"Book('{self.title}', '{self.author}', {self.year})"

# 創建 Book 類的實例
book1 = Book("1984", "喬治·奧威爾", 1949)

# 使用 repr() 獲取字符串表示
book_repr = repr(book1)
print(f"字符串表示：{book_repr}")

# 使用 eval() 重建對象
book2 = eval(book_repr)

# 演示功能
print(f"原始對象：{book1}")  # 調用 __str__()
print(f"重建對象：{book2}")  # 調用 __str__()
print(f"它們是否相等？{book1.title == book2.title and book1.author == book2.author and book1.year == book2.year}")

# 比較身份
print(f"book1 和 book2 是同一對象嗎？{book1 is book2}")
```
