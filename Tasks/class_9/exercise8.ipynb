### **練習一：在 `ColoredTurtle` 類別中實現運算子過載**

**任務：** 創建一個繼承自 Python `turtle` 模組的 `Turtle` 類別的 `ColoredTurtle` 類別，並為顏色和比較實現運算子過載。

#### **要求：**

1. **類別實現：**
   - 創建一個名為 `ColoredTurtle` 的類別，繼承自 `turtle.Turtle`。
   - 實現以下運算子過載：
     - `__eq__(self, other)`：根據顏色比較兩個 `ColoredTurtle` 物件的相等性。
     - `__le__(self, other)`：比較兩個 `ColoredTurtle` 物件到原點（0, 0）的距離。
     - `__add__(self, other)`：返回一個新的 `ColoredTurtle` 物件，其位置是兩個 `ColoredTurtle` 物件的 x 和 y 坐標的和。

2. **創建實例：**
   - 創建 `ColoredTurtle` 類別的實例，使用不同的顏色和位置。

3. **演示功能：**
   - 演示過載運算子的工作原理：
     - 比較兩個 `ColoredTurtle` 物件的顏色相等性。

     - 結合兩個 `ColoredTurtle` 物件的位置。

### **Example Code Snippet:**
```python
import turtle

class ColoredTurtle(turtle.Turtle):
    def __init__(self, color):
        super().__init__()
        self.color(color)

    def __eq__(self, other):
        """Check if two turtles are the same color."""
        return self.fillcolor() == other.fillcolor()

    def __le__(self, other):
        """Compare the distance of two turtles from the origin."""
        return self.distance(0, 0) <= other.distance(0, 0)

    def __add__(self, other):
        """Create a new ColoredTurtle at the sum of their positions."""
        new_turtle = ColoredTurtle("orange")
        new_turtle.goto(self.xcor() + other.xcor(), self.ycor() + other.ycor())
        return new_turtle

# Create ColoredTurtle instances
turtle1 = ColoredTurtle("red")
turtle2 = ColoredTurtle("blue")

# Move turtles to different positions
turtle1.goto(-100, 100)
turtle2.goto(50, 50)

# Demonstrating equality
print(f"Is turtle1 the same color as turtle2? {turtle1 == turtle2}")

# Demonstrating less than or equal comparison based on distance from origin
print(f"Is turtle1 closer to the origin than or equal to turtle2? {turtle1 <= turtle2}")

# Demonstrating addition (creating a new turtle at the summed position)
new_turtle = turtle1 + turtle2
print(f"New Turtle Color: {new_turtle.fillcolor()}")
print(f"New Turtle Position: ({new_turtle.xcor()}, {new_turtle.ycor()})")

# Safe exit
try:
    turtle.done()
    turtle.bye()
except:
    print("Graphics terminated")
```

### **示例代碼片段：**
```python
import turtle

class ColoredTurtle(turtle.Turtle):
    def __init__(self, color):
        super().__init__()
        self.color(color)

    def __eq__(self, other):
        """檢查兩隻烏龜是否顏色相同。"""
        return self.fillcolor() == other.fillcolor()

    def __le__(self, other):
        """比較兩隻烏龜到原點的距離。"""
        return self.distance(0, 0) <= other.distance(0, 0)

    def __add__(self, other):
        """在兩隻烏龜的坐標和位置上創建一個新的 ColoredTurtle。"""
        new_turtle = ColoredTurtle("橙色")
        new_turtle.goto(self.xcor() + other.xcor(), self.ycor() + other.ycor())
        return new_turtle

# 創建 ColoredTurtle 實例
turtle1 = ColoredTurtle("紅色")
turtle2 = ColoredTurtle("藍色")

# 移動烏龜到不同的位置
turtle1.goto(-100, 100)
turtle2.goto(50, 50)

# 演示相等性
print(f"turtle1 和 turtle2 顏色相同嗎？ {turtle1 == turtle2}")

# 演示小於或等於比較（根據到原點的距離）
print(f"turtle1 到原點的距離是否小於或等於 turtle2？ {turtle1 <= turtle2}")

# 演示加法（創建一個新的烏龜在和的位置上）
new_turtle = turtle1 + turtle2
print(f"新烏龜顏色：{new_turtle.fillcolor()}")
print(f"新烏龜位置：({new_turtle.xcor()}, {new_turtle.ycor()})")

# 安全退出
try:
    turtle.done()
    turtle.bye()
except:
    print("圖形已終止")
```
