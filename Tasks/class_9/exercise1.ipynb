### **練習三：重新定義狗類別的建構函式**  
**任務：** 重新定義 `Dog` 類別以包含自己的 `__init__()` 方法，並新增屬性 `breed`。

#### **要求：**  
1. **類別實作：**  
   - 使用提供的 `Animal` 類別定義：
     ```python
     # 父類：動物
     class Animal:
         def __init__(self, name):
             self.name = name  # 物件的屬性
         
         def speak(self):

     ```

2. **重新定義狗類別：**  
   - 重新定義 `Dog` 類別，包含自己的 `__init__()` 方法並新增 `breed` 屬性：
     ```python
     # 子類：狗（定義自己的 __init__() 方法）
     class Dog(Animal):
         def __init__(self, name, breed):
             super().__init__(name)  # 呼叫父類的 __init__ 方法
             self.breed = breed  # 新增屬性
     ```

3. **物件創建與方法調用：**  
   - 創建一個名為 `my_dog` 的 `Dog` 類別實例，傳入你選擇的名字和品種。

4. **不使用 `super().__init__()` 的測試：**  
   - 修改 `Dog` 類別，將 `super().__init__(name)` 行註解掉。

#### **Example Code Snippet:**  
```python
# Parent class: Animal
class Animal:
    def __init__(self, name):
        self.name = name  # Object attribute
    
    def speak(self):
        return "The animal makes a sound"

# Derived class: Dog (define its own __init__() method)
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # Call the parent class's __init__ method
        self.breed = breed  # New attribute

# Using the example
my_dog = Dog("Buddy", "Golden Retriever")
print(my_dog.name)  # Output: Buddy
print(my_dog.breed)  # Output: Golden Retriever
print(my_dog.speak())  # Output: The animal makes a sound
```

#### **Using `super().__init__()`**  
- The `super()` function is used to call parent class methods, especially for initializing parent class attributes.
- This ensures the parent class's initialization logic is executed correctly, allowing the subclass to utilize inherited attributes.
- In the subclass's `__init__()` method, calling the parent class's `__init__()` is necessary for the subclass instance to possess attributes defined in the parent class.

---

#### **範例代碼片段：**  
```python
# 父類：動物
class Animal:
    def __init__(self, name):
        self.name = name  # 物件的屬性
    
    def speak(self):
        return "動物發出聲音"

# 子類：狗（定義自己的 __init__() 方法）
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # 呼叫父類的 __init__ 方法
        self.breed = breed  # 新增屬性

# 使用示例
my_dog = Dog("小白", "金毛")
print(my_dog.name)  # 輸出：小白
print(my_dog.breed)  # 輸出：金毛
print(my_dog.speak())  # 輸出：動物發出聲音
```

#### **使用 `super().__init__()` 的說明**  
- `super()` 函數用於呼叫父類的方法，特別是在子類中需要初始化父類的屬性時。
- 確保父類的初始化邏輯被正確執行，使得子類可以使用繼承自父類的屬性。

### **練習四：方法覆蓋與貓類別**  
**任務：** 實作 `Cat` 類別，覆蓋 `Animal` 類別中的 `speak` 方法。

#### **要求：**  
1. **類別實作：**  
   - 使用提供的 `Animal` 類別定義：
     ```python
     # 父類：動物
     class Animal:
         def __init__(self, name):
             self.name = name  # 物件的屬性
         
         def speak(self):
             return "動物發出聲音"
     ```

2. **定義貓類別：**  
   - 創建一個 `Cat` 類別，繼承自 `Animal` 並覆蓋 `speak` 方法：
     ```python
     # 子類：貓
     class Cat(Animal):
         def speak(self):
             return "喵喵！"  # 覆蓋 speak() 方法
     ```

3. **物件創建與方法調用：**  
   - 創建一個名為 `my_cat` 的 `Cat` 類別實例，傳入你選擇的名字。
   - 打印貓的名字並調用 `speak` 方法。

---

#### **Example Code Snippet:**  
```python
# Parent class: Animal
class Animal:
    def __init__(self, name):
        self.name = name  # Object attribute
    
    def speak(self):
        return "The animal makes a sound"

# Derived class: Cat
class Cat(Animal):
    def speak(self):
        return "Meow!"  # Override the speak() method

# Using the example
my_cat = Cat("Whiskers")
print(my_cat.name)  # Output: Whiskers
print(my_cat.speak())  # Output: Meow!
```

#### **Method Overriding Explanation:**  
- Subclasses can redefine methods in the parent class, which is known as method overriding.
- When a subclass overrides a method, the subclass version replaces the parent class version, allowing developers to customize behavior as needed.

---

#### **範例代碼片段：**  
```python
# 父類：動物
class Animal:
    def __init__(self, name):
        self.name = name  # 物件的屬性
    
    def speak(self):
        return "動物發出聲音"

# 子類：貓
class Cat(Animal):
    def speak(self):
        return "喵喵！"  # 覆蓋 speak() 方法

# 使用示例
my_cat = Cat("小黑")
print(my_cat.name)  # 輸出：小黑
print(my_cat.speak())  # 輸出：喵喵！
```

#### **方法覆蓋說明：**  
- 子類可以重新定義父類中的方法，這稱為方法覆蓋（Method Overriding）。
- 當子類覆蓋父類的方法時，子類的版本將取代父類的版本，這使得開發者可以根據需求自定義行為。
