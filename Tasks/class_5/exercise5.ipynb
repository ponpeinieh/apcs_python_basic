### **Exercise 1: Dictionary Sorting (Student Scores)**  
**Task:** Sort student records by different criteria  

#### **Test Data:**  
```python
students = {
    'Alice': 88, 
    'Bob': 76,
    'Charlie': 92,
    'Diana': 85
}
```

#### **Sorting Tasks:**  
1. Sort by student name (key)  
2. Sort by score (value)  

#### **Example Output:**  
```
Original: {'Alice': 88, 'Bob': 76, 'Charlie': 92, 'Diana': 85}
Sorted by name: {'Alice': 88, 'Bob': 76, 'Charlie': 92, 'Diana': 85}
Sorted by score: {'Bob': 76, 'Diana': 85, 'Alice': 88, 'Charlie': 92}
```

---

### **練習一：字典排序（學生成績）**  
**任務：** 按不同條件排序學生成績  

#### **測試數據:**  
```python
students = {
    'Alice': 88, 
    'Bob': 76,
    'Charlie': 92,
    'Diana': 85
}
```

#### **排序任務:**  
1. 按學生姓名排序 (鍵)  
2. 按分數排序 (值)  

#### **範例輸出:**  
```
原始數據: {'Alice': 88, 'Bob': 76, 'Charlie': 92, 'Diana': 85}
按姓名排序: {'Alice': 88, 'Bob': 76, 'Charlie': 92, 'Diana': 85}
按分數排序: {'Bob': 76, 'Diana': 85, 'Alice': 88, 'Charlie': 92}
```

---

### **Sample Solution Code**  
```python
# English Version
def sort_students(students, by='name'):
    if by == 'name':
        return dict(sorted(students.items()))
    elif by == 'score':
        return dict(sorted(students.items(), key=lambda item: item[1]))

# Test
print("Original:", students)
print("Sorted by name:", sort_students(students, 'name')) 
print("Sorted by score:", sort_students(students, 'score'))
 
```

### **Exercise 2: Sorting List of Dictionaries**  
**Task:** Sort student records by grade  

#### **Test Data:**  
```python
students = [
    {'name': 'Alice', 'grade': 85},
    {'name': 'Bob', 'grade': 92},
    {'name': 'Charlie', 'grade': 78},
    {'name': 'Diana', 'grade': 95},
    {'name': 'Ellen', 'grade': 92}
]
```

#### **Requirements:**  
1. Sort by grade (ascending and descending)  
2. When grades are equal, sort by name alphabetically  

#### **Example Output:**  
```
Original: [Alice:85, Bob:92, Charlie:78, Diana:95, Ellen:92]
Ascending: [Charlie:78, Alice:85, Bob:92, Ellen:92, Diana:95]
Descending: [Diana:95, Bob:92, Ellen:92, Alice:85, Charlie:78]
```

#### **Solution Code:**  
```python
# Basic sorting by grade
def sort_by_grade(students, reverse=False):
    return sorted(students, key=lambda x: x['grade'], reverse=reverse)

# Multi-level sorting (grade then name)
def sort_by_grade_then_name(students, reverse=False):
    return sorted(students, key=lambda x: (x['grade'], x['name']), reverse=reverse)

# Testing
print("Original:", [f"{s['name']}:{s['grade']}" for s in students])
print("Ascending:", [f"{s['name']}:{s['grade']}" for s in sort_by_grade(students)])
print("Descending:", [f"{s['name']}:{s['grade']}" for s in sort_by_grade(students, reverse=True)])
print("Grade+Name:", [f"{s['name']}:{s['grade']}" for s in sort_by_grade_then_name(students)])
```

### **練習二：字典列表排序**  
**任務：** 按成績排序學生記錄  

#### **測試數據:**  
```python
students = [
    {'name': 'Alice', 'grade': 85},
    {'name': 'Bob', 'grade': 92},
    {'name': 'Charlie', 'grade': 78},
    {'name': 'Diana', 'grade': 95},
    {'name': 'Ellen', 'grade': 92}
]
```

#### **要求:**  
1. 按成績排序 (升序和降序)  
2. 成績相同時按姓名排序  

#### **範例輸出:**  
```
原始數據: [Alice:85, Bob:92, Charlie:78, Diana:95, Ellen:92]
升序排序: [Charlie:78, Alice:85, Bob:92, Ellen:92, Diana:95]
降序排序: [Diana:95, Bob:92, Ellen:92, Alice:85, Charlie:78]
```

#### **解決方案:**  
```python
# 按成績排序
def sort_by_grade(students, reverse=False):
    return sorted(students, key=lambda x: x['grade'], reverse=reverse)

# 多級排序 (成績後按姓名)
def sort_by_grade_then_name(students, reverse=False):
    return sorted(students, key=lambda x: (x['grade'], x['name']), reverse=reverse)

# 測試
print("原始數據:", [f"{s['name']}:{s['grade']}" for s in students])
print("升序排序:", [f"{s['name']}:{s['grade']}" for s in sort_by_grade(students)])
print("降序排序:", [f"{s['name']}:{s['grade']}" for s in sort_by_grade(students, reverse=True)])
print("成績+姓名:", [f"{s['name']}:{s['grade']}" for s in sort_by_grade_then_name(students)])
```
