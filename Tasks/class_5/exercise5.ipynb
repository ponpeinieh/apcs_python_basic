### **練習一：字典排序（學生成績）**  
**任務：** 按不同條件排序學生成績  

#### **測試數據:**  
```python
students = {
    'Alice': 88, 
    'Bob': 76,
    'Charlie': 92,
    'Diana': 85
}
```

#### **排序任務:**  
1. 按學生姓名排序 (鍵)  
2. 按分數排序 (值)  

#### **範例輸出:**  
```
原始數據: {'Alice': 88, 'Bob': 76, 'Charlie': 92, 'Diana': 85}
按姓名排序: {'Alice': 88, 'Bob': 76, 'Charlie': 92, 'Diana': 85}
按分數排序: {'Bob': 76, 'Diana': 85, 'Alice': 88, 'Charlie': 92}
```

---

### **Sample Solution Code**  
```python
# English Version
def sort_students(students, by='name'):
    if by == 'name':
        return dict(sorted(students.items()))
    elif by == 'score':
        return dict(sorted(students.items(), key=lambda item: item[1]))

# Test
print("Original:", students)
print("Sorted by name:", sort_students(students, 'name')) 
print("Sorted by score:", sort_students(students, 'score'))
 
```

### **Exercise 2: Sorting List of Dictionaries**  
**Task:** Sort student records by grade  

#### **Test Data:**  
```python
students = [
    {'name': 'Alice', 'grade': 85},
    {'name': 'Bob', 'grade': 92},
    {'name': 'Charlie', 'grade': 78},
    {'name': 'Diana', 'grade': 95},
    {'name': 'Ellen', 'grade': 92}
]
```

#### **Requirements:**  
1. Sort by grade (ascending and descending)  
2. When grades are equal, sort by name alphabetically  

#### **Example Output:**  
```
Original: [Alice:85, Bob:92, Charlie:78, Diana:95, Ellen:92]
Ascending: [Charlie:78, Alice:85, Bob:92, Ellen:92, Diana:95]
Descending: [Diana:95, Bob:92, Ellen:92, Alice:85, Charlie:78]
```

#### **Solution Code:**  
```python
# Basic sorting by grade
def sort_by_grade(students, reverse=False):
    return sorted(students, key=lambda x: x['grade'], reverse=reverse)

# Multi-level sorting (grade then name)
def sort_by_grade_then_name(students, reverse=False):
    return sorted(students, key=lambda x: (x['grade'], x['name']), reverse=reverse)

# Testing
print("Original:", [f"{s['name']}:{s['grade']}" for s in students])
print("Ascending:", [f"{s['name']}:{s['grade']}" for s in sort_by_grade(students)])
print("Descending:", [f"{s['name']}:{s['grade']}" for s in sort_by_grade(students, reverse=True)])
print("Grade+Name:", [f"{s['name']}:{s['grade']}" for s in sort_by_grade_then_name(students)])
```

### **練習二：字典列表排序**  
**任務：** 按成績排序學生記錄  

#### **測試數據:**  
```python
students = [
    {'name': 'Alice', 'grade': 85},
    {'name': 'Bob', 'grade': 92},
    {'name': 'Charlie', 'grade': 78},
    {'name': 'Diana', 'grade': 95},
    {'name': 'Ellen', 'grade': 92}
]
```

#### **要求:**  
1. 按成績排序 (升序和降序)

### **練習：英文單字出現次數分析**  
**任務：** 分析歌詞中的單字頻率  

#### **要求:**  
1. 處理歌詞:  
   - 轉換為小寫  
   - 去除標點符號    
   - 分割為單字列表  
2. 使用字典儲存結果:  
   - `單字`作為鍵  
   - `出現次數`作為值  
3. 顯示出現頻率最高的5個單字(若頻率相同, 則按照字母順序)  

#### **範例歌詞:**  
```python
lyrics = """
Hey, Jude, don't make it bad
Take a sad song and make it better
Remember to let her into your heart
Then you can start to make it better
Hey, Jude, don't be afraid
You were made to go out and get her
The minute you let her under your skin
Then you begin to make it better
"""
```

#### **預期輸出:**  
```
出現頻率最高的5個單字:
make: 2
it: 2
hey: 1
jude: 1
dont: 1
```

### Solution 1: Using Basic Dictionary 
```python
lyrics = """
Hey Jude, don't make it bad
Take a sad song and make it better
"""

def word_frequency_basic(lyrics):
    # Process the lyrics
    lyrics_lower = lyrics.lower()
    
    # Remove punctuation by replacing each punctuation with space
    punctuation = '''!()-[]{};:'"\,<>./?@#$%^&*_~'''
    for char in punctuation: # or just use `for char in string.punctuation:`
        lyrics_lower = lyrics_lower.replace(char, ' ')
   
    # Split into words
    words = lyrics_lower.split()
    
    # Count word frequency
    freq = {}
    for word in words:
        if word in freq: # or replace if-else with 'freq[word] = freq.get(word, 0) + 1'
            freq[word] += 1
        else:
            freq[word] = 1
    
    # Sort by frequency (descending) and then alphabetically (ascending)
    sorted_words = sorted(freq.items(), key=lambda x: (-x[1], x[0]))
    
    # Display top 5
    print("Top 5 words:")
    for word, count in sorted_words[:5]:
        print(f"{word}: {count}")

word_frequency_basic(lyrics)
```

### Solution 2: Using Counter from collections
```python
from collections import Counter

lyrics = """
Hey Jude, don't make it bad
Take a sad song and make it better
"""

def word_frequency_counter(lyrics):
    # Process the lyrics
    lyrics_lower = lyrics.lower()
    
    # Remove punctuation by replacing each punctuation with space
    punctuation = '''!()-[]{};:'"\,<>./?@#$%^&*_~'''
    for char in punctuation:
        lyrics_lower = lyrics_lower.replace(char, ' ')
    
    # Split into words and count
    words = lyrics_lower.split()
    word_counts = Counter(words)
    
    # Display top 5
    print("Top 5 words:")
    for word, count in word_counts.most_common(5):
        print(f"{word}: {count}")

word_frequency_counter(lyrics)
```

### Solution 3: Using Regular Expressions
```python
import re
from collections import Counter

lyrics = """
Hey Jude, don't make it bad
Take a sad song and make it better
"""

def word_frequency_regex(lyrics):
    # Use regex to find all words (only alphabetic characters)
    words = re.findall(r"[a-zA-Z]+", lyrics.lower())
    
    # Count word frequency
    word_counts = Counter(words)
    
    # Display top 5
    print("Top 5 words:")
    for word, count in word_counts.most_common(5):
        print(f"{word}: {count}")

word_frequency_regex(lyrics)
```

All three solutions will produce the same expected output:
```
Top 5 words:
make: 2
it: 2
a: 1
and: 1
bad: 1
```

### **Final Optimized Solution (Using `str.maketrans()` and `str.translate()`)**
```python
from collections import Counter
import string

lyrics = """
Hey Jude, don't make it bad
Take a sad song and make it better
"""

def word_frequency_translate(lyrics):
    # Create a translation table to remove all punctuation
    translator = str.maketrans('', '', string.punctuation)
    
    # Process lyrics:
    # 1. Convert to lowercase
    # 2. Remove punctuation using translate()
    # 3. Split into words
    words = lyrics.lower().translate(translator).split()
    
    # Count word frequency
    word_counts = Counter(words)
    
    # Display top 5
    print("Top 5 words:")
    for word, count in word_counts.most_common(5):
        print(f"{word}: {count}")

word_frequency_translate(lyrics)
```

### Differences of above 4 solutions

The **regex-based solution** (`r"[a-zA-Z]+"`) can produce **different results** compared to the other three solutions (`replace()`, `Counter`, or `str.translate()`). Here’s why:

---

### **Key Difference: Handling Apostrophes in Contractions**
1. **Other Solutions (replace/translate):**
   - Convert `"don't"` → `"dont"` (apostrophe **removed**).
   - Example:  
     ```python
     "don't" → "dont" (treated as one word)
     ```

2. **Regex Solution (`r"[a-zA-Z]+"`):**
   - Splits `"don't"` into **two words**: `"don"` and `"t"` (apostrophe **breaks the word**).  
   - Example:  
     ```python
     "don't" → ["don", "t"] (treated as two separate words)
     ```

---

### **Example Output Comparison**
#### **Input Lyrics:**
```python
lyrics = "don't stop believin'"
```

#### **Results:**
| Method               | Output Words                     | Explanation |
|----------------------|----------------------------------|-------------|
| `replace()`/`translate()` | `["dont", "stop", "believin"]` | Apostrophes removed, contractions become single words. |
| **Regex (`r"[a-zA-Z]+"`)** | `["don", "t", "stop", "believin"]` | Apostrophes split contractions into fragments. |

---

### **Why Does This Happen?**
- **`replace()`/`translate()`:**  
  - First **remove all punctuation** (including apostrophes), then split.  
  - `"don't"` → `"dont"` → `["dont"]`.

- **Regex (`r"[a-zA-Z]+"`):**  
  - Matches **alphabetic sequences only**, so apostrophes **act as separators**.  
  - `"don't"` → `"don" + "t"` → `["don", "t"]`.

---

### **Which One is "Correct"?**
It depends on your **use case**:
1. **For word frequency analysis** (e.g., "don't" = 1 word):  
   - Use `replace()`/`translate()` (or include apostrophes in regex: `r"[a-zA-Z']+"` + `replace("'", "")`).  
   - *Better for linguistics* (treat contractions as single words).

2. **For strict alphabetical matching** (e.g., ignore all non-letters):  
   - Use `r"[a-zA-Z]+"`.  
   - *Better for cases where apostrophes are noise* (e.g., OCR cleanup).
