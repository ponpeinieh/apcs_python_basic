### **Exercise 1: OOP Basics - Dog Class**  
**Task:** Compare procedural and object-oriented approaches with a simple Dog example.

#### **Part 1: Procedural Style**
```python
# Data
dog_name = "Buddy"
dog_age = 3
dog_breed = "Golden Retriever"

# Functions
def bark(name, age):
    print(f"{name} (age {age}) says: Woof!")

def eat(name, food, breed):
    print(f"{breed} {name} is eating {food}")

# Usage
bark(dog_name, dog_age)          # Buddy (age 3) says: Woof!
eat(dog_name, "steak", dog_breed) # Golden Retriever Buddy is eating steak
```

**Key Issues:**  
1. Need to pass multiple parameters manually  
2. Data becomes fragmented with more attributes  

---

### **Part 2: Object-Oriented Style**
```python
class Dog:
    def __init__(self, name, age, breed):
        self.name = name
        self.age = age
        self.breed = breed
    
    def bark(self):
        print(f"{self.name} (age {self.age}) says: Woof!")
    
    def eat(self, food):
        print(f"{self.breed} {self.name} is eating {food}")
    
    def have_birthday(self):
        self.age += 1
        print(f"Happy birthday! {self.name} is now {self.age}")

# Usage
dog1 = Dog("Buddy", 3, "Golden Retriever")
dog2 = Dog("Max", 5, "Bulldog")

dog1.bark()          # Buddy (age 3) says: Woof!
dog2.eat("bone")     # Bulldog Max is eating bone
dog1.have_birthday() # Happy birthday! Buddy is now 4
```

**Advantages:**  
1. All related data bundled together  
2. Methods automatically access all attributes  
3. Easier to add new behaviors  

---

### **Comparison Table**
| Aspect         | Procedural                 | OOP                      |
|----------------|----------------------------|--------------------------|
| **Organization**| Scattered variables        | Unified object           |
| **Extensibility**| Hard to add new features | Easy to add methods      |
| **Data Access** | Manual parameter passing   | Automatic via `self`     |

### **練習一：物件導向基礎 - 狗狗類別**  
**任務：** 用簡單的狗狗例子比較程序式與物件導向寫法。

#### **第一部分：程序式寫法**
```python
# 數據
dog_name = "阿毛"
dog_age = 3
dog_breed = "黃金獵犬"

# 函數
def bark(name, age):
    print(f"{name}（{age}歲）在叫：汪汪！")

def eat(name, food, breed):
    print(f"{breed} {name} 正在吃 {food}")

# 使用
bark(dog_name, dog_age)           # 阿毛（3歲）在叫：汪汪！
eat(dog_name, "牛排", dog_breed)  # 黃金獵犬 阿毛 正在吃 牛排
```

**問題點：**  
1. 需手動傳遞多個參數  
2. 新增屬性時代碼會變得混亂  

---

### **第二部分：物件導向寫法**
```python
class Dog:
    def __init__(self, name, age, breed):
        self.name = name
        self.age = age
        self.breed = breed
    
    def bark(self):
        print(f"{self.name}（{self.age}歲）在叫：汪汪！")
    
    def eat(self, food):
        print(f"{self.breed} {self.name} 正在吃 {food}")
    
    def have_birthday(self):
        self.age += 1
        print(f"生日快樂！{self.name} 現在 {self.age} 歲了")

# 使用
dog1 = Dog("阿毛", 3, "黃金獵犬")
dog2 = Dog("小黑", 5, "鬥牛犬")

dog1.bark()          # 阿毛（3歲）在叫：汪汪！
dog2.eat("骨頭")     # 鬥牛犬 小黑 正在吃 骨頭
dog1.have_birthday() # 生日快樂！阿毛 現在 4 歲了
```

**優勢：**  
1. 相關數據自動綁定在一起  
2. 方法可直接存取所有屬性  
3. 更容易擴充新功能

### **Exercise 2: OOP vs Procedural Programming - Bank Account**  
**Task:** Compare OOP and procedural styles by implementing a bank account system.

#### **Part 1: Procedural Style**
```python
# Bank Account System (Procedural)
def create_account(name, initial_balance):
    return {"name": name, "balance": initial_balance}

def deposit(account, amount):
    account["balance"] += amount
    return account["balance"]

def withdraw(account, amount):
    if account["balance"] >= amount:
        account["balance"] -= amount
        return account["balance"]
    else:
        print("Insufficient balance!")
        return account["balance"]

def check_balance(account):
    return account["balance"]

# Testing
account1 = create_account("Alice", 1000)
print(f"{account1['name']}'s balance: ${check_balance(account1)}")  # Alice's balance: $1000
deposit(account1, 500)
print(f"After deposit: ${check_balance(account1)}")  # After deposit: $1500
withdraw(account1, 200)
print(f"After withdrawal: ${check_balance(account1)}")  # After withdrawal: $1300
```

**Procedural Advantages:**  
- Separates data and functions  
- Good for simple scripts  

---

### **Part 2: OOP Style**
```python
class BankAccount:
    def __init__(self, name, initial_balance):
        self.name = name
        self.balance = initial_balance
    
    def deposit(self, amount):
        self.balance += amount
        return self.balance
    
    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            return self.balance
        print("Insufficient balance!")
        return self.balance
    
    def check_balance(self):
        return self.balance

# Testing
account2 = BankAccount("Bob", 2000)
print(f"{account2.name}'s balance: ${account2.check_balance()}")  # Bob's balance: $2000
account2.deposit(300)
print(f"After deposit: ${account2.check_balance()}")  # After deposit: $2300
account2.withdraw(400)
print(f"After withdrawal: ${account2.check_balance()}")  # After withdrawal: $1900
```

**OOP Advantages:**  
- Bundles data and behavior  
- Easier to extend with new features  

---

### **Comparison Table**
| Aspect         | Procedural              | OOP                   |
|----------------|-------------------------|-----------------------|
| **Organization**| Data + functions separate | Integrated objects   |
| **Extensibility** | Hard to maintain      | Easy to add features |
| **Usage**      | `deposit(account, 100)` | `account.deposit(100)` |

### **練習二：物件導向 vs 程序式程式設計 - 銀行帳戶**  
**任務：** 比較兩種程式風格實現的銀行帳戶系統。

#### **第一部分：程序式風格**
```python
# 銀行帳戶系統（程序式）
def create_account(name, initial_balance):
    return {"name": name, "balance": initial_balance}

def deposit(account, amount):
    account["balance"] += amount
    return account["balance"]

def withdraw(account, amount):
    if account["balance"] >= amount:
        account["balance"] -= amount
        return account["balance"]
    else:
        print("餘額不足！")
        return account["balance"]

def check_balance(account):
    return account["balance"]

# 測試
account1 = create_account("Alice", 1000)
print(f"{account1['name']} 的餘額: {check_balance(account1)}")  # Alice 的餘額: 1000
deposit(account1, 500)
print(f"存款後餘額: {check_balance(account1)}")  # 1500
withdraw(account1, 200)
print(f"取款後餘額: {check_balance(account1)}")  # 1300
```

**程序式優點：**  
- 數據與函數分離  
- 適合小型專案  

---

### **第二部分：物件導向風格**
```python
class BankAccount:
    def __init__(self, name, initial_balance):
        self.name = name
        self.balance = initial_balance
    
    def deposit(self, amount):
        self.balance += amount
        return self.balance
    
    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            return self.balance
        print("餘額不足！")
        return self.balance
    
    def check_balance(self):
        return self.balance

# 測試
account2 = BankAccount("Bob", 2000)
print(f"{account2.name} 的餘額: {account2.check_balance()}")  # Bob 的餘額: 2000
account2.deposit(300)
print(f"存款後餘額: {account2.check_balance()}")  # 2300
account2.withdraw(400)
print(f"取款後餘額: {account2.check_balance()}")  # 1900
```

**物件導向優點：**  
- 封裝數據與行為  
- 更容易擴展  

---

### **對照表**
| 項目          | 程序式                 | 物件導向             |
|---------------|------------------------|----------------------|
| **組織方式**  | 數據與函數分離         | 整合為物件           |
| **擴展性**    | 維護困難               | 易於新增功能         |
| **呼叫方式**  | `deposit(account, 100)` | `account.deposit(100)` |
