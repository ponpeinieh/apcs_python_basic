### **Exercise 2: List Method Chaining Challenges**  
**Task:** Explore method chaining limitations and alternatives  

#### **Part 1: Method Chaining Attempts**  
```python
my_list = [1, 2, 3]  
# These methods return None, so chaining fails:
my_list.append(4).insert(1, 'a').extend([5, 6])  # TypeError!
```

#### **Part 2: Traditional Approach**  
```python
my_list = [1, 2, 3]  
my_list.append(4)          # Modifies in-place
my_list.insert(1, 'a')    # Modifies in-place 
my_list.extend([5, 6])    # Modifies in-place
```

#### **Part 3: Delegation-Based Solution**  
**Implement ChainableList:**  
```python
class ChainableList:
    def __init__(self, items=None):
        self._list = list(items) if items else []
    
    def append(self, item):
        self._list.append(item)
        return self  # ← Key difference
    
    def insert(self, index, item):
        self._list.insert(index, item)
        return self
    
    def extend(self, items):
        self._list.extend(items)
        return self
    
    def to_list(self):
        return self._list.copy()
```

#### **Example Usage:**  
```python
result = ChainableList([1, 2]).append(3).insert(1, 'a').extend([4, 5]).to_list()
# Output: [1, 'a', 2, 3, 4, 5]
```

---

### **Key Learning Table**  
| Concept | Python Behavior | Solution |  
|---------|-----------------|----------|  
| Method Chaining | Fails (returns None) | Return `self` |  
| In-place Modification | Standard list behavior | Delegation pattern |  
| API Design | Common need for chaining | Builder pattern |

### **練習二：列表方法鏈式操作挑戰**  
**任務：** 探索鏈式調用的限制與解決方案  

#### **第一部分：鏈式調用嘗試**  
```python
my_list = [1, 2, 3]  
# 這些方法返回None，導致鏈式調用失敗:
my_list.append(4).insert(1, 'a').extend([5, 6])  # 類型錯誤!
```

#### **第二部分：傳統做法**  
```python
my_list = [1, 2, 3]  
my_list.append(4)          # 原地修改
my_list.insert(1, 'a')    # 原地修改
my_list.extend([5, 6])    # 原地修改
```

#### **第三部分：委託模式解決方案**  
**實作可鏈式列表:**  
```python
class ChainableList:
    def __init__(self, items=None):
        self._list = list(items) if items else []
    
    def append(self, item):
        self._list.append(item)
        return self
    
    def insert(self, index, item):
        self._list.insert(index, item)
        return self
    
    def extend(self, items):
        self._list.extend(items)
        return self
    
    def to_list(self):
        return self._list.copy()
```

#### **使用範例:**  
```python
result = ChainableList([1, 2]).append(3).insert(1, 'a').extend([4, 5]).to_list()
# 輸出: [1, 'a', 2, 3, 4, 5]
```

### **練習四：找出所有重複出現的值**  
**任務：** 返回所有重複出現的值（包含多次重複）  

#### **核心功能:**  
1. 保持原始出現順序  
2. 包含首次出現後的所有重複值  
3. 使用集合達到高效O(n)時間複雜度  

#### **測試案例:**  
```python
[1, 2, 2, 3]          → [2]
```

### **Optimized Solution Code**  
```python
# English Version
def find_all_duplicates(items):
    seen = set()
    duplicates = []
    for item in items:
        if item in seen:  # Only check against seen items
            duplicates.append(item)
        else:
            seen.add(item)  # First occurrence
    return duplicates

# Test Cases with Edge Cases
test_cases = [
    ([1, 2, 2, 3], [2]),
    ([1, 2, 2, 2, 3], [2, 2]),
    (['a','b','a','a','c'], ['a', 'a']),
    ([], []),  # Empty list
    ([1, 2, 3], []),  # No duplicates
    ([1, 1, 1, 1], [1, 1, 1])  # All duplicates
]

for input_list, expected in test_cases:
    result = find_all_duplicates(input_list)
    print(f"Input: {input_list} → Result: {result} (Expected: {expected})")
```
