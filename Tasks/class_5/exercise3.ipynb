### **練習一：字串搜索方法比較**  
**任務：** 比較 `find()`, `index()`, 和 `count()` 的差異  

#### **要求：**  
1. 使用相同測試字串  
2. 對每個子字串應用三種方法  
3. 以下列格式輸出結果：  
   ```
   子字串: [substring]
   - find(): [結果]
   - index(): [結果]
   - count(): [結果]
   ```

#### **範例輸出：**  
```
測試字串: "Python程式設計很有趣，Python也很強大"

子字串: "Python"
- find(): 0
- index(): 0
- count(): 2

子字串: "Java"
- find(): -1
- index(): 找不到
- count(): 0

子字串: "有趣"
- find(): 8
- index(): 8
- count(): 1
```

---

### **Sample Solution Code**  
```python
# English Version
text = "Python programming is fun and Python is powerful"
subs = ["Python", "Java", "fun"]

print(f'Test String: "{text}"\n')
for sub in subs:
    print(f'Sub: "{sub}"')
    print(f'- find(): {text.find(sub)}')
    
    try:
        print(f'- index(): {text.index(sub)}')
    except ValueError:
        print('- index(): Not found')
    
    print(f'- count(): {text.count(sub)}\n')

# 繁體中文版
text = "Python程式設計很有趣，Python也很強大"
subs = ["Python", "Java", "有趣"]

print(f'測試字串: "{text}"\n')
for sub in subs:
    print(f'子字串: "{sub}"')
    print(f'- find(): {text.find(sub)}')
    
    try:
        print(f'- index(): {text.index(sub)}')
    except ValueError:
        print('- index(): 找不到')
    
    print(f'- count(): {text.count(sub)}\n')
```

---

### **Key Observations Table**  
| Scenario          | `find()` | `index()` | `count()` | 中文對比               |
|-------------------|----------|-----------|-----------|------------------------|
| Substring exists  | position | position  | count     | 子字串存在時            |
| Substring missing | -1       | Error     | 0         | 子字串不存在時          |
| Empty string      | 0        | 0         | length+1  | 空字串的特殊情況        |

**Teaching Points:**
1. `find()` vs `index()`: Safe vs strict searching
2. `count()` always returns a number
3. All methods are case-sensitive
4. Methods work the same with non-ASCII characters

### **練習二：查找所有子字串位置**  
**任務：** 不使用`count()`查找所有子字串出現位置  

#### **要求：**  
1. 僅使用`find()`實現  
2. 妥善處理空字串情況  
3. 返回索引列表  

#### **範例輸出：**  
```
文字: "你好你好世界", 子字串: "你好" → [0, 2]
文字: "測試測試", 子字串: "測試" → [0, 2]
文字: "無匹配", 子字串: "XYZ" → []
```

---

### **Sample Solution Code**  
```python
# English Version - Explicit Implementation
def find_all(text, sub):
    positions = []
    if len(sub) == 0:  # Handle empty substring case
        return positions
        
    index = text.find(sub)
    while index != -1:
        positions.append(index)
        index = text.find(sub, index + len(sub))
    return positions

# Test Cases
test_cases = [
    ("hello hello world", "hello"),
    ("ababab", "ab"),
    ("no match", "xyz"),
    ("", "test")  # Edge case: empty text
]

for text, sub in test_cases:
    result = find_all(text, sub)
    print(f'Text: "{text}", Sub: "{sub}" → {result}')

# 繁體中文版 - 明確條件語句實現
def find_all(text, sub):
    positions = []
    if len(sub) == 0:  # 處理空子字串情況
        return positions
        
    index = text.find(sub)
    while index != -1:
        positions.append(index)
        next_index = index + len(sub)
        index = text.find(sub, next_index)
    return positions

# 測試案例
test_cases = [
    ("你好你好世界", "你好"),
    ("測試測試", "測試"),
    ("無匹配", "XYZ"),
    ("", "測試")  # 邊界案例: 空文字
]

for text, sub in test_cases:
    result = find_all(text, sub)
    print(f'文字: "{text}", 子字串: "{sub}" → {result}')
```

---

**Teaching Notes:**  
1. The empty substring check prevents infinite loops  
2. Using `len(sub)` for jump size handles multi-character substrings  
3. Explicit variable names make the logic self-documenting

### **練習四：用戶名驗證**  
**任務：** 使用字串方法創建用戶名驗證器  

#### **要求：**  
1. 驗證規則：  
   - 長度3-16個字符  
   - 僅含字母數字（`isalnum()`）  
   - 首字符必須為字母（`isalpha()`）  
2. 創建函式 `validate_username(username)` 返回 `True`/`False`  
3. 測試案例：  
   - 有效: `"小明123"`, `"阿華"`  
   - 無效: `"123小美"`, `"a!"`, `"過長的用戶名超過十六字"`  

#### **範例輸出：**  
```
用戶名: "小明123" → 有效
用戶名: "123小美" → 無效 (以數字開頭)
用戶名: "a!" → 無效 (含特殊符號)
用戶名: "過長的用戶名超過十六字" → 無效 (過長)
```

---

### **Sample Solution Code**  
```python
# English Version
def validate_username(username):
    return (3 <= len(username) <= 16 and 
            username.isalnum() and 
            username[0].isalpha())

test_cases = [
    "Alice123",    # Valid
    "bob",        # Valid
    "123Alice",   # Invalid (starts with number)
    "a!",         # Invalid (symbol)
    "way_too_long_username"  # Invalid (length)
]
for username in test_cases:
    if validate_username(username):
        result = "Valid"
    else:
        if username[0].isdigit():
            reason = "starts with number"
        elif not username.isalnum():
            reason = "contains symbol"
        elif len(username) > 16:
            reason = "too long"
        else:
            reason = "too short"
        result = f"Invalid ({reason})"
    print(f"Username: {username} → {result}")

# 繁體中文版
def validate_username(username):
    return (3 <= len(username) <= 16 and 
            username.isalnum() and 
            username[0].isalpha())

test_cases = [
    "小明123",    # 有效
    "阿華",      # 有效
    "123小美",   # 無效 (以數字開頭)
    "a!",       # 無效 (特殊符號)
    "過長的用戶名超過十六字"  # 無效 (長度)
]

for username in test_cases:
    if validate_username(username):
        result = "有效"
    else:
        if username[0].isdigit():
            reason = "以數字開頭"
        elif not username.isalnum():
            reason = "含特殊符號"
        elif len(username) > 16:
            reason = "過長"
        else:
            reason = "過短"
        result = f"無效 ({reason})"
    print(f"用戶名: {username} → {result}")

```

|-------|-------------|----------|  
| Length | `len()` | 長度檢查 |  
| Alphanumeric | `isalnum()` | 字母數字檢查 |  
| First character | `[0].isalpha()` | 首字符檢查 |  

**Teaching Tips:**

**Advanced Extension:**  
```python
def validate_username_with_reasons(username):
    checks = {
        "length": 3 <= len(username) <= 16,
        "alphanumeric": username.isalnum(),
        "first_char": username[0].isalpha() if username else False
    }
    return all(checks.values()), checks
```
