### **Exercise 1: Creating Turtle Objects**  
**Objective:** Learn object creation and property modification  

#### **Step-by-Step Instructions:**
1. **Import the turtle module**
   ```python
   import turtle
   ```

2. **Create turtle objects**  
   ```python
   t1 = turtle.Turtle()  # Constructor creates new object
   t2 = turtle.Turtle()
   ```

3. **Set properties for t1**  
   ```python
   t1.color("red")      # Set line color
   t1.shape("turtle")   # Change appearance
   t1.speed(3)          # 1=slowest, 10=fastest
   ```

4. **Set properties for t2**  
   ```python
   t2.color("blue")
   t2.shape("arrow")
   t2.speed(5)
   ```

5. **Make them move**  
   ```python
   t1.forward(100)      # Move 100 pixels
   t2.left(90)          # Turn 90 degrees
   t2.forward(150)
   ```

6. **Jupyter-safe termination**  
   ```python
   try:
       turtle.done()    # Complete rendering
       turtle.bye()     # Close window
   except:
       print("Graphics terminated")
   ```

#### **Full Solution:**
```python
import turtle

# Create and customize turtles
t1 = turtle.Turtle()
t2 = turtle.Turtle()

t1.color("red")
t1.shape("turtle")
t1.speed(3)

t2.color("blue")
t2.shape("arrow")
t2.speed(5)

# Movement
t1.forward(100)
t2.left(90)
t2.forward(150)

# Safe exit
try:
    turtle.done()
    turtle.bye()
except:
    print("Graphics terminated")
```

### **Exercise 3: Interactive Methods**  
**Objective:** Object-to-object interaction  

#### **Step-by-Step Instructions:**
1. **Setup turtles**  
   ```python
   explorer = turtle.Turtle()
   guard = turtle.Turtle()
   guard.goto(100, 0)
   ```

2. **Create collision logic**  
   ```python
   def check_collision(t1, t2):
       if t1.distance(t2) < 50:
           t1.color("red")
           print("Alert: Collision!")
   ```

3. **Simulate interaction**  
   ```python
   check_collision(explorer, guard)
   explorer.goto(30, 0)
   check_collision(explorer, guard)
   ```

#### **Full Solution:**
```python
import turtle

explorer = turtle.Turtle()
guard = turtle.Turtle()
guard.goto(100, 0)

def check_collision(t1, t2):
    if t1.distance(t2) < 50:
        t1.color("red")
        print("Alert: Collision!")

check_collision(explorer, guard)
explorer.goto(30, 0)
check_collision(explorer, guard)

try:
    turtle.done()
    turtle.bye()
except:
    print("Graphics terminated")
```

### **練習一：創建Turtle物件**  
**目標：** 學習物件創建與屬性設定  

#### **完整解答:**
```python
import turtle

# 創建烏龜物件
t1 = turtle.Turtle()
t2 = turtle.Turtle()

# 設定屬性
t1.color("red")      # 設定顏色
t1.shape("turtle")   # 設定形狀
t1.speed(3)          # 設定速度

t2.color("blue")
t2.shape("arrow")
t2.speed(5)

# 移動指令
t1.forward(100)      # 前進100像素
t2.left(90)          # 左轉90度
t2.forward(150)

# 安全結束
try:
    turtle.done()
    turtle.bye()
except:
    print("圖形視窗已關閉")
```

### **練習二：自訂屬性**  
**目標：** 追蹤物件狀態  

#### **完整解答:**
```python
import turtle

runner = turtle.Turtle()
runner.steps = 0  # 自訂步數屬性

def move(t, distance):
    t.forward(distance)
    t.steps += distance
    print(f"總移動步數: {t.steps}")

move(runner, 50)
move(runner, 30)

try:
    turtle.done()
    turtle.bye()
except:
    print("圖形視窗已關閉")
```

### **練習三：互動方法**  
**目標：** 物件間互動  

#### **完整解答:**
```python
import turtle

explorer = turtle.Turtle()
guard = turtle.Turtle()
guard.goto(100, 0)

def check_collision(t1, t2):
    if t1.distance(t2) < 50:
        t1.color("red")
        print("警告：發生碰撞！")

check_collision(explorer, guard)
explorer.goto(30, 0)
check_collision(explorer, guard)

try:
    turtle.done()
    turtle.bye()
except:
    print("圖形視窗已關閉")
```

### **Exercise 4: Drawing a Hexagon**  
**Objective:** Create a turtle program that draws a regular hexagon using loops and object methods.

```python
import turtle

# Create and configure turtle
artist = turtle.Turtle()
artist.color("teal")
artist.pensize(3)

# Draw hexagon
for _ in range(6):
    artist.forward(100)
    artist.left(60)

# Window management
try:
    turtle.done()
    turtle.bye()
except:
    print("Display completed")
```
