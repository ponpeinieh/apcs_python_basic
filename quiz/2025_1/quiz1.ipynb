### q181. 1. 等紅綠燈 - 2025年1月APCS

#### 問題描述

操場起跑線上有一個紅綠燈，綠燈為 `a` 秒，紅燈為 `b` 秒，依照綠燈→紅燈的順序循環。有 `n` 個小朋友，從操場的起跑線騎腳踏車一起起跑，他們分別騎完一圈的時間為 `t1, t2, ..., tn`。若騎到終點時為紅燈，需要等待紅燈結束變為綠燈才可以停止騎車。求出這 `n` 個小朋友共需要等待幾秒的紅燈秒數。

**輸入說明：**
- 第一行輸入兩個正整數 `a, b`（1 ≤ a, b ≤ 100），分別表示綠燈時間和紅燈時間
- 接下來一行有一個正整數 `n`（1 ≤ n ≤ 30），表示小朋友數量
- 最後一行有 `n` 個正整數數字（每個小朋友騎一圈的時間），數字不超過 1000

**評分標準：**
- (60 分): `a = b = 10, n = 1`
- (40 分): 無限制

---

#### Problem Description

There is a traffic light at the starting line of a playground. The green light lasts for `a` seconds and the red light lasts for `b` seconds, cycling in the order: green → red → green → ... There are `n` children who start riding bicycles from the starting line at the same time. The time each child takes to complete one lap is `t1, t2, ..., tn`. If a child arrives at the finish line during a red light, they must wait until the red light ends and turns green. Find the total number of seconds all children need to wait for red lights.

**Input Format:**
- First line: two positive integers `a, b` (1 ≤ a, b ≤ 100), green and red light durations
- Second line: one positive integer `n` (1 ≤ n ≤ 30), number of children
- Third line: `n` positive integers (time each child takes), each number ≤ 1000

**Scoring:**
- (60 points): `a = b = 10, n = 1`
- (40 points): no restrictions

---

#### 解答

```python
def calculate_waiting_time(a, b, n, times):
    """
    計算所有小朋友的紅燈等待時間總和
    Calculate total waiting time for all children
    
    Args:
        a: 綠燈時間 / Green light duration
        b: 紅燈時間 / Red light duration
        n: 小朋友數量 / Number of children
        times: 每個小朋友騎一圈的時間列表 / List of times for each child
    
    Returns:
        總等待時間 / Total waiting time
    """
    total_waiting = 0
    cycle_length = a + b  # 一個完整的紅綠燈周期 / One complete traffic light cycle
    
    for t in times:
        # 計算到達時間在紅綠燈周期中的位置 / Calculate position in traffic light cycle
        position_in_cycle = t % cycle_length
        
        # 如果到達時間在紅燈期間，需要等待 / If arriving during red light, need to wait
        if position_in_cycle >= a:  # 注意：這裡使用 >= a，因為綠燈區間是 [0, a-1]
            # 計算需要等待的時間 / Calculate waiting time
            waiting_time = cycle_length - position_in_cycle
            total_waiting += waiting_time
    
    return total_waiting

def main():
    # 讀取第一行：綠燈和紅燈時間 / Read first line: green and red light durations
    a, b = map(int, input().split())
    
    # 讀取第二行：小朋友數量 / Read second line: number of children
    n = int(input().strip())
    
    # 讀取第三行：每個小朋友的時間 / Read third line: times for each child
    times = list(map(int, input().split()))
    
    # 計算總等待時間 / Calculate total waiting time
    result = calculate_waiting_time(a, b, n, times)
    
    # 輸出結果 / Output result
    print(result)

if __name__ == "__main__":
    main()
```

#### 程式驗證一

```python
# a = b = 10, n = 1
a, b = 10, 10
n = 1
times = [14]  # 或其他不超過1000的數字

# position_in_cycle = 14 % 20 = 14
# 14 >= 10 → 需要等待
# waiting_time = 20 - 14 = 6
# 輸出：6 ✓
```

#### 程式驗證二

```python
# 任意 a, b, n 值
a, b = 4, 3
n = 3
times = [12, 16, 25]

# 周期 = 7秒
# 第一個小朋友：12 % 7 = 5 → 5 >= 4 → 等待 7-5=2秒
# 第二個小朋友：16 % 7 = 2 → 2 < 4 → 不等待
# 第三個小朋友：25 % 7 = 4 → 4 >= 4 → 等待 7-4=3秒
# 總等待：2 + 0 + 3 = 5秒 ✓
```

````python
a,b = map(int, input().split())
n = int(input())
ts = list(map(int, input().split()))

# a,b = 10,10
# n = 1
# ts = [14]

# a,b = 4, 3
# n = 3
# ts = [12,16,25]

total_time = 0
for t in ts:
    r = t % (a+b)
    if r>=a:
        total_time += (b-(r-a)) 
print(total_time)
````
