### q182. 2. 字串操作 - 2025年1月APCS

#### 問題描述 

給一個字串 S，有三種字串操作：

1. **兩兩交換**：將字串內相鄰兩個字元交換，例如字串 "apcsntnu" 先分組成 (ap)(cs)(nt)(nu)，將會交換為 (pa)(sc)(tn)(un)，即得到 "pasctnun"。

2. **兩兩排序**：將字串內相鄰兩個字元按照字典序排序，例如字串 "family" 先分組成 (fa)(mi)(ly)，將會排序成 (af)(im)(ly)，即得到 "afimly"。

3. **完美重排**：假設字串長度為 n，將字串分成兩半為 [0, n/2-1] 和 [n/2, n-1]，並且組合成 0, n/2, 1, n/2+1, ..., n/2-1, n-1。例如 "apcsntnu" 拆成 "apcs" 和 "ntnu"，然後交錯成 "anptcnsu"。

給定 k 個操作，請依序操作字串，輸出最後的字串結果。

**輸入說明：**
- 第一行輸入一個字串 S (2 ≤ |S| ≤ 100)，字串長度保證為偶數
- 接下來一行輸入一個正整數 k
- 接下來有 k 行，每一行有一個數字，種類為 0, 1, 2

**評分標準：**
- (60 分): |S| < 10, k = 1 且保證操作為完美重排
- (40 分): 無限制

---

#### Problem Description

Given a string S, there are three string operations:

1. **Pairwise Swap**: Swap adjacent pairs of characters. For example, string "apcsntnu" is grouped as (ap)(cs)(nt)(nu) and becomes (pa)(sc)(tn)(un), resulting in "pasctnun".

2. **Pairwise Sort**: Sort adjacent pairs of characters in lexicographical order. For example, string "family" is grouped as (fa)(mi)(ly) and becomes (af)(im)(ly), resulting in "afimly".

3. **Perfect Rearrangement**: For a string of length n, split into two halves [0, n/2-1] and [n/2, n-1], then interleave them as 0, n/2, 1, n/2+1, ..., n/2-1, n-1. For example, "apcsntnu" becomes "anptcnsu".

Given k operations, apply them in sequence and output the final string result.

**Input Format:**
- First line: string S (2 ≤ |S| ≤ 100), length is guaranteed to be even
- Second line: positive integer k
- Next k lines: each containing a number 0, 1, or 2

**Scoring:**
- (60 points): |S| < 10, k = 1 and guaranteed to be perfect rearrangement
- (40 points): no restrictions

---

#### 解答

```python
def pairwise_swap(s):
    """兩兩交換：相鄰兩個字元交換"""
    result = []
    for i in range(0, len(s), 2):
        # 交換相鄰兩個字元
        result.append(s[i+1])
        result.append(s[i])
    return ''.join(result)

def pairwise_sort(s):
    """兩兩排序：相鄰兩個字元按字典序排序"""
    result = []
    for i in range(0, len(s), 2):
        # 對相鄰兩個字元排序
        pair = sorted([s[i], s[i+1]])
        result.extend(pair)
    return ''.join(result)

def perfect_rearrangement(s):
    """完美重排：交錯組合兩半字串"""
    n = len(s)
    half = n // 2
    first_half = s[:half]
    second_half = s[half:]
    
    result = []
    for i in range(half):
        result.append(first_half[i])
        result.append(second_half[i])
    
    return ''.join(result)

def main():
    # 讀取輸入字串
    s = input().strip()
    
    # 讀取操作數量
    k = int(input().strip())
    
    # 讀取操作序列
    operations = []
    for _ in range(k):
        op = int(input().strip())
        operations.append(op)
    
    # 依序執行操作
    current_string = s
    for op in operations:
        if op == 0:
            current_string = pairwise_swap(current_string)
        elif op == 1:
            current_string = pairwise_sort(current_string)
        elif op == 2:
            current_string = perfect_rearrangement(current_string)
    
    # 輸出結果
    print(current_string)

if __name__ == "__main__":
    main()
```

#### 測試範例 #2

**輸入：**
```

```

**計算過程：**
1. 原始字串: "facebook"
2. 操作1 (2): 完美重排
   - 分成兩半: "face" 和 "book"
   - 交錯組合: f, b, a, o, c, o, e, k → "fbaocoek"
3. 操作2 (0): 兩兩交換
   - 分組: (fb)(ao)(co)(ek)
   - 交換: (bf)(oa)(oc)(ke) → "bfoaocke"
4. 操作3 (2): 完美重排
   - 分成兩半: "bfoa" 和 "ocke"
   - 交錯組合: b, o, f, c, o, k, a, e → "bofcokae"
5. 操作4 (1): 兩兩排序
   - 分組: (bo)(fc)(ok)(ae)
   - 排序: (bo)(cf)(ko)(ae) → "bocfkoae"
- **最終輸出：bocfkoae**

````python
def swapit(clist): # char list
    for i in range(0, len(clist), 2):
        clist[i], clist[i+1] = clist[i+1], clist[i]
def sortit(clist):
    for i in range(0, len(clist), 2):
        if clist[i]>clist[i+1]:
            clist[i], clist[i+1] = clist[i+1], clist[i]  
def rearrangeit(clist):
    # first make a copy of the last half
    copylist = clist[len(clist)//2:]
    for i in range(len(clist)//2-1, -1, -1):
        clist[2*i]=clist[i]
    for i in range(len(copylist)):
        clist[2*i+1]=copylist[i]

s = input()
k = int(input())
ops =[]
for _ in range(k):
    ops.append(int(input()))

# s = 'apcsntnu'
# k = 1
# ops = [2]

# s= 'facebook'
# k = 4
# ops= [2, 0, 2, 1]

clist = [c for c in s]  
for op in ops:
    if op==0:
        swapit(clist)
    elif op==1:
        sortit(clist)
    else:
        rearrangeit(clist)
print("".join(clist))
````
