### i399. 1. 數字遊戲

#### 題目描述

給三個介於 1 ~ 9 的整數 A₁, A₂, A₃。

先輸出一個正整數 P 表示眾數數量，也就是出現最多次的數字的次數。

接下來將輸入的三個數字去除重複（剩下一個）後由大到小依序輸出。

**輸入說明**
- 三個介於 1~9 的整數

**子題配分**
- (60%): 每個數字都不相同
- (40%): 無額外限制

**輸出說明**
1. 先輸出一個數值 P 表示出現最多次的數字個數
2. 接下來將輸入的三個數字去除重複（剩下一個）後由大到小依序輸出

#### 測試範例

**範例輸入 #1**
```
6 6 6
```

**範例輸出 #1**
```
3 6
```

**範例輸入 #2**
```
7 9 7
```

**範例輸出 #2**
```
2 9 7
```

**範例輸入 #3**
```
4 1 8
```

**範例輸出 #3**
```
1 8 4 1
```

#### 程式碼解答

```python
# 讀取輸入
numbers = list(map(int, input().split()))

# 計算每個數字出現的次數
from collections import Counter
count = Counter(numbers)

# 找出眾數數量（出現最多次的次數）
max_count = max(count.values())

# 去除重複並由大到小排序
unique_sorted = sorted(set(numbers), reverse=True)

# 輸出結果
print(max_count, end=' ')
print(' '.join(map(str, unique_sorted)))
```

### i399. 1. Number Game

#### Problem Description

Given three integers A₁, A₂, A₃ between 1 ~ 9.

First output a positive integer P representing the mode count, which is the frequency of the most frequently occurring number.

Then output the three input numbers after removing duplicates (keeping one of each) in descending order.

**Input Specification**
- Three integers between 1~9

**Scoring**
- (60%): All numbers are distinct
- (40%): No additional constraints

**Output Specification**
1. First output a value P representing the frequency of the most frequently occurring number
2. Then output the three input numbers after removing duplicates (keeping one of each) in descending order

#### Test Examples

**Example Input #1**
```
6 6 6
```

**Example Output #1**
```
3 6
```

**Example Input #2**
```
7 9 7
```

**Example Output #2**
```
2 9 7
```

**Example Input #3**
```
4 1 8
```

**Example Output #3**
```
1 8 4 1
```

#### Code Solution

```python
# Read input
numbers = list(map(int, input().split()))

# Calculate frequency of each number
from collections import Counter
count = Counter(numbers)

# Find the mode count (maximum frequency)
max_count = max(count.values())

# Remove duplicates and sort in descending order
unique_sorted = sorted(set(numbers), reverse=True)

# Output results
print(max_count, end=' ')
print(' '.join(map(str, unique_sorted)))
```
