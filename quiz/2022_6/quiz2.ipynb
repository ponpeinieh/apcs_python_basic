### i400. 2. 字串解碼

#### 題目描述

有一個字串加密系統，對於一個長度為 `n` 的字串 `S` 和一個長度為 `n` 的 01 字串 `e`，會產生一個加密過後的字串 `Y`，其中加密流程為以下兩個步驟：

1. **第一步驟**：
   - 如果字串 `e` 中 `1` 的出現次數是偶數，則直接進第二步驟
   - 如果出現次數是奇數，則將字串 `S` 平分成兩等份，兩份順序交換後再接起來
   - 如果字串長度是奇數，則最中間的字元不動位置

2. **第二步驟**：
   - 讓 `i` 從 `1` 到 `n` 迭代
   - 如果 `e[i] = 0`，就將 `S` 的第一個字元切掉並接到字串 `T` 的最後
   - 如果 `e[i] = 1`，就將 `S` 的最後一個字元切掉並接到字串 `T` 的最後

以範例 1 為例，e
 陣列為 10110，其中 1
 出現的次數為奇數，因此需要交換字串 S
 從 BCAAD 變為 ADABC。接下來執行第二階段，其中過程如下。

| i    |  1   |   2  |   3  |   4  | 5    |
|-----|-----|-----|-----|-----|-----|
|  e   |  1   |   0  |  1   |   1  |   0  |
|  S   |  ADAB   | DAB    |   DA  | D    |   空字串  |
|   T  |   C  | CA    |  CAB   |   CABA  |  CABAD   | 


現在給定 `m` 個 01 字串的加密表和按照順序加密原字串後得到加密過後的字串，請嘗試還原出原本的字串。

**輸入說明**
- 第一行輸入兩個數字 `m, n` (1<=m,n<=100)
- 接下來有 `m` 個長度為 `n` 的 01 字串
- 最後輸入一個長度為 `n` 的加密字串（字串均由大寫字母組成）

**子題配分**
- (60%): m=1
- (40%): 無額外限制

**輸出說明**
- 輸出解密後的字串

#### 測試範例

**範例輸入 #1**
```
1 5
10110
CABAD
```

**範例輸出 #1**
```
BCAAD
```

**範例輸入 #2**
```
3 6
111110
101101
000000
RETYWQ
```

**範例輸出 #2**
```
QWERTY
```

#### 程式碼解答

```python
from collections import deque

def decrypt_string(m, n, e_list, encrypted):
    # 從加密字串開始反向操作
    current = deque(encrypted)
    
    # 反向處理每個加密步驟（從最後一個到第一個）
    for e in reversed(e_list):
        S = deque()
        T = current  # 當前加密結果
        
        # 反向第二步驟
        for i in range(n-1, -1, -1):
            if e[i] == '0':
                # 原操作：從S前面取字元加到T後面
                # 反向：從T後面取字元放回S前面
                S.appendleft(T.pop())
            else:  # e[i] == '1'
                # 原操作：從S後面取字元加到T後面  
                # 反向：從T後面取字元放回S後面
                S.append(T.pop())
        
        current = S
        
        # 反向第一步驟：檢查1的個數是否為奇數
        ones_count = e.count('1')
        if ones_count % 2 == 1:
            # 原操作：交換前後半
            # 反向：再做一次交換即可還原
            mid = n // 2
            if n % 2 == 0:
                # 將deque轉為list進行切片，再轉回deque
                temp_list = list(current)
                current = deque(temp_list[mid:] + temp_list[:mid])
            else:
                temp_list = list(current)
                current = deque(temp_list[mid+1:] + [temp_list[mid]] + temp_list[:mid])
    
    return ''.join(current)

# 讀取輸入
m, n = map(int, input().split())
e_list = []
for _ in range(m):
    e_list.append(input().strip())
encrypted = input().strip()

# 解密並輸出結果
result = decrypt_string(m, n, e_list, encrypted)
print(result)
```

### i400. 2. String Decoding

#### Problem Description

There is a string encryption system. For a string `S` of length `n` and a 01 string `e` of length `n`, it produces an encrypted string `Y` through the following two steps:

1. **Step 1**:
   - If the number of `1`s in string `e` is even, proceed directly to step 2
   - If the count is odd, split string `S` into two equal parts, swap their order, and concatenate them
   - If the string length is odd, the middle character remains in position

2. **Step 2**:
   - Iterate `i` from `1` to `n`
   - If `e[i] = 0`, remove the first character from `S` and append it to the end of string `T`
   - If `e[i] = 1`, remove the last character from `S` and append it to the end of string `T`

Taking Example 1 as an example, the array `e` is 10110, where the number of `1`s is odd, so string `S` needs to be swapped from BCAAD to ADABC. Then proceed to the second stage, with the process as follows:

| i    |  1   |   2  |   3  |   4  | 5    |
|-----|-----|-----|-----|-----|-----|
|  e   |  1   |   0  |  1   |   1  |   0  |
|  S   |  ADAB   | DAB    |   DA  | D    |   Empty string  |
|   T  |   C  | CA    |  CAB   |   CABA  |  CABAD   | 

Now given `m` encryption 01 strings and the final encrypted string obtained by sequentially applying the encryption process, try to recover the original string.

**Input Specification**
- First line: two numbers `m, n` (1<=m,n<=100)
- Next `m` lines: each contains a 01 string of length `n`
- Last line: an encrypted string of length `n` (string consists of uppercase letters only)

**Scoring**
- (60%): m=1
- (40%): No additional constraints

**Output Specification**
- Output the decrypted string

#### Test Examples

**Example Input #1**
```
1 5
10110
CABAD
```

**Example Output #1**
```
BCAAD
```

**Example Input #2**
```
3 6
111110
101101
000000
RETYWQ
```

**Example Output #2**
```
QWERTY
```

#### Code Solution

```python
from collections import deque

def decrypt_string(m, n, e_list, encrypted):
    # Start from the encrypted string and reverse operations
    current = deque(encrypted)
    
    # Process each encryption step in reverse order (from last to first)
    for e in reversed(e_list):
        S = deque()
        T = current  # Current encrypted result
        
        # Reverse step 2
        for i in range(n-1, -1, -1):
            if e[i] == '0':
                # Original: take from front of S, append to T
                # Reverse: take from back of T, put to front of S
                S.appendleft(T.pop())
            else:  # e[i] == '1'
                # Original: take from back of S, append to T
                # Reverse: take from back of T, put to back of S
                S.append(T.pop())
        
        current = S
        
        # Reverse step 1: Check if count of 1s is odd
        ones_count = e.count('1')
        if ones_count % 2 == 1:
            # Original: swap first and second halves
            # Reverse: apply the same swap to recover
            mid = n // 2
            if n % 2 == 0:
                # Convert deque to list for slicing, then back to deque
                temp_list = list(current)
                current = deque(temp_list[mid:] + temp_list[:mid])
            else:
                temp_list = list(current)
                current = deque(temp_list[mid+1:] + [temp_list[mid]] + temp_list[:mid])
    
    return ''.join(current)

# Read input
m, n = map(int, input().split())
e_list = []
for _ in range(m):
    e_list.append(input().strip())
encrypted = input().strip()

# Decrypt and output result
result = decrypt_string(m, n, e_list, encrypted)
print(result)
```
