### k732. 2. 特殊位置計算

#### 問題描述

給定一個 `n × m` 的二維矩陣 `a`，設 `x = a[i][j]`，離 `(i, j)` 曼哈頓距離為 `x` 內的點數值總和 `% 10` 恰為 `x` 的稱之為特殊位置。定義兩個點 `(a, b)` 和 `(c, d)` 的曼哈頓距離為 `|a - c| + |b - d|`。

請寫一個程式，輸出共有幾個特殊位置，並按照字典序由小到大輸出這些位置的座標。

#### 子問題
- 子問題一 (60%) `n=1`
- 子問題二 (40%) `n ≤ 50, m ≤ 50`

#### 輸入說明
- 第一行輸入兩個正整數 `n, m` (`1≤n, m ≤ 50`)，接下來有 `n` 行，每行有 `m` 個數字，每一個數字介於 `0` 到 `9`。

#### 輸出說明
- 第一行輸出共有幾個特殊位置，接下來輸出 `k` 行，每一行輸出兩個正整數代表座標。特殊位置請按照字典順序由小到大輸出。

---

### 測試範例 #1 驗證

**輸入：**
```
1 8
1 2 3 4 5 6 7 8
```

**輸出：**
```
1
0 5
```

### 測試範例 #2 驗證

**輸入：**
```
2 3
5 2 3
4 5 6
```

**輸出：**
```
2
0 0 
1 1
```

---

#### 解答

```python
def manhattan_special_positions(n, m, matrix):
    special_positions = []
    
    for i in range(n):
        for j in range(m):
            k = matrix[i][j]
            total_sum = 0
            
            # Calculate the sum of values within Manhattan distance k
            for x in range(n):
                for y in range(m):
                    if abs(i - x) + abs(j - y) <= k:
                        total_sum += matrix[x][y]
            
            # Check if the condition for special position holds
            if total_sum % 10 == k:
                special_positions.append((i, j))
    
    # Sort special positions lexicographically
    special_positions.sort()
    
    # Output the result
    print(len(special_positions))
    for pos in special_positions:
        print(pos[0], pos[1])

def main():
    n, m = map(int, input().split())
    matrix = [list(map(int, input().split())) for _ in range(n)]
    manhattan_special_positions(n, m, matrix)

if __name__ == "__main__":
    main()
```

### k732. 2. Special Position Calculation

#### Problem Description

Given an `n × m` two-dimensional matrix `a`, let `x = a[i][j]`. A position is called a special position if the sum of the values of points within a Manhattan distance of `x` from `(i, j)` is `sum % 10` equal to `x`. The Manhattan distance between two points `(a, b)` and `(c, d)` is defined as `|a - c| + |b - d|`.

Please write a program that outputs the total number of special positions and the coordinates of these positions in lexicographical order from smallest to largest.

#### Subproblems
- Subproblem 1 (60%) `n=1`
- Subproblem 2 (40%) `n ≤ 50, m ≤ 50`

#### Input Description
- The first line contains two positive integers `n, m` (`1≤n, m ≤ 50`). The following `n` lines each have `m` numbers, where each number is between `0` and `9`.

#### Output Description
- The first line outputs the total number of special positions, followed by `k` lines, each outputting two positive integers representing the coordinates. Special positions should be output in lexicographical order from smallest to largest.
