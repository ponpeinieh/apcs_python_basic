### k731. 1. 路徑偵測

#### 問題描述

給一個二維平面，座標如同數學的二維座標（Y 正為北，X 正為東）。起始位置在 (0, 0)，接下來會有 $ n $ 個座標，你需要按照這些座標點的順序移動，保證僅會垂直或水平方向上移動，不會斜向移動，且第一個點保證一定是 X 軸正的位置（初始方向向右）。

請輸出這條路徑中，左轉、右轉、迴轉的個數分別為多少。

#### 子問題
- 子問題一 (60%) $ n=2 $
- 子問題二 (40%) $ n \leq 100 $

#### 輸入說明
- 第一行輸入一個正整數 $ n $。
- 接下來有 $ n $ 行，每一行都有兩個正整數 $ x, y $。保證的是相鄰兩個點的座標差值不超過 100。

#### 輸出說明
- 輸出三個正整數，分別代表左轉、右轉、迴轉的次數。

---

def count_turns(points):
    left_turns = 0
    right_turns = 0
    u_turns = 0

    # 定義方向變化
    direction = [(1, 0), (0, 1), (-1, 0), (0, -1)]  # 右, 上, 左, 下
    current_direction = 0  # 初始方向向右

    for i in range(1, len(points)):
        x1, y1 = points[i - 1]
        x2, y2 = points[i]

        if x1 == x2:  # 垂直移動
            new_direction = 1 if y2 > y1 else 3  # 向上或向下
        else:  # 水平移動
            new_direction = 0 if x2 > x1 else 2  # 向右或向左

        # 計算方向變化
        turn = (new_direction - current_direction) % 4
        if turn == 1:  # 右轉
            right_turns += 1
        elif turn == 3:  # 左轉
            left_turns += 1
        elif turn == 2:  # 迴轉
            u_turns += 1

        current_direction = new_direction  # 更新當前方向

    return left_turns, right_turns, u_turns

def main():
    n = int(input())
    points = [tuple(map(int, input().split())) for _ in range(n)]
    turns = count_turns(points)
    print(turns[0], turns[1], turns[2])

if __name__ == "__main__":
    main()

````python
def count_turns(coordinates):
    """
    Count left turns, right turns, and U-turns in a path
    
    Args:
        coordinates: List of tuples [(x1, y1), (x2, y2), ...]
    
    Returns:
        tuple: (left_turns, right_turns, u_turns)
    """
    if len(coordinates) < 2:
        return 0, 0, 0
    
    left_turns = 0
    right_turns = 0
    u_turns = 0
    
    # Initial direction is right (east) as specified
    current_direction = 'E'  # E: east, N: north, W: west, S: south
    
    # Start from the first movement (from origin to first coordinate)
    for i in range(len(coordinates) - 1):
        current_point = coordinates[i]
        next_point = coordinates[i + 1]
        
        # Calculate the direction of movement to next point
        dx = next_point[0] - current_point[0]
        dy = next_point[1] - current_point[1]
        
        # Determine the new direction
        if dx > 0:
            new_direction = 'E'
        elif dx < 0:
            new_direction = 'W'
        elif dy > 0:
            new_direction = 'N'
        elif dy < 0:
            new_direction = 'S'
        else:
            # Same point, no turn
            continue
        
        # Check if this is a turn (direction changed)
        if new_direction != current_direction:
            # Determine the type of turn
            if (current_direction == 'E' and new_direction == 'N') or \
               (current_direction == 'N' and new_direction == 'W') or \
               (current_direction == 'W' and new_direction == 'S') or \
               (current_direction == 'S' and new_direction == 'E'):
                left_turns += 1
            elif (current_direction == 'E' and new_direction == 'S') or \
                 (current_direction == 'S' and new_direction == 'W') or \
                 (current_direction == 'W' and new_direction == 'N') or \
                 (current_direction == 'N' and new_direction == 'E'):
                right_turns += 1
            elif (current_direction == 'E' and new_direction == 'W') or \
                 (current_direction == 'W' and new_direction == 'E') or \
                 (current_direction == 'N' and new_direction == 'S') or \
                 (current_direction == 'S' and new_direction == 'N'):
                u_turns += 1
        
        # Update current direction for next iteration
        current_direction = new_direction
    
    return left_turns, right_turns, u_turns

def solve_path_problem():
    """
    Main function to solve the path tracking problem
    """
    # Example input format
    coordinates = [(0, 0), (2, 0), (2, 3), (0, 3), (0, 0)]
    
    # For actual use, you would read input like this:
    # n = int(input().strip())
    # coordinates = []
    # for _ in range(n):
    #     x, y = map(int, input().split())
    #     coordinates.append((x, y))
    
    left, right, u_turn = count_turns(coordinates)
    
    print(f"左轉: {left}")
    print(f"右轉: {right}")
    print(f"迴轉: {u_turn}")

# Test with some examples
def test_examples():
    print("=== 測試範例 ===")
    
    # Example 1: Rectangle path (right, up, left, down)
    coords1 = [(0, 0), (2, 0), (2, 3), (0, 3), (0, 0)]
    left1, right1, u1 = count_turns(coords1)
    print(f"矩形路徑 {coords1}:")
    print(f"  左轉: {left1}, 右轉: {right1}, 迴轉: {u1}")
    
    # Example 2: Simple path with right turns
    coords2 = [(0, 0), (1, 0), (1, 1), (0, 1)]
    left2, right2, u2 = count_turns(coords2)
    print(f"簡單路徑 {coords2}:")
    print(f"  左轉: {left2}, 右轉: {right2}, 迴轉: {u2}")
    
    # Example 3: Path with U-turn
    coords3 = [(0, 0), (2, 0), (2, 1), (0, 1), (0, 0)]
    left3, right3, u3 = count_turns(coords3)
    print(f"包含迴轉的路徑 {coords3}:")
    print(f"  左轉: {left3}, 右轉: {right3}, 迴轉: {u3}")

if __name__ == "__main__":
    test_examples()
    print("\n=== 主要解答 ===")
    solve_path_problem()
````
