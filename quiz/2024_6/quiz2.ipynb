### o077. 2. 電子畫布 - 2024年6月APCS

#### 問題描述

有一個 H×W 的電子畫布，一開始數值都是 0 代表未填色。模擬 N 次畫筆操作。

每次畫筆操作為選一個座標 (r, c) 停留 t 秒，會將**曼哈頓距離 ≤ t** 的區塊染上顏色 x。若有多個顏色重複填到相同區塊，顏色的數值會累加起來。

請輸出 N 次操作後的畫布狀態。

#### 輸入說明

- 第一行輸入三個正整數 H, W, N (1 ≤ H, W ≤ 20, 1 ≤ N ≤ 100)
- 接下來有 N 行，每一行有四個整數 r, c, t, x (0 ≤ r < H, 0 ≤ c < W, 0 ≤ t ≤ 20, 1 ≤ x ≤ 10)

#### 評分標準

- (60 分): H = 1
- (40 分): 無限制

---

#### Problem Description 

There is an H×W electronic canvas initially filled with zeros (uncolored). Simulate N brush operations.

Each brush operation selects a coordinate (r, c) and stays for t seconds. It will color all blocks with **Manhattan distance ≤ t** with color x. If multiple colors are applied to the same block, the color values are accumulated.

Output the state of the canvas after N operations.

#### Input Format

- First line: three positive integers H, W, N (1 ≤ H, W ≤ 20, 1 ≤ N ≤ 100)
- Next N lines: four integers r, c, t, x (0 ≤ r < H, 0 ≤ c < W, 0 ≤ t ≤ 20, 1 ≤ x ≤ 10)

#### Scoring

- (60 points): H = 1
- (40 points): no restrictions

---

#### 解答
```python
def manhattan_distance(r1, c1, r2, c2):
    """
    計算曼哈頓距離
    Calculate Manhattan distance
    """
    return abs(r1 - r2) + abs(c1 - c2)

def main():
    # 讀取畫布尺寸和操作次數
    H, W, N = map(int, input().split())
    
    # 初始化畫布 (全為0)
    canvas = [[0] * W for _ in range(H)]
    
    # 處理每個畫筆操作
    for _ in range(N):
        r, c, t, x = map(int, input().split())
        
        # 遍歷畫布上的每個像素
        for i in range(H):
            for j in range(W):
                # 計算曼哈頓距離
                dist = manhattan_distance(r, c, i, j)
                
                # 如果距離小於等於t，則加上顏色x
                if dist <= t:
                    canvas[i][j] += x
    
    # 輸出畫布狀態
    for i in range(H):
        # 將每行轉換為字串輸出
        row_str = ' '.join(str(canvas[i][j]) for j in range(W))
        print(row_str)

if __name__ == "__main__":
    main()
```

## 算法說明

### 曼哈頓距離說明

曼哈頓距離計算公式：
```
distance = |r1 - r2| + |c1 - c2|
```

對於一個操作 (r, c, t, x)，受影響的區域是以 (r, c) 為中心，曼哈頓距離小於等於 t 的所有像素。

### 複雜度分析

- **時間複雜度**: O(N × H × W)
  - N 次操作，每次操作需要遍歷 H×W 個像素
  - 最大計算量: 100 × 20 × 20 = 40,000 次操作
- **空間複雜度**: O(H × W)
  - 需要存儲 H×W 的畫布狀態

### 優化建議

對於大規模數據，可以考慮優化（但本題規模很小，不需要）：
1. **區域計算**: 只遍歷受影響的區域，而不是整個畫布
2. **差分數組**: 使用**二維差分數組**來批量更新區域

但對於本題的約束條件 (H, W ≤ 20)，直接遍歷整個畫布已經足夠高效。

````python
# Ryan Note: 
# Below implementation is using an optimized(smaller) area checking,
# ie a rectangle area restricted by the required manhatten distance 't'

# H,W,N = map(int, input().split())
# ops = []
# for _ in range(N):
#     ops.append(list(map(int, input().split())))

# H,W,N = 1,20,3
# ops =[[0,13,5,7],[0,6,4,4],[0,13,12,6]]

H,W,N = 6,7,3
ops =[[3,2,2,1],[1,6,1,2],[1,3,2,5]]

grids = [[0]*W for _ in range(H)]
for op in ops:
    r,c,t,x = op[0], op[1], op[2], op[3]
    for i in range(max(0, r-t), min(H-1, r+t)+1):
        for j in range(max(0, c-t), min(W-1, c+t)+1):
            if abs(r-i)+abs(c-j)<=t:
                grids[i][j]+=x
for i in range(H):
    for j in range(W): 
        print(grids[i][j], end=" ")
    print()
    
````

````python
# Ryan note:
# Below implementation is using BFS traversal. 
# We need a data structure to record the visited grid points. 
# Here a simple 2d list is used, ie. with the same dimension as the original grids
# A tuple of (r,c,t) is stored in the queue, 

from collections import deque

H,W,N = map(int, input().split())
ops = []
for _ in range(N):
    ops.append(list(map(int, input().split())))

# H,W,N = 1,20,3
# ops =[[0,13,5,7],[0,6,4,4],[0,13,12,6]]

# H,W,N = 6,7,3
# ops =[[3,2,2,1],[1,6,1,2],[1,3,2,5]]

directions = ((0,1),(0,-1),(1,0),(-1,0))

def BFS(grids, r,c,t,x):
    visited = [[False]*W for _ in range(H)]
    queue = deque([(r,c,0)])
    visited[r][c] = True # mark the node as visited
    grids[r][c]+=x # update node
    while queue:
        current_r,current_c,current_t = queue.popleft()
        if current_t < t:
            for dr, dc in directions:
                new_r = current_r + dr
                new_c = current_c + dc
                if new_r>=0 and new_r<H and new_c>=0 and new_c<W:
                    if not visited[new_r][new_c]:
                        queue.append((new_r, new_c, current_t+1))
                        visited[new_r][new_c] = True # mark as visited
                        grids[new_r][new_c]+=x # update node
       
grids = [[0]*W for _ in range(H)]
for r, c, t, x in ops:
    BFS(grids, r, c, t, x)

for i in range(H):
    for j in range(W): 
        print(grids[i][j], end=" ")
    print()
     
````
