### Problem Description 

There is a town with n tall buildings with heights h₁, h₂, ..., hₙ. The mayor wants to hold a high-altitude stunt performance in the town center, where the performer will glide from a building to the ground on the right side.

For the safety of the performer, the gliding path must have decreasing building heights, and the performer must glide continuously to the right (cannot skip intermediate buildings). Find the longest possible gliding path.

### Input Format

- First line: positive integer n (5 < n < 100)
- Second line: n positive integers h₁, h₂, ..., hₙ (1 < hᵢ < 1000) representing building heights

### Scoring

- (60 points): n = 5
- (40 points): no restrictions

## 解決方案 1: O(n²) 暴力解法

```python
def longest_continuous_decreasing_path(heights):
    """
    O(n²) 暴力解法：對每個起始位置向右找出最長連續遞減路徑
    O(n²) brute force: for each starting position, find the longest continuous decreasing path to the right
    """
    n = len(heights)
    if n == 0:
        return 0
    
    max_length = 0
    
    # 對每個起始位置
    for i in range(n):
        current_length = 1
        current_height = heights[i]
        
        # 向右檢查連續遞減
        for j in range(i + 1, n):
            if heights[j] < current_height:
                current_length += 1
                current_height = heights[j]
            else:
                break  # 不再遞減，停止檢查
        
        max_length = max(max_length, current_length)
    
    return max_length

def main():
    n = int(input().strip())
    heights = list(map(int, input().split()))
    result = longest_continuous_decreasing_path(heights)
    print(result)

if __name__ == "__main__":
    main()
```

### 解決方案 2: O(n) 動態規劃解法

```python
def longest_continuous_decreasing_path(heights):
    """
    O(n) 動態規劃解法：dp[i] 表示以位置i結尾的最長連續遞減路徑長度
    O(n) DP solution: dp[i] represents the length of longest continuous decreasing path ending at position i
    """
    n = len(heights)
    if n == 0:
        return 0
    
    # dp[i] 表示以位置i結尾的最長連續遞減路徑長度
    dp = [1] * n
    max_length = 1
    
    # 從左向右計算
    for i in range(1, n):
        if heights[i] < heights[i-1]:
            dp[i] = dp[i-1] + 1
        else:
            dp[i] = 1
        max_length = max(max_length, dp[i])
    
    return max_length

def main():
    n = int(input().strip())
    heights = list(map(int, input().split()))
    result = longest_continuous_decreasing_path(heights)
    print(result)

if __name__ == "__main__":
    main()
```

### 解決方案 3: O(n) 空間優化解法

```python
def longest_continuous_decreasing_path(heights):
    """
    O(n) 空間優化解法：單次遍歷找出最長連續遞減段
    O(n) space-optimized solution: find longest continuous decreasing segment in one pass
    """
    n = len(heights)
    if n == 0:
        return 0
    
    max_length = 1
    current_length = 1
    
    # 遍歷數組，檢查連續遞減關係
    for i in range(1, n):
        if heights[i] < heights[i-1]:
            current_length += 1
            max_length = max(max_length, current_length)
        else:
            current_length = 1  # 重置當前序列長度
    
    return max_length

def main():
    n = int(input().strip())
    heights = list(map(int, input().split()))
    result = longest_continuous_decreasing_path(heights)
    print(result)

if __name__ == "__main__":
    main()
```

## 算法比較

| 解決方案 | 時間複雜度 | 空間複雜度 | 優點 | 缺點 |
|----------|------------|------------|------|------|
| 方案1: 暴力解法 | O(n²) | O(1) | 直觀易懂 | 效率較低 |
| 方案2: 動態規劃 | O(n) | O(n) | 時間高效 | 需要額外空間 |
| 方案3: 空間優化 | O(n) | O(1) | 時間空間都高效 | 需要理解狀態轉移 |

## 推薦方案

**方案3: O(n) 空間優化解法** 是最佳選擇，因為：
1. **時間效率**: O(n) 線性時間複雜度
2. **空間效率**: O(1) 常數空間複雜度  
3. **代碼簡潔**: 邏輯清晰易懂
4. **性能優越**: 適合各種規模的輸入數據

所有三種方案都能正確解決問題，但方案3在時間和空間效率上都達到最優。

````python
# n = int(input())
# hs = list(map(int, input().split()))

# n = 5
# hs = [6, 2, 5, 3, 1]

n = 10
hs = [31, 41, 97, 93, 23, 89, 59, 26, 15, 58]

longest_path = 0
path = 1
for i in range(n-1):
    if hs[i]>hs[i+1]:
        path+=1
    else:
        longest_path = max(longest_path, path)
        path=1
longest_path = max(longest_path, path)
print(longest_path)
````

````python
def longest_decreasing_path(heights):
    """
    使用動態規劃找出最長的連續遞減子數組
    Find the longest contiguous decreasing subarray using dynamic programming
    """
    n = len(heights)
    if n == 0:
        return 0
    
    # dp[i] 表示以位置i結尾的最長連續遞減子數組長度
    dp = [1] * n 
    max_length = 1
    
    for i in range(1, n):
        if heights[i] < heights[i-1]:
            dp[i] = dp[i-1] + 1
        max_length = max(max_length, dp[i])
    
    return max_length


# n = 5
# hs = [6, 2, 5, 3, 1]

n = 10
hs = [31, 41, 97, 93, 23, 89, 59, 26, 15, 58]

print(longest_decreasing_path(hs))
````
