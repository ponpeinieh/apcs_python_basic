我們可以使用圖形遍歷算法（如 BFS 或 DFS）來訪問在 `N × M` 網格中與點 `P(r, c)` 曼哈頓距離 within `t` 的所有點。曼哈頓距離的定義為：兩點 `(r₁, c₁)` 和 `(r₂, c₂)` 之間的距離 = `|r₁ - r₂| + |c₁ - c₂|`。

### 方法：
1. **圖形表示**：網格自然是一個圖形，其中每個單元格都連接到它的四個相鄰單元格（如果它們存在）。將網格建模為圖形，每個單元格為節點，相鄰單元格（上下左右）之間存在邊。

2. **遍歷選擇**：您可以使用 BFS（廣度優先搜索）或 DFS（深度優先搜索）來遍歷圖形。然而，**BFS 在這裡更自然，因為它先探索所有距離為 `d` 的點，然後再移動到 `d+1`，這與曼哈頓距離的約束完全一致**。BFS 按層級擴散的特性完美匹配曼哈頓距離的探索需求。

3. **終止條件**：當從 `P` 的曼哈頓距離超過 `t` 時停止搜索。由於您正在遍歷相鄰點，從 `P` 到任意點的移動步數正好等於曼哈頓距離（因為每步在一個方向上移動一個單位）。因此，當當前層級（距離）大於 `t` 時即可停止。

### BFS 實現步驟：
- **初始化**：從起點 `P(r, c)` 開始，距離為 0
- **佇列(Queue)操作**：使用佇列存儲待訪問點，維護已訪問標記避免重複
- **層級遍歷**：處理每個點時，若距離 ≤ `t` 則記錄該點，並將未訪問的相鄰點加入佇列（距離+1）
- **終止時機**：當佇列為空或當前距離超過 `t` 時結束

### 程式碼實例：
```python
from collections import deque

def points_within_manhattan_bfs(grid, r, c, t):
    n, m = len(grid), len(grid[0])
    visited = [[False] * m for _ in range(n)]
    directions = [(1,0), (-1,0), (0,1), (0,-1)]
    queue = deque([(r, c, 0)]) # not only store the r and c, but also manhatten distance 
    visited[r][c] = True # mark as visited when adding into the queue
    result = [] # to store all points within manhatten distance t
    
    while queue:
        curr_r, curr_c, dist = queue.popleft()
        result.append((curr_r, curr_c))
        
        if dist >= t:  # 已達最大距離，停止擴展
            continue
            
        for dr, dc in directions:
            nr, nc = curr_r + dr, curr_c + dc
            if 0 <= nr < n and 0 <= nc < m and not visited[nr][nc]: # no need to check if within t
                visited[nr][nc] = True
                queue.append((nr, nc, dist + 1))
    
    return result
```

### 替代方案：
對於較大的 `t` 值，可採用範圍迭代法，直接計算邊界並檢查曼哈頓距離：
```python
def points_within_manhattan_direct(grid, r, c, t):
    n, m = len(grid), len(grid[0])
    result = []
    min_row, max_row = max(0, r - t), min(n-1, r + t)
    min_col, max_col = max(0, c - t), min(m-1, c + t)
    
    for i in range(min_row, max_row + 1):
        for j in range(min_col, max_col + 1):
            if abs(i - r) + abs(j - c) <= t:
                result.append((i, j))
    return result
```

### 選擇建議：
- **小範圍搜索** (`t` ≤ 10)：使用 BFS，時間複雜度 O(t²)
- **大範圍搜索** (`t` 較大)：使用範圍迭代法，避免 BFS 的額外開銷
- **通用場景**：範圍迭代法更簡單高效
