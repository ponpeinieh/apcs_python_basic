### 3. 給定右側函式 `f()`，已知 `f(14)`、`f(10)`、`f(6)` 分別回傳 `25`、`18`、`10`，函式中的 `(?)` 應為下列何者？

```python
def f(n):
    if n < 2:
        return n
    else:
        return n + f( ? )
```

#### 選項

(A) `(n+1)//2`  
(B) `n//2`  
(C) `(n-1)//2`  
(D) `(n//2)+1`  


### 逐步分析

我們已知：
- `f(6) = 10`
- `f(10) = 18`
- `f(14) = 25`

我們需要找出遞迴呼叫的參數 `f(?)` 是什麼。

#### 步驟 1: 理解函式結構
這是一個遞迴函式：
- 基準情況：如果 `n < 2`，直接回傳 `n`。
- 遞迴情況：回傳 `n + f(某個比 n 小的值)`。

所以 `f(n) = n + f(g(n))`，其中 `g(n)` 就是題目中的 `?`。


#### 步驟 2: 正確的推導方法
 

**測試選項 (B) `n//2`:**
- `f(6) = 6 + f(3) = 10` ⇒ `f(3) = 4`
- `f(3) = 3 + f(1) = 3 + 1 = 4` ✓
- `f(10) = 10 + f(5) = 18` ⇒ `f(5) = 8`
- `f(5) = 5 + f(2) = 5 + f(2)`
- `f(2) = 2 + f(1) = 2 + 1 = 3`
- 所以 `f(5) = 5 + 3 = 8` ✓
- `f(14) = 14 + f(7) = 25` ⇒ `f(7) = 11`
- `f(7) = 7 + f(3) = 7 + 4 = 11` ✓

完美！選項 (B) 符合所有條件。

### 結論

唯一能讓 `f(14) = 25`、`f(10) = 18`、`f(6) = 10` 都成立的條件是 `n//2`。

**最終答案：(B) `n//2`**
