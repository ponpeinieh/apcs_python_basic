### 2. 給定右側程式，若已知輸出的結果為 `[1][2][3][5][4][6]`，程式中的 `(?)` 應為下列何者？

```python
for i in range(5):
    j = 0
    while (?):
        print(f'[{i + j}]', end='')
        j = j + 2
```

#### 選項
(A) `j < i`  
(B) `j > i`  
(C) `j <= i`  
(D) `j >= i`  


### 逐步分析

我們的目標是找出 `while` 迴圈的條件 `(?)`，以產生確切的輸出序列：`[1][2][3][5][4][6]`。

#### 步驟 1: 理解程式結構
-   外層的 `for` 迴圈會讓 `i` 從 `0` 到 `4` 依序執行（因為 `range(5)` 產生 `0, 1, 2, 3, 4`）。
-   對於每個 `i`，內層的 `while` 迴圈會開始執行，變數 `j` 從 `0` 開始，每次增加 `2` (`j = j + 2`)。
-   `print` 陳述式會輸出 `i + j` 的值，並用括號 `[]` 包起來。

#### 步驟 2: 推導輸出序列的來源
最終輸出是一個字串：`[1][2][3][5][4][6]`。這表示 `print` 陳述式總共執行了 6 次，產生了數字 `1, 2, 3, 5, 4, 6`（請注意這個順序）。

我們需要將這些輸出值，對應回產生它們的 `(i, j)` 組合（`輸出的值 = i + j`）。

讓我們為每個可能的 `i` 追蹤迴圈執行情況：

*   **`i = 0`**：`j` 從 `0` 開始。
    *   如果迴圈執行，第一個印出的值會是 `0 + 0 = 0`。
    *   但序列的第一個輸出是 `[1]`，不是 `[0]`。
    *   **結論：** 當 `i=0` 時，`while` 迴圈**不能執行**。這是一個關鍵線索。

*   **`i = 1`**：`j` 從 `0` 開始。
    *   如果迴圈執行，第一個印出的值是 `1 + 0 = 1`。這符合第一個輸出 `[1]`。
    *   `j` 變成 `2`。再次檢查條件。
    *   如果迴圈執行第二次，下一個印出的值會是 `1 + 2 = 3`。但下一個輸出是 `[2]`，不是 `[3]`。
    *   **結論：** 當 `i=1` 時，`while` 迴圈必須**剛好執行一次**。

*   **`i = 2`**：`j` 從 `0` 開始。
    *   第一次印出：`2 + 0 = 2`。這符合第二個輸出 `[2]`。
    *   `j` 變成 `2`。再次檢查條件。
    *   如果迴圈執行第二次，下一個印出的值會是 `2 + 2 = 4`。但下一個輸出是 `[3]`，不是 `[4]`。
    *   **結論：** 當 `i=2` 時，`while` 迴圈必須**剛好執行一次**。

*   **`i = 3`**：`j` 從 `0` 開始。
    *   第一次印出：`3 + 0 = 3`。這符合第三個輸出 `[3]`。
    *   `j` 變成 `2`。再次檢查條件。
    *   如果迴圈執行第二次，下一個印出的值會是 `3 + 2 = 5`。這符合第四個輸出 `[5]`。
    *   `j` 變成 `4`。再次檢查條件。
    *   如果迴圈執行第三次，下一個印出的值會是 `3 + 4 = 7`。但下一個輸出是 `[4]`，不是 `[7]`。
    *   **結論：** 當 `i=3` 時，`while` 迴圈必須**剛好執行兩次**。

*   **`i = 4`**：`j` 從 `0` 開始。
    *   第一次印出：`4 + 0 = 4`。這符合第五個輸出 `[4]`。
    *   `j` 變成 `2`。再次檢查條件。
    *   如果迴圈執行第二次，下一個印出的值會是 `4 + 2 = 6`。這符合第六個輸出 `[6]`。
    *   `j` 變成 `4`。再次檢查條件。
    *   最終輸出到此結束，所以迴圈必須停止。
    *   **結論：** 當 `i=4` 時，`while` 迴圈必須**剛好執行兩次**。

#### 步驟 3: 總結迴圈行為
讓我們列出每個 `i` 值所需執行的 `while` 迴圈次數：

| `i` | 使用到的 `j` 值 | 迴圈執行次數 | 印出的數值 |
| :-- | :-------------- | :----------- | :--------- |
| 0   | (無)            | 0            | (無)       |
| 1   | 0               | 1            | 1          |
| 2   | 0               | 1            | 2          |
| 3   | 0, 2            | 2            | 3, 5       |
| 4   | 0, 2            | 2            | 4, 6       |

#### 步驟 4: 找出正確的條件
條件 `(?)` 必須在我們希望迴圈停止時為 `False`。迴圈會在 `(?)` 為 `True` 時繼續。

讓我們用上面的表格來測試每個選項。迴圈條件會在每次迭代開始時檢查，`j` 的值會是 `0, 2, 4, ...`。

*   **(A) `j < i`**
    *   `i=0`：`j=0`。`0 < 0` 是 `False`。迴圈不執行。✅
    *   `i=1`：`j=0`。`0 < 1` 是 `True`。迴圈執行。`j=2`。`2 < 1` 是 `False`。迴圈停止。（執行 1 次）✅
    *   `i=2`：`j=0`。`0 < 2` 是 `True`。迴圈執行。`j=2`。`2 < 2` 是 `False`。迴圈停止。（執行 1 次）✅
    *   `i=3`：`j=0`。`0 < 3` 是 `True`。迴圈執行。`j=2`。`2 < 3` 是 `True`。迴圈再次執行。`j=4`。`4 < 3` 是 `False`。迴圈停止。（執行 2 次）✅
    *   `i=4`：`j=0`。`0 < 4` 是 `True`。迴圈執行。`j=2`。`2 < 4` 是 `True`。迴圈再次執行。`j=4`。`4 < 4` 是 `False`。迴圈停止。（執行 2 次）✅

這完全符合我們要求的行為。

*   **(B) `j > i`**
    *   `i=0`：`j=0`。`0 > 0` 是 `False`。迴圈不執行。✅
    *   `i=1`：`j=0`。`0 > 1` 是 `False`。迴圈不執行。❌ (我們需要執行 1 次)
    到此已可判定錯誤。

*   **(C) `j <= i`**
    *   `i=0`：`j=0`。`0 <= 0` 是 `True`。迴圈執行，印出 `0`。❌ (輸出應從 `1` 開始)

*   **(D) `j >= i`**
    *   `i=1`：`j=0`。`0 >= 1` 是 `False`。迴圈不執行。❌ (我們需要執行 1 次)

### 結論

唯一能產生確切輸出 `[1][2][3][5][4][6]` 的條件是 `j < i`。

**最終答案：(A) `j < i`**
