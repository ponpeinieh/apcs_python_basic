### 1. 給定右側函式 f( )，當執行 `f(10)` 時，最終回傳結果為何？

```python
def f(i):
    if i > 0:
        if (i // 2) % 2 == 0:
            return f(i - 2) * i
        return f(i - 2) * (-i)
    else:
        return 1
```

#### 選項
(A) 1  
(B) 3840  
(C) -3840  
(D) 執行時導致無窮迴圈，不會停止執行  


#### 函式運作流程

1. **基礎情況**：當 `i <= 0` 時，函式返回 `1`。
2. **遞迴情況**：
   - 如果 `i > 0` 且 `(i // 2) % 2 == 0` 為真，則執行 `return f(i - 2) * i`。
   - 否則，執行 `return f(i - 2) * (-i)`。

這意味著函式的行為取決於 `i` 的值以及 `(i // 2) % 2` 的結果。

#### 計算過程

讓我們逐步計算 `f(10)`：

- `f(10)`:
  - `10 > 0`，且 `(10 // 2) % 2 == 0`（即 `5 % 2 == 0`，為假），因此執行 `return f(8) * (-10)`。

- `f(8)`:
  - `8 > 0`，且 `(8 // 2) % 2 == 0`（即 `4 % 2 == 0`，為真），因此執行 `return f(6) * 8`。

- `f(6)`:
  - `6 > 0`，且 `(6 // 2) % 2 == 0`（即 `3 % 2 == 0`，為假），因此執行 `return f(4) * (-6)`。

- `f(4)`:
  - `4 > 0`，且 `(4 // 2) % 2 == 0`（即 `2 % 2 == 0`，為真），因此執行 `return f(2) * 4`。

- `f(2)`:
  - `2 > 0`，且 `(2 // 2) % 2 == 0`（即 `1 % 2 == 0`，為假），因此執行 `return f(0) * (-2)`。

- `f(0)`:
  - `0 <= 0`，因此返回 `1`。

#### 回溯計算

現在我們來回溯計算每個函式調用的結果：
- `f(2) = 1 * (-2) = -2`
- `f(4) = -2 * 4 = -8`
- `f(6) = -8 * (-6) = 48`
- `f(8) = 48 * 8 = 384`
- `f(10) = 384 * (-10) = -3840`

#### 最終結果

因此，當執行 `f(10)` 時，最終回傳的結果為 `-3840`。

#### 結論

最終回傳結果為 **(C) -3840**。
