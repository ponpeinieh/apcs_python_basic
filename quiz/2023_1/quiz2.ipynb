### j606. 2. 造字程式

#### 問題描述 

有一個長度 `K` 的初始字串 `S`，每個字元都是小寫的英文字母。

接下來有 `Q` 次的修改，每次的修改會把舊的字串重新排列成一個新的字串。
更具體來講，每次修改時會給一個 `1` 到 `K` 的排列 `P = [P₁, P₂, ..., Pₖ]`，要將舊字串的第 `i` 的字元複製到新字串的第 `Pᵢ` 個字元。 

**例子：**
- 若舊字串是 “abac”，且 `P = [4, 1, 3, 2]`，可以得到新字串 “bcaa”。
- 若舊字串是 “bcaa”，且 `P = [1, 2, 3, 4]`，可以得到新字串 “bcaa”。
- 若舊字串是 “bcaa”，且 `P = [2, 3, 4, 1]`，可以得到新字串 “abca”。

在 `Q` 的修改中，每次修改出來的新字串會被當成下一次修改中的舊字串，而第一次修改時使用的舊字串就是初始字串。

題目另外會給一個數字 `R`，請依照下面定義的順序輸出 `R` 行，每行 `Q` 個字元：
- 輸出操作 `1` 到 `Q` 的新字串的第 `1` 個字元。
- 輸出操作 `1` 到 `Q` 的新字串的第 `2` 個字元。
- ...
- 輸出操作 `1` 到 `Q` 的新字串的第 `R` 個字元。

#### 輸入說明 
- 第一行有三個整數 `K`，`Q`，`R`。
- 第二行是長度 `K` 的初始字串。
- 接下來有 `Q` 行，每行是一個 `1` 到 `K` 的排列。

(60 分)：`R=1`，`1 ≤ K, Q ≤ 20`  
(40 分)：`1 ≤ R ≤ K < 20`，`1 ≤ Q ≤ 20`

#### 輸出說明 
- 請依照題目敘述輸出 `R x Q` 個字元。

### 測試範例 #2 驗證 

**輸入：**
```
4 3 4
abac
4 1 3 2
1 2 3 4
2 3 4 1
```

**輸出：**
```
bba
ccb
aac
aaa
```

---

#### 解答 

```python
def rearrange_string(s, permutations):
    for perm in permutations:
        new_string = [''] * len(s)
        for i in range(len(s)):
            new_string[perm[i] - 1] = s[i]
        s = ''.join(new_string)
    return s

def main():
    K, Q, R = map(int, input().split())
    s = input().strip()
    permutations = [list(map(int, input().split())) for _ in range(Q)]
    
    results = []
    for i in range(R):
        new_string = rearrange_string(s, permutations)
        results.append([new_string[j] for j in range(Q)])
        s = new_string
    
    for row in results:
        print(''.join(row))

if __name__ == "__main__":
    main()
```

### ### j606. 2. Character Rearrangement Program

#### Problem Description 

You are given an initial string `S` of length `K`, where each character is a lowercase English letter. 

There are `Q` modifications to be made, where each modification rearranges the old string into a new string. Specifically, each modification provides a permutation `P = [P₁, P₂, ..., Pₖ]` (with values ranging from 1 to K), which indicates that the `i`-th character of the old string should be copied to the  `Pᵢ`-th position of the new string. 

**Examples:**
- If the old string is “abac” and `P = [4, 1, 3, 2]`, the new string will be “bcaa”.
- If the old string is “bcaa” and `P = [1, 2, 3, 4]`, the new string will be “bcaa”.
- If the old string is “bcaa” and `P = [2, 3, 4, 1]`, the new string will be “abca”.

For each of the `Q` modifications, the new string generated will be used as the old string for the next modification, with the first modification using the initial string. 

Additionally, a number `R` is given, and you are to output `R` lines of characters, where each line contains `Q` characters:
- Output the first character of the new strings from operations 1 to Q.
- Output the second character of the new strings from operations 1 to Q.
- ...
- Output the `R`-th character of the new strings from operations 1 to Q.

#### Input Description 
- The first line contains three integers `K`, `Q`, `R`.
- The second line contains the initial string of length `K`.
- The next `Q` lines each contain a permutation of `1` to `K`.

(60 points): `R=1`, `1 ≤ K, Q ≤ 20`  
(40 points): `1 ≤ R ≤ K < 20`, `1 ≤ Q ≤ 20`

#### Output Description 
- Output `R x Q` characters according to the problem description.
