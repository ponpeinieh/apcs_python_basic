### j605. 1. 程式考試

#### 問題描述

給定 `n` 個提交紀錄，第 `i` 個提交紀錄有兩個整數 `t_i` 和 `s_i` 代表上傳時間和該次上傳的分數。若第 `i` 次的提交結果為嚴重錯誤，則 `s_i` 為 `-1`。

計算總分的公式為：
$$\text{總分} = \text{提交紀錄中的最高分} - \text{總提交次數} - \text{總嚴重錯誤次數} \times 2$$
若計算出來的分數為負數，則計為 `0`。

請輸出總分和第一次獲得最高分的時間點。

#### 輸入說明
- 第一行有一個正整數 `K` (`1 ≤ K ≤ 6`) 代表提交的次數，接下來有 `K` 行，每一行有兩個整數 `t_i` (`1 ≤ t_i ≤ 100`) 和 `s_i` (`-1 ≤ s_i ≤ 100`)。保證提交紀錄按照時間點嚴格遞增排序，且第一個提交紀錄不會是嚴重錯誤。

- (60 分)： `K=3`
- (40 分)：無其他限制

#### 輸出說明
- 輸出兩個整數，代表總分和第一次獲得最高分的時間點。

### 測試範例 #2 驗證

#### 解答

```python
def calculate_total_score(submissions):
    highest_score = 0
    severe_error_count = 0
    total_submissions = len(submissions)
    first_highest_time = 0

    for t, s in submissions:
        if s >= 0:
            if s > highest_score:
                highest_score = s
                first_highest_time = t
        else:
            severe_error_count += 1

    # 計算總分
    total_score = highest_score - total_submissions - (severe_error_count * 2)
    total_score = max(total_score, 0)  # 確保分數不為負數

    return total_score, first_highest_time


def main():
    K = int(input())
    submissions = [tuple(map(int, input().split())) for _ in range(K)]
    total_score, first_highest_time = calculate_total_score(submissions)
    print(total_score, first_highest_time)

if __name__ == "__main__":
    main()
```
 

### j605. 1. Program Exam

#### Problem Description

Given `n` submission records, the `i`-th submission record has two integers `t_i` and `s_i` representing the upload time and the score of that submission. If the `i`-th submission results in a severe error, then `s_i` will be `-1`. 

The formula for calculating the total score is:
$$\text{total score} = \text{highest score in submission records} - \text{total number of submissions} - \text{total severe error count} \times 2$$

If the calculated score is negative, it counts as `0`.

Please output the total score and the time point when the highest score was first achieved.

#### Input Description
- The first line contains a positive integer `K` (`1 ≤ K ≤ 6`) representing the number of submissions. The next `K` lines each contain two integers `t_i` (`1 ≤ t_i ≤ 100`) and `s_i` (`-1 ≤ s_i ≤ 100`). It is guaranteed that the submission records are strictly sorted by time, and the first submission record will not be a severe error.

- (60 points): `K=3`
- (40 points): No other restrictions

#### Output Description
- Output two integers representing the total score and the time point when the highest score was first achieved.

def calculate_total_score(submissions):
    highest_score = 0
    severe_error_count = 0
    total_submissions = len(submissions)
    first_highest_time = 0

    for t, s in submissions:
        if s >= 0:
            if s > highest_score:
                highest_score = s
                first_highest_time = t
        else:
            severe_error_count += 1

    # Calculate total score
    total_score = highest_score - total_submissions - (severe_error_count * 2)
    total_score = max(total_score, 0)  # Ensure score is not negative

    return total_score, first_highest_time


def main():

````python
# k = 5
# data =[(3,89),(5,-1),(10,90),(15,0),(20,90)]
# k = 3
# data=[(3,0),(5,-1),(6,-1)]
k = int(input())
data= [tuple(map(int, input().split())) for _ in range(k)]
import heapq
def calc(k, data):
    arr = []
    serious_error_count=0
    for d in data:
        if d[1] == -1:
            serious_error_count+=1
        heapq.heappush(arr, (-1*d[1], d[0]))
    max_record = heapq.heappop(arr)
    max_score, time = max_record[0]*-1, max_record[1]  
    total_score =  max_score - k - serious_error_count*2
    if total_score<0:
        total_score = 0
    return total_score, time
result = calc(k,data)
print(result[0],result[1])
````
