### m370. 1. 機械鼠

#### 問題描述

有 n 個位置上有食物，一隻老鼠一開始位於位置 x。老鼠在開始覓食前要選擇今天要往左邊或往右移動去尋找食物，經過食物時可以停下來吃食物，吃完後可以選擇繼續往相同方向移動，或者是結束今天的覓食。

請問老鼠最多能吃到多少個食物，以及最後停下來吃食物的位置。

#### 輸入說明
- 第一行：兩個整數 x 和 n，以空格分隔
- 第二行：n 個整數，以空格分隔，表示每個食物的位置

#### 輸出說明
- 兩個整數：最多能吃到的食物數目和最後一個吃的食物停下的位置

#### 解題思路

1. **問題分析**：
   - 老鼠可以選擇向左或向右移動
   - 一旦選擇方向，只能沿該方向移動，但可以在任何食物位置停止
   - 需要找到能吃到最多食物的路徑

2. **關鍵觀察**：
   - 將食物位置排序
   - 對於每個可能的方向（左或右），計算能吃到多少食物
   - 老鼠可以選擇在任何點停止，所以需要考慮所有可能的停止點

3. **算法選擇**：
   - 將食物位置排序
   - 分別計算向左和向右移動能獲得的最大食物數
   - 比較兩個方向的结果，選擇較優解

#### 解答

```python
def main():
    x, n = map(int, input().split())
    positions = list(map(int, input().split()))
    
    # 排序食物位置
    positions.sort()
    
    # 使用二分查找找到初始位置應該插入的位置
    left, right = 0, n - 1
    
    while left <= right:
        mid = (left + right) // 2
        if positions[mid] < x:
            left = mid + 1
        else:
            right = mid - 1
    
    # left 就是初始位置應該插入的索引
    left_count = left  # 左邊的食物數量
    right_count = n - left  # 右邊的食物數量
    
    if left_count > right_count:
        # 向左移動可以吃到更多食物
        max_food = left_count
        final_position = positions[0]  # 最左邊的食物
    else:
        # 向右移動可以吃到更多食物
        max_food = right_count
        final_position = positions[-1]  # 最右邊的食物
    
    print(max_food, final_position)

if __name__ == "__main__":
    main()
```

import bisect

def main():
    x, n = map(int, input().split())
    positions = sorted(map(int, input().split()))
    
    # 使用 bisect_left 找到插入位置
    idx = bisect.bisect_left(positions, x)
    left_count = idx  # 左邊的食物數量
    right_count = n - idx  # 右邊的食物數量
    
    if left_count > right_count:
        print(left_count, positions[0])
    else:
        print(right_count, positions[-1])

    main()

## 算法說明

### 關鍵洞察
1. **排序後二分查找**: 通過排序和二分查找，可以快速確定初始位置左右兩邊的食物數量
2. **簡單比較**: 只需要比較左邊和右邊的食物數量，選擇數量多的一邊
3. **最終位置**: 選擇某個方向後，最終位置就是該方向最遠端的食物位置

### 時間複雜度
- **排序**: O(n log n)
- **二分查找**: O(log n)
- **總複雜度**: O(n log n)

### 空間複雜度
- O(n) - 存儲食物位置

這種方法比之前的解法簡單得多，而且效率更高！

````python
x,n = map(int, input().split())
positions = list(map(int, input().split()))

# x,n = 10, 3
# positions = [1, 5, 13]

# x,n = 10, 9
# positions = [-1, 13, 12, 16, 100, -9, 7, 8, 25]
# first sort the positions
positions.sort()
# do the binary search for the 'starting' position
from bisect import bisect_left
pos = bisect_left(positions, x)
if pos > n-pos:
    print(pos, positions[0])
else:
    print(n-pos, positions[-1])
````

### m370. 1. Mechanical Mouse

#### Problem Description

There are **n** positions with food, and a mouse starts at position **x**. Before foraging for food, the mouse must choose to move left or right today. When passing food, it can stop to eat, and after eating, it can choose to continue moving in the same direction or end today's foraging.

What is the maximum amount of food the mouse can eat, and what position does it stop at after eating?

#### Input Format
- First line: Two integers **x** and **n**, separated by a space.
- Second line: **n** integers, separated by spaces, indicating the positions of the food.

#### Output Format
- Two integers: the maximum number of food items that can be eaten and the last position where food was eaten.

#### Solution Approach

1. **Problem Analysis**:
   - The mouse can choose to move left or right.
   - Once a direction is chosen, it can only move in that direction but can stop at any food position.
   - We need to find the path that allows the mouse to eat the most food.

2. **Key Observations**:
   - Sort the food positions.
   - For each possible direction (left or right), calculate how much food can be eaten.
   - The mouse can stop at any point, so all possible stopping points need to be considered.

3. **Algorithm Choice**:
   - Sort the food positions.
   - Calculate the maximum number of food items that can be obtained by moving left and right respectively.
   - Compare the results from both directions and choose the better option.

#### Solution

```python
def main():
    x, n = map(int, input().split())
    positions = list(map(int, input().split()))
    
    # Sort food positions
    positions.sort()
    
    # Use binary search to find the initial position to insert
    left, right = 0, n - 1
    
    while left <= right:
        mid = (left + right) // 2
        if positions[mid] < x:
            left = mid + 1
        else:
            right = mid - 1
    
    # left is the index where the initial position should be inserted
    left_count = left  # Number of food items to the left
    right_count = n - left  # Number of food items to the right
    
    if left_count > right_count:
        # More food can be eaten by moving left
        max_food = left_count
        final_position = positions[0]  # Leftmost food
    else:
        # More food can be eaten by moving right
        max_food = right_count
        final_position = positions[-1]  # Rightmost food
    
    print(max_food, final_position)

if __name__ == "__main__":
    main()
```

3. **Final Position**: After choosing a direction, the final position is simply the furthest food item in that direction.
