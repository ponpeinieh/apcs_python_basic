### m371. 2. 卡牌遊戲

#### 問題描述

你有一個 n×m 大小的表格，你可以從中消除具有相同數值且之間沒有障礙物的兩個元素，並獲得分數。請問你可以獲得的最大得分是多少？每一種數字在表格中出現恰好兩次。消除兩個相同的数字時，可以獲得該数字的分數。

消除規則：你可以垂直或水平地將兩個相同數值的元素消除，但消除的兩個元素之間不能有其他尚未消除的元素。

#### 輸入說明
- 第一行：兩個整數 n 和 m，以空格分隔，分別代表表格的行數和列數
- 接下來 n 行：每行包含 m 個整數，以空格分隔，表示表格中的元素
- 每個元素的數值範圍介於 [0,1000] 之內，且每種數字在表格中出現恰好兩次。
- 輸入保證表格上的每種數字恰好出現兩次，且表格的格數為偶數。

#### 輸出說明
- 一個整數：代表你可以獲得的最大得分 

#### 解答

```python
def checkrow(grids, n):
    """
    檢查單一行中可能的消除並返回該行的分數。
    
    Args:
        grids: 代表遊戲板的二維列表
        n: 要檢查的行索引
    
    Returns:
        score: 從該行消除中獲得的分數
    """
    length = len(grids[n])  # 行中的列數
    score = 0
    p1 = 0  # 第一個指針
    
    while True:
        # 跳過已消除的單元格（標記為 -1）
        while grids[n][p1] == -1:
            p1 += 1
            if p1 >= length-1:  # 已到達行尾
                return score
        
        p2 = p1 + 1  # 第二個指針從 p1 後開始
        if p2 == length:  # 沒有更多單元格可比較
            return score
        
        # 為 p2 跳過已消除的單元格
        while grids[n][p2] == -1:
            p2 += 1
            if p2 == length:  # 已到達行尾
                return score
        
        # 檢查兩個單元格是否有相同的值
        if grids[n][p1] == grids[n][p2]:
            # 消除配對並添加到分數
            score += grids[n][p1]  # 添加值（兩個單元格有相同的值）
            grids[n][p1] = grids[n][p2] = -1  # 標記為已消除
            p1 = p2 + 1  # 將 p1 移動到配對後的下一個位置
            
            if p1 == length:  # 已到達行尾
                return score
        else:
            # 值不匹配，將 p1 移動到 p2 的位置
            p1 = p2
```

```python
def checkrow(row_data):
    """
    Check a single row for possible eliminations and return the score.
    
    Args:
        row_data: 1D list representing a row of the game board
    
    Returns:
        score: points earned from eliminations in this row
    """
    length = len(row_data)  # Get the length of the row
    score = 0  # Initialize score counter
    p1 = 0  # First pointer starts at position 0
    
    while True:
        # Skip over already eliminated cells (marked as -1)
        while row_data[p1] == -1:
            p1 += 1  # Move to next position
            # If we reach the end or second-to-last position, return current score
            if p1 >= length - 1:
                return score
        
        # Set second pointer to position right after first pointer
        p2 = p1 + 1
        # If second pointer is beyond row length, return current score
        if p2 == length:
            return score
        
        # Skip over eliminated cells for the second pointer
        while row_data[p2] == -1:
            p2 += 1  # Move to next position
            # If second pointer reaches end of row, return current score
            if p2 == length:
                return score
        
        # Check if the two cells have matching values
        if row_data[p1] == row_data[p2]:
            # Eliminate the matching pair
            score += row_data[p1]  # Add the value to score (both are same)
            row_data[p1] = -1  # Mark first cell as eliminated
            row_data[p2] = -1  # Mark second cell as eliminated
            
            # Move first pointer to position after the eliminated pair
            p1 = p2 + 1
            # If we've reached the end of the row, return final score
            if p1 == length:
                return score
        else:
            # Values don't match, move first pointer to second pointer's position
            p1 = p2
```

### m371. 2. Card Game

#### Problem Description

You have an n×m grid from which you can eliminate two elements with the same value that have no obstacles between them, and earn points. What is the maximum score you can obtain? Each number appears exactly twice in the grid. When eliminating two identical numbers, you earn points equal to that number.

Elimination Rule: You can eliminate two elements with the same value vertically or horizontally, but there must be no other non-eliminated elements between the two elements being eliminated.

#### Input Specification
- First line: Two integers n and m, separated by a space, representing the number of rows and columns of the grid
- Next n lines: Each line contains m integers, separated by spaces, representing the elements in the grid
- The value range of each element is between [0, 1000], and each number appears exactly twice in the grid.
- The input guarantees that each number appears exactly twice on the grid, and the total number of grid cells is even.

#### Output Specification
- One integer: Represents the maximum score you can obtain

#### Solution

```python
def checkrow(grids, n):
    """
    Check a single row for possible eliminations and return the score from that row.
    
    Args:
        grids: 2D list representing the game board
        n: row index to check
    
    Returns:
        score: points earned from eliminations in this row
    """
    length = len(grids[n])  # Number of columns in the row
    score = 0
    p1 = 0  # First pointer
    
    while True:
        # Skip already eliminated cells (marked as -1)
        while grids[n][p1] == -1:
            p1 += 1
            if p1 >= length-1:  # Reached end of row
                return score
        
        p2 = p1 + 1  # Second pointer starts right after p1
        if p2 == length:  # No more cells to compare with
            return score
        
        # Skip already eliminated cells for p2
        while grids[n][p2] == -1:
            p2 += 1
            if p2 == length:  # Reached end of row
                return score
        
        # Check if the two cells have the same value
        if grids[n][p1] == grids[n][p2]:
            # Eliminate the pair and add to score
            score += grids[n][p1]  # Add the value (both cells have same value)
            grids[n][p1] = grids[n][p2] = -1  # Mark as eliminated
            p1 = p2 + 1  # Move p1 to next position after the pair
            
            if p1 == length:  # Reached end of row
                return score
        else:
            # Values don't match, move p1 to p2 position
            p1 = p2
```
