### h081. 1. 程式交易

#### 題目描述

小明最近想要用程式做股票交易，給一個股票的歷史價格 a[1], a[2], ..., a[n]。  
他的投資策略如下：

1. 同一個時間最多只會持有一張股票，並會在時間點 a 花 a[1] 買進。
2. 若當下持有股票且該股票買進價格為 x，當遇到價格 y 大於等於 x + D 時即賣出，並轉得利潤 y - x。
3. 若當下沒有持有股票且上一次的賣出價格為 x，當遇到價格 y 小於等於 x - D 時則會買進股票。

輸出依照上述規則買賣後所得到的利潤和。若交易結束仍持有股票，則不考慮該股票買進的成本，直接無視該股票即可。

**輸入說明**
- 第一行輸入兩個正整數 n, D
- 接下來有 n 個正整數，代表每個時間點股票的價格

**數字範圍**
- 1 <= n <= 100, 1 <= D <= 100
- 1 <= a[i] <= 100

**子題配分**
- (50%): n = 3
- (50%): 無額外限制

**輸出說明**
- 輸出一個正整數，代表總利潤

#### 測試範例

**範例輸入 #1**
```
3 10
50 20 45
```

**範例輸出 #1**
```
0
```

**範例輸入 #2**
```
6 10
30 20 45 38 10 20
```

**範例輸出 #2**
```
25
```

#### 程式碼解答

```python
# 讀取輸入
n, D = map(int, input().split())
prices = list(map(int, input().split()))

# 初始化變數
purchase_price = prices[0]  # 初始買入價格
sell_price = None           # 上一次賣出價格
profit = 0                  # 總利潤

# 從第二天開始遍歷價格
for p in prices[1:]:
    # 如果持有股票且當前價格達到賣出條件
    if purchase_price and p >= purchase_price + D:
        profit += p - purchase_price  # 計算利潤
        sell_price = p                # 記錄賣出價格
        purchase_price = None         # 清空持有狀態
    
    # 如果沒有持有股票且當前價格達到買入條件
    elif sell_price and p <= sell_price - D:
        purchase_price = p            # 記錄買入價格
        sell_price = None             # 清空賣出記錄

# 輸出總利潤
print(profit)
```

### h081. 1. Program Trading

#### Problem Description

Xiao Ming recently wants to use programs for stock trading. Given historical stock prices a[1], a[2], ..., a[n], his investment strategy is as follows:

1. At most one stock is held at the same time, and the first purchase is made at time point a with price a[1].
2. If currently holding a stock bought at price x, and encountering a price y ≥ x + D, sell immediately and realize profit y - x.
3. If not holding any stock and the last selling price was x, when encountering a price y ≤ x - D, buy the stock.

Output the total profit according to the above rules. If still holding a stock when trading ends, ignore the cost of that stock (disregard it completely).

**Input Specification**
- First line: two positive integers n, D
- Next line: n positive integers representing stock prices at each time point

**Number Range**
- 1 <= n <= 100, 1 <= D <= 100
- 1 <= a[i] <= 100

**Scoring**
- (50%): n = 3
- (50%): No additional constraints

**Output Specification**
- Output a positive integer representing the total profit

#### Test Examples

**Example Input #1**
```
3 10
50 20 45
```

**Example Output #1**
```
0
```

**Example Input #2**
```
6 10
30 20 45 38 10 20
```

**Example Output #2**
```
25
```

#### Code Solution

```python
# Read input
n, D = map(int, input().split())
prices = list(map(int, input().split()))

# Initialize variables
purchase_price = prices[0]  # Initial purchase price
sell_price = None           # Last selling price
profit = 0                  # Total profit

# Process prices starting from day 2
for p in prices[1:]:
    # If holding stock and current price meets selling condition
    if purchase_price and p >= purchase_price + D:
        profit += p - purchase_price  # Calculate profit
        sell_price = p                # Record selling price
        purchase_price = None         # Clear holding state
    
    # If not holding stock and current price meets buying condition
    elif sell_price and p <= sell_price - D:
        purchase_price = p            # Record purchase price
        sell_price = None             # Clear selling record

# Output total profit
print(profit)
```

````python
# n=3 
# D=10
# prices =[50, 20, 45]
# n=6
# D=10
# prices =[30,20,45,38,10,20]
n,D = map(int, input().split())
prices = list(map(int,input().split()))
purchase_price = prices[0]
sell_price=None
profit = 0
for p in prices[1:]:
    if purchase_price and p>=purchase_price+D:
        profit += p - purchase_price
        sell_price = p
        purchase_price = None
    elif sell_price and p<=sell_price-D:
        purchase_price = p
        sell_price = None

print(profit)
        
````
