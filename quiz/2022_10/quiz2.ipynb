### j123. 2. 運貨站

#### 題目描述

貨運站需要管理 `n` 個五種不同形狀的貨物，下圖標示出貨物的形狀以及對應的英文代碼。

![](ShowImage.png)

現在這 `n` 個貨物要按照順序堆放在一個容量大小為 `R × C` 的倉庫內，第 `i` 個貨物的形狀為 `ti`，並且和倉庫的頂部距離為 `yi` (見圖一)。貨物堆放置倉庫內時必須維持和倉庫頂端的高度由右向左推到不能前進為止，並且過程中不行將貨物的方向做旋轉。若有一個貨物不能完整放入倉庫內，則該貨物會被貨運站丟棄。

請輸出依序放完這 `n` 個貨物後，倉庫內有多少剩餘空格，以及被丟棄的貨物有幾個。

![](ShowImage_2.png)

**圖一**：該貨物類別為 B，並且離倉庫頂端距離為 2。

保證輸入內貨物距離倉庫頂部的高度不會讓貨物底部低於地面，並且不會有任何貨物卡在倉庫門口的情形。

**輸入**：
- 第一行：三個數字 `R, C, n`，代表倉庫大小為 `R × C` 以及有 `n` 個貨物
- 接下來 `n` 行：每行有一個大寫英文字母 `ti` 和一個數字 `yi` 代表貨物的種類以及和倉庫頂部的距離
- 1<=R<=30, 1<=C<=50, 1<=n<=200

**輸出**：
- 倉庫剩餘的空格數量，以及被丟棄的貨物數量

**子題配分**：
- (20%)：只會出現Ｂ類型
- (40%)：只會出現Ａ,Ｂ,Ｃ類型  
- (40%)：５種類型都會出現

#### 程式碼解答

```python
R, C, n = map(int, input().split())

# 定義形狀座標 (行, 列) 相對於左上角位置
shapes = {
    'A': [(0,0), (1,0), (2,0), (3,0)],  # 4x1 垂直長條
    'B': [(0,0), (0,1), (0,2)],         # 1x3 水平長條
    'C': [(0,0), (1,0), (0,1), (1,1)],  # 2x2 正方形
    'D': [(0,2), (1,0), (1,1), (1,2)],  # 2x3 L形
    'E': [(0,1), (1,0), (1,1), (2,0), (2,1)]  # 3x2 階梯形
}

# 倉庫網格: 0 = 空, 1 = 已佔用
warehouse = [[0] * C for _ in range(R)]
discard_count = 0

def get_width_height(shape_coords):
    """計算形狀的寬度和高度"""
    rows = [r for r, c in shape_coords]
    cols = [c for r, c in shape_coords]
    height = max(rows) + 1
    width = max(cols) + 1
    return width, height

def can_place(shape_coords, top_row, left_col):
    """檢查形狀是否可以放在指定位置"""
    for dr, dc in shape_coords:
        r = top_row + dr
        c = left_col + dc
        # 檢查是否超出邊界或已被佔用
        if r < 0 or r >= R or c < 0 or c >= C or warehouse[r][c] == 1:
            return False
    return True

def place_shape(shape_coords, top_row, left_col):
    """將形狀放在指定位置"""
    for dr, dc in shape_coords:
        r = top_row + dr
        c = left_col + dc
        warehouse[r][c] = 1

# 處理每個貨物
for _ in range(n):
    t, y = input().split()
    y = int(y)
    shape_coords = shapes[t]
    W, H = get_width_height(shape_coords)
    
    # 檢查底部是否超出倉庫地板
    if y + H - 1 >= R:
        discard_count += 1
        continue
    
    # 嘗試從右到左放置
    placed = False
    final_col = -1
    
    # 從最右邊可能的位置開始向左移動
    for col in range(C - W, -1, -1):
        if can_place(shape_coords, y, col):
            final_col = col
        else:
            # 在第一個不能放置的位置停止
            break
    
    if final_col != -1:
        place_shape(shape_coords, y, final_col)
    else:
        discard_count += 1

# 計算剩餘空格數量
empty_count = sum(cell == 0 for row in warehouse for cell in row)
print(empty_count, discard_count)
```

### j123. 2. Freight Station

#### Problem Description

A freight station needs to manage `n` cargo items of five different shapes. The diagram below shows the shapes and their corresponding English codes.

![](ShowImage.png)

These `n` cargo items need to be stacked in order in a warehouse with capacity `R × C`. The shape of the `i`-th cargo item is `ti`, and its distance from the top of the warehouse is `yi` (see Figure 1). When placing cargo in the warehouse, items must be pushed from right to left until they cannot move further, maintaining their distance from the top, and the orientation of cargo cannot be rotated during the process. If a cargo item cannot be completely placed in the warehouse, it will be discarded by the freight station.

Please output how many empty spaces remain in the warehouse after placing all `n` cargo items in order, and how many cargo items were discarded.

![](ShowImage_2.png)

**Figure 1**: The cargo type is B, and the distance from the top of the warehouse is 2.

It is guaranteed that the input cargo distance from the top will not cause the bottom of the cargo to be below the ground, and no cargo will get stuck in the warehouse entrance.

**Input**:
- First line: three numbers `R, C, n` representing warehouse size `R × C` and `n` cargo items
- Next `n` lines: each line contains an uppercase letter `ti` and a number `yi` representing the cargo type and distance from the top
- Constraints: 1<=R<=30, 1<=C<=50, 1<=n<=200

**Output**:
- Number of remaining empty spaces in the warehouse, and the number of discarded cargo items

**Scoring**:
- (20%): Only type B appears
- (40%): Only types A, B, C appear  
- (40%): All five types appear

#### Test Example #1 Verification

```
B 2
```

**Output:**
```
8 2
```

#### Code Solution

```python
R, C, n = map(int, input().split())

# Define shape coordinates (row, col) relative to top-left position
shapes = {
    'A': [(0,0), (1,0), (2,0), (3,0)],  # 4x1 vertical bar
    'B': [(0,0), (0,1), (0,2)],         # 1x3 horizontal bar  
    'C': [(0,0), (1,0), (0,1), (1,1)],  # 2x2 square
    'D': [(0,2), (1,0), (1,1), (1,2)],  # 2x3 L-shape
    'E': [(0,1), (1,0), (1,1), (2,0), (2,1)]  # 3x2 staircase
}

# Warehouse grid: 0 = empty, 1 = occupied
warehouse = [[0] * C for _ in range(R)]
discard_count = 0

def get_width_height(shape_coords):
    """Calculate width and height of a shape"""
    rows = [r for r, c in shape_coords]
    cols = [c for r, c in shape_coords]
    height = max(rows) + 1
    width = max(cols) + 1
    return width, height

def can_place(shape_coords, top_row, left_col):
    """Check if shape can be placed at given position"""
    for dr, dc in shape_coords:
        r = top_row + dr
        c = left_col + dc
        # Check if out of bounds or already occupied
        if r < 0 or r >= R or c < 0 or c >= C or warehouse[r][c] == 1:
            return False
    return True

def place_shape(shape_coords, top_row, left_col):
    """Place shape at given position"""
    for dr, dc in shape_coords:
        r = top_row + dr
        c = left_col + dc
        warehouse[r][c] = 1

# Process each cargo item
for _ in range(n):
    t, y = input().split()
    y = int(y)
    shape_coords = shapes[t]
    W, H = get_width_height(shape_coords)
    
    # Check if bottom exceeds warehouse floor
    if y + H - 1 >= R:
        discard_count += 1
        continue
    
    # Try to place from right to left
    final_col = -1
    
    # Start from rightmost possible position and move left
    for col in range(C - W, -1, -1):
        if can_place(shape_coords, y, col):
            final_col = col
        else:
            # Stop at first position that cannot place
            break
    
    if final_col != -1:
        place_shape(shape_coords, y, final_col)
    else:
        discard_count += 1

# Count remaining empty spaces
empty_count = sum(cell == 0 for row in warehouse for cell in row)
print(empty_count, discard_count)
```

````python
def check_shape(ro, co, t):
    for r,c in shapes[t]:
        if warehouse[ro+r][co+c] !=0:
            return False
    return True     
   
def update_warehouse(ro,co,t):
    for r,c in shapes[t]:
        warehouse[ro+r][co+c] = 1

def space_available():
    space =0
    for r in range(R):
        for c in range(C):
            if warehouse[r][c] ==0:
                space+=1
    return space
            
R, C, n = map(int, input().split())
# R,C,n = 5,4,6
# items = (('B',0),('B',3),('B',1),('B',3),('B',1),('B',2))
# R,C,n = 5,6,6
# items = (('C',1),('A',1),('E',0),('E',0),('B',0),('A',0))

# relative coordinates of each shape
shapes = {
    'A': ((0,0), (1,0), (2,0), (3,0)),  # 4x1
    'B': ((0,0), (0,1), (0,2)),         # 1x3
    'C': ((0,0), (1,0), (0,1), (1,1)),  # 2x2
    'D': ((0,2), (1,0), (1,1), (1,2)),  # 2x3
    'E': ((0,1), (1,0), (1,1), (2,0), (2,1))  # 3x2
}
# only width for each shape is needed, because the quiz description says that:
# It is guaranteed that the input cargo distance from the top will not cause 
# the bottom of the cargo to be below the ground, and no cargo will get stuck 
# in the warehouse entrance.
widths = {'A': 1, 'B': 3, 'C': 2, 'D': 3, 'E': 2} 
warehouse = [[0]*C for _ in range(R)]
discard_count = 0

for i in range(n):
    t, y = input().split() # type - item shape type
    y = int(y) # y: row position
    # t,y = items[i]
    # calculate right-most column position for upper left corner of the object
    c_max = C-widths[t]
    containable = False
    c_best = None # to store the leftmost position possible
    for c_pos in range(c_max, -1, -1):
        if check_shape(y, c_pos, t):
            containable = True
            c_best = c_pos
        else:
            break    
    if containable:        
        update_warehouse(y, c_best, t)
    else:
        discard_count+=1
    
space = space_available()        
print(space, discard_count)
````
