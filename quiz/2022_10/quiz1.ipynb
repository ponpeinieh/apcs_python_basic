### i428. 1. 巴士站牌

#### 題目描述

平面上有 `n` 個巴士站，第 `i` 個巴士站的位置可以用座標點 `(xi, yi)` 來表示。

兩個巴士站之間行進的時間是這兩個巴士站座標的曼哈頓距離。曼哈頓距離的定義如下：
對於兩個座標點 `(x1,y1)` 與 `(x2,y2)`，兩點之間曼哈頓距離為 `|x1-x2| + |y1-y2|`。

你今天要從巴士站 `1` 坐車到巴士站 `n`，中間依序經過巴士站 `2,3,...,(n-1)`。
請計算過程中相鄰兩站的行進時間的 **最大值** 與 **最小值**。

**輸入說明**
- 第 1 行有一個正整數 `n` 表示路程中總共會經過幾個巴士站。
- 在輸入的第 2 行到第 `n+1` 行，每行有兩個整數標示巴士站的座標。
- 嚴格地說，輸入第 `i+1` 行的兩個數字依序是 `xi, yi`。

**子題配分**：
- (60%): n=4 and -100<=xi, yi<=100
- (40%): 4<=n<=100 and -100<=xi, yi<=100

**輸出說明**
- 在一行輸出兩個整數並以空格分開。
- 第一個整數表示相鄰兩站的行進時間的 **最大值**。
- 第二個整數表示相鄰兩站的行進時間的 **最小值**。

#### 測試範例

**範例輸入 #1**
```
4
1 1
1 3
4 5
2 6
```

**範例輸出 #1**
```
5 2
```

**範例輸入 #2**
```
4
1 2
-1 -1
1 3
0 0
```

**範例輸出 #2**
```
6 4
```

#### 程式碼解答

```python
n = int(input())

# 讀取所有巴士站座標
stations = []
for _ in range(n):
    x, y = map(int, input().split())
    stations.append((x, y))

# 計算相鄰兩站之間的曼哈頓距離
distances = []
for i in range(1, n):
    x1, y1 = stations[i-1]
    x2, y2 = stations[i]
    distance = abs(x1 - x2) + abs(y1 - y2)
    distances.append(distance)

# 找出最大值和最小值
max_distance = max(distances)
min_distance = min(distances)

print(max_distance, min_distance)
```

### i428. 1. Bus Stops

#### Problem Description

There are `n` bus stops on a plane. The position of the `i`-th bus stop can be represented by coordinates `(xi, yi)`.

The travel time between two bus stops is the Manhattan distance between their coordinates. The definition of Manhattan distance is as follows:
For two coordinate points `(x1,y1)` and `(x2,y2)`, the Manhattan distance between them is `abs(x1-x2) + abs(y1-y2)`.

You need to travel from bus stop `1` to bus stop `n`, passing through bus stops `2,3,...,(n-1)` in order.
Please calculate the **maximum** and **minimum** travel times between adjacent stops during the journey.

**Input Specification**
- Line 1: A positive integer `n` indicating the total number of bus stops along the route.
- Lines 2 to `n+1`: Each line contains two integers representing the coordinates of a bus stop.
- Strictly speaking, the two numbers in line `i+1` are `xi, yi` in order.

**Scoring**:
- (60%): n=4 and -100<=xi, yi<=100
- (40%): 4<=n<=100 and -100<=xi, yi<=100

**Output Specification**
- Output two integers separated by a space on one line.
- The first integer represents the **maximum** travel time between adjacent stops.
- The second integer represents the **minimum** travel time between adjacent stops.

#### Test Examples

**Example Input #1**
```
4
1 1
1 3
4 5
2 6
```

**Example Output #1**
```
5 2
```

**Example Input #2**
```
4
1 2
-1 -1
1 3
0 0
```

**Example Output #2**
```
6 4
```

#### Code Solution

```python
n = int(input())

# Read all bus stop coordinates
stations = []
for _ in range(n):
    x, y = map(int, input().split())
    stations.append((x, y))

# Calculate Manhattan distances between adjacent stops
distances = []
for i in range(1, n):
    x1, y1 = stations[i-1]
    x2, y2 = stations[i]
    distance = abs(x1 - x2) + abs(y1 - y2)
    distances.append(distance)

# Find maximum and minimum
max_distance = max(distances)
min_distance = min(distances)

print(max_distance, min_distance)
```

````python
n = int(input())
positions = []
for _ in range(n):
    positions.append(list(map(int, input().split()))) 
# n=4
# positions = ((1,1), (1,3), (4,5), (2,6))
# n=4
# positions=((1,2),(-1,-1),(1,3),(0,0))
max_d = 0
min_d = float('inf')
for i in range(n-1):
    x,y = positions[i]
    x2,y2 = positions[i+1]
    d = abs(x-x2)+abs(y-y2)
    if d>max_d:
        max_d = d
    if d<min_d:
        min_d = d
print(max_d, min_d)
````
