### m932. 2. 蜜蜂觀察

#### 問題描述

蜜蜂 Bob 在一個大小是 m×n 的蜂巢，每個蜂巢格上有一個代表字母（大小寫英文字母）。

Bob 一開始在蜂巢左下角，行走方向定義：
- 0: 往右上
- 1: 往右邊  
- 2: 往右下
- 3: 往左下
- 4: 往左邊
- 5: 往左上

![](grid.png)

輸入每步移動的方向，請輸出經過的路徑每格的代表字母，以及經過字元的種類數（大小寫相異）。若移動時碰到牆壁，該行動會停在原地。

#### 輸入說明
- 第一行：三個整數 m, n, k (1 ≤ m, n < 20, 1 ≤ k < 100)
- 接下來的 m 行：每行 n 個字母（大小寫英文字母），代表蜂巢的狀態
- 最後一行：k 個整數，表示 Bob 的移動路徑方向

#### 輸出說明
- 輸出一行：包含 Bob 每步經過的字母組成的字串 + 經過字元的種類數

#### 測試範例 #1 驗證

**輸入：**
```
2 4 5
TyuI
ABaB
0 1 2 3 0
```

**網格：**
```
行 0: T y u I
行 1: A B a B
```

![](grid1.png)

**計算過程：**
- 起始位置: (1, 0) → 'A' (不記錄)
- 方向 0 (-1, 0): (0, 0) → 'T' (記錄)
- 方向 1 (0, 1): (0, 1) → 'y' (記錄)
- 方向 2 (1, 1): (1, 2) → 'a' (記錄)
- 方向 3 (1, 0): (2, 2) → 無效，停留在 (1, 2) → 'a' (記錄)
- 方向 0 (-1, 0): (0, 2) → 'u' (記錄)

**路徑字串:** "T" + "y" + "a" + "a" + "u" = "Tyaau"
**獨特字符:** 'T', 'y', 'a', 'u' → 4個

這完全符合範例輸出 #1！

#### 測試範例 #2 驗證

**輸入：**
```
4 6 11
rMmnis
LRveEX
ZexDoc
HAdbHA
0 1 5 1 1 0 3 0 0 1 0
```

![](grid2.png)

**計算過程：**
- 起始位置: (3, 0) → 'H' (不記錄)
- 按照11個方向依次移動並記錄字符

**輸出：** "ZeLRvmvmmnn" 和 7


#### 解答

```python
def is_valid_position(r, c, m, n):
    """檢查位置是否在蜂巢範圍內"""
    return 0 <= r < m and 0 <= c < n

def get_next_position(r, c, direction):
    """根據當前位置和方向計算下一個位置（統一移動規則）"""
    # 統一的移動方向增量
    moves = [
        (-1, 0),   # 0: 右上
        (0, 1),    # 1: 右
        (1, 1),    # 2: 右下
        (1, 0),    # 3: 左下
        (0, -1),   # 4: 左
        (-1, -1)   # 5: 左上
    ]
    
    dr, dc = moves[direction]
    return r + dr, c + dc

def main():
    # 讀取輸入
    m, n, k = map(int, input().split())
    
    # 讀取蜂巢
    grid = []
    for i in range(m):
        row = input().strip()
        grid.append(row)
    
    # 讀取移動方向
    directions = list(map(int, input().split()))
    
    # 初始化位置 (左下角，即第 m-1 行，第 0 列)
    r, c = m - 1, 0
    visited_chars = set()
    path_chars = []
    
    # 起始位置不記錄在路徑中，只記錄移動後的位置
    # 模擬移動
    for direction in directions:
        nr, nc = get_next_position(r, c, direction)
        
        # 檢查新位置是否有效，如果無效則停留在原地
        if is_valid_position(nr, nc, m, n):
            r, c = nr, nc
        
        char = grid[r][c]
        visited_chars.add(char)
        path_chars.append(char)
    
    # 輸出結果
    path_str = ''.join(path_chars)
    unique_count = len(visited_chars)
    print(path_str)
    print(unique_count)

if __name__ == "__main__":
    main()
```

### m932. 2. Bee Observation

#### Problem Description

Bee Bob is in a beehive of size m×n, where each cell has a representative letter (uppercase and lowercase letters).

Bob starts at the lower-left corner of the beehive, and the movement directions are defined as follows:
- 0: Up-Right
- 1: Right  
- 2: Down-Right
- 3: Down-Left
- 4: Left
- 5: Up-Left

![](grid.png)

Given the movement directions, output the letters representing each cell visited along the path, as well as the number of unique characters (case-sensitive) encountered. If a move hits a wall, the action stops at the original position.

#### Input Format
- First line: three integers m, n, k (1 ≤ m, n < 20, 1 ≤ k < 100)
- Next m lines: each line contains n letters (uppercase and lowercase letters), representing the state of the beehive.
- Last line: k integers indicating Bob's movement directions.

#### Output Format
- Output a single line: a string composed of the letters Bob passed through + the number of unique characters.

#### Sample Test Case #1

**Input:**
```
2 4 5
TyuI
ABaB
0 1 2 3 0
```

**Grid:**
```
Row 0: T y u I
Row 1: A B a B
```

![](grid1.png)

**Calculation Process:**
- Starting position: (1, 0) → 'A' (not recorded)
- Direction 0 (-1, 0): (0, 0) → 'T' (recorded)
- Direction 1 (0, 1): (0, 1) → 'y' (recorded)
- Direction 2 (1, 1): (1, 2) → 'a' (recorded)
- Direction 3 (1, 0): (2, 2) → invalid, stay at (1, 2) → 'a' (recorded)
- Direction 0 (-1, 0): (0, 2) → 'u' (recorded)

**Path String:** "T" + "y" + "a" + "a" + "u" = "Tyaau"
**Unique Characters:** 'T', 'y', 'a', 'u' → 4

This matches Sample Output #1 perfectly!

#### Sample Test Case #2

**Input:**
```
4 6 11
rMmnis
LRveEX
ZexDoc
HAdbHA
0 1 5 1 1 0 3 0 0 1 0
```

![](grid2.png)

**Calculation Process:**
- Starting position: (3, 0) → 'H' (not recorded)
- Move according to the 11 directions and record characters.

**Output:** "ZeLRvmvmmnn" and 7


#### Solution

```python
def is_valid_position(r, c, m, n):
    """Check if the position is within the beehive range."""
    return 0 <= r < m and 0 <= c < n

def get_next_position(r, c, direction):
    """Calculate the next position based on current position and direction (uniform movement rules)."""
    # Directional movement increments
    moves = [
        (-1, 0),   # 0: Up-Right
        (0, 1),    # 1: Right
        (1, 1),    # 2: Down-Right
        (1, 0),    # 3: Down-Left
        (0, -1),   # 4: Left
        (-1, -1)   # 5: Up-Left
    ]
    
    dr, dc = moves[direction]
    return r + dr, c + dc

def main():
    # Read input
    m, n, k = map(int, input().split())
    
    # Read beehive
    grid = []
    for i in range(m):
        row = input().strip()
        grid.append(row)
    
    # Read movement directions
    directions = list(map(int, input().split()))
    
    # Initialize position (lower left corner, i.e., row m-1, column 0)
    r, c = m - 1, 0
    visited_chars = set()
    path_chars = []
    
    # Starting position is not recorded in the path, only record after moving
    # Simulate movement
    for direction in directions:
        nr, nc = get_next_position(r, c, direction)
        
        # Check if the new position is valid; if not, stay in place
        if is_valid_position(nr, nc, m, n):
            r, c = nr, nc
        
        char = grid[r][c]
        visited_chars.add(char)
        path_chars.append(char)
    
    # Output result
    path_str = ''.join(path_chars)
    unique_count = len(visited_chars)
    print(path_str)
    print(unique_count)

if __name__ == "__main__":
    main()
```

````python
m,n,k = map(int, input().split())
grids = []
for _ in range(m):
    grids.append(input())
ops = list(map(int, input().split()))

# m,n,k = [2,4,5]
# grids =['TyuI','ABaB']
# ops = [0, 1, 2, 3, 0]

# m,n,k = 4, 6, 11
# grids = ['rMmnis','LRveEX','ZexDoc','HAdbHA']
# ops = [0, 1, 5, 1, 1, 0, 3, 0, 0, 1, 0]

directions = [(-1,0), (0,1), (1,1), (1,0), (0,-1), (-1,-1)]
path = []
current_r = m - 1
current_c = 0
for op in ops:
    next_r = current_r + directions[op][0]
    next_c = current_c + directions[op][1]
    if 0<= next_r < m and 0<= next_c < n:
        current_r = next_r
        current_c = next_c
    path.append(grids[current_r][current_c])
print("".join(path))
print(len(set(path)))        
````

### m932. 2. 蜜蜂觀察(Advanced) - with different direction movement for even and odd rows

#### 問題描述

蜜蜂 Bob 在一個大小是 m×n 的蜂巢，每個蜂巢格上有一個代表字母（大小寫英文字母）。

Bob 一開始在蜂巢左下角，行走方向定義：
- 0: 往右上
- 1: 往右邊  
- 2: 往右下
- 3: 往左下
- 4: 往左邊
- 5: 往左上

輸入每步移動的方向，請輸出經過的路徑每格的代表字母，以及經過字元的種類數（大小寫相異）。若移動時碰到牆壁，該行動會停在原地。

#### 輸入說明
- 第一行：三個整數 m, n, k (1 ≤ m, n < 20, 1 ≤ k < 100)
- 接下來的 m 行：每行 n 個字母（大小寫英文字母），代表蜂巢的狀態
- 最後一行：k 個整數，表示 Bob 的移動路徑方向

#### 輸出說明
- 輸出一行：包含 Bob 每步經過的字母組成的字串 + 經過字元的種類數

#### 蜂巢坐標系統說明

蜂巢坐標系統需要特別處理，因為奇數行和偶數行的鄰接關係不同：

```
偶數行 (0-based, 偶數索引):    奇數行 (0-based, 奇數索引):
   ↗ → ↘                         ↗ → ↘
   ↖ ← ↙                         ↖ ← ↙

移動方向對應的坐標變化:
方向 0 (右上): 偶數行: (r-1, c+0), 奇數行: (r-1, c+1)
方向 1 (右):   (r+0, c+1)
方向 2 (右下): 偶數行: (r+1, c+0), 奇數行: (r+1, c+1)
方向 3 (左下): 偶數行: (r+1, c-1), 奇數行: (r+1, c+0)
方向 4 (左):   (r+0, c-1)
方向 5 (左上): 偶數行: (r-1, c-1), 奇數行: (r-1, c+0)
```

#### 解答

- Note that below solution mistakenly start the route from the starting position. It should instead skip the start position.

```python
def is_valid_position(r, c, m, n):
    """檢查位置是否在蜂巢範圍內"""
    return 0 <= r < m and 0 <= c < n

def get_next_position(r, c, direction, m, n):
    """根據當前位置和方向計算下一個位置"""
    # 判斷當前行是偶數行還是奇數行 (0-based)
    is_even_row = (r % 2 == 0)
    
    if direction == 0:  # 右上
        if is_even_row:
            nr, nc = r - 1, c
        else:
            nr, nc = r - 1, c + 1
    elif direction == 1:  # 右
        nr, nc = r, c + 1
    elif direction == 2:  # 右下
        if is_even_row:
            nr, nc = r + 1, c
        else:
            nr, nc = r + 1, c + 1
    elif direction == 3:  # 左下
        if is_even_row:
            nr, nc = r + 1, c - 1
        else:
            nr, nc = r + 1, c
    elif direction == 4:  # 左
        nr, nc = r, c - 1
    elif direction == 5:  # 左上
        if is_even_row:
            nr, nc = r - 1, c - 1
        else:
            nr, nc = r - 1, c
    
    # 檢查新位置是否有效，如果無效則停留在原地
    if is_valid_position(nr, nc, m, n):
        return nr, nc
    else:
        return r, c

def main():
    # 讀取輸入
    m, n, k = map(int, input().split())
    
    # 讀取蜂巢
    grid = []
    for i in range(m):
        row = input().strip()
        grid.append(row)
    
    # 讀取移動方向
    directions = list(map(int, input().split()))
    
    # 初始化位置 (左下角，即第 m-1 行，第 0 列)
    r, c = m - 1, 0
    visited_chars = set()
    path_chars = []
    
    # 記錄起始位置
    visited_chars.add(grid[r][c])
    path_chars.append(grid[r][c])
    
    # 模擬移動
    for direction in directions:
        r, c = get_next_position(r, c, direction, m, n)
        char = grid[r][c]
        visited_chars.add(char)
        path_chars.append(char)
    
    # 輸出結果
    path_str = ''.join(path_chars)
    unique_count = len(visited_chars)
    print(path_str)
    print(unique_count)

if __name__ == "__main__":
    main()
```
