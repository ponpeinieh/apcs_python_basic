### m931. 1. 遊戲選角

#### 問題描述

有 n 個角色，每個角色有攻擊力和防禦力。角色的能力值是攻擊力和防禦力的平方和，請輸出能力值第二大的角色的攻擊力和防禦力數值。

保證每個角色的能力值相異。

#### 輸入說明

- 第一行包含一個整數 n（3 ≤ n ≤ 20），表示有多少個角色
- 接下來的 n 行，每行包含兩個整數 ai 和 di（1 ≤ ai, di ≤ 100），表示第 i 個角色的攻擊力和防禦力

#### 評分標準

- (60 分): 滿足 n = 3
- (40 分): 一般情況

---

#### Problem Description

There are n characters, each with attack and defense values. A character's ability value is the sum of squares of their attack and defense values. Output the attack and defense values of the character with the second largest ability value.

It is guaranteed that all characters have distinct ability values.

#### Input Format

- First line: an integer n (3 ≤ n ≤ 20), representing the number of characters
- Next n lines: two integers ai and di (1 ≤ ai, di ≤ 100), representing the attack and defense values of the i-th character

#### Scoring

- (60 points): n = 3
- (40 points): no restrictions

---

#### 解答
```python
def main():
    n = int(input())
    characters = []
    
    # 讀取每個角色的攻擊力和防禦力，並計算能力值
    for _ in range(n):
        a, d = map(int, input().split())
        ability = a * a + d * d
        characters.append((a, d, ability))
    
    # 按能力值降序排序
    characters.sort(key=lambda x: x[2], reverse=True)
    
    # 輸出能力值第二大的角色的攻擊力和防禦力
    print(characters[1][0], characters[1][1])

if __name__ == "__main__":
    main()
```

## 算法說明

### 能力值計算

每個角色的能力值計算公式：
```
ability = ai² + di²
```

### 解題思路

1. **讀取輸入**: 讀取角色數量 n 和每個角色的攻擊力與防禦力
2. **計算能力值**: 為每個角色計算攻擊力和防禦力的平方和
3. **排序角色**: 根據能力值對所有角色進行降序排序
4. **選擇結果**: 輸出排序後列表中第二個角色的攻擊力和防禦力

### 複雜度分析

- **時間複雜度**: O(n log n)
  - 主要來自排序操作，n 最大為 20，效率足夠
- **空間複雜度**: O(n)
  - 需要存儲 n 個角色的信息

### 替代方案

對於這道題目，由於 n 的範圍很小（3-20），也可以使用以下方法：
1. **部分排序**: 只找到前兩個最大值，而不需要完全排序
2. **手動比較**: 遍歷一次找到最大值，再遍歷一次找到第二大的值

但使用內置排序函數代碼更簡潔，且對於小規模數據效率足夠。

# 遊戲選角 - 使用堆積佇列 (Heap Queue) 解決方案

```python
import heapq

def main():
    n = int(input())
    
    # 使用最大堆來存儲能力值和對應的角色信息
    # 由於Python的heapq只實現最小堆，我們通過存儲負值來模擬最大堆
    max_heap = []
    
    for _ in range(n):
        a, d = map(int, input().split())
        ability = a * a + d * d
        # 存儲負的能力值來模擬最大堆
        heapq.heappush(max_heap, (-ability, a, d))
    
    # 彈出最大的元素（能力值第一的角色）
    heapq.heappop(max_heap)
    
    # 現在堆頂是第二大的元素
    second_ability, a, d = max_heap[0]
    
    # 輸出能力值第二大的角色的攻擊力和防禦力
    print(a, d)

if __name__ == "__main__":
    main()
```

## 替代方案：使用 nlargest 函數

```python
import heapq

def main():
    n = int(input())
    characters = []
    
    for _ in range(n):
        a, d = map(int, input().split())
        ability = a * a + d * d
        characters.append((-ability, a, d))  # 存儲負值用於最大堆
    
    # 使用heapq.nlargest獲取前兩個最大的元素
    second_largest = heapq.nlargest(2, characters)[1]
    
    # 輸出結果
    print(second_largest[1], second_largest[2])

if __name__ == "__main__":
    main()
```
