### q836. 1. 小心陷阱 - 2025年6月APCS

#### 問題描述

在一維數線上進行一場模擬遊戲。遊戲規則如下：

- 初始位置為 0
- 初始生命值為 k（1 ≤ k ≤ 20）
- 當前生命值為 v 時，角色會往右跳 v 格，也就是從位置 p 移動到 p + v
- 每次移動後，如果落在 x1 或 x2 的倍數上，會損失生命值：
  - 若落在 x1 的倍數，生命值減少 y1
  - 若落在 x2 的倍數，生命值減少 y2
  - 若同時是 x1 和 x2 的倍數，生命值總共減少 y1 + y2
- 當生命值小於等於 0 時，遊戲結束，輸出當前所在的位置

**輸入格式：**
- 第一行：初始生命值 k
- 第二行：x1 和 y1（空格分隔）
- 第三行：x2 和 y2（空格分隔）

**輸入範圍：**
- k：1 ≤ k ≤ 20
- x1, x2：1 ≤ x1, x2 ≤ 10
- y1, y2：1 ≤ y1, y2 ≤ 20

---

#### Problem Description 

A simulation game is played on a 1D number line. The rules are as follows:

- Initial position is 0
- Initial health points is k (1 ≤ k ≤ 20)
- When current health is v, the character jumps v steps to the right, moving from position p to p + v
- After each move, if landing on multiples of x1 or x2, health points are deducted:
  - If landing on a multiple of x1, health decreases by y1
  - If landing on a multiple of x2, health decreases by y2
  - If landing on a multiple of both x1 and x2, health decreases by y1 + y2
- When health points become 0 or less, the game ends and outputs the current position

**Input Format:**
- First line: initial health k
- Second line: x1 and y1 (space separated)
- Third line: x2 and y2 (space separated)

**Input Range:**
- k: 1 ≤ k ≤ 20
- x1, x2: 1 ≤ x1, x2 ≤ 10
- y1, y2: 1 ≤ y1, y2 ≤ 20

---

#### 解答

```python
def main():
    # 讀取初始生命值 / Read initial health
    k = int(input().strip())
    
    # 讀取第一個陷阱的參數 / Read first trap parameters
    data1 = input().split()
    x1 = int(data1[0])
    y1 = int(data1[1])
    
    # 讀取第二個陷阱的參數 / Read second trap parameters
    data2 = input().split()
    x2 = int(data2[0])
    y2 = int(data2[1])
    
    # 初始化位置和當前生命值 / Initialize position and current health
    position = 0
    current_hp = k
    
    # 模擬遊戲過程 / Simulate the game process
    while current_hp > 0:
        # 向右跳躍當前生命值的距離 / Jump right by current health value
        position += current_hp
        
        # 檢查是否落在陷阱上 / Check if landing on traps
        is_x1_multiple = (position % x1 == 0)
        is_x2_multiple = (position % x2 == 0)
        
        # 計算傷害 / Calculate damage
        damage = 0
        if is_x1_multiple:
            damage += y1
        if is_x2_multiple:
            damage += y2
        
        # 扣除生命值 / Deduct health
        current_hp -= damage
        
    # 輸出最終位置 / Output final position
    print(position)

if __name__ == "__main__":
    main()
```

## 程式說明

1. **讀取輸入**：讀取初始生命值和兩個陷阱的參數
2. **初始化**：設置起始位置為0，當前生命值為k
3. **遊戲循環**：當生命值大於0時繼續遊戲
4. **移動機制**：每次向右跳躍當前生命值的距離
5. **陷阱檢測**：檢查新位置是否是x1或x2的倍數
6. **傷害計算**：根據陷阱規則計算傷害值
7. **生命值更新**：扣除相應的生命值
8. **遊戲結束條件**：當生命值≤0時結束循環
9. **輸出結果**：顯示最終位置

---

### 範例輸入 #1 
```
1
3 1
4 1
```

### 範例輸出 #1 
```
3
```

### 驗證過程 
- 初始生命值: 1
- 第一次跳躍: 0 → 1 (移動距離 = 1)
  - 位置1: 不是3或4的倍數，不扣血
  - 剩餘生命值: 1
- 第二次跳躍: 1 → 2 (移動距離 = 1)
  - 位置2: 不是3或4的倍數，不扣血
  - 剩餘生命值: 1
- 第三次跳躍: 2 → 3 (移動距離 = 1)
  - 位置3: 是3的倍數，扣1點生命值
  - 生命值: 1 - 1 = 0 (遊戲結束)
- 最終位置: 3 ✓

### 範例輸入 #2 
```
7
3 2
2 3
```

### 範例輸出 #2 
```
18
```

### 驗證過程 
- 初始生命值: 7
- 第一次跳躍: 0 → 7 (移動距離 = 7)
  - 位置7: 不是2或3的倍數，不扣血
  - 剩餘生命值: 7
- 第二次跳躍: 7 → 14 (移動距離 = 7)
  - 位置14: 是2的倍數，扣3點生命值
  - 剩餘生命值: 7 - 3 = 4
- 第三次跳躍: 14 → 18 (移動距離 = 4)
  - 位置18: 是2和3的倍數 (18÷2=9, 18÷3=6)
  - 扣血: 3 + 2 = 5點
  - 生命值: 4 - 5 = -1 (遊戲結束)
- 最終位置: 18 ✓
