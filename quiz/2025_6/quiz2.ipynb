### q837. 2. 轉盤得分 - 2025年6月APCS

#### 問題描述 

你有 m 個輪盤，每個輪盤上有 n 格，每格上寫有一個小寫英文字母（a ~ z）。一場遊戲有 k 個回合。

**遊戲規則：**
- 每個回合，每個輪盤會基於上一個回合結束的狀態做轉動
- 轉動可以是正數（順時針）、負數（逆時針）或 0（不轉）
- 當所有輪盤都轉動完畢後，觀察它們對齊位置上的字元
- 對於每一個位置（從上到下的每一格），統計出現次數最多的字元
- 將該字元的出現次數計入分數
- 每個回合的得分為這 n 格的對齊統計加總
- 最終總得分為所有回合的得分總和

**轉動規則：**
- 順時針轉動：字母向右移動
- 逆時針轉動：字母向左移動
- 轉動距離會取模 n 來處理循環

**輸入格式：**
- 第一行：m n k（輪盤數量、每輪盤格數、回合數）
- 接下來 m 行：每個輪盤的初始字母字串
- 接下來 k 行：每行 m 個整數，表示該回合每個輪盤的轉動距離

**輸入範圍：**
- m, n, k ≤ 30
- -100 ≤ 轉動距離 ≤ 100

---

#### Problem Description 

You have m roulette wheels, each with n slots containing lowercase English letters (a ~ z). The game has k rounds.

**Game Rules:**
- Each round, each wheel rotates based on its state from the previous round
- Rotation can be positive (clockwise), negative (counterclockwise), or 0 (no rotation)
- After all wheels have rotated, observe the aligned characters
- For each position (each slot from top to bottom), count the most frequently occurring character
- Add the count of that character to the score
- Each round's score is the sum of these counts across all n positions
- The final total score is the sum of all rounds' scores

**Rotation Rules:**
- Clockwise rotation: letters move to the right
- Counterclockwise rotation: letters move to the left
- Rotation distance is taken modulo n to handle circular movement

**Input Format:**
- First line: m n k (number of wheels, slots per wheel, number of rounds)
- Next m lines: initial letter strings for each wheel
- Next k lines: each containing m integers representing rotation distances for each wheel in that round

**Input Range:**
- m, n, k ≤ 30
- -100 ≤ rotation distance ≤ 100

---

#### 範例說明 

**初始輪盤:**
- 輪盤1: "apcsie"
- 輪盤2: "taiwan" 
- 輪盤3: "icpeda"

**轉動距離:** [1, 0, -4]

**轉動後結果:**
- 輪盤1: "eapcsi" ← "apcsie" 右轉1
- 輪盤2: "taiwan" ← 不動
- 輪盤3: "daicpe" ← "icpeda" 左轉4

**分數計算:**
- 第1格: (e, t, d) → 最多出現1次
- 第2格: (a, a, a) → 最多出現3次  
- 第3格: (p, i, i) → 最多出現2次
- 第4格: (c, w, c) → 最多出現2次
- 第5格: (s, a, p) → 最多出現1次
- 第6格: (i, n, e) → 最多出現1次

**該回合得分:** 1 + 3 + 2 + 2 + 1 + 1 = 10

## 解答

```python
def rotate_wheel(wheel, rotation, n):
    """
    轉動單個輪盤
    Rotate a single wheel
    
    Args:
        wheel: 輪盤的字母列表 / List of characters for the wheel
        rotation: 轉動距離 / Rotation distance
        n: 輪盤格數 / Number of slots
    
    Returns:
        轉動後的輪盤 / Rotated wheel
    """
    if rotation == 0:
        return wheel.copy()
    
    actual_rotation = rotation % n
    if actual_rotation == 0:
        return wheel.copy()
    
    if rotation > 0:
        # 順時針轉動 / Clockwise rotation
        return wheel[-actual_rotation:] + wheel[:-actual_rotation]
    else:
        # 逆時針轉動 / Counterclockwise rotation
        actual_rotation = -actual_rotation
        return wheel[actual_rotation:] + wheel[:actual_rotation]

def rotate_all_wheels(wheels, rotations, n):
    """
    轉動所有輪盤
    Rotate all wheels
    
    Args:
        wheels: 所有輪盤的列表 / List of all wheels
        rotations: 轉動距離列表 / List of rotation distances
        n: 輪盤格數 / Number of slots
    
    Returns:
        轉動後的輪盤列表 / List of rotated wheels
    """
    rotated_wheels = []
    for i in range(len(wheels)):
        rotated_wheel = rotate_wheel(wheels[i], rotations[i], n)
        rotated_wheels.append(rotated_wheel)
    return rotated_wheels

def calculate_round_score(wheels, n, m):
    """
    計算單回合得分
    Calculate score for a single round
    
    Args:
        wheels: 所有輪盤的列表 / List of all wheels
        n: 輪盤格數 / Number of slots
        m: 輪盤數量 / Number of wheels
    
    Returns:
        該回合得分 / Score for this round
    """
    round_score = 0
    for j in range(n):  # 對於每個位置 / For each position
        char_count = {}
        for i in range(m):  # 對於每個輪盤 / For each wheel
            char = wheels[i][j]
            char_count[char] = char_count.get(char, 0) + 1
        
        # 找出出現次數最多的字元 / Find the most frequent character
        max_count = max(char_count.values()) if char_count else 0
        round_score += max_count
    
    return round_score

def main():
    # 讀取第一行輸入 / Read first line input
    m, n, k = map(int, input().split())
    
    # 讀取輪盤初始狀態 / Read initial wheel states
    wheels = []
    for _ in range(m):
        wheel_str = input().strip()
        wheels.append(list(wheel_str))
    
    total_score = 0
    current_wheels = wheels.copy()
    
    # 處理每個回合 / Process each round
    for _ in range(k):
        # 讀取該回合的轉動距離 / Read rotation distances for this round
        rotations = list(map(int, input().split()))
        
        # 轉動所有輪盤 / Rotate all wheels
        current_wheels = rotate_all_wheels(current_wheels, rotations, n)
        
        # 計算該回合得分 / Calculate score for this round
        round_score = calculate_round_score(current_wheels, n, m)
        total_score += round_score
    
    # 輸出總得分 / Output total score
    print(total_score)

if __name__ == "__main__":
    main()
```

## 函式說明 / Function Descriptions

### 1. `rotate_wheel(wheel, rotation, n)`
**功能**：轉動單個輪盤
**參數**：
- `wheel`: 輪盤的字母列表
- `rotation`: 轉動距離（正數順時針，負數逆時針，0不轉）
- `n`: 輪盤格數
**回傳值**：轉動後的輪盤

### 2. `rotate_all_wheels(wheels, rotations, n)`
**功能**：轉動所有輪盤
**參數**：
- `wheels`: 所有輪盤的列表
- `rotations`: 轉動距離列表（每個元素對應一個輪盤）
- `n`: 輪盤格數
**回傳值**：轉動後的所有輪盤列表

### 3. `calculate_round_score(wheels, n, m)`
**功能**：計算單回合得分
**參數**：
- `wheels`: 所有輪盤的列表
- `n`: 輪盤格數
- `m`: 輪盤數量
**回傳值**：該回合的得分

### 4. `main()`
**功能**：主程式，協調所有功能
**流程**：
1. 讀取輸入參數
2. 初始化輪盤狀態
3. 處理每個回合：
   - 讀取轉動距離
   - 轉動所有輪盤
   - 計算回合得分
4. 累計總分並輸出

````python
m,n,k = [int(x) for x in input().split()]
texts = []
for _ in range(m): # m strings with length of n
    texts.append(input()) 
rounds = []
for _ in range(k):
    turns = [int(x) for x in input().split()]
    rounds.append(turns)
# print(m,n,k)
# print(texts)
# print(rounds)

# m,n,k = 4,3,3
# texts = ['abc','bab','cbc','abc']
# rounds = [[-1,-6,-6,-7], [5,-3,4,0],[-7,4,2,8]]

def turn(text, start, t): #正數（順時針）、負數（逆時針）或 0（不轉）
    global n
    return (start - t) % n
def calc(positions): # positions of m discs
    global n
    score = 0
    for j, _ in enumerate(range(n)):
        atoz = [0]*26
        for i, pos in enumerate(positions):
            #print(texts[i][(positions[i] + j)%n])
            atoz[ord(texts[i][(pos + j)%n])-ord('a')] +=1
        score += max(atoz)
    return score

total_score=0
positions=[0]*m
for turns in rounds:
    for i,t in enumerate(turns):
        positions[i] = turn(texts[i], positions[i], t)  
    score = calc(positions)
    total_score+=score
print(total_score)
````

### Another implementation

```python
# 讀取輸入參數：輪盤數量m，每個輪盤格數n，回合數k
m, n, k = [int(x) for x in input().split()]

# 讀取所有輪盤的初始文字
texts = []
for _ in range(m):  # m 個長度為 n 的字串
    texts.append(input()) 

# 讀取所有回合的轉動指令
rounds = []
for _ in range(k):
    turns = [int(x) for x in input().split()]
    rounds.append(turns)

# 轉動函式：計算轉動後的位置指針
def turn(text, start, t):  # 正數（順時針）、負數（逆時針）或 0（不轉）
    global n
    # 使用模運算處理循環轉動，計算新的起始位置
    return (start - t) % n

# 計算得分函式：根據當前所有輪盤的位置計算該回合得分
def calc(positions):  # positions of m discs
    global n
    score = 0
    
    # 遍歷每個位置（從上到下的每一格）
    for j, _ in enumerate(range(n)):
        # 初始化26個字母的計數器
        atoz = [0] * 26
        
        # 對於每個輪盤，統計在當前位置j的字元
        for i, pos in enumerate(positions):
            # 計算實際顯示的字元：使用模運算處理循環
            # (pos + j) % n 計算在位置j實際顯示的字元索引
            char_index = (pos + j) % n
            char = texts[i][char_index]
            
            # 對應字母計數加1
            atoz[ord(char) - ord('a')] += 1
        
        # 找出該位置出現最多次的字元次數，加入得分
        score += max(atoz)
    
    return score

# 主程式執行
total_score = 0
positions = [0] * m  # 初始化每個輪盤的起始位置指針（都從0開始）

# 處理每個回合
for turns in rounds:
    # 更新每個輪盤的位置指針（基於轉動指令）
    for i, t in enumerate(turns):
        positions[i] = turn(texts[i], positions[i], t)  
    
    # 計算當前回合得分
    score = calc(positions)
    total_score += score

# 輸出總得分
print(total_score)
```

## 主要步驟摘要 / Main Steps Summary

### 1. **輸入讀取階段**
- 讀取基本參數：輪盤數量(m)、格數(n)、回合數(k)
- 讀取所有輪盤的初始文字內容
- 讀取所有回合的轉動指令

### 2. **核心演算法設計**
- **位置指針法**：每個輪盤用一個位置指針表示當前起始點
- **模運算處理循環**：使用 `% n` 確保指針在有效範圍內
- **延遲計算**：不實際轉動字串，只在需要時計算顯示的字元

### 3. **轉動處理機制**
- `turn()` 函式：更新位置指針，負數表示逆時針轉動
- 公式：`(start - t) % n` 確保指針正確循環

### 4. **得分計算邏輯**
- 對於每個位置，統計所有輪盤在該位置顯示的字元
- 使用長度26的陣列統計a-z的出現次數
- 取最大值作為該位置的貢獻分數

### 5. **執行流程**
- 初始化所有輪盤位置指針為0
- 對於每個回合：更新指針 → 計算得分 → 累計總分
- 輸出最終總得分

## 方法特點 / Approach Characteristics

1. **空間效率**：不實際複製或轉動字串，只維護位置指針
2. **時間效率**：使用預先計算的位置指針，避免字串操作開銷
3. **數學優雅**：使用模運算優雅地處理循環轉動
4. **簡潔性**：程式碼相對簡短，邏輯清晰

這種方法通過數學計算而非實際字串操作來模擬轉動，在效能和記憶體使用上都有優勢。
