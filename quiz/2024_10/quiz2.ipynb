### o712. 2. 蒐集寶石 - 2024年10月APCS

#### 問題描述 

有一個 M×N 的地圖，每一格的數字紀錄著寶石的數量，如果數字是 -1 代表牆壁。有一位機器人一開始位於 (r, c) 的位置上且方向朝右邊，他遵循著以下規則行走：

1. 若機器人位於的格子內寶石數量為 0，則機器人程式終止。
2. 機器人維護著一個分數 score，將 score 加上當前格的寶石數量，並且撿起一顆寶石。
3. 若 score 是 k 的倍數，則向右轉 90 度。
4. 若機器人面向的格子是牆壁或是超出邊界，則繼續向右轉 90 度直到面向的格子非牆壁或非超出邊界，並回到第 1 步。

**範例說明**

例如機器人一開始在座標
（2，1）且k=2，向右走二步之後分数為3+2+3=8，由於8是2（k=2）的倍數所以向右轉90度。接下來往下走一步分數變為11，需要向右轉2次90度才不會面向牆壁或是邊界外的格子。
接下來向前走一步走到座標
（2，3），由於先前已經拿走一顆寶石，該位置的寶石數量變為2，因此分數變為13，再繼續往上走兩步到（0，3）處分數為16，由於16為2（k=2）的倍數所以向右轉90度。
向前走一格到（0，4）後需要向右轉兩次90度，回到（0，3）後由於寶石數量為0，機器人停止。過程中機器人總共撿了8顆寶石。

#### Problem Description 

There is an M×N map where each cell contains the number of gems, and -1 represents a wall. A robot starts at position (r, c) facing right, and follows these rules:

1. If the gem count in the current cell is 0, the robot terminates.
2. The robot maintains a score, adds the current cell's gem count to score, and picks up one gem.
3. If score is a multiple of k, turn right 90 degrees.
4. If the facing cell is a wall or out of bounds, continue turning right 90 degrees until facing a valid cell, then go back to step 1.

Input Format:
- First line: 5 integers M, N, k, r, c
- Next M lines: each with N numbers representing the map

Scoring:
- (60%): M = 1
- (40%): no restrictions

---

#### 解答

```python
def main():
    # 讀取第一行輸入
    M, N, k, r, c = map(int, input().split())
    
    # 讀取地圖
    grid = []
    for _ in range(M):
        row = list(map(int, input().split()))
        grid.append(row)
    
    # 方向：右(0), 下(1), 左(2), 上(3)
    directions = [(0, 1), (1, 0), (0, -1), (-1, 0)]
    dir_idx = 0  # 初始方向：右
    
    score = 0
    gems_collected = 0
    current_r, current_c = r, c
    
    # 模擬機器人移動
    while True:
        # 檢查當前格子寶石數量
        if grid[current_r][current_c] == 0:
            break
        
        # 撿起寶石並更新分數
        score += grid[current_r][current_c]
        gems_collected += 1
        grid[current_r][current_c] -= 1  # 減少一顆寶石
        
        # 檢查是否需要轉向（分數是k的倍數）
        if score % k == 0:
            dir_idx = (dir_idx + 1) % 4  # 向右轉90度
        
        # 嘗試移動
        next_r = current_r + directions[dir_idx][0]
        next_c = current_c + directions[dir_idx][1]
        
        # 檢查是否遇到牆壁或超出邊界
        turn_count = 0
        while (next_r < 0 or next_r >= M or next_c < 0 or next_c >= N or 
               grid[next_r][next_c] == -1):
            # 向右轉90度
            dir_idx = (dir_idx + 1) % 4
            next_r = current_r + directions[dir_idx][0]
            next_c = current_c + directions[dir_idx][1]
            turn_count += 1
            
            # 如果轉了4次都沒找到合法方向，表示被困住
            if turn_count >= 4:
                break
        
        # 如果被困住，終止程式
        if turn_count >= 4:
            break
        
        # 移動到新位置
        current_r, current_c = next_r, next_c
    
    # 輸出撿到的寶石總數
    print(gems_collected)

if __name__ == "__main__":
    main()
```

#### 測試範例 #1

**輸入：**
```
1 7 3 0 4
1 -1 2 1 2 1 0
```

**計算過程：**
- 地圖: [1, -1, 2, 1, 2, 1, 0]
- 起始位置: (0, 4)，寶石數量: 2
- 初始方向: 右(0)

**步驟：**
1. 位置(0,4): 寶石=2, score=2, gems=1
   - score=2 不是3的倍數，不轉向
   - 向右移動到(0,5): 寶石=1

2. 位置(0,5): 寶石=1, score=3, gems=2
   - score=3是3的倍數，向右轉（方向變下）
   - 向下移動：超出邊界，轉向右（方向變左）
   - 向左移動到(0,4): 寶石=1

3. 位置(0,4): 寶石=1, score=4, gems=3
   - score=4不是3的倍數，不轉向
   - 向左移動到(0,3): 寶石=1

4. 位置(0,3): 寶石=1, score=5, gems=4
   - score=5不是3的倍數，不轉向
   - 向左移動到(0,2): 寶石=2

5. 位置(0,2): 寶石=2, score=7, gems=5
   - score=7不是3的倍數，不轉向
   - 向左移動到(0,1): 牆壁(-1)，轉向
   - 轉向下→右→上，最終向右移動到(0,3): 寶石=0

6. 位置(0,3): 寶石=0，終止
- **輸出：5 ✓**

#### 測試範例 #2

**輸入：**
```
4 5 4 2 1
2 0 1 1 1
2 -1 0 2 -1 
0 3 2 3 0
1 1 -1 3 1
```

**計算過程：**
- 複雜的路徑模擬，最終收集8顆寶石
- **輸出：8 ✓**

````python
M, N, k, r, c = map(int, input().split())
# grids = []
# for _ in range(M):
#     grids.append(list(map(int, input().split())))
grids = [list(map(int, input().split())) for _ in range(M)]

# M, N, k, r, c = 1, 7, 3, 0, 4
# grids = [[1, -1, 2, 1, 2, 1, 0]]

# M, N, k, r, c = 4, 5, 4, 2, 1
# grids = [[2, 0, 1, 1, 1],[2, -1, 0, 2, -1],[0, 3, 2, 3, 0],[1, 1, -1, 3, 1]]

right = (0, 1)
down = (1, 0)
left = (0, -1)
up = (-1, 0)
directions =(right, down, left, up)
def gen():
    while True:
        for d in directions:
            yield d
direction_generator = gen()
def get_next_direction():
    return next(direction_generator)

def check_next_move(r,c,direction):
    next_r, next_c = r+direction[0], c+direction[1]
    if next_r==M or next_r==-1 or next_c==N or next_c==-1:
        return False
    elif grids[next_r][next_c] == -1:
        return False
    else:
        return True
        
score = 0
stone_count = 0 
direction  = get_next_direction()
while grids[r][c]!=0:
    score += grids[r][c]
    grids[r][c]-=1
    stone_count+=1
    
    if score%k==0 :
        direction = get_next_direction()
    while not check_next_move(r,c,direction):
        direction=get_next_direction()
    r,c = r+direction[0], c+direction[1]
print(stone_count)   
        
````
