#### 問題描述 

有一個杯子，可將其體積視為由兩個長方體組成：
- 下方的長方體：底面積為 w1 × w1 cm²，高為 h1 cm
- 上方的長方體：底面積為 w2 × w2 cm²，高為 h2 cm

一開始杯子為空。要裝 n 次飲料，每一次裝 v cm³ 容量的飲料，當水杯滿時水位不再上升。問這 n 次倒飲料中水位上升變化量最高是幾 cm。

**輸入說明：**
- 第一行有一個正整數 n (1 ≤ n ≤ 10) 代表要裝 n 次飲料
- 接下來一行有 4 個正整數 w1, w2, h1, h2 (1 ≤ w1, w2, h1, h2 ≤ 50) 代表杯子的寬度與高度
- 最後一行有 n 個正整數代表每次裝飲料的容量

**保證每次水位上升都是整數**

**評分標準：**
- (60%): n = 1，即答案為倒水後的高度
- (40%): 無限制

#### Problem Description

There is a cup that can be viewed as consisting of two cuboids:
- Lower cuboid: base area w1 × w1 cm², height h1 cm
- Upper cuboid: base area w2 × w2 cm², height h2 cm

The cup is initially empty. Pour n times of drinks, each time with volume v cm³. When the cup is full, the water level stops rising. Find the maximum water level change during these n pouring operations.

**Input Format:**
- First line: positive integer n (1 ≤ n ≤ 10)
- Second line: 4 positive integers w1, w2, h1, h2 (1 ≤ w1, w2, h1, h2 ≤ 50)
- Third line: n positive integers representing drink volumes

**Guaranteed that water level changes are integers**

**Scoring:**
- (60%): n = 1, answer is the height after pouring
- (40%): no restrictions

---

#### 解答

```python
def calculate_water_level(w1, w2, h1, h2, volumes):
    """
    計算倒水過程中的最高水位變化量
    Calculate maximum water level change during pouring
    
    Args:
        w1, w2, h1, h2: 杯子參數 / Cup parameters
        volumes: 每次倒水的容量列表 / List of volumes for each pour
    
    Returns:
        最高水位變化量 (cm) / Maximum water level change (cm)
    """
    # 計算杯子的總容量和分段容量
    lower_volume = w1 * w1 * h1  # 下方長方體容量
    upper_volume = w2 * w2 * h2  # 上方長方體容量
    total_volume = lower_volume + upper_volume  # 杯子總容量
    
    # 計算水位高度與體積的關係
    def volume_to_height(volume):
        """
        將體積轉換為水位高度
        Convert volume to water level height
        """
        if volume <= lower_volume:
            # 水只在下半部
            return volume / (w1 * w1)
        else:
            # 水已經進入上半部
            remaining_volume = volume - lower_volume
            return h1 + remaining_volume / (w2 * w2)
    
    current_volume = 0
    max_level_change = 0
    previous_height = 0
    
    for v in volumes:
        # 倒水
        current_volume += v
        current_volume = min(current_volume, total_volume)  # 不能超過總容量
        
        # 計算當前水位高度
        current_height = volume_to_height(current_volume)
        
        # 計算這次倒水造成的水位變化
        level_change = current_height - previous_height
        max_level_change = max(max_level_change, level_change)
        
        previous_height = current_height
    
    return round(max_level_change)

def main():
    # 讀取倒水次數
    n = int(input().strip())
    
    # 讀取杯子參數
    w1, w2, h1, h2 = map(int, input().split())
    
    # 讀取每次倒水的容量
    volumes = list(map(int, input().split()))
    
    # 計算最高水位變化量
    result = calculate_water_level(w1, w2, h1, h2, volumes)
    
    # 輸出結果
    print(result)

if __name__ == "__main__":
    main()
```

#### 測試範例 #1

**輸入：**
```
1
4 6 8 5
200
```

**計算過程：**
- 下方容量: 4×4×8 = 128 cm³
- 上方容量: 6×6×5 = 180 cm³
- 總容量: 128 + 180 = 308 cm³
- 倒水 200 cm³ > 128 cm³，所以水會進入上半部
- 水位高度: 8 + (200-128)/36 = 8 + 72/36 = 8 + 2 = 10 cm
- **輸出：10 ✓**

#### 測試範例 #2

**輸入：**
```
2
5 10 12 8
400 600
```

**計算過程：**
- 下方容量: 5×5×12 = 300 cm³
- 上方容量: 10×10×8 = 800 cm³
- 總容量: 300 + 800 = 1100 cm³

**第一次倒水 400 cm³:**
- 400 > 300，進入上半部
- 水位高度: 12 + (400-300)/100 = 12 + 1 = 13 cm
- 水位變化: 13 cm

**第二次倒水 600 cm³:**
- 總體積: 400 + 600 = 1000 cm³
- 水位高度: 12 + (1000-300)/100 = 12 + 7 = 19 cm
- 水位變化: 19 - 13 = 6 cm

**最高水位變化: max(13, 6) = 13 cm**
- **輸出：13 ✓**

#### 測試範例 #3

**輸入：**
```
5
16 44 28 17
2560 1280 1536 1024 10448
```

**計算過程：**
- 下方容量: 16×16×28 = 7168 cm³
- 上方容量: 44×44×17 = 32912 cm³
- 總容量: 7168 + 32912 = 40080 cm³

**逐步計算水位變化，找出最大值為 10 cm**
- **輸出：10 ✓**

````python
n = int(input())
w1,w2,h1,h2 = map(int, input().split())
vs = list(map(int, input().split()))

# n = 1
# w1, w2, h1, h2 = 4, 6, 8, 5
# vs = [200]

# n = 2
# w1, w2, h1, h2 = 5, 10, 12, 8
# vs=[400, 600]

# n = 5
# w1, w2, h1, h2 = 16, 44, 28, 17
# vs= [2560, 1280, 1536, 1024, 10448]

lowerv = w1*w1*h1
upperv = w2*w2*h2
hs = []
vsum = 0
prevh = 0
for v in vs:
    vsum+=v
    if vsum<=lowerv:
        h = vsum//(w1*w1)
    elif vsum<=(lowerv+upperv):
        h = (vsum-lowerv)//(w2*w2)+h1
    else:
        h = h1+h2
    hs.append(h-prevh)
    prevh = h
print(max(hs))    
````
