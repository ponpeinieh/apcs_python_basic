### 練習題 1 ：攝氏溫度轉換華氏溫度程式

**描述**：編寫一個程式，將攝氏溫度（C）轉換為華氏溫度（F）。使用Lambda函式來進行轉換，並利用`map()`函數將一組攝氏溫度轉換為華氏溫度。

**攝氏轉華氏的公式**：
$$
F = \frac{9}{5}C + 32
$$

**要求**：
1. 定義一個Lambda函式，接受攝氏溫度並返回對應的華氏溫度。
2. 使用`map()`函數將一組攝氏溫度的清單轉換為華氏溫度的清單。
3. 提供一組測試的攝氏溫度數據，並顯示轉換後的華氏溫度清單。

**示例攝氏溫度數據**：
```python
celsius_temps = [0, 20, 37, 100]
```

**範例程式碼**：
```python
# 定義 Lambda 函式
c_to_f = lambda c: (9/5) * c + 32

# 使用 map() 進行轉換
fahrenheit_temps = list(map(c_to_f, celsius_temps))

# 顯示結果
print(fahrenheit_temps)
```

**範例輸出**：
```plaintext
[32.0, 68.0, 98.6, 212.0]
```

這段程式碼將攝氏溫度 `[0, 20, 37, 100]` 轉換為華氏溫度 `[32.0, 68.0, 98.6, 212.0]`。

### 練習題 2 ：計算圓的面積和周長

**描述**：編寫一個程式，計算給定半徑的圓的面積和周長。使用Lambda函式來進行計算，並利用`map()`函數將一組半徑轉換為對應的面積和周長。

**圓的面積和周長公式**：
- 面積（A）： 
  $$
  A = \pi r^2
  $$
- 周長（C）：
  $$
  C = 2\pi r
  $$

**要求**：
1. 定義兩個Lambda函式，一個用於計算圓的面積，另一個用於計算圓的周長。
2. 使用`map()`函數將一組半徑的清單轉換為對應的面積和周長的清單。
3. 提供一組測試的半徑數據，並顯示轉換後的面積和周長清單。

**示例半徑數據**：
```python
radii = [1, 2, 3, 4, 5]
```

**範例程式碼**：
```python
import math

# 定義 Lambda 函式
calculate_area = lambda r: math.pi * r**2
calculate_circumference = lambda r: 2 * math.pi * r

# 使用 map() 進行計算
areas = list(map(calculate_area, radii))
circumferences = list(map(calculate_circumference, radii))

# 顯示結果
print("Areas:", areas)
print("Circumferences:", circumferences)
```

**範例輸出**：
```plaintext
Areas: [3.141592653589793, 12.566370614359172, 28.274333882308138, 50.26548245743669, 78.53981633974483]
Circumferences: [6.283185307179586, 12.566370614359172, 18.84955592153876, 25.132741228718345, 31.41592653589793]
```

### 練習題 3 ：學生組隊系統

**描述**：編寫一個程式來組建學生團隊。每個團隊由一名男學生和一名女學生組成，目標是將最高分的男學生與最低分的女學生配對，反之亦然。這樣的組隊方式可以促進學習上的互補，讓學生們互相幫助。

**要求**：
1. 定義包含 10 名學生的清單，其中包括 5 名男學生和 5 名女學生，每位學生都有姓名、平均成績和性別屬性。
2. 使用 `filter()` 函式分別篩選出男學生和女學生。
3. 使用 `sorted()` 函式對男學生按成績降序排序，對女學生按成績升序排序。
4. 形成 5 組，每組包含一名男學生和一名女學生，並顯示每組的學生姓名。

**示例學生資料**：
```python
students = [
    {"name": "Alice", "average_score": 78, "gender": "female"},
    {"name": "Bob", "average_score": 92, "gender": "male"},
    {"name": "Charlie", "average_score": 85, "gender": "male"},
    {"name": "David", "average_score": 65, "gender": "male"},
    {"name": "Eve", "average_score": 88, "gender": "female"},
    {"name": "Frank", "average_score": 95, "gender": "male"},
    {"name": "Grace", "average_score": 70, "gender": "female"},
    {"name": "Hank", "average_score": 60, "gender": "male"},
    {"name": "Ivy", "average_score": 90, "gender": "female"},
    {"name": "Jill", "average_score": 75, "gender": "female"}
]
```
**範例輸出**：
```plaintext 
Team 1: Frank and Grace
Team 2: Bob and Jill
Team 3: Charlie and Alice
Team 4: David and Eve
Team 5: Hank and Ivy
```

**範例程式碼**：
```python
# 定義學生資料清單
students = [
    {"name": "Alice", "average_score": 78, "gender": "female"},
    {"name": "Bob", "average_score": 92, "gender": "male"},
    {"name": "Charlie", "average_score": 85, "gender": "male"},
    {"name": "David", "average_score": 65, "gender": "male"},
    {"name": "Eve", "average_score": 88, "gender": "female"},
    {"name": "Frank", "average_score": 95, "gender": "male"},
    {"name": "Grace", "average_score": 70, "gender": "female"},
    {"name": "Hank", "average_score": 60, "gender": "male"},
    {"name": "Ivy", "average_score": 90, "gender": "female"},
    {"name": "Jill", "average_score": 75, "gender": "female"}
]

# 使用 filter() 分別篩選出男學生和女學生
male_students = list(filter(lambda student: student["gender"] == "male", students))
female_students = list(filter(lambda student: student["gender"] == "female", students))

# 使用 sorted() 對男學生按成績降序排序，對女學生按成績升序排序
sorted_males = sorted(male_students, key=lambda student: student["average_score"], reverse=True)
sorted_females = sorted(female_students, key=lambda student: student["average_score"])

# 形成 5 組
teams = []
for i in range(5):
    teams.append((sorted_males[i], sorted_females[i]))

# 顯示每組的學生姓名
for id, team in enumerate(teams):
    male, female = team
    print(f"Team {id+1}: {male['name']} and {female['name']}")
```

**描述**：編寫一個程式來檢查台灣身份證號碼的有效性。身份證號碼遵循特定的檢查規則，規則如下:

**檢查規則**：

1. **英文代號轉換為數字**：

   |  英文代號  |  數字  |  地區  |
   |:----------:|:------:|:------:|
   |     A     |   10   | 台北市 |
   |     B     |   11   | 台中市 |
   |     C     |   12   | 基隆市 |
   |     D     |   13   | 台南市 |
   |     E     |   14   | 高雄市 |
   |     F     |   15   | 台北縣 |
   |     G     |   16   | 宜蘭縣 |
   |     H     |   17   | 桃園縣 |
   |     I     |   34   | 嘉義市 |
   |     J     |   18   | 新竹縣 |
   |     K     |   19   | 苗栗縣 |
   |     M     |   21   | 南投縣 |
   |     N     |   22   | 彰化縣 |
   |     O     |   35   | 新竹市 |
   |     P     |   23   | 雲林縣 |
   |     Q     |   24   | 嘉義縣 |
   |     T     |   27   | 屏東縣 |
   |     U     |   28   | 花蓮縣 |
   |     V     |   29   | 台東縣 |
   |     W     |   32   | 金門縣 |
   |     X     |   30   | 澎湖縣 |
   |     Z     |   33   | 連江縣 |

2. **數字處理**：將英文轉換為的數字，個位數乘以 9，再加上十位數的數字。

3. **乘法計算**：除了最後一個數字為檢查碼，其餘各數字從右到左依次乘以 1、2、3、4、...、8，然後計算總和。

4. **求和**: 將第 (2) 步驟與第 (3) 步驟的結果相加，再加上最後一碼（檢查碼）。

5. **有效性檢查**：將 (4) 的結果除以 10，若能夠整除則為正確的編碼，否則為錯誤的編碼。

**例子**：身份證號碼 T112663836

計算過程：
```
2 + 7*9 + 1*8 + 1*7 + 2*6 + 6*5 + 6*4 + 3*3 + 8*2 + 3*1 + 6(檢查碼) = 180
```

結果：
```
180 除以 10 整除，因此為正確編碼。
```

**要求**： 

1. 確保身份證號碼長度為 10，第一個字元為字母，接下來的字元為數字。
   - 如何檢查一個字元是否為字母或數字? 可以使用 `str.isalpha()` 和 `str.isdigit()`。
2. 將第一個字母轉換為對應的數字。
3. 根據上述的規則計算總和，確保總和能被 10 整除。

**示例身份證號碼**：
```plaintext
A123456789   
B234567890   
C345678901   
D456789012   
Z987654321   
```

**範例輸出**：
```plaintext 
A123456789 是有效的身份證號碼。
B234567890 不是有效的身份證號碼。
C345678901 不是有效的身份證號碼。
D456789012 不是有效的身份證號碼。
Z987654321 不是有效的身份證號碼。
```

**範例程式碼**：
```python
# 定義英文代號轉換為數字的字典
letter_to_num = {
    'A': 10, 'B': 11, 'C': 12, 'D': 13, 'E': 14, 'F': 15, 'G': 16, 'H': 17,
    'I': 34, 'J': 18, 'K': 19, 'M': 21, 'N': 22, 'O': 35, 'P': 23, 'Q': 24,
    'T': 27, 'U': 28, 'V': 29, 'W': 32, 'X': 30, 'Z': 33
}

def check_taiwan_id(id_number):
    # 檢查長度是否為 10，且第一個字元是字母，其餘是數字
    if len(id_number) != 10 or not id_number[0].isalpha() or not id_number[1:].isdigit():
        return False
    
    # 將第一個字母轉換為數字
    first_char = id_number[0].upper()
    if first_char not in letter_to_num:
        return False
    
    # 取得對應的數字
    num = letter_to_num[first_char]
    
    # 計算第 (2) 步驟：個位數乘以 9，加上十位數
    step2 = (num // 10) + (num % 10) * 9
    
    # 計算第 (3) 步驟：其餘數字從右到左乘以 1, 2, 3, ..., 8
    step3 = 0
    for i in range(1, 9):
        step3 += int(id_number[i]) * (9 - i)
    
    # 計算總和：step2 + step3 + 檢查碼
    total = step2 + step3 + int(id_number[-1])
    
    # 檢查是否能被 10 整除
    return total % 10 == 0

# 測試身份證號碼
id_numbers = [
    "A123456789",
    "B234567890",
    "C345678901",
    "D456789012",
    "Z987654321"
]

# 檢查並輸出結果
for id_num in id_numbers:
    if check_taiwan_id(id_num):
        print(f"{id_num} 是有效的身份證號碼。")
    else:
        print(f"{id_num} 不是有效的身份證號碼。")
```
