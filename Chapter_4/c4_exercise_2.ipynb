### 練習題 1 ：用戶資料驗證系統

**描述**：編寫一個函數，接受用戶資料的字典清單，並對每個用戶進行驗證。用戶資料包括姓名、年齡、驗證狀態和信用。根據年齡的不同，更新用戶的驗證狀態和信用。

**用戶資料格式**：
```python
{
    "name": "用戶姓名",
    "age": 年齡 (整數),
    "validated": 驗證狀態 (布林值，初始為 False),
    "credit": 信用 (整數，初始為 0)
}
```

**要求**：
1. 定義一個函數 `validate_users`，接受用戶資料字典的清單作為參數。
2. 對清單中的每個用戶進行驗證：
   - 如果年齡大於或等於 18，則將 `validated` 設置為 `True`，並將 `credit` 增加 10。
3. 返回更新後的用戶資料字典的清單。
4. 提供 4 個用戶測試數據，並使用函數進行驗證，顯示每個用戶的驗證結果。

**示例用戶資料**：
```python
users = [
    {"name": "Alice", "age": 17, "validated": False, "credit": 0},
    {"name": "Bob", "age": 20, "validated": False, "credit": 0},
    {"name": "Charlie", "age": 19, "validated": False, "credit": 0},
    {"name": "David", "age": 15, "validated": False, "credit": 0}
]
```

**範例輸出**：
```plaintext
User: Alice, Validation Status: False, Credit: 0
User: Bob, Validation Status: True, Credit: 10
User: Charlie, Validation Status: True, Credit: 10
User: David, Validation Status: False, Credit: 0
```

### 練習題 3 ：寵物資訊管理系統

**描述**：編寫一個函數，使用一個字典來儲存一個寵物的資訊，並回傳這個字典。該函數需要接受寵物的種類和名字，並可以接受其他相關資訊作為名稱引數。

**寵物資訊格式**：
```python
{
    "species": "寵物的種類",
    "name": "寵物的名字",
    # 其他資訊，例如：
    "age": 年齡 (整數),
    "color": 顏色 (字串),
    "breed": 品種 (字串)
}
```

**要求**：
1. 定義一個函數 `make_pet()`，接受必須傳入的位置引數（`species` 和 `name`），以及任意數量的名稱引數。
2. 使用字典來儲存寵物的資訊，並將其他名稱引數添加到字典中。
3. 呼叫該函數，創建至少三個寵物的字典，並顯示這些資訊。

**範例輸出**：
```plaintext
{'species': 'Dog', 'name': 'Buddy', 'age': 5, 'color': 'Brown'}
{'species': 'Cat', 'name': 'Whiskers', 'age': 3, 'color': 'Black', 'breed': 'Siamese'}
{'species': 'Hamster', 'name': 'Nibbles', 'color': 'Golden'}
```
