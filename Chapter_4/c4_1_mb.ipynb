### 練習一：使用函數顯示心形

#### 描述
在這個練習中，您將創建一個簡單的程式，持續在 micro:bit 上顯示心形。心形將顯示一段時間，然後清除，這個過程將無限重複。

#### 要求
- 使用函數來組織您的代碼。
- 創建一個函數來顯示心形並等待特定的時間。
- 創建一個主循環函數來不斷調用顯示心形的函數。

#### 示例程式碼

```python
from microbit import *

def heart_wait():
    """顯示心形圖案，等待一段時間，然後清除顯示。"""
    display.show(Image.HEART)
    sleep(1000)
    display.clear()
    sleep(1000)

def main_loop():
    """主循環，不斷顯示心形圖案。"""
    while True:
        heart_wait()

# 啟動程式
main_loop()
```

#### 程式說明：
- **`heart_wait()`**: 顯示心形圖案，等待一段時間，然後清除顯示。
- **`main_loop()`**: 無限循環，不斷調用 `heart_wait()` 來顯示心形。 

### 練習二：使用參數顯示心形

#### 描述
在這個練習中，您將擴展之前的程式，創建一個可以接受參數的函數。這個函數將根據傳入的時間顯示心形，然後清除顯示。

#### 要求
- 修改 `heart_wait()` 函數，使其接受一個整數參數，表示顯示心形的時間（以毫秒為單位）。
- 在主循環中調用這個函數，並傳入不同的顯示時間。

#### 示例程式碼

```python
from microbit import *

def heart_wait(display_time):
    """顯示心形圖案，等待指定的時間，然後清除顯示。"""
    display.show(Image.HEART)
    sleep(display_time)
    display.clear()
    sleep(1000)

def main_loop():
    """主循環，不斷顯示心形圖案，顯示時間為不同的值。"""
    while True:
        heart_wait(1000)  # 顯示1秒
        heart_wait(500)   # 顯示0.5秒
        heart_wait(1500)  # 顯示1.5秒

# 啟動程式
main_loop()
```

#### 程式說明：
- **`heart_wait(display_time)`**: 根據傳入的時間參數顯示心形圖案，然後清除顯示。
- **`main_loop()`**: 無限循環，不斷調用 `heart_wait()`，傳入不同的顯示時間。
