### 練習題 1 ：計算單字頻率

**描述**：編寫一個函數，接受一段英文文字，並計算每個英文單字的出現頻率。請使用以下示例文字作為輸入：

**示例文字**：
```
It was the best of times, it was the worst of times, it was the age of wisdom, it was the age of foolishness, it was the epoch of belief, it was the epoch of incredulity, it was the season of Light, it was the season of Darkness, it was the spring of hope, it was the winter of despair, we had everything before us, we were all going direct to Heaven, we were all going direct the other way—in short, the period was so far like the present period, that some of its noisiest authorities insisted on its being received, for good or for evil, in the superlative degree of comparison only.
```

**要求**：
1. 將文字轉換為小寫(亦即單字不考慮大小寫的差別)，並忽略標點符號(見以下備注:如何去除掉標點符號)。
2. 使用 `Counter` 來計算每個單字的頻率。
3. 輸出頻率最高的前 N 個單字及其出現次數，N 由用戶輸入。

**範例**：
- 輸入 N: `3`
- 輸出 `('the', 14), ('of', 12), ('was', 11)`

**備注**: 如何去除掉標點符號 - 使用 str.maketrans(x[, y[, z]]) + str.translate(table) 

- static str.maketrans(x[, y[, z]])

此靜態方法返回一個可以用於 str.translate() 的翻譯表(Translation Table)。

如果只有一個參數，它必須是一個字典，將 Unicode 編碼（整數）或字元（長度為 1 的字串）映射到 Unicode 編碼、字串（任意長度）或 None。字元鍵將被轉換為編碼。

如果有兩個參數，它們必須是等長的字串，並且在生成的字典中，x 中的每個字元將映射到 y 中相同位置的字元。

如果有第三個參數，它必須是一個字串，其字元將在結果中映射到 None。

- str.translate(table)

返回一個字串的副本，其中每個字元都通過給定的翻譯表進行映射。您可以使用 `str.maketrans()` 來產生翻譯表。

- 範例程式

```python
text='It was the best of times, it was the worst of times, it was the age of wisdom,'
translator = str.maketrans('', '', string.punctuation)
text = text.translate(translator)
print(text)
```
- **str.maketrans()方法參數**:
   - **第一個參數 (`''`)**: 這是要替換的字串。在這裡，它是一個空字串，表示我們不想用其他字元替換任何字元。
   - **第二個參數 (`''`)**: 這也是一個空字串，表示沒有對應的字元來替換第一個參數中的字元。
   - **第三個參數 (`string.punctuation`)**: 這是一個包含所有標點符號的字串。通過指定這個字串，我們告訴 `maketrans` 方法將所有標點符號映射到 `None`。這意味著，在翻譯過程中，**所有標點符號都將被移除**。

### 練習題 3：基本計算器

**描述**：
編寫一個基本計算器函數，該函數可以對兩個數字進行加法、減法、乘法或除法運算。

#### 參數說明：
1. **num1** (數字)：第一個操作數，可以使用位置或名稱參數傳遞。
2. **num2** (數字)：第二個操作數，可以使用位置或名稱參數傳遞。
3. **operation** (字串，可選)：指定要執行的運算類型，預設值為 `'add'`。可接受的值包括：
   - `'add'`：加法
   - `'subtract'`：減法
   - `'multiply'`：乘法
   - `'divide'`：除法

#### 要求：

需要完成以下幾個任務：
1. **使用位置參數**：首先僅使用位置參數呼叫函數，例如 `calculator(5, 3)`。
2. **使用名稱參數**：僅使用名稱參數呼叫函數，例如 `calculator(num1=5, num2=3)`。
3. **混合使用位置參數和名稱參數**：同時使用位置和名稱參數呼叫函數，例如 `calculator(5, 3, operation='multiply')`。
4. **操作參數的選擇**：
   - 在呼叫函數時，可以選擇不傳遞 `operation` 參數，使用預設值。
   - 也可以在呼叫時顯式傳遞 `operation` 參數。
5. 如果傳遞了無效的'operation'，應返回錯誤消息：`"Error: Invalid operation."`
