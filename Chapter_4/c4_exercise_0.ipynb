### 練習題 1 ：增強折扣應用

**描述**：編寫一個函數 `apply_discount`，接受價格列表和折扣列表，對每個價格依次應用折扣，返回最終折扣後的價格列表。

**價格格式**：
```python
prices = [價格1, 價格2, 價格3, ...]
```

**折扣格式**：
```python
discounts = [折扣1, 折扣2, 折扣3, ...]  # 每個折扣為百分比
```

**要求**：
1. 定義函數 `apply_discount(prices, discounts)`，接受兩個參數：價格列表和折扣列表。
2. 對於每個價格，依次應用提供的折扣。
3. 返回最終折扣後的價格列表，並確保原始價格列表不受影響。
4. 提供測試數據，並使用函數顯示折扣後的價格。

**示例價格和折扣**：
```python
my_original_data = [100, 90, 200]
discounts = [10, 5]  # 先應用 10% 然後 5%
```

**範例輸出**：
```plaintext
Final Prices: [85.5, 76.5, 171.0]
```

### 練習題 2 ：價格增長模擬

**價格格式**：
```python
prices = [價格1, 價格2, 價格3, ...]
```

**要求**：
1. 定義函數 `simulate_price_increase(prices, annual_increase, years)`，接受三個參數：

   - `annual_increase`：每年的增長百分比（例如，5 表示每年增長 5%）。
   - `years`：模擬的年數。
2. 計算每個價格在指定年數後的最終價格，並返回一個新的價格列表。
3. 確保原始價格列表保持不變。
4. 提供測試數據，並使用函數顯示經過指定年數後的價格。

**示例價格和年增長**：
```python
my_original_data = [100, 90, 200]
annual_increase = 5  # 每年 5% 增加
years = 3
```

**範例輸出**：
```plaintext
Final Prices After 3 Years: [115.76, 104.54, 231.84]
```
