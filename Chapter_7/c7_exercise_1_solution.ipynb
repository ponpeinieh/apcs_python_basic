### 作業 1: 基本模組導入

```python
import math

def calculate_circle_area():
    try:
        radius = float(input("請輸入圓的半徑："))
        area = math.pi * (radius ** 2)
        print(f"圓的面積是：{area}")
    except ValueError:
        print("請輸入有效的數字！")

calculate_circle_area()
```

### 作業 2: 使用除法的例外處理

```python
def divide_numbers():
    try:
        num1 = float(input("請輸入第一個數字："))
        num2 = float(input("請輸入第二個數字："))
        result = num1 / num2
        print(f"結果是：{result}")
    except ValueError:
        print("請輸入有效的數字！")
    except ZeroDivisionError:
        print("不能除以零！")

divide_numbers()
```

### 作業 3: 自定義例外

```python
class NegativeNumberError(Exception):
    pass

def check_positive_number(number):
    if number < 0:
        raise NegativeNumberError("數字不能是負數！")

def main():
    try:
        number = float(input("請輸入一個數字："))
        check_positive_number(number)
        print("您輸入的數字是正數。")
    except ValueError:
        print("請輸入有效的數字！")
    except NegativeNumberError as e:
        print(e)

main()
```

### 作業 4: 隨機數生成

```python
import random

def guess_number():
    target_number = random.randint(1, 100)
    while True:
        try:
            guess = int(input("猜一個介於 1 到 100 之間的數字："))
            if guess < target_number:
                print("太低了！")
            elif guess > target_number:
                print("太高了！")
            else:
                print("恭喜你，猜對了！")
                break
        except ValueError:
            print("請輸入有效的整數！")

guess_number()
```
