### 作業 1: 基本模組導入

**任務：**
創建一個 Python 腳本，導入 `math` 模組並使用它計算圓的面積。腳本應提示用戶輸入半徑，並處理任何潛在的錯誤（例如，非數值輸入）。

**要求：**
- 導入 `math` 模組。
- 使用公式 \( \text{Area} = \pi \times r^2 \) 計算面積。
- 處理 `ValueError` 以應對無效輸入。

### 作業 2: 使用除法的例外處理

**任務：**
編寫一個程序，提示用戶輸入兩個數字並執行除法。使用例外處理來管理用戶輸入非數值或嘗試除以零的情況。

**要求：**
- 提示用戶輸入兩個數字。
- 執行除法並顯示結果。
- 處理 `ValueError` 和 `ZeroDivisionError`。

### 作業 3: 自定義例外

**任務：**
創建一個名為 `NegativeNumberError` 的自定義例外。編寫一個函數，接受一個數字作為輸入，如果該數字為負數，則引發此例外。在主程序中包含例外處理，捕獲並顯示友好的消息。

**要求：**
- 定義自定義例外類。
- 實現引發例外的函數。
- 在主程序中處理該例外。

### 作業 4: 隨機數生成

**任務：**
編寫一個程序，使用 `random` 模組生成一個介於 1 到 100 之間的隨機整數。提示用戶猜測該數字，並提供有關猜測是否太高、太低或正確的反饋。包括對非整數輸入的例外處理。

**要求：**
- 導入 `random` 模組。
- 生成隨機整數。
- 提示用戶進行猜測並提供反饋。
- 處理 `ValueError` 以應對無效輸入。
