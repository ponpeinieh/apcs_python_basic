### **Exercise 1: Sum of Even Numbers**  
**Task:** Write a Python program that calculates the sum of all even numbers from 1 to a given number `n` (inclusive).  

**Example:**  
- If `n = 10`, the output should be `30` (because 2 + 4 + 6 + 8 + 10 = 30).  

**Skills Tested:**  
- Loops (`for` or `while`)  
- Conditionals (`if` statements)  
- Basic arithmetic operations  

---

### **Exercise 2: Check Palindrome**  
**Task:** Write a Python function that checks if a given string is a palindrome (reads the same backward as forward, ignoring case).  

**Example:**  
- `"Madam"` → `True`  
- `"Python"` → `False`  

**Skills Tested:**  
- String manipulation (e.g., `str.lower()`, slicing `[::-1]`)  
- Functions  
- Boolean logic  

---

### **Exercise 3: Find the Largest Number in a List**  
**Task:** Write a Python function that takes a list of numbers and returns the largest number without using built-in functions like `max()`.  

**Example:**  
- Input: `[4, 9, 2, 7]` → Output: `9`  

**Skills Tested:**  
- Lists and iteration  
- Comparison operators  
- Basic function definition

### **練習 1：計算偶數的和**  
**任務：** 寫一個 Python 程式，計算從 1 到給定數字 `n`（包含 `n`）之間所有偶數的總和。  

**範例：**  
- 輸入 `n = 10`，輸出應為 `30`（因為 2 + 4 + 6 + 8 + 10 = 30）。  

**考察知識點：**  
- 迴圈（`for` 或 `while`）  
- 條件判斷（`if` 語句）  
- 基本算術運算

### **練習 2：檢查回文字串**  
**任務：** 寫一個 Python 函數，判斷給定的字串是否是回文（正讀反讀都一樣，忽略大小寫）。  

**範例：**  
- `"Madam"` → `True`  
- `"Python"` → `False`  

**考察知識點：**  
- 字串操作（如 `str.lower()`、切片 `[::-1]`）  
- 函數定義  
- 布林邏輯

### **練習 3：找出列表中的最大數**  
**任務：** 寫一個 Python 函數，接受一個數字列表，返回其中的最大數（不能使用內建函數如 `max()`）。  

**範例：**  
- 輸入 `[4, 9, 2, 7]`，輸出 `9`。  

**考察知識點：**  
- 列表和迭代  
- 比較運算符  
- 基礎函數定義
