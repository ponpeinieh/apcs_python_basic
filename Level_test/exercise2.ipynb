### **Exercise 3: Set Operations – Mutual Friends**  
**Task:**  
Given two sets `friends1` and `friends2` representing two people's friend lists, write a function `common_friends(friends1, friends2)` that returns their **mutual friends** and **unique friends**.  
- **Output format:** A dictionary with keys `'common'` and `'unique'`.  
- The value for `'unique'` should be a list of two sets (unique friends of person 1 and person 2).  

**Example:**  
```python
Input: 
friends1 = {'Alice', 'Bob', 'Charlie'}  
friends2 = {'Bob', 'David', 'Alice'}  
Output: 
{
    'common': {'Alice', 'Bob'},  
    'unique': [{'Charlie'}, {'David'}]
}
```

**Key Concepts Tested:**  
- Set operations (intersection `&`, difference `-`)  
- Structured dictionary output  
- Data formatting
