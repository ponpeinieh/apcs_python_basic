### `__init__()`

- `__init__` 方法就像一般的方法，可以被子類別繼承。
- 當你創建 `A` 類的實例時，它會自動使用 `B` 類的 `__init__` 方法，因為 `A` 沒有自己定義的 `__init__`。

### 代碼範例

```python
class B:
    def __init__(self):
        print('呼叫 B 的 __init__')

class A(B):
    pass

a = A()  # 這將輸出: 呼叫 B 的 __init__
```  

### `__init__()`
- 由於 `B` 的 `__init__` 方法需要一個 `value` 參數，所以在創建 `A` 的實例時必須傳入這個參數。

### 代碼範例

```python
class B:
    def __init__(self, value):
        print('呼叫 B 的 __init__')
        self.value = value

class A(B):
    pass

# a = A()  # 這會報錯: B.__init__() 缺少 1 個必需的位置參數: 'value'
a = A(100)  # 現在正確地傳入了參數
``` 

 ### `__init__()`

- 如果子類別的 `__init__()` 沒有特別呼叫其父類別的 `__init__()`，那麼直譯器**不會**自動呼叫父類別的 `__init__()` 方法。
- 這意味著**父類別的初始化邏輯不會執行**，可能會導致一些屬性未正確設置。

### 代碼範例

```python
class B:
    def __init__(self):
        print('呼叫 B 的 __init__')

class A(B):
    def __init__(self):
        print('呼叫 A 的 __init__')
        # 如果這裡沒有 super().__init__()，B 的 __init__() 不會被呼叫。

# 創建 A 的物件
a = A()  # 這將只輸出: 呼叫 A 的 __init__
``` 

 ### `__init__()`
  
- 在這個範例中，`A` 類的 `__init__()` 方法明確呼叫了 `B` 類的 `__init__()` 方法，這樣父類別的初始化邏輯會正常執行。

### 代碼範例 

```python
class B:
    def __init__(self):
        print('呼叫 B 的 __init__')

class A(B):
    def __init__(self):
        print('呼叫 A 的 __init__')
        super().__init__()  # 明確呼叫 B 的 __init__()
        print('A 的 __init__呼叫完畢')

# 創建 A 的物件
a = A()  # 這將輸出: 呼叫 A 的 __init__、呼叫 B 的 __init__、A 的 __init__呼叫完畢
```
