### Python 繼承

#### 介紹
- 繼承是一種物件導向程式設計的概念，允許一個類別（子類）從另一個類別（父類）獲取屬性和方法。
- 繼承使代碼重用成為可能，並促進類別之間的關係建模。

#### 為什麼需要繼承？
- **代碼重用**：通過繼承，可以重用父類的屬性和方法，減少代碼冗餘。
- **可擴展性**：可以在子類中擴展或修改父類的功能，以滿足特定需求。
- **組織性**：繼承幫助組織類別結構，使其更易於理解和維護。

#### 繼承的語法
- 使用小括號來指定父類：

```python
class ChildClass(ParentClass):
    pass
```

### 代碼示例

#### 父類：動物
- 首先定義父類 `Animal`，這個類別包含一個屬性和一個方法。

```python
# 父類：動物
class Animal:
    def __init__(self, name):
        self.name = name  # 物件的屬性
    
    def speak(self):
        return "動物發出聲音"
```

#### 子類：狗（無內容）
- 下面的 `Dog` 類別繼承自 `Animal` 類別，但沒有新增任何內容，這意味著它將完全繼承父類的屬性、方法以及 `__init__()` 方法。

```python
# 子類：狗（無內容）
class Dog(Animal):
    pass  # 直接繼承 Animal 的所有屬性和方法

# 使用示例
my_dog = Dog("小白")  # 仍然可以使用父類的 __init__ 方法
print(my_dog.name)  # 小白
print(my_dog.speak())  # 動物發出聲音
```

#### 改寫子類：狗（定義自己的 `__init__()` 方法）
- 下面改寫的 `Dog` 類別，定義了自己的 `__init__()` 方法，並新增了一個屬性 `breed`。

```python
# 子類：狗（定義自己的 __init__() 方法）
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # 呼叫父類的 __init__ 方法
        self.breed = breed  # 新增屬性

# 使用示例
my_dog = Dog("小白", "金毛")
print(my_dog.name)  # 小白
print(my_dog.breed)  # 金毛
print(my_dog.speak())  # 動物發出聲音
```

### 代碼說明
- **原始 `Dog` 類別**：繼承了 `Animal` 類別的所有屬性和方法。當呼叫 `my_dog.speak()` 時，將返回 "動物發出聲音"。
- **改寫的 `Dog` 類別**：繼承了 `Animal` 類別的屬性和方法，並定義了自己的 `__init__()` 方法以初始化 `breed` 屬性。當呼叫 `my_dog.speak()` 時，將返回 "動物發出聲音"。

#### `super().__init__()` 的使用
- `super()` 函數用於呼叫父類的方法，特別是在子類中需要初始化父類的屬性時。
- 這確保了父類的初始化邏輯被正確地執行，使得子類可以使用繼承自父類的屬性。
- 在子類的 `__init__()` 方法中，必須呼叫父類的 `__init__()` 方法，以便讓子類的實例擁有父類中定義的所有屬性。如果不呼叫父類的 `__init__()`，子類將不會擁有父類的屬性。
- 在 Python 3 中，建議使用 `super().__init__()` 來呼叫父類的 `__init__()` 方法，這是一種更現代和安全的寫法。也可以使用父類名稱進行呼叫，例如 `Animal.__init__(self, name)`，但這種用法在多重繼承的情況下不夠彈性，且在類別階層發生變化時容易出現問題。

#### 方法覆蓋
- 子類可以重新定義父類中的方法，這稱為方法覆蓋（Method Overriding）。
- 當子類覆蓋父類的方法時，子類的版本將取代父類的版本，這使得開發者可以根據需求自定義行為。

```python
# 子類：貓
class Cat(Animal):
    def speak(self):
        return "喵喵！"  # 覆蓋 speak() 方法

# 使用示例
my_cat = Cat("小黑")
print(my_cat.name)  # 小黑
print(my_cat.speak())  # 喵喵！
```
