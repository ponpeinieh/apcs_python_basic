### 練習題 1：商品價格記錄與排序系統

**描述**：編寫一個程式，使用字典來記錄一系列商品的價格歷史。每個商品可能有多個價格記錄（例如不同時間點的價格）。程式需要計算每個商品的平均價格，並根據平均價格對商品進行排序。

**商品價格記錄格式**：
```python
("商品名稱", 價格)
```

**要求**：
1. 使用字典來存儲每個商品的價格歷史，字典的鍵是商品名稱，值是一個列表，包含該商品的所有價格記錄。
2. 計算每個商品的平均價格。
3. 根據平均價格對商品進行排序，從低到高。
4. 輸出排序後的商品列表，顯示每個商品的名稱及其平均價格。

**示例商品價格記錄**：
```python
price_records = [
    ('milk', 86),
    ('butter', 120),
    ('bread', 50),
    ('milk', 90),
    ('butter', 110),
    ('bread', 55),
    ('milk', 88),
    ('butter', 115),
    ('bread', 60)
]
```

**範例輸出**：
```plaintext
按平均價格排序的商品：
bread: 55.00
milk: 88.00
butter: 115.00
```

### 範例程式

```python
def calculate_average_prices(price_records):
    # 步驟 1：使用字典存儲每個商品的價格歷史
    price_history = {}
    for item, price in price_records:
        if item not in price_history:
            price_history[item] = []
        price_history[item].append(price)
    
    # 步驟 2：計算每個商品的平均價格
    average_prices = {}
    for item, prices in price_history.items():
        average_prices[item] = sum(prices) / len(prices)
    
    # 步驟 3：根據平均價格對商品進行排序
    sorted_items = sorted(average_prices.items(), key=lambda x: x[1])
    
    return sorted_items

# 主程式
if __name__ == "__main__":
    # 示例商品價格記錄
    price_records = [
        ('milk', 86),
        ('butter', 120),
        ('bread', 50),
        ('milk', 90),
        ('butter', 110),
        ('bread', 55),
        ('milk', 88),
        ('butter', 115),
        ('bread', 60)
    ]
    
    # 計算並排序商品
    sorted_items = calculate_average_prices(price_records)
    
    # 輸出結果
    print("按平均價格排序的商品：")
    for item, avg_price in sorted_items:
        print(f"{item}: {avg_price:.2f}")
```


- 使用 `defaultdict` 改寫的程式碼解決方案。`defaultdict` 是 Python 中 `collections` 模組提供的一個非常有用的工具，它可以自動為字典中的新鍵初始化一個預設值，這樣我們就不需要手動檢查鍵是否存在並初始化列表了。

```python
from collections import defaultdict

def calculate_average_prices(price_records):
    # 步驟 1：使用 defaultdict 存儲每個商品的價格歷史
    price_history = defaultdict(list)  # 預設值為空列表
    for item, price in price_records:
        price_history[item].append(price)  # 直接追加價格，無需檢查鍵是否存在
    
    # 步驟 2：計算每個商品的平均價格
    average_prices = {}
    for item, prices in price_history.items():
        average_prices[item] = sum(prices) / len(prices)
    
    # 步驟 3：根據平均價格對商品進行排序
    sorted_items = sorted(average_prices.items(), key=lambda x: x[1])
    
    return sorted_items

# 主程式
if __name__ == "__main__":
    # 示例商品價格記錄
    price_records = [
        ('milk', 86),
        ('butter', 120),
        ('bread', 50),
        ('milk', 90),
        ('butter', 110),
        ('bread', 55),
        ('milk', 88),
        ('butter', 115),
        ('bread', 60)
    ]
    
    # 計算並排序商品
    sorted_items = calculate_average_prices(price_records)
    
    # 輸出結果
    print("按平均價格排序的商品：")
    for item, avg_price in sorted_items:
        print(f"{item}: {avg_price:.2f}")
```
