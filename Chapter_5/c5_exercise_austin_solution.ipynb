### 練習一 ：繪製六邊形

**描述**：  
編寫一個程式，使用 Python 的 `turtle` 模組繪製一個六邊形。

![](hex.png)

**要求**:

1. **創建烏龜**：畫筆顏色設為紅色。

2. **繪製六邊形**：邊長為 100 單位。

3. **完成繪製**：繪製完成後隱藏烏龜 (呼叫 `hideturtle()`)，並關閉畫面。

### Python 程式碼範例

```python
import turtle

# 創建烏龜
t = turtle.Turtle()
t.color("red")

# 繪製六邊形
for _ in range(6):
    t.forward(100)  # 移動 100 單位
    t.right(60)     # 轉 60 度

# 完成繪製
t.hideturtle()  # 隱藏烏龜

# 使用 try-except 來結束
try:
    turtle.done()  # 等待用戶關閉窗口
    turtle.bye()   # 關閉所有窗口
except:
    print('good bye')
``` 

### 練習二 ：繪製五角星

**描述**：  
編寫一個程式，使用 Python 的 `turtle` 模組繪製一個五角星。

![](star1.png)

**要求**:

1. **創建烏龜**：畫筆顏色設為藍色。

2. **繪製五角星**：邊長為 100 單位。

3. **完成繪製**：繪製完成後隱藏烏龜 (呼叫 `hideturtle()`)，並使用 `try-except` 來結束程式。

### Python 程式碼範例

```python
import turtle

# 創建烏龜
star_turtle = turtle.Turtle()
star_turtle.color("blue")

# 繪製五角星
for _ in range(5):
    star_turtle.forward(100)  # 移動 100 單位
    star_turtle.right(144)     # 轉 144 度

# 完成繪製
star_turtle.hideturtle()  # 隱藏烏龜

# 使用 try-except 來結束
try:
    turtle.done()  # 等待用戶關閉窗口
    turtle.bye()   # 關閉所有窗口
except:
    print('good bye')
```

### 練習三 ：繪製五角星(進階) - 僅繪製五角星的輪廓

**描述**：  
編寫一個程式，使用 Python 的 `turtle` 模組繪製一個五角星的輪廓，確保不會有交叉的線條。

![](star2.png)

**要求**:

1. **創建烏龜**：畫筆顏色設為紅色。

2. **繪製五角星輪廓**：每個直綫的綫段長為 40 單位。

3. **完成繪製**：繪製完成後隱藏烏龜 (呼叫 `hideturtle()`)，並使用 `try-except` 來結束程式。


### Python 程式碼範例

```python
import turtle

# 創建烏龜
star_turtle = turtle.Turtle()
star_turtle.color("red")

# 繪製五角星輪廓
for _ in range(5):
    star_turtle.forward(40)  # 移動 40 單位
    star_turtle.right(144)     # 轉 144 度
    star_turtle.forward(40)  # 再次移動 40 單位
    star_turtle.left(72)       # 轉 72 度

# 完成繪製
star_turtle.hideturtle()  # 隱藏烏龜

# 使用 try-except 來結束
try:
    turtle.done()  # 等待用戶關閉窗口
    turtle.bye()   # 關閉所有窗口
except:
    print('good bye')
```

### 練習四：電梯類別設計

**描述**：  
編寫一個 `Lift` 類別，該類別應包含電梯的基本屬性和操作方法。

**要求**:

1. **屬性**：
   - `total_floors`：總樓層數
   - `door_open`：門是否打開（布林值）
   - `current_floor`：當前樓層

2. **方法**：
   - `__init__(self, total_floors)`：初始化總樓層數，設置當前樓層為 1，門為關閉狀態。
   - `open_door(self)`：打開電梯門。
   - `close_door(self)`：關閉電梯門。
   - `goto_floor(self, floor)`：前往指定樓層，並在樓層之間移動時改變當前樓層。
   - `up_one_floor(self)`：向上移動一層樓。
   - `down_one_floor(self)`：向下移動一層樓。

3. **創建物件**：在主程式中，創建一個 `Lift` 物件並進行樓層移動和門的開關操作。

### Python 程式碼範例

```python
class Lift:
    def __init__(self, total_floors):
        self.total_floors = total_floors
        self.current_floor = 1
        self.door_open = False

    def open_door(self):
        if not self.door_open:
            self.door_open = True
            print("The door is now open.")
        else:
            print("The door is already open.")

    def close_door(self):
        if self.door_open:
            self.door_open = False
            print("The door is now closed.")
        else:
            print("The door is already closed.")

    def goto_floor(self, floor):
        if 1 <= floor <= self.total_floors:
            self.close_door()
            print(f"Going to floor {floor}...")
            self.current_floor = floor
            self.open_door()
        else:
            print("Invalid floor selection.")

    def up_one_floor(self):
        if self.current_floor < self.total_floors:
            self.goto_floor(self.current_floor + 1)
        else:
            print("You are already on the top floor.")

    def down_one_floor(self):
        if self.current_floor > 1:
            self.goto_floor(self.current_floor - 1)
        else:
            print("You are already on the ground floor.")

# 主程式
if __name__ == "__main__":
    lift = Lift(5)  # 創建一個有 5 層的電梯

    lift.open_door()  # 開門
    lift.goto_floor(3)  # 前往 3 樓
    lift.up_one_floor()  # 向上移動一層
    lift.down_one_floor()  # 向下移動一層
    lift.close_door()  # 關門
```
