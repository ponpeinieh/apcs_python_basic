### 練習一 ：繪製五角星

**描述**：  
編寫一個程式，使用 Python 的 `turtle` 模組繪製一個五角星。

![](star1.png)

**要求**:

1. **創建烏龜**：畫筆顏色設為藍色。

2. **繪製五角星**：邊長為 100 單位。

3. **完成繪製**：繪製完成後隱藏烏龜 (呼叫 `hideturtle()`)，並使用 `try-except` 來結束程式。

### 程式碼解決方案

```python
import turtle

# 創建烏龜
star_turtle = turtle.Turtle()
star_turtle.color("blue")

# 繪製五角星
for _ in range(5):
    star_turtle.forward(100)  # 移動 100 單位
    star_turtle.right(144)     # 轉 144 度

# 完成繪製
star_turtle.hideturtle()  # 隱藏烏龜

# 使用 try-except 來結束
try:
    turtle.done()  # 等待用戶關閉窗口
    turtle.bye()   # 關閉所有窗口
except:
    print('good bye')
```

### 練習二 ：繪製五角星(進階) - 僅繪製五角星的輪廓

**描述**：  
編寫一個程式，使用 Python 的 `turtle` 模組繪製一個五角星的輪廓，確保不會有交叉的線條。

![](star2.png)

**要求**:

1. **創建烏龜**：畫筆顏色設為紅色。

2. **繪製五角星輪廓**：每個直綫的綫段長為 40 單位。

3. **完成繪製**：繪製完成後隱藏烏龜 (呼叫 `hideturtle()`)，並使用 `try-except` 來結束程式。

### 程式碼解決方案

```python
import turtle

# 創建烏龜
star_turtle = turtle.Turtle()
star_turtle.color("red")

# 繪製五角星輪廓
for _ in range(5):
    star_turtle.forward(40)  # 移動 40 單位
    star_turtle.right(144)     # 轉 144 度
    star_turtle.forward(40)  # 再次移動 40 單位
    star_turtle.left(72)       # 轉 72 度

# 完成繪製
star_turtle.hideturtle()  # 隱藏烏龜

# 使用 try-except 來結束
try:
    turtle.done()  # 等待用戶關閉窗口
    turtle.bye()   # 關閉所有窗口
except:
    print('good bye')
```

### 練習三 ：繪製帶內切圓的正多邊形

**描述**：  
編寫一個程式，使用 Python 的 `turtle` 模組繪製一個正多邊形和一個內切圓，該圓與多邊形的所有邊相切。用戶將輸入多邊形的邊數（邊的數量）。


![](inscribed_circle.png)

**要求**:

1. **輸入邊數**：  
   - 用戶將被提示輸入多邊形的邊數（邊的數量）。例如，您可以輸入 8、12 等值。

2. **繪製多邊形**：  
   - 正多邊形是所有邊和角相等的多邊形。  
   - 外部角可以使用以下公式計算：  
     $$
     \text{外部角} = \frac{360}{n}
     $$  
     其中 $n$ 是邊的數量。

3. **繪製內切圓**：  
   - 內切圓的半徑可以使用以下公式計算：  
     $$
     r = \frac{s}{2 \cdot \tan\left(\frac{\pi}{n}\right)}
     $$  
     其中 $s$ 是多邊形的邊長，$n$ 是邊的數量。

4. **結合兩個形狀**：  
   - 首先，根據用戶輸入繪製多邊形。  
   - 然後，在多邊形內部繪製內切圓，確保該圓與多邊形的所有邊相切。


### 程式碼解決方案

```python
import turtle
import math

def draw_polygon(t, sides, side_length):
    t.color("black")  # 固定為黑色
    exterior_angle = 360 / sides  # 計算多邊形的外部角
    for _ in range(sides):
        t.forward(side_length)
        t.right(exterior_angle)

def draw_inscribed_circle(t, sides, side_length):
    # 計算內切圓的半徑
    radius = side_length / (2 * math.tan(math.pi / sides))
    
    # 移動烏龜到圓的起始位置
    t.penup()
    t.goto(side_length / 2, 0)  # 移動到圓起始位置(位於最上方正多邊形的邊的中央位置)
    t.pendown()
    t.color("black")  # 固定為黑色
    t.circle(-radius)  # 以逆時針方向繪製內切圓（半徑為負）

# 主執行
if __name__ == "__main__":
    # 建立一個 Turtle 物件
    my_turtle = turtle.Turtle()

    # 設置烏龜的屬性
    my_turtle.speed(1)  # 設置烏龜速度為慢
    my_turtle.pensize(2)  # 設置筆的粗細

    # 輸入邊數
    while True:
        try:
            sides = int(input("輸入邊數（例如：8、12）："))
            if sides > 2:
                break
            else:
                print("邊數必須大於 2，請重新輸入！")
        except ValueError:
            print("請輸入有效的數字！")

    # 動態調整邊長
    side_length = 400 / sides  # 根據邊數調整邊長

    # 繪製多邊形
    draw_polygon(my_turtle, sides, side_length)

    # 繪製內切圓
    draw_inscribed_circle(my_turtle, sides, side_length)

    # 完成繪圖
    my_turtle.hideturtle()  # 隱藏烏龜
    turtle.done()  # 完成繪圖

    try:
        turtle.bye()
    except:
        print('再見！')
```
